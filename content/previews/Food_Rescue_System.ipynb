### ភស្តុតាងរបស់ខ្ញុំ 3 · ត្រូវបំពេញ

ព្យាករស្លាកសម្រាប់ចំនួន 19, 20 និង 21 ដែលបានអនុម័ត និងចំនួន 20 ដែលមិនទាន់អនុម័ត។ ដំណើរការហើយប្រៀបធៀប។ ពន្យល់ថាហេតុអ្វីករណីស្មើ 20 សំខាន់ ហើយហេតុអ្វីការពិនិត្យត្រូវមកមុន។

```python
def pickup_label(total_portions, reviewed):
    if type(total_portions) is not int or total_portions < 0:
        raise ValueError("Use nonnegative whole portions / ប្រើចំណែកអាហារជាចំនួនគត់មិនអវិជ្ជមាន")
    if type(reviewed) is not bool:
        raise ValueError("Review must be True or False / ការពិនិត្យត្រូវជា True ឬ False")
    if not reviewed:
        return "hold"
    return "team" if total_portions >= 20 else "standard"

for total, reviewed in ((19, True), (20, True), (21, True), (20, False)):
    print(total, reviewed, "=>", pickup_label(total, reviewed))
```

```python
def new_db():
    db = sqlite3.connect(":memory:")
    db.execute("PRAGMA foreign_keys=ON")
    db.executescript("""
        CREATE TABLE batches(
            batch_id TEXT PRIMARY KEY, description TEXT NOT NULL,
            ppc INTEGER NOT NULL CHECK(ppc > 0),
            opening INTEGER NOT NULL CHECK(opening >= 0),
            available INTEGER NOT NULL CHECK(available >= 0));
        CREATE TABLE reservations(
            request_id TEXT PRIMARY KEY, label TEXT NOT NULL,
            planned_portions INTEGER NOT NULL CHECK(planned_portions > 0));
        CREATE TABLE lines(
            request_id TEXT REFERENCES reservations(request_id),
            line_no INTEGER, batch_id TEXT REFERENCES batches(batch_id),
            containers INTEGER NOT NULL CHECK(containers > 0),
            ppc_snapshot INTEGER NOT NULL CHECK(ppc_snapshot > 0),
            PRIMARY KEY(request_id, line_no));
        CREATE TABLE events(
            request_id TEXT REFERENCES reservations(request_id),
            event_type TEXT CHECK(event_type IN ('collected', 'delivered')),
            PRIMARY KEY(request_id, event_type));
    """)
    with db:
        db.executemany("INSERT INTO batches VALUES (?, ?, ?, ?, ?)",
                       [(b, name, ppc, opening, opening)
                        for b, name, ppc, opening in BATCHES])
    return db

def snapshot(db):
    return tuple(tuple(db.execute(query).fetchall()) for query in (
        "SELECT * FROM batches ORDER BY batch_id",
        "SELECT * FROM reservations ORDER BY request_id",
        "SELECT * FROM lines ORDER BY request_id, line_no",
        "SELECT * FROM events ORDER BY request_id, event_type",
    ))

print("Setup functions ready / អនុគមន៍រៀបចំរួចរាល់")
```

## 5 · What if only the first line succeeds?
**Question:** Can we leave half a request reserved?

**Sample decision:** No. A transaction must accept the whole request or restore the prior state. We first check approval and positive counts. Inside `with db`, we create the header, guard each stock update and store each line's unit snapshot. If any later line fails, the exception leaves the block and SQLite rolls back. Catching the exception outside the transaction is intentional.

Request IDs are unique. Retrying an already accepted ID is refused with unchanged state, not automatically counted as a new success. In a real uncertain retry, inspect existing records before doing anything else. This single-operator sample is not a production concurrency design.

### ភស្តុតាងរបស់ខ្ញុំ 4 · ត្រូវបំពេញ

ព្យាករស្ថានភាពមុន និងក្រោយ P04។ បន្ទាប់ពីដំណើរការ កត់ត្រាបរិមាណ B01/B03 ដែលអាចកក់បាន និងថាតើមានកំណត់ត្រាមេ ឬបន្ទាត់ P04 នៅសល់ឬទេ។ ពន្យល់ភាពខុសគ្នារវាងការចាប់ផ្តើមថ្មី និងការសាកល្បងប្រតិបត្តិការដដែលម្តងទៀត។

**ចម្លើយ / ការព្យាកររបស់ខ្ញុំ៖** [សរសេរនៅទីនេះ]

**លទ្ធផលដែលបានឃើញ និងការពន្យល់៖** [សរសេរនៅទីនេះ]

```python
def reserve(db, request_id, reviewed, lines, inject_failure=False):
    if type(reviewed) is not bool or not reviewed:
        raise ValueError("Coordinator approval required / ត្រូវការការអនុម័តពីអ្នកសម្របសម្រួល")
    if not isinstance(request_id, str) or not request_id.strip():
        raise ValueError("Request ID required / ត្រូវការលេខសម្គាល់សំណើ")
    if not isinstance(lines, (list, tuple)) or not lines:
        raise ValueError("At least one line required / ត្រូវមានយ៉ាងតិចមួយបន្ទាត់")
    prepared = []
    for line in lines:
        if not isinstance(line, (list, tuple)) or len(line) != 2:
            raise ValueError("Use (batch_id, containers) / ប្រើ (batch_id, containers)")
        batch_id, count = line
        if not isinstance(batch_id, str) or type(count) is not int or count <= 0:
            raise ValueError("Use a batch ID and positive whole count / ប្រើលេខសម្គាល់ និងចំនួនគត់វិជ្ជមាន")
        batch = db.execute("SELECT ppc FROM batches WHERE batch_id=?", (batch_id,)).fetchone()
        if batch is None:
            raise ValueError("Unknown batch / ក្រុមអាហារមិនស្គាល់: " + batch_id)
        prepared.append((batch_id, count, batch[0]))
    total = sum(count * ppc for batch_id, count, ppc in prepared)
    label = pickup_label(total, reviewed)
    with db:
        db.execute("INSERT INTO reservations VALUES (?, ?, ?)", (request_id, label, total))
        for number, (batch_id, count, ppc) in enumerate(prepared, 1):
            change = db.execute(
                "UPDATE batches SET available=available-? WHERE batch_id=? AND available>=?",
                (count, batch_id, count))
            if change.rowcount != 1:
                raise ValueError("Insufficient containers / ឯកតាផ្ទុកមិនគ្រប់គ្រាន់: " + batch_id)
            db.execute("INSERT INTO lines VALUES (?, ?, ?, ?, ?)",
                       (request_id, number, batch_id, count, ppc))
            if inject_failure and number == 1:
                raise RuntimeError("Simulated interruption / ការផ្អាកសន្មត")
    return label


def show_atomic_example():
    db = new_db()
    try:
        reserve(db, *REQUESTS[0])
        reserve(db, *REQUESTS[2])
        before = snapshot(db)
        print("Before / មុន:", [(row[0], row[4]) for row in before[0]])
        try:
            reserve(db, *REQUESTS[3])
        except ValueError as error:
            print("Refused / បដិសេធ:", error)
        after = snapshot(db)
        assert after == before
        print("After / ក្រោយ:", [(row[0], row[4]) for row in after[0]])
        print("P04 headers / កំណត់ត្រាមេ:", db.execute(
            "SELECT * FROM reservations WHERE request_id='P04'").fetchall())
        print("P04 lines / បន្ទាត់:", db.execute(
            "SELECT * FROM lines WHERE request_id='P04'").fetchall())
    finally:
        db.close()

show_atomic_example()
```

### ភស្តុតាងរបស់ខ្ញុំ 6 · ត្រូវបំពេញ

សរសេរសរុបដែលរំពឹងទុក និងសំណើរង់ចាំ រួចប្រៀបធៀបនឹងរបាយការណ៍។ បង្ហាញទាំង 94 = 34 + 60 និង 94 = 34 + 18 + 13 + 29 ដោយដាក់ស្លាកឱ្យមានន័យ។ ហេតុអ្វី P05 មិនត្រូវបង្ហាញក្នុងបញ្ជីរង់ចាំការប្រមូល?

**លទ្ធផលដែលបានឃើញ និងការពន្យល់៖** [សរសេរនៅទីនេះ]

```python
def request_rows(db):
    return db.execute("""
        SELECT r.request_id, r.label,
               SUM(l.containers * l.ppc_snapshot) AS planned_portions,
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='collected'),
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='delivered')
        FROM reservations r JOIN lines l ON l.request_id=r.request_id
        GROUP BY r.request_id, r.label ORDER BY r.request_id
    """).fetchall()


def report(db):
    rows = request_rows(db)
    offered, available = db.execute(
        "SELECT SUM(opening * ppc), SUM(available * ppc) FROM batches").fetchone()
    return {
        "offered": offered,
        "available": available,
        "reserved": sum(row[2] for row in rows),
        "collected": sum(row[2] for row in rows if row[3]),
        "delivered": sum(row[2] for row in rows if row[4]),
        "reserved_only": sum(row[2] for row in rows if not row[3]),
        "collected_not_delivered": sum(row[2] for row in rows if row[3] and not row[4]),
    }


def pending_collections(db):
    return [(request_id, portions, label)
            for request_id, label, portions, collected, delivered in request_rows(db)
            if not collected]


def show_report(db):
    labels = {
        "offered": "Opening offered / បានផ្តល់ដំបូង",
        "available": "Available / នៅអាចកក់បាន",
        "reserved": "Reserved / បានកក់",
        "collected": "Collection recorded / បានកត់ត្រាការប្រមូល",
        "delivered": "Delivery recorded / បានកត់ត្រាការប្រគល់",
        "reserved_only": "Reserved, no collection record / បានកក់ គ្មានកំណត់ត្រាប្រមូល",
        "collected_not_delivered": "Collected, no delivery record / បានប្រមូល គ្មានកំណត់ត្រាប្រគល់",
    }
    print("PLANNING PORTIONS / ចំណែកអាហារប៉ាន់ស្មាន")
    for key, value in report(db).items():
        print(labels[key] + ":", value)
    print("Available containers / ឯកតាផ្ទុកដែលអាចកក់បាន:", db.execute(
        "SELECT batch_id, available FROM batches ORDER BY batch_id").fetchall())
    print("Pending collection: ID, portions, label / រង់ចាំការប្រមូល៖ លេខសម្គាល់ ចំណែកអាហារ ស្លាក:",
          pending_collections(db))


def build_day(requests=REQUESTS, events=EVENTS):
    db = new_db()
    decisions = []
    try:
        for request_id, reviewed, lines in requests:
            before = snapshot(db)
            try:
                label = reserve(db, request_id, reviewed, lines)
            except (ValueError, sqlite3.IntegrityError) as error:
                assert snapshot(db) == before
                decisions.append((request_id, "held" if reviewed is False else "rejected", str(error)))
            else:
                decisions.append((request_id, "accepted", label))
        for request_id, event_type in events:
            record_event(db, request_id, event_type)
        return db, decisions
    except Exception:
        db.close()
        raise


def run_canonical_day():
    db, decisions = build_day()
    try:
        print("Request outcomes / លទ្ធផលសំណើ:")
        for row in decisions:
            print(row)
        show_report(db)
    finally:
        db.close()

run_canonical_day()
```
