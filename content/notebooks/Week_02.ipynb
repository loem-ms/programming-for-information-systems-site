### The idea: business rules

A **business rule** is a decision an organisation makes about how work is done: *"An exam mark of 50 or more is a pass."* *"Children under 6 travel free."* In our food donation system the coordinator says:

> *"If an offer is not approved yet, it waits (**hold**). If it is approved and has **20 portions or more**, we send a **team** to collect it. Otherwise one volunteer is enough (**standard**)."*

People choose the rule. The program applies it the same way to every offer, at any hour, so a wrong rule is wrong every time. Rules must be **written exactly** and **tested at the edges**.

**How a ✍️ Try it works.** Each Try it is a tiny **function**: a name, its inputs in brackets, and a `return` line with the answer. Change only the lines inside; `bridge.check` then tries your examples. Section 6 explains functions.

### 4 · `elif`: more than two answers, checked in order

Python checks the conditions **from top to bottom**, runs only the **first** branch that is `True`, and then stops. So the order of the branches is part of the rule.

```{iframe} https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/anim_if_elif.html

```

🤔 **Predict** the three words. Can the `"large"` line ever run?

````python
def size_word(portions):
    if portions >= 20:
        return "medium"
    elif portions >= 50:
        return "large"
    else:
        return "small"

print(size_word(10), size_word(30), size_word(80))
````

**Prediction:** …  **Result:** …

30 and 80 both pass the first test (`>= 20`), so `"large"` can never run. Put the **strictest** condition first: `>= 50` before `>= 20`. A wrong order never crashes; it quietly gives some values the wrong answer, so only a test finds it.
