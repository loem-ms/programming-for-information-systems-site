## 00 · Running code, comments, indentation

Python runs the lines of a cell from top to bottom when you press ▶ or **Shift+Enter**. Editing a cell changes nothing until you run it again. Text after `#` is a **comment**: Python ignores it, so it is a note for people. A line that ends with a colon `:` starts a **block**, and the lines that belong to it are indented by 4 spaces.

**Watch for:** indentation decides what belongs to what. It is not decoration.

````python
battery_percent = 15          # a comment: Python ignores this part
print("Battery:", battery_percent)
if battery_percent < 20:
    print("Low battery: please charge your phone")   # inside the if block
print("Check finished")       # not indented: always runs
````

## 02 · Arithmetic

`+ - * /` work as in maths, and `/` always gives a decimal (`float`). `//` divides and drops the remainder, `%` gives the remainder, and `**` is a power. Python does `*` and `/` before `+` and `-`, so use parentheses to make the order clear.

**Watch for:** `"3" * 4` is `"3333"`, because the quoted 3 is text.

````python
print(2 + 3 * 4, (2 + 3) * 4)     # multiplication first, unless you use parentheses
print(7 / 2, 7 // 2, 7 % 2)       # divide, whole groups, remainder
print(2 ** 3)                      # power: 2 x 2 x 2
print(round(10 / 3, 2))            # round to 2 decimal places
print(3 * 4, "3" * 4)              # a number versus text
````

## 04 · Comparisons and `if`

A comparison (`==`, `!=`, `<`, `<=`, `>`, `>=`) gives `True` or `False`, and `and`, `or`, `not` combine comparisons. `if` / `elif` / `else` checks the conditions from top to bottom and runs only the **first** branch that is `True`.

**Watch for:** `>=` includes the boundary but `>` does not. One `=` stores a value; `==` compares two values.

````python
battery_percent = 20
print(battery_percent > 20, battery_percent >= 20, battery_percent == 20)

if battery_percent <= 10:
    status = "critical"
elif battery_percent <= 20:
    status = "low"
else:
    status = "ok"
print(status)

is_charging = False
print(battery_percent < 50 and not is_charging)   # low and not charging
````

## 05 · Lists, dictionaries, tuples, sets

These four types hold many values in one name. A list of dictionaries, one dictionary per thing, is a common way to keep many records.

| Type | Looks like | Use it for |
|---|---|---|
| list | `["rice", "eggs"]` | items in order; get one by position, starting at 0 |
| dictionary | `{"name": "Dara", "mark": 78}` | values you find by a key (a name) |
| tuple | `("Dara", 78)` | a small fixed group; a function can return two values this way |
| set | `{"rice", "eggs"}` | each value only once |

**Watch for:** `student["email"]` crashes when the key is missing, but `student.get("email")` gives `None`. `{}` is an empty dictionary; an empty set is `set()`.

````python
shopping = ["rice", "eggs", "soap"]               # list
shopping.append("eggs")                           # add at the end
print(shopping[0], shopping[-1], len(shopping))   # first, last, how many
print("soap" in shopping)                         # is it in the list?

student = {"name": "Dara", "mark": 78}            # dictionary
student["mark"] = 80                              # change a value
print(student["name"], student["mark"], student.get("email"))

students = [student, {"name": "Sophea", "mark": 85}]   # list of dictionaries
print(students[1]["name"])

name, mark = ("Vanna", 64)                        # tuple, unpacked into two names
print(name, mark)
print(set(shopping))                              # set: "eggs" appears only once
````

## 06 · Loops and adding things up

A `for` loop runs its indented block once for each item in a list. To add things up, create the total **before** the loop and add to it **inside** the loop with `+=`. To add up by group, use a dictionary: `totals[key] = totals.get(key, 0) + amount`.

**Watch for:** `total = value` inside the loop replaces the total each time instead of adding to it.

````python
cart = [
    {"item": "pen", "price": 0.5, "quantity": 4},
    {"item": "notebook", "price": 1.5, "quantity": 2},
    {"item": "bag", "price": 12.0, "quantity": 1},
]
total_dollars = 0                          # start BEFORE the loop
for cart_line in cart:
    total_dollars += cart_line["price"] * cart_line["quantity"]
print("Cart total:", total_dollars)

sales = [("drinks", 3), ("snacks", 5), ("drinks", 2)]
items_by_category = {}                     # add up by group
for category, items_sold in sales:
    items_by_category[category] = items_by_category.get(category, 0) + items_sold
print(items_by_category)
````

## 09 · `try`/`except` (and `raise`)

Some errors are expected, for example when someone types `two` where a number should be. Put the risky line inside `try:`. If a `ValueError` happens, Python jumps to `except ValueError:` instead of stopping. Catch only the error you expect, so that real bugs still show.

**`raise`:** you can create an error yourself with `raise ValueError("message")` when a value has the right type but breaks a rule, such as a quantity below 1.

````python
for typed in ["3", "two", "0"]:
    try:
        quantity = int(typed)                 # may raise ValueError
        if quantity < 1:
            raise ValueError("quantity must be at least 1")
        print("OK:", quantity)
    except ValueError as error:
        print("Not accepted:", typed, "->", error)
print("The program is still running")
````

## 10 · Tests with `assert`

A test runs your code with one input and compares the result with the answer you decided **before** running it. `assert condition` does nothing when the condition is `True`, and stops with an `AssertionError` when it is `False`. Test three kinds of case: **normal** (typical values), **boundary** (exactly at a limit, and just on each side) and **odd** (zero, empty, negative or unusual).

**Watch for:** when a test fails, fix the code, not the expected answer, unless the expected answer was wrong.

````python
FREE_DELIVERY_FROM_DOLLARS = 25

def free_delivery(order_dollars):
    return order_dollars >= FREE_DELIVERY_FROM_DOLLARS

assert free_delivery(40) == True        # normal
assert free_delivery(10) == False       # normal
assert free_delivery(25) == True        # boundary: exactly at the limit
assert free_delivery(24.99) == False    # boundary: just below
assert free_delivery(0) == False        # odd: an empty order
print("All free_delivery tests passed")
````

## 11 · Files and CSV

A file keeps data after the notebook or program stops. `with open(name, mode, encoding="utf-8") as file:` opens a file and closes it again automatically: mode `"w"` writes a new file, `"a"` adds to the end, and `"r"` (the default) reads. A **CSV** file is a table saved as plain text, one row per line, with commas between the values; `csv.DictReader` gives you each row as a dictionary.

**Watch for:** every value read from a CSV file is text, so convert numbers with `int()` or `float()`.

````python
import csv

with open("todo.txt", "w", encoding="utf-8") as file:     # "w": write a new file
    file.write("Buy rice\nCharge phone\n")                  # \n starts a new line
with open("todo.txt", encoding="utf-8") as file:          # reading is the default
    print(file.read())

rows = [["name", "mark"], ["Dara", 78], ["Sophea", 85], ["Vanna", 64]]
with open("class_marks.csv", "w", newline="", encoding="utf-8") as file:
    csv.writer(file).writerows(rows)

total_marks = 0
with open("class_marks.csv", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        print(row)                         # the mark comes back as text
        total_marks += int(row["mark"])    # convert before adding
print("Total:", total_marks)
````

## 12 · JSON and JSON Lines

**JSON** is a text format for dictionaries and lists that almost every system can read. `json.dump(data, file)` saves Python data to a file and `json.load(file)` reads it back; `json.dumps` and `json.loads` do the same with a string. **JSON Lines** (`.jsonl`) stores one JSON record per line, so you can add a record at the end (mode `"a"`) without rewriting the file. This makes it useful for a history log.

**Watch for:** JSON writes `true`, `false` and `null` where Python writes `True`, `False` and `None`.

````python
import json

phone = {"owner": "Dara", "battery_percent": 80, "charging": False, "apps": ["maps", "camera"]}
print(json.dumps(phone))                  # JSON text: false instead of False

with open("phone.json", "w", encoding="utf-8") as file:
    json.dump(phone, file, indent=2)      # save the whole dictionary
with open("phone.json", encoding="utf-8") as file:
    loaded = json.load(file)              # read it back
print(loaded["apps"], loaded == phone)

with open("calls.jsonl", "w", encoding="utf-8") as file:    # JSON Lines: one record per line
    file.write(json.dumps({"to": "Sophea", "minutes": 3}) + "\n")
    file.write(json.dumps({"to": "Vanna", "minutes": 12}) + "\n")
with open("calls.jsonl", encoding="utf-8") as file:
    for line in file:
        print(json.loads(line))
````

## 13 · Classes, objects, methods

A **class** describes a kind of thing, such as a phone. An **object** is one real phone made from that class, and each object keeps its own data. `__init__` runs when you create an object and stores its data as **attributes** on `self`. A **method** is a function inside the class; `self` means "this object".

````python
class Phone:
    def __init__(self, owner, battery_percent):
        self.owner = owner
        self.battery_percent = battery_percent

    def use(self, minutes):
        """Each minute of use costs 1% battery, but never below 0."""
        self.battery_percent = max(0, self.battery_percent - minutes)

    def is_low(self):
        return self.battery_percent < 20

dara_phone = Phone("Dara", 50)
sophea_phone = Phone("Sophea", 90)
dara_phone.use(35)
print(dara_phone.owner, dara_phone.battery_percent, dara_phone.is_low())
print(sophea_phone.owner, sophea_phone.battery_percent, sophea_phone.is_low())
````

## 14 · Sorting with a key function

`sorted(items)` returns a new sorted list and leaves the original unchanged. For a list of dictionaries, give a **key function**: it receives one item and returns the value to sort by. Add `reverse=True` for the largest first.

**Watch for:** write `key=by_price` (the function's name), not `key=by_price()`.

````python
products = [
    {"name": "pen", "price": 0.5},
    {"name": "bag", "price": 12.0},
    {"name": "notebook", "price": 1.5},
]

def by_price(product):
    """Key function: take one product, return the value to sort by."""
    return product["price"]

print(sorted([3, 1, 2]))
for product in sorted(products, key=by_price):                    # cheapest first
    print(product["name"], product["price"])
print(sorted(products, key=by_price, reverse=True)[0]["name"])    # the most expensive
````

## 15 · Modules, `import`, `if __name__ == "__main__":`

A **module** is a `.py` file of functions that other code can reuse with `import`. Python's standard library has many ready-made modules, such as `math`, `json` and `csv`. When you start a file as a program, Python sets its `__name__` to `"__main__"`; when another file imports it, `__name__` is the module's name. So code under `if __name__ == "__main__":` runs only when the file is started as a program, and importing the file (for example, to test its functions) starts nothing.

````python
import importlib
import math                                   # a module from the standard library
import os
import sys

print(math.sqrt(16), math.ceil(4.2))

shop_code = '''TAX_RATE = 0.10

def price_with_tax(price_dollars):
    return round(price_dollars * (1 + TAX_RATE), 2)

def main():
    print("Demo: 10 dollars with tax is", price_with_tax(10))

if __name__ == "__main__":
    main()
'''
with open("shop_tools.py", "w", encoding="utf-8") as file:
    file.write(shop_code)

# Notebook housekeeping: look for modules in this folder and load the newest version.
# In a normal .py program you only need the import line.
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())
importlib.invalidate_caches()
sys.modules.pop("shop_tools", None)

import shop_tools                             # main() does not run: __name__ is "shop_tools"
print(shop_tools.price_with_tax(20))
print(shop_tools.__name__)
````

## 16 · Running a `.py` file and command-line arguments

A finished program is a `.py` file that you start from a terminal, for example `python3 greet.py Dara`. Each run is a new process that starts with empty memory. The words after the file name arrive in the list `sys.argv` as text (`sys.argv[0]` is the file name). In a notebook you can start a program the same way with `subprocess.run([sys.executable, "greet.py", "Dara"], ...)`: `sys.executable` is the Python that runs this notebook, and an exit code of 0 means success.

````python
import subprocess
import sys

program = '''import sys

def main():
    names = sys.argv[1:]          # the words after the file name, as text
    if len(names) == 0:
        print("Usage: python3 greet.py NAME ...")
        sys.exit(1)               # a non-zero exit code means "something went wrong"
    for name in names:
        print("Hello,", name)

if __name__ == "__main__":
    main()
'''
with open("greet.py", "w", encoding="utf-8") as file:
    file.write(program)

result = subprocess.run([sys.executable, "greet.py", "Dara", "Sophea"], capture_output=True, text=True)
print(result.stdout.strip())
print("Exit code:", result.returncode)

result = subprocess.run([sys.executable, "greet.py"], capture_output=True, text=True)
print(result.stdout.strip())
print("Exit code:", result.returncode)
````

## 17 · Running tests in a fresh process

A notebook remembers everything you ran, even a function you have since deleted or renamed. So tests can pass here and still fail on another computer. A **fresh process** starts with nothing remembered, so it is the honest check: in a notebook use **Runtime → Restart session and run all**; for a program, run the test file from a terminal. A test file imports the code it tests, has `test_` functions that use `assert`, and ends with a non-zero exit code when a test fails.

````python
import subprocess
import sys

with open("shop_rules.py", "w", encoding="utf-8") as file:
    file.write('''FREE_DELIVERY_FROM_DOLLARS = 25

def free_delivery(order_dollars):
    return order_dollars >= FREE_DELIVERY_FROM_DOLLARS
''')

with open("test_shop_rules.py", "w", encoding="utf-8") as file:
    file.write('''from shop_rules import free_delivery

def test_normal():
    assert free_delivery(40) == True
    assert free_delivery(10) == False

def test_boundary():
    assert free_delivery(25) == True
    assert free_delivery(24.99) == False

if __name__ == "__main__":
    test_normal()
    test_boundary()
    print("All tests passed")
''')

result = subprocess.run([sys.executable, "test_shop_rules.py"], capture_output=True, text=True)
print((result.stdout + result.stderr).strip())
print("Exit code:", result.returncode)
````

## 18 · Terminal cheat-sheet

A **terminal** is a window where you type commands instead of clicking. Type these commands in a terminal, not in a code cell: they are not Python. On your own computer, use **Terminal** (macOS or Linux) or **PowerShell** (Windows). In Colab, use the **Terminal** button in the left sidebar.

| I want to… | macOS / Linux | Windows (PowerShell) |
|---|---|---|
| see which folder I am in | `pwd` | `pwd` |
| list the files here | `ls` | `ls` |
| go into a folder | `cd food_rescue_kit` | `cd food_rescue_kit` |
| go up one folder | `cd ..` | `cd ..` |
| check which Python I have | `python3 --version` | `py --version` |
| run a program | `python3 greet.py` | `py greet.py` |
| run a program with arguments | `python3 greet.py Dara Sophea` | `py greet.py Dara Sophea` |
| run one test file | `python3 test_shop_rules.py` | `py test_shop_rules.py` |

| stop a running program | **Ctrl+C** | **Ctrl+C** |
| repeat an earlier command | **↑** (up arrow) | **↑** (up arrow) |
| complete a long file name | **Tab** | **Tab** |

- If `python3` is not found, try `python` or `py -3`.
- Run a program from the folder that contains it, or give its path: `python3 programs/greet.py`.
