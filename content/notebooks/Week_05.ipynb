### 2 · Methods make new text: `strip()` and `lower()`

📖 **What it is.** A **method** is a function that belongs to a value. You write it after a dot: `typed.strip()`. Text has many methods. `strip()` removes spaces at both ends. `lower()` makes every letter lowercase.

**Why it exists.** People add spaces and mix capitals: `"Bread"`, `" bread"`, `"BREAD"`. A computer compares text letter by letter, so `"Bread" == "bread"` is `False`. Tidy text first, then compare.

**How Python runs it.** Text in Python **never changes in place**. A method reads the text and makes **new** text; the original stays exactly as it was. To keep the new text, give it a name: `donor = donor.strip()`. You can chain methods: in `" Bread ".strip().lower()`, each dot works on the result of the step before.

**Where you meet it.** A login form strips the spaces around your email address; a search box ignores capitals.

| Code | Result (new text) | The original |
|---|---|---|
| `"  Dara ".strip()` | `"Dara"` | still `"  Dara "` |
| `"Rice Meals".lower()` | `"rice meals"` | still `"Rice Meals"` |
| `" Bread ".strip().lower()` | `"bread"` | still `" Bread "` |

### 3 · Exceptions: when Python cannot go on

📖 **What it is.** `int()` turns text into a whole number: `int("30")` gives `30`, and `int(" 12 ")` gives `12` (it ignores spaces at both ends). When the text is not a whole number, `int()` has no correct answer to give. So it **raises an exception**: a signal that says what went wrong, with a type such as `ValueError` and a message. `int("-3")` works too, because negative numbers are whole numbers. Whether a number fits your rule (at least 1000 riel, 1 to 200 portions) is a separate step.

**Why it exists.** A quiet wrong answer, such as 0, would slip into your totals. A loud stop is safer: everyone sees it.

**How Python runs it.** The exception starts at the line that raised it. Python stops that function at once and the exception **travels up**: back to the line that called the function, then to the line that called that one, and so on. If nothing on the way **catches** it, the program stops and Python prints a **traceback**: the path the exception travelled.

![An exception from int travels up through the calls: caught by except it becomes a message, not caught it stops the program with a traceback](../assets/diagrams/exception_travel.svg)

**Where you meet it.** Every program has exceptions. The "Something went wrong, please try again" screen in an app is usually an exception that the programmer caught and turned into a message.

| Exception | When Python raises it | Example |
|---|---|---|
| `ValueError` | the type is right, but the value makes no sense | `int("many")` |
| `KeyError` | a dictionary has no such key | `row["phone"]` |
| `NameError` | a name does not exist (often a typo) | `print(totl)` |
| `TypeError` | the wrong type for an operation | `"30" + 5` |

Beginners read from the top, see three blocks of code and give up. **Program thinking: read the error message, from the bottom up.** Every programmer does this first.

| Read | It says | It means |
|---|---|---|
| 1 · the **last line** | `ValueError: invalid literal for int() with base 10: 'many'` | the type, then the message: `int()` was given `'many'`, which is not a whole number ("base 10" means our normal digits 0–9) |
| 2 · the **arrow just above** it | `----> 2     return int(text)` in `read_portions` | the line that raised the exception |
| 3 · the **arrows higher up** | line 7 in `total_portions`, line 10 in your cell | the path back: who called whom. "Most recent call last" means the oldest call is at the top |

**How to fix it:** here the code is right and the **data** is bad. You cannot stop people from typing `many`, so the program must decide what to do with it. That is the next block.

### 4 · `try` / `except ValueError`: catch it and decide

📖 **What it is.** `try` / `except` catches an exception you expect, on its way up, so your program can decide what to do instead of stopping.

**How Python runs it.** Python runs the lines under `try` one by one. If they all work, it skips `except`. If one of them raises a `ValueError`, Python leaves the `try` block at once: the lines after it do not run. Python jumps to `except ValueError:`, runs those lines, and then goes on after the whole statement. An exception of another type is not caught: it keeps travelling.

**Why it exists.** In an information system, bad input is normal, not rare. The person who typed it needs a clear message, and everyone else needs the program to keep running.

**Where you meet it.** A phone top-up app that says *please type a number*; a web server that turns one bad request into an error page while the site stays up for everyone else.

```text
try:
    quantity = int(text)     ── works?      ──►  the rest of try runs, except is skipped
                             └─ ValueError  ──►  jump to except ValueError:
except ValueError:
    a message for the person
```

Two good habits: keep `try` **small** (only the lines that can fail), and catch **only** the exception you expect.

#### 🐞 Common mistake · `except:` that hides every error

```python
def read_top_up(text):
    try:
        return int(txt)        # a typo: txt instead of text
    except:
        return None

print(read_top_up("5000"))     # None
```

**What you see:** `None`, even for `"5000"`, and no error message.

**What it means:** `txt` does not exist, so Python raised a `NameError`. A bare `except:` catches **every** exception, so it caught your typo too and hid it. The app now refuses every top-up, and nobody knows why.

**How to fix it:** catch only what you expect: `except ValueError:`. Now the typo shows at once as `NameError: name 'txt' is not defined`, and you fix it in a minute.

### 5 · Collect every problem, then send back two values

📖 **What it is.** A good check does not stop at the first problem. It starts with an empty list, `errors = []`, checks every field, and `append`s one message for each problem. An empty list at the end means "no problems".

**Why it exists.** A form that shows only the first problem makes people send it again and again. A check that collects every problem lets the person fix everything at once.

**Where you meet it.** A registration website marks every wrong field at once, in red, next to each field.

How `errors` grows in a sign-up check:

| After this step | `errors` |
|---|---|
| start | `[]` |

#### Two values come back as a tuple

📖 A check often needs to send back two answers: the clean value, and the list of problems. `return price, errors` packs the two values into one **tuple**: a fixed group of values in round brackets, such as `(2500, [])`. The caller **unpacks** the tuple into two names, in the same order: `price, errors = read_price("2500")`. A bank transfer works the same way: it comes back as "done" or "refused: not enough money".

![return offer, errors sends one tuple back; the caller unpacks it into two names](../assets/diagrams/tuple_return.svg)

The shape of this week's door: a good value with an empty list, **or** `None` with the problems. Never both. `None` means the caller keeps nothing: **change nothing when the input is bad**.

| The input is … | The function returns |
|---|---|
| good | `(value, [])` |
| bad | `(None, ["a plain message", …])` |

### Part B · Under the hood (about 45 min)

#### 🧑‍🏫 How `clean_offer` works: a five-step pipeline

`clean_offer` is too big to write in one go. **Program thinking: decompose.** Split one big job into small steps, so that each step does one job. This is how programmers think about every large function. Each raw row flows through the same five steps, like goods on a conveyor belt. Every step may add a message to `errors`. Only at the end does the function decide what to give back.

```text
 raw row: a dictionary of text, as it came from the file or the form
   {"donor": " Sunrise Bakery ", "category": " Bread ", "portions": "30", "hours_left": "5", …}
        │
        ▼
 1 · required fields   each blank one        → "donor is required."               ─┐
        │                                                                         │
        ▼                                                                         │
 2 · tidy text         strip() every text, lower() the category → offer           │
        │                                                                         │  errors
        ▼                                                                         │  (one list,
 3 · numbers           int() inside try; ValueError → "… must be a whole number." ─┤   growing)
        │                                                                         │
        ▼                                                                         │
 4 · limits            in CATEGORIES? 1–200 portions? at least 1 hour left?       ─┘
        │
        ▼
 5 · return            errors empty → (offer, [])         errors → (None, errors)
```

You know every tool already: steps 1 and 2 are Try its 1, 3 and 4; steps 3 and 4 are the ladder; step 5 is a tuple.

**Read the starter.** Scroll to *Write your file* in Session 3 and read the starter next to this table.

| In the starter | What it takes or holds | What it gives back or does | Why it is so |
|---|---|---|---|
| the five settings at the top | your rules: required fields, categories, limits | the values every step compares with | a coordinator can change a limit without touching the function |
| `clean_offer(raw)` | one raw row: a dictionary of text | `(offer, [])` or `(None, errors)` | the caller needs both: the clean offer to save, or the reasons to show |
| `errors` | starts as `[]`; any step may `append` | the list in the answer | the donor sees every problem at once |
| `offer` | the tidy values, built in steps 2 and 3 | given back only when `errors` is empty | change nothing when input is bad |

In the code, the category check runs right after step 2, and steps 3 and 4 sit together inside each `try`.

Why does `clean_offer` **return** its problems, instead of printing them or raising an exception? Because three different callers use the same door. `bridge.show("sheet")` colours the row red and shows your messages. The web page shows them under the form. `bridge.check` tests them. A `print` would only reach one screen. An exception would stop at the first problem and leave every row after it unread, which is exactly yesterday's lost import.

### 🧭 Your decision · what does the door accept?

Every limit protects someone and refuses someone. A limit is a guess until you test it on data. `semester.csv` holds last semester's 404 offers. The prepared cell below counts what each limit **would** have refused, and draws when the offers arrived. Run it and read it; there is nothing to write.

Each bar of the chart counts the offers that arrived with that many hours left. Everything to the **left** of a limit would be refused.

````python
# @title Last semester's offers: what would each limit refuse? { display-mode: "form" }
# A prepared cell: it reads semester.csv, prints two small tables and draws one chart. Just run it.
import csv
import matplotlib.pyplot as plt

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    semester = list(csv.DictReader(file))
hours = []
portions = []
for row in semester:
    hours.append(int(row["hours_left"]))
    portions.append(int(row["portions"]))

print("REJECT_IF_HOURS_LEFT_BELOW → offers the door would refuse (of " + str(len(semester)) + ")")
for limit in [1, 2, 3, 4, 5]:
    refused = 0
    for value in hours:
        if value < limit:
            refused = refused + 1
    print("   " + str(limit) + " → " + str(refused) + " (" + str(round(100 * refused / len(semester))) + "%)")
print()
print("MAX_PORTIONS → offers the door would refuse (of " + str(len(semester)) + ")")
for limit in [35, 50, 100, 200]:
    refused = 0
    for value in portions:
        if value > limit:
            refused = refused + 1
    print("   " + str(limit) + " → " + str(refused))
print("   (the largest offer last semester: " + str(max(portions)) + " portions)")

plt.figure(figsize=(8, 3))
plt.hist(hours, bins=range(0, 68))
plt.title("Hours left when each offer arrived (last semester, 404 offers)")
plt.xlabel("Hours left")
plt.ylabel("Number of offers")
plt.show()
````
