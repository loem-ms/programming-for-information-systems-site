### The idea: parts, and one way in

At a noodle stall in the market, you talk to one person: the one who takes your order. You never walk into the kitchen. Behind that person, each cook has one job: one makes the broth, one cooks the noodles, one cuts the herbs. The cooks never talk to you; they pass their work on. When the owner changes the recipe (less salt in the broth), the person who takes your order does not change at all.

A program is built the same way. Its **parts** (modules) each do one job: they take values and **return** answers, and they never talk to a person. **One file** is the way in: a person starts it, it asks what they want, calls the parts and shows the answers. This file is the program's **entry point**. Your six files from Weeks 4–8 are the parts. This week you write the entry point, `app.py`.

```text
a person ── types "1" ──▶  app.py     the entry point: asks, calls the parts, shows the answer
                             │ calls
                             ▼
         rules.py · records.py · report.py · storage.py     the parts: return answers
```

This is **Principle 9 · Separate what changes from what stays.** The recipe (your settings and rules) changes often; the way in and the machinery (storage, screens) stay the same. This week's habit of program thinking is to **decompose**: split a big job into small functions that call each other, each with one job, one name and its own test.

### 1 · Functions that call functions

📖 **What it is.** Inside a function you can call another function, just as you call one in a cell. In `riel_text(with_bonus(20000))`, Python calls `with_bonus` first, then gives its answer to `riel_text`.

**Why it exists.** A big job is hard to write and harder to test. So you **decompose** it into small functions, each with one job and a clear name. The big function then reads like a list of steps, and each small function can be tested and changed on its own. Your `app.py` works like this: its menu calls `urgency`, `pickup_label`, `most_urgent_first` and others.

**How Python runs it.** When Python meets a call, it pauses what it was doing, runs the called function, and puts the returned value in the place of the call. Then it goes on. The innermost call runs first, like the brackets in maths. **Trace by hand** with your finger on the code:

| Step | Python runs … | The line now reads |
|---|---|---|
| 1 | `with_bonus(20000)`, which returns 22000 | `print(riel_text(22000))` |
| 2 | `riel_text(22000)`, which returns `"22000 riel"` | `print("22000 riel")` |
| 3 | `print` | the screen shows `22000 riel` |

**Where you meet it.** A banking app's "send money" button calls small functions: check the balance, take the fee, write the history, send the SMS. Each one is tested alone, and a new fee changes one function only.

### 2 · `import`: use a function from another file

📖 **What it is.** `from prices import with_bonus` loads the file `prices.py` and gives you its function `with_bonus`. You have imported Python's own files since Week 5 (`import csv`, `import json`). Your own `.py` files work the same way. A file you import is called a **module**; its name is the file name without `.py`.

**Why it exists.** A rule should be written **once**, in one place, and every file that needs it imports it. When the rule changes, you change one file, and every file that imports it follows. A copied rule gives you two copies that disagree after the first change.

**How Python runs it.** The first time a program meets `from prices import with_bonus`, Python:

1. looks for `prices.py` in the folder of the file you started (then in its own library folders);
2. **runs `prices.py` once**, from top to bottom: the constants get their values, each `def` makes a function, and a `print` at the top level prints;
3. keeps the result as a module (a box of names) and gives you the names you asked for.

When another file in the same program imports `prices` again, Python does not run it again: it reuses the box.

![app.py imports four parts: Python finds each file, runs it once and hands over the names app.py asked for](../assets/diagrams/import_graph.svg)

| You write | You get | You call |
|---|---|---|
| `from prices import with_bonus` | the name `with_bonus` | `with_bonus(5000)` |
| `import prices` | the module `prices` | `prices.with_bonus(5000)` |

**Where you meet it.** Every real program is many files. A shop's app has one file for prices, one for the cart and one for payments, and each imports what it needs. Your `app.py` starts with four `from … import …` lines, one for each part it uses.

### 3 · `if __name__ == "__main__":` · a part and a program in one file

Look back at your run of `till.py`: when it imported `prices.py`, `__name__` was `prices`, and the `Demo:` line did not print. Here is why.

📖 **What it is.** While a file runs, Python keeps its name in the variable `__name__` (two underscores on each side). When a person **starts** the file, `__name__` is `"__main__"`. When another file **imports** it, `__name__` is the module's name, such as `"prices"`. So `if __name__ == "__main__":` means: *run these lines only when this file is the one a person started.*

**Why it exists.** One file can be used in two ways: as a part that other files import, and as a program you start to try it. The lines under the guard (a demo, a question to the user, a menu) must not run when another file only wants the functions. The kit imports your files to check them: a part that waits for typing when it is imported would stop the check.

**How Python runs it.**

| How the file is used | `__name__` | the lines under the guard |
|---|---|---|
| started: `python prices.py` | `"__main__"` | run |
| imported: `from prices import with_bonus` | `"prices"` | skipped |

**Where you meet it.** Almost every Python program you download ends with the same two lines: `if __name__ == "__main__":` and a call such as `main()`. Python's own library files use the guard for small demos.

A `print` at the top level of `report.py` would print every time `app.py` starts. Fix: a part holds only imports, settings, functions and the guard. Move demos and `input()` under `if __name__ == "__main__":`, or into the entry point.

### 4 · The entry point: the file a person starts

📖 **What it is.** The **entry point** is the one file a person starts to use a program. For your system it is `app.py`. In many other projects it is called `main.py`. It ends with the guard and one call: `main()`.

**Why it exists.** A person, or a new teammate, must know where to begin: there is one clear way in. And only the entry point talks to people. The parts never ask (`input()`) and never print their answers; they **return** them. So every part can be checked without a person typing, and the same parts can serve a menu today and a web page tomorrow.

**How Python runs it.** When a person types `python app.py`:

1. Python runs `app.py` with `__name__` set to `"__main__"`.
2. Its `import` lines run each part once and bring in their functions.
3. The `def` lines create `handle_choice` and `main`; nothing inside them has run yet.
4. At the bottom the guard is true, so `main()` is called: it loads the records, shows the menu and waits for a choice.

![__name__ is &quot;__main__&quot; in the file a person starts and the module&#x27;s name when another file imports it, so the guarded main() runs only in the entry point](../assets/diagrams/entry_point.svg)

| | a part (`rules.py`, `report.py`, …) | the entry point (`app.py`) |
|---|---|---|
| who starts it | another file imports it | a person: `python app.py` |
| talks to the person? | never: it returns answers | yes: `input()` and `print` |
| what the kit checks | every function | `handle_choice` (it returns text) |

Inside `app.py` the work is split once more. `handle_choice(choice, records)` **returns** the text for one choice; `main()` does the asking and the printing. So `handle_choice` can be checked like any part.

**Where you meet it.** A mobile banking app opens on one start screen, with hundreds of parts behind it. Web projects start from a file such as `main.py` or `app.py`, and many tutorials begin with `python main.py`.

👀 **Example.** Here is `till.py` rewritten as an entry point. Read it; you do not need to run it. When a person starts `python till.py`, the guard is true and `main()` asks and prints. When another file imports `till.py`, nothing runs.

```python
from prices import with_bonus

def main():
    amount = int(input("Top-up amount in riel: "))
    print("Credit:", with_bonus(amount))

    main()
```

### 5 · A menu loop: `while True` and `break`

📖 **What it is.** `while True:` repeats its block for ever. `break` leaves the loop at once; Python goes on with the first line after the loop.

**Why it exists.** A menu does not know how many choices a person will make. It must ask, answer and ask again, until the person says stop.

**How Python runs it.** In Week 6 you saw that `while` checks its condition before every round. Here the condition is `True`, which is always true, so only a `break` can end the loop. One round of your menu: ask → if the choice is `"q"`, say goodbye and `break` → otherwise show `handle_choice(choice, records)` → back to the top.

```{iframe} https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/anim_menu_loop.html
:width: 100%

[The menu loop: `while True`, `input`, `handle_choice`, `break`](https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/anim_menu_loop.html)
```

**Where you meet it.** An ATM, the `*123#` menu on your phone and a chat bot all run a loop like this: wait for input, answer, wait again.

👀 **Example.** Nobody can type into a cell that runs on its own, so here a list stands in for the keys a person types. 🤔 **Predict:** how many lines are printed? Is `"3"` ever used?

````python
typed = ["1", "x", "q", "3"]          # pretend someone typed these, one by one
position = 0
while True:
    choice = typed[position]
    position = position + 1
    if choice == "q":
        print("Goodbye!")
        break
    print("You chose", choice)
````

````python
# @title One semester: offers per week (line chart) { display-mode: "form" }
# Reads semester.csv (invented data), counts the offers in each week and draws them as a line.
import csv
import matplotlib.pyplot as plt

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    semester = list(csv.DictReader(file))


def count_by(rows, field):
    """Count the rows for each value of the field (Week 6)."""
    counts = {}
    for row in rows:
        counts[row[field]] = counts.get(row[field], 0) + 1
    return counts


def rate(part, whole):
    """The part as a percentage of the whole, rounded to 1 decimal place. 0.0 when the whole is 0."""
    if whole == 0:
        return 0.0
    return round(100 * part / whole, 1)


per_week = count_by(semester, "week")
weeks = list(range(1, 17))
offers_per_week = []
for week in weeks:
    offers_per_week.append(per_week.get(str(week), 0))      # the keys are text: "1", "2", …

print(len(semester), "offers in", len(weeks), "weeks")
print("Offers per week:", offers_per_week)
plt.figure(figsize=(8, 3))
plt.plot(weeks, offers_per_week, marker="o")
plt.title("Offers per week, one semester")
plt.xlabel("Week")
plt.ylabel("Number of offers")
plt.ylim(0, 40)
plt.xticks(weeks)
plt.show()
````

#### Counts and rates

A **count** says how many. A **rate** says how many **out of how many**. In class A, 3 of 20 students were late; in class B, 4 of 50. Class B has more late students (a bigger count), but class A has the lateness problem: 15% against 8% (a bigger rate). Always ask: *out of how many?*

The prepared cells use this function. Read it: its docstring is the promise, and the empty case is handled before the division.

```python
def rate(part, whole):
    """The part as a percentage of the whole, rounded to 1 decimal place. 0.0 when the whole is 0."""
    if whole == 0:
        return 0.0                       # nothing to count: no division by 0
    return round(100 * part / whole, 1)
```

The next prepared cell counts the rejected offers in each category, divides by all the offers in that category, and draws both measures side by side.

````python
# @title Rejections by category: the count and the rate { display-mode: "form" }
# Counts the rejected offers per category, divides by all offers per category, and draws both.
rejected = []
for row in semester:
    if row["status"] == "rejected":
        rejected.append(row)
rejections = count_by(rejected, "category")
offers_by_category = count_by(semester, "category")

categories = list(offers_by_category)
rejection_counts = []
rejection_rates = []
print("category".ljust(12), "rejected".rjust(9), "offers".rjust(7), "rate".rjust(7))
for category in categories:
    count = rejections.get(category, 0)
    share = rate(count, offers_by_category[category])
    rejection_counts.append(count)
    rejection_rates.append(share)
    print(category.ljust(12), str(count).rjust(9), str(offers_by_category[category]).rjust(7),
          (str(share) + "%").rjust(7))

figure, (left, right) = plt.subplots(1, 2, figsize=(10, 3))
left.bar(categories, rejection_counts)
left.set_title("Rejections (count)")
left.set_ylabel("rejected offers")
right.bar(categories, rejection_rates)
right.set_title("Rejection rate")
right.set_ylabel("% of offers rejected")
plt.tight_layout()
plt.show()
````

🤔 Read the two charts side by side. Rice meals have the **most** rejections, but how many rice offers came in? Which category has the highest **rate**? If you only count, which food do you blame? If you divide?

The last prepared cell answers Week 5's question. It splits the semester at 3 hours left (the default `URGENT_HOURS`) and compares how often each group was cancelled.

````python
# @title Short notice: how often offers were cancelled { display-mode: "form" }
# Splits the semester at 3 hours left and compares the cancel rates of the two groups.
short_notice = []
longer_notice = []
for row in semester:
    if int(row["hours_left"]) <= 3:
        short_notice.append(row)
    else:
        longer_notice.append(row)

for label, rows in [("3 hours or less", short_notice), ("more than 3 hours", longer_notice)]:
    cancelled = count_by(rows, "status").get("cancelled", 0)
    print(label.ljust(18), len(rows), "offers ·", cancelled, "cancelled ·", rate(cancelled, len(rows)), "%")
````

#### What the data can and cannot show

- **The data is about offers, not about hunger.** It cannot tell you how many people needed food, or whether a partner got enough.

The data alone cannot tell. To find out, change one thing (for example, ask donors to call earlier for a month) and measure again.
- **Who got food is not who needed it.** The data shows which partner received food, not which partner needed it most. To judge fairness, ask the coordinator and the partners.
