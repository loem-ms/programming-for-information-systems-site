### 1 · `import` between your own files

A `.py` file can use the functions in another `.py` file in the same folder: `from prices import with_bonus`. The function is written **once**, in its own file, and every file that needs it imports it. The two cells below write two tiny files for a phone shop. The third defines `run_file`, which runs a `.py` file in a new Python, as if you typed `python till.py` in a terminal (you do not need to remember how).

````python
%%writefile prices.py
# prices.py · the shop's decisions live here
BONUS_PERCENT = 10


def with_bonus(amount):
    """Return the top-up amount plus the bonus."""
    return amount + amount * BONUS_PERCENT // 100


# __name__ is explained in section 2 below
print("prices.py is running. Its __name__ is", __name__)

if __name__ == "__main__":
    print("Demo:", with_bonus(5000))
````

````python
import subprocess
import sys


def run_file(name):
    """Run a .py file in a new Python and show what it printed."""
    result = subprocess.run([sys.executable, name], capture_output=True, text=True)
    print(result.stdout + result.stderr)


run_file("prices.py")
````

### 2 · `if __name__ == "__main__":`

Python gives every running file a name, `__name__`. When you **run** the file itself, its `__name__` is `"__main__"`. When another file **imports** it, its `__name__` is the file's own name (`"prices"`). So `if __name__ == "__main__":` means: *only when this file is run, not when it is imported.*
