### Values, conditions and complete decisions

A program stores values, evaluates expressions and chooses which statements to run. Start with a small score example that needs no application knowledge. `score = 20` stores a number. `enabled = False` stores a Boolean. `"False"` is text and is not the Boolean `False`; do not put quotes around Boolean values. Numbers and Boolean flags represent different kinds of information.

**Assignment and comparison.** One `=` assigns a value to a name. `==` asks whether two values are equal. `>=` includes equality, so `20 >= 20` is true and `20 > 20` is false. Evaluate the expression on the right before storing its result. A name such as `at_least_20` can store the Boolean result of a comparison.

**A branch is a block of statements.** Put a colon after the condition and indent every statement in that branch by four spaces. `if` checks the first condition. `elif` checks another condition only when all earlier conditions were false. `else` covers the remaining cases. A statement aligned with `if` runs after the selected block. Assign the output in every branch so a previous notebook run cannot supply an old result.

**Order expresses a rule.** Suppose a small practice program must report "disabled" whenever enabled is false. Otherwise it reports "high" for a score of at least 20 and "low" below 20. Check the enabling condition first, then the score. An `if / elif / else` chain chooses one result. Several separate `if` statements can assign the result more than once. Read `not enabled` as “enabled is false.” Read `and` as “both conditions,” and `or` as “at least one condition.” This is a stated practice rule, not a rule Python chooses for you.

**Notebook state and evidence.** Editing an input cell does not rerun the decision below it. Rerun the input and decision cells together. Use `assert actual == expected` only after writing the expected result from the rule. A failed assertion is information, not a reason to change the expectation. At this stage you can repeat short checks manually; loops and functions arrive later.

**Optional application example.** Food Rescue uses the same branch structure to combine a supplied review flag with a planned amount. Its own rule gives review priority. The Python concepts above can be understood and tested without opening that project.

### Small example 1 · Compare before branching

= assigns, while ==, > and >= ask different questions and return Booleans. Predict True, False, True for this exact boundary. This example has no conditional block yet.


```python
score = 20
print(score == 20)
print(score > 20)
print(score >= 20)
```

### Small example 2 · Choose between two blocks

A colon starts the block and indentation groups its statements. Exactly one branch assigns message, then the unindented print runs. Predict disabled. Try True only after predicting enabled.


```python
enabled = False
if enabled:
    message = "enabled"
else:
    message = "disabled"
print(message)
```

### Small example 3 · Combine Boolean questions

and needs both operands to be true; or needs at least one; not reverses a Boolean. Predict False, True, True. These expressions compute values without yet choosing a block.


```python
ready = True
connected = False
print(ready and connected)
print(ready or connected)
print(not connected)
```

### Worked example: predict, then trace

```python
score = 20
enabled = True
if not enabled:
    status = "disabled"
elif score >= 20:
    status = "high"
else:
    status = "low"
print("Status:", status)
assert status == "high"
```
