### Function contracts, local names and returned data

A function groups instructions behind a name so you can call them with different inputs. A function definition stores instructions. It does not calculate an answer until you call the function. `def rectangle_area(width, height):` names two parameters. In `rectangle_area(3, 4)`, the arguments 3 and 4 become the local values of width and height for that call. The result is the area of a rectangle, measured in square units.

**A returned value travels back to the caller.** `return width * height` ends that call and gives its result to the surrounding expression. `answer = rectangle_area(3, 4)` therefore stores 12. `print(...)` displays text but does not supply a returned numeric result. A function reaching its end without return gives `None`. Arithmetic on that result will fail. Keep display at the caller when several other calculations need the same helper.

**Local names and dependencies.** A local total should start inside the function each time. It should not depend on an old global total from a notebook cell. Call the helper twice with different arguments to reveal accidental dependence on earlier state. Name the parameter for what it means and include the unit in the contract. This week's functions assume already-valid nonnegative integer inputs; text validation arrives next week.

**A contract says what callers can rely on.** Write the input meaning, output type, and one example before implementation. A function can return one dictionary containing several named results. The caller reads those fields by key, just as it reads an input record. Returning a dictionary is different from printing several lines. The keys are part of the contract and must be spelled consistently.

### Small example 1 · Define first, then call

def stores the function instructions. They run only when the function is called. The argument 4 becomes the local parameter number, and return supplies 8 to the caller. Predict only 8 on screen.


````python
def double(number):
    return number * 2

answer = double(4)
print(answer)
assert answer == 8
````

### Small example 2 · A returned value can be reused

The caller receives a value it can store, add or test. Keep print outside the helper. Each call has a fresh local result. Predict 5, 9, 14; the second call does not overwrite first.


````python
def add_one(number):
    result = number + 1
    return result

first = add_one(4)
second = add_one(8)
print(first, second, first + second)
````

### Small example 3 · Return one dictionary with named fields

A function can return the same dictionary shape you learned in Week 3. It is one returned value containing several fields. The caller reads a key. Predict Ada and 4. This is the syntax needed by P2, without supplying its implementation.


````python
def make_score(name, points):
    return {"name": name, "points": points}

note = make_score("Ada", 4)
print(note["name"])
print(note["points"])
````

````python
def rectangle_area(width, height):
    area = width * height
    return area

first = rectangle_area(3, 4)
second = rectangle_area(0, 5)
print(first, second)
assert first == 12
assert second == 0
````
