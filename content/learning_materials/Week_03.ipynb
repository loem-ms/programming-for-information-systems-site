## Session 1 · Python foundations


### A plain list

A list keeps items in order. Positions start at zero. len counts items.

```python
counts = [2, 4, 1]
print(counts[0], len(counts))  # 2 3
```


### Changing a list

Index assignment replaces one item. append adds at the end and changes the list in place.

```python
counts = [2, 4]
counts[1] = 3
counts.append(1)
print(counts)  # [2, 3, 1]
```


### Removing an item

pop removes and returns the last item. append returns None, so do not assign its result back to the list.

```python
counts = [2, 3, 1]
print(counts.pop(), counts)  # 1 [2, 3]
```


### A loop

for visits each item. The indented body repeats once per item. An empty list has no iterations.

```python
for count in [2, 4, 1]:
    print(count)
```


### A running total

Initialize once before the loop. += adds the current contribution to the earlier total.

```python
total = 0
for count in [2, 4, 1]:
    total += count
```


### One dictionary

A dictionary groups values under named keys. A missing or misspelled key raises KeyError.

```python
record = {"name": "A", "points": 2}
print(record["points"])  # 2
```


### A list of records

First select a dictionary by position, then a field by key. Record count and total points are different measures.

```python
records = [{"name": "A", "points": 2}, {"name": "B", "points": 1}]
print(len(records), records[0]["points"])  # 2 records, 2 points in A
```

### Example 3 · Visit a plain list with a loop

### Example 4 · Accumulate numbers before records

Initialize total once before the loop, then update it for each number. += means add to the previous value. Predict running totals 2, 6, 7 and final total 7. The length is 3, which answers a different question.


````python
counts = [2, 4, 1]
total = 0
for count in counts:
    total += count
    print("Running:", total)
print("Final:", total)
assert total == 7
````

### Example 5 · Read one dictionary before a list of dictionaries

A dictionary uses named keys, not positions. Curly braces create one record. The quoted key selects its value. Predict A, 2, then 3. Only after this example do we put several dictionaries into a list.


````python
record = {"name": "A", "points": 2}
print(record["name"])
print(record["points"])
record["points"] = 3
print(record["points"])
````

### Combined example


````python
records = [{"name": "A", "points": 2},
           {"name": "B", "points": 1}]
total = 0
names = []
for row in records:
    total += row["points"]
    names.append(row["name"])
    print(row["name"], total)
assert total == 3
assert names == ["A", "B"]
````
