### Lists, records and accumulation

A list stores several items in order. A dictionary groups related values under named keys. For example, a list of score records can store each learner's name and points. In `records[0]["points"]`, `[0]` selects the first dictionary because Python list positions start at zero; `["points"]` then selects a value inside that dictionary. A misspelled key raises `KeyError`. The key is quoted text; the value can have another type.

**A loop visits one record at a time.** In `for row in records:`, `row` is the current dictionary, not the entire list. All indented statements repeat for that dictionary. Read the current points inside the loop. Print a trace inside the loop when you want one observation per input. Print the final summary after the loop when you want one overall result. Walk through two rows by hand before running the code.

**An accumulator remembers earlier work.** Set a numeric total to zero before the loop. `total += contribution` means `total = total + contribution`. Putting the zero assignment inside the loop forgets earlier rows. Using `total = contribution` also loses earlier work. An empty list runs the body zero times; a correctly initialized total remains zero. Rerunning a cell should initialize the accumulator again rather than add to an old total.

**The measure determines the calculation.** `len(records)` counts records. Adding each points value gives total points. These are different measures and should have different labels. If you split one record with 4 points into two records with 2 points each, the record count changes but the point total stays the same. Such a comparison checks that the program measures what you intend. Repeated names do not automatically tell you whether records should be merged.

**Building a list is another accumulation.** Start `selected = []`, test each row with `if`, then call `selected.append(value)` to add a result at the end. `append` changes the list; it does not return the updated list. Do not write `selected = selected.append(value)`. A list of selected names can be built without changing the original score records.

### Small example 4 · Accumulate numbers before records

Initialize total once before the loop, then update it for each number. += means add to the previous value. Predict running totals 2, 6, 7 and final total 7. The length is 3, which answers a different question.


```python
counts = [2, 4, 1]
total = 0
for count in counts:
    total += count
    print("Running:", total)
print("Final:", total)
assert total == 7
```

### Worked example: predict, then trace

Read the complete example once. Write each expected result before running. Point to the statement responsible for each value.

```python
records = [{"name": "A", "points": 2},
           {"name": "B", "points": 1}]
total = 0
names = []
for row in records:
    total += row["points"]
    names.append(row["name"])
    print(row["name"], total)
assert total == 3
assert names == ["A", "B"]
```

The trace is A 2, then B 3. The current row changes each iteration; total and names keep earlier contributions. Before rerunning, explain why both accumulators reset outside the loop. No project-specific record format is needed.

### Complete teaching example: three measures

This independent practice plan uses T01 and T02. Predict 2 lines, 6 containers and 20 planned portions, then inspect the output. The review flag comes from the coordinator separately; this loop does not infer it.

```python
pickup_lines = [
    {"batch_id": "T01", "portions_per_container": 3, "containers": 4},
    {"batch_id": "T02", "portions_per_container": 4, "containers": 2},
]
container_count = 0
total_portions = 0
for line in pickup_lines:
    container_count += line["containers"]
    total_portions += line["portions_per_container"] * line["containers"]
print("Lines:", len(pickup_lines))
print("Containers:", container_count)
print("Planned portions:", total_portions)
assert (len(pickup_lines), container_count, total_portions) == (2, 6, 20)
```
