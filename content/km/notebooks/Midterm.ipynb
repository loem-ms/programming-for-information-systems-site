### A3 · ស្លាក

```python
def label(portions):
    if portions >= 5:
        return "standard"
    elif portions >= 20:
        return "team"
    return "too small"

print(label(25))
```

✍️ **A3** · វា print … ព្រោះ …  តើអ្នកនឹងកែវាដោយរបៀបណា?

### A4 · សរុប

```python
def total(records):
    count = 0
    for record in records:
        count = count + record["portions"]
    print(count)

result = total([{"portions": 10}, {"portions": 5}])
print("Result:", result)
```

✍️ **A4** · វា print … ព្រោះ …  ប្ដូរតែមួយបន្ទាត់ ដើម្បីឱ្យ `result` ស្មើ 15។

## ផ្នែក E · វាយតម្លៃ code ដែល AI សរសេរ (10 ពិន្ទុ ប្រហែល 10 នាទី)

មិត្តរួមថ្នាក់ម្នាក់ បានសុំឱ្យជំនួយការ AI សរសេរ `plan_day`។ នេះជាចម្លើយ។ វា run ដោយគ្មាន error។ សំណើថ្មីមួយទើបតែមកដល់ (F005); មិនទាន់មាននរណាសម្រេចចិត្តលើវានៅឡើយ ដូច្នេះវាគ្មាន value `"approved"` ទេ។ Run cell នេះ បន្ទាប់មកមើលឱ្យល្អិត។

```python
def plan(x):
    t = 0
    v = 0
    w = []
    for o in x:
        if o.get("approved", True):
            t = t + o["portions"]
            v = v + o["portions"] // 15 + 1
        else:
            w.append(o["food"])
    return "Approved: " + str(t) + " portions, " + str(v) + " volunteers. Waiting: " + ", ".join(w)

today = bridge.midterm_offers() + [{"id": "F005", "food": "Bananas", "portions": 20}]
result = plan(today)
print(result)
print("Type of the result:", type(result).__name__)
```
