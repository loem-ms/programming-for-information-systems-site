## 00 · ការ run code, comment, indentation

Python run បន្ទាត់នៃ cell មួយពីលើចុះក្រោម ពេលអ្នកចុច ▶ ឬ **Shift+Enter**។ ការកែ cell មួយមិនប្ដូរអ្វីទេ រហូតដល់អ្នក run វាម្ដងទៀត។ អក្សរនៅក្រោយ `#` គឺជា **comment**: Python មិនអើពើវាទេ ដូច្នេះវាជាកំណត់ចំណាំសម្រាប់មនុស្ស។ បន្ទាត់ដែលបញ្ចប់ដោយសញ្ញា `:` ចាប់ផ្ដើម **block** មួយ ហើយបន្ទាត់ដែលជារបស់ block នោះ ត្រូវចូលបន្ទាត់ (indent) 4 ដកឃ្លា។

**ប្រយ័ត្ន:** indentation កំណត់ថាបន្ទាត់ណាជារបស់អ្វី។ វាមិនមែនសម្រាប់លម្អទេ។

````python
battery_percent = 15          # comment មួយ: Python មិនអើពើផ្នែកនេះទេ
print("Battery:", battery_percent)
if battery_percent < 20:
    print("Low battery: please charge your phone")   # នៅក្នុង block របស់ if
print("Check finished")       # មិនបានចូលបន្ទាត់: run ជានិច្ច
````

**លំហាត់។** សារ "Unplug when full" គួរតែបង្ហាញ តែពេលទូរស័ព្ទកំពុងសាកថ្មប៉ុណ្ណោះ។ កែ indentation។ បន្ទាប់មក run cell ជាមួយ `is_charging = True` និងជាមួយ `False`។

````python
is_charging = False
if is_charging:
    print("Phone is charging")
print("Unplug when full")      # TODO: បន្ទាត់នេះត្រូវនៅក្នុង block របស់ if
print("Check finished")
````

ចូលបន្ទាត់ `print("Unplug when full")` 4 ដកឃ្លា ដើម្បីឱ្យវាតម្រង់ជួរជាមួយបន្ទាត់ខាងលើវា។ ជាមួយ `False` ឥឡូវអ្នកឃើញតែ `Check finished` ប៉ុណ្ណោះ។ ជាមួយ `True` អ្នកឃើញទាំងបីបន្ទាត់។

## 01 · Value, type, ការបំប្លែង type

value នីមួយៗមាន **type** មួយ: `int` (ចំនួនគត់), `float` (ចំនួនទសភាគ), `str` (text នៅក្នុងសញ្ញាសម្រង់) ឬ `bool` (`True` ឬ `False`)។ `name = value` រក្សាទុក value មួយក្រោមឈ្មោះមួយ ហើយ `type(value)` បង្ហាញ type របស់វា។ text ដែលមើលទៅដូចលេខ នៅតែជា text ដដែល ដូច្នេះត្រូវបំប្លែងវាដោយ `int()` ឬ `float()` មុនពេលគណនា ហើយប្រើ `str()` ដើម្បីបំប្លែងលេខទៅជា text។

**ប្រយ័ត្ន:** `"False"` ក្នុងសញ្ញាសម្រង់ គឺជា text មិនមែនជា Boolean `False` ទេ។

````python
students_in_class = 32          # int: ចំនួនគត់
average_mark = 71.5             # float: ចំនួនទសភាគ
class_name = "MIS 101"          # str: text នៅក្នុងសញ្ញាសម្រង់ជានិច្ច
exam_finished = True            # bool: True ឬ False
print(type(students_in_class), type(average_mark), type(class_name), type(exam_finished))

typed_age = "19"                # text ឧទាហរណ៍ ដែលវាយបញ្ចូលក្នុង form
age = int(typed_age)            # បំប្លែង text ទៅជាចំនួនគត់
print(age + 1)
print("Next year you will be " + str(age + 1))
print(bool("False"))            # True: text ណាដែលមិនទទេ រាប់ថាជា True
````

**លំហាត់។** form លើ web របស់ហាងមួយ ផ្ញើបរិមាណ និងតម្លៃជា text។ បំប្លែងទាំងពីរ ហើយគណនាតម្លៃសរុបជាលេខ។

````python
quantity_text = "3"
price_text = "1.50"
# TODO: បំប្លែង quantity_text ដោយ int() និង price_text ដោយ float() រួចគុណវាជាមួយគ្នា
total_dollars = 0
print("Total:", total_dollars, "dollars")
````

## 02 · ការគណនា (arithmetic)

`+ - * /` ដំណើរការដូចក្នុងគណិតវិទ្យា ហើយ `/` តែងតែផ្ដល់ចំនួនទសភាគ (`float`)។ `//` ចែក ហើយបោះសំណល់ចោល `%` ផ្ដល់សំណល់ ហើយ `**` ជាស្វ័យគុណ។ Python ធ្វើ `*` និង `/` មុន `+` និង `-` ដូច្នេះប្រើវង់ក្រចក ដើម្បីឱ្យលំដាប់ច្បាស់។

**ប្រយ័ត្ន:** `"3" * 4` គឺ `"3333"` ព្រោះលេខ 3 ក្នុងសញ្ញាសម្រង់ជា text។

````python
print(2 + 3 * 4, (2 + 3) * 4)     # គុណមុន លើកលែងតែអ្នកប្រើវង់ក្រចក
print(7 / 2, 7 // 2, 7 % 2)       # ចែក, ចំនួនក្រុមពេញ, សំណល់
print(2 ** 3)                      # ស្វ័យគុណ: 2 x 2 x 2
print(round(10 / 3, 2))            # បង្គត់ត្រឹម 2 ខ្ទង់ទសភាគ
print(3 * 4, "3" * 4)              # លេខ ធៀបនឹង text
````

**លំហាត់។** និស្សិត 50 នាក់ ត្រូវធ្វើការជាក្រុម ក្រុមនីមួយៗ 6 នាក់។ តើមានក្រុមពេញប៉ុន្មាន ហើយនៅសល់និស្សិតប៉ុន្មាននាក់? ប្រើ `//` និង `%`។

````python
class_size = 50
group_size = 6
# TODO: គណនាចំនួនក្រុមពេញ និងចំនួននិស្សិតដែលនៅសល់
full_groups = 0
students_left_over = 0
print(full_groups, "full groups,", students_left_over, "students left over")
````

```python

students_left_over = class_size % group_size
```

លទ្ធផលដែលរំពឹងទុក: `8 full groups, 2 students left over`។ ពិនិត្យមើល: 8 × 6 + 2 = 50។

## 03 · Text និង f-string

Text (**string**) ត្រូវនៅក្នុងសញ្ញាសម្រង់។ method របស់ string ដូចជា `.strip()`, `.lower()`, `.upper()`, `.title()` និង `.replace()` return string **ថ្មី** មួយ ហើយ string ដើមមិនប្រែប្រួលទេ។ **f-string** មាន `f` នៅមុនសញ្ញាសម្រង់ ហើយដាក់ value ចូលក្នុង `{}` ហើយ `{price:.2f}` បង្ហាញលេខដែលមានពីរខ្ទង់ទសភាគ។

**ប្រយ័ត្ន:** នៅក្នុង `{}` របស់ f-string ត្រូវប្រើសញ្ញាសម្រង់ប្រភេទផ្សេង ឧទាហរណ៍ `f"{student['name']}"`។

````python
typed_name = "  sok DARA  "
clean_name = typed_name.strip().title()     # លុបដកឃ្លាខាងក្រៅ រួចដាក់អក្សរធំនៅដើមពាក្យនីមួយៗ
print(clean_name)
print(len(typed_name), len(clean_name))     # 12 8: string ដើមមិនបានប្រែប្រួលទេ
print(clean_name.lower(), clean_name.replace("Sok", "Chan"))

item = "notebook"
price_dollars = 1.5
quantity = 3
print(f"{quantity} x {item} = {price_dollars * quantity:.2f} dollars")
````

## 04 · ការប្រៀបធៀប និង `if`

ការប្រៀបធៀបមួយ (`==`, `!=`, `<`, `<=`, `>`, `>=`) ផ្ដល់ `True` ឬ `False` ហើយ `and`, `or`, `not` ផ្សំការប្រៀបធៀបចូលគ្នា។ `if` / `elif` / `else` ពិនិត្យលក្ខខណ្ឌពីលើចុះក្រោម ហើយ run តែ branch **ទីមួយ** ដែលជា `True` ប៉ុណ្ណោះ។

**ប្រយ័ត្ន:** `>=` រាប់បញ្ចូល boundary ប៉ុន្តែ `>` មិនរាប់ទេ។ `=` មួយ រក្សាទុក value មួយ។ `==` ប្រៀបធៀប value ពីរ។

````python
battery_percent = 20
print(battery_percent > 20, battery_percent >= 20, battery_percent == 20)

if battery_percent <= 10:
    status = "critical"
elif battery_percent <= 20:
    status = "low"
else:
    status = "ok"
print(status)

is_charging = False
print(battery_percent < 50 and not is_charging)   # ថ្មខ្សោយ ហើយមិនកំពុងសាក
````

## 06 · Loop និងការបូកសរុប

error ខ្លះអាចរំពឹងទុកបាន ឧទាហរណ៍ ពេលនរណាម្នាក់វាយ `two` នៅកន្លែងដែលគួរជាលេខ។ ដាក់បន្ទាត់ដែលមានហានិភ័យនៅក្នុង `try:`។ បើ `ValueError` កើតឡើង Python លោតទៅ `except ValueError:` ជំនួសឱ្យការឈប់។ ចាប់ (catch) តែ error ដែលអ្នករំពឹងទុកប៉ុណ្ណោះ ដើម្បីឱ្យ bug ពិតៗនៅតែបង្ហាញ។

**`raise`:** អ្នកអាចបង្កើត error ដោយខ្លួនឯងដោយ `raise ValueError("message")` ពេល value មួយមាន type ត្រឹមត្រូវ ប៉ុន្តែបំពានវិធាន ដូចជាបរិមាណតិចជាង 1។

## 12 · JSON និង JSON Lines

**JSON** គឺជាទម្រង់ text សម្រាប់ dictionary និង list ដែលស្ទើរតែគ្រប់ប្រព័ន្ធអាចអានបាន។ `json.dump(data, file)` រក្សាទុកទិន្នន័យ Python ទៅក្នុង file ហើយ `json.load(file)` អានវាមកវិញ។ `json.dumps` និង `json.loads` ធ្វើដូចគ្នា ប៉ុន្តែជាមួយ string។ **JSON Lines** (`.jsonl`) រក្សាទុក record JSON មួយក្នុងមួយបន្ទាត់ ដូច្នេះអ្នកអាចបន្ថែម record មួយនៅខាងចុង (mode `"a"`) ដោយមិនចាំបាច់សរសេរ file ឡើងវិញ។ ហេតុនេះហើយ វាមានប្រយោជន៍សម្រាប់កំណត់ត្រាប្រវត្តិ (history log)។

**ប្រយ័ត្ន:** JSON សរសេរ `true`, `false` និង `null` នៅកន្លែងដែល Python សរសេរ `True`, `False` និង `None`។

````python
import json

phone = {"owner": "Dara", "battery_percent": 80, "charging": False, "apps": ["maps", "camera"]}
print(json.dumps(phone))                  # text JSON: false ជំនួសឱ្យ False

with open("phone.json", "w", encoding="utf-8") as file:
    json.dump(phone, file, indent=2)      # រក្សាទុក dictionary ទាំងមូល
with open("phone.json", encoding="utf-8") as file:
    loaded = json.load(file)              # អានវាមកវិញ
print(loaded["apps"], loaded == phone)

with open("calls.jsonl", "w", encoding="utf-8") as file:    # JSON Lines: record មួយក្នុងមួយបន្ទាត់
    file.write(json.dumps({"to": "Sophea", "minutes": 3}) + "\n")
    file.write(json.dumps({"to": "Vanna", "minutes": 12}) + "\n")
with open("calls.jsonl", encoding="utf-8") as file:
    for line in file:
        print(json.loads(line))
````

**លំហាត់។** បន្ថែមការហៅទូរស័ព្ទមួយទៅក្នុង log ដោយប្រើ mode `"a"`។ បន្ទាប់មកអាន log ម្ដងមួយបន្ទាត់ ហើយបូកសរុបនាទី។

````python
products = [
    {"name": "pen", "price": 0.5},
    {"name": "bag", "price": 12.0},
    {"name": "notebook", "price": 1.5},
]

def by_price(product):
    """Key function: take one product, return the value to sort by."""
    return product["price"]

print(sorted([3, 1, 2]))
for product in sorted(products, key=by_price):                    # ថោកបំផុតមុនគេ
    print(product["name"], product["price"])
print(sorted(products, key=by_price, reverse=True)[0]["name"])    # ថ្លៃបំផុត
````

## 15 · Module, `import`, `if __name__ == "__main__":`

**module** មួយ គឺជា file `.py` ដែលមាន function ដែល code ផ្សេងទៀតអាចយកមកប្រើម្ដងទៀតដោយ `import`។ standard library របស់ Python មាន module ស្រាប់ៗជាច្រើន ដូចជា `math`, `json` និង `csv`។ ពេលអ្នកចាប់ផ្ដើម file មួយជា program នោះ Python កំណត់ `__name__` របស់វាជា `"__main__"`។ ពេល file ផ្សេងទៀត import វា `__name__` គឺជាឈ្មោះរបស់ module។ ដូច្នេះ code នៅក្រោម `if __name__ == "__main__":` run តែពេល file ត្រូវបានចាប់ផ្ដើមជា program ប៉ុណ្ណោះ ហើយការ import file នោះ (ឧទាហរណ៍ ដើម្បី test function របស់វា) មិនចាប់ផ្ដើមអ្វីទេ។

````python
import importlib
import math                                   # module មួយពី standard library
import os
import sys

print(math.sqrt(16), math.ceil(4.2))

shop_code = '''TAX_RATE = 0.10

def price_with_tax(price_dollars):
    return round(price_dollars * (1 + TAX_RATE), 2)

def main():
    print("Demo: 10 dollars with tax is", price_with_tax(10))

if __name__ == "__main__":
    main()
'''
with open("shop_tools.py", "w", encoding="utf-8") as file:
    file.write(shop_code)

# ការរៀបចំសម្រាប់ notebook: រក module ក្នុង folder នេះ ហើយ load កំណែថ្មីបំផុត។
# ក្នុង program .py ធម្មតា អ្នកត្រូវការតែបន្ទាត់ import ប៉ុណ្ណោះ។
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())
importlib.invalidate_caches()
sys.modules.pop("shop_tools", None)

import shop_tools                             # main() មិន run ទេ: __name__ គឺ "shop_tools"
print(shop_tools.price_with_tax(20))
print(shop_tools.__name__)
````

**លំហាត់។** ការ import `class_tools` print បន្ទាត់ demo មួយ ព្រោះបន្ទាត់ចុងក្រោយរបស់វាហៅ `main()` ដោយផ្ទាល់។ កែ text របស់ module ដើម្បីឱ្យ `main()` run តែពេល file ត្រូវបានចាប់ផ្ដើមជា program ប៉ុណ្ណោះ។ ពេលនោះ ការ import នឹងមិន print អ្វីទេ។

````python
import importlib
import os
import sys

tools_code = '''def average(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

def main():
    print("Demo: the average of 70, 80 and 90 is", average([70, 80, 90]))

main()   # TODO: run main() only when the file is started as a program
'''
with open("class_tools.py", "w", encoding="utf-8") as file:
    file.write(tools_code)

if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())
importlib.invalidate_caches()
sys.modules.pop("class_tools", None)

import class_tools                 # ក្រោយពេលអ្នកកែ បន្ទាត់នេះមិន print អ្វីទេ
print("Imported. average([50, 60]) =", class_tools.average([50, 60]))
````

## 16 · ការ run file `.py` និង command-line argument

program ដែលបញ្ចប់រួច គឺជា file `.py` ដែលអ្នកចាប់ផ្ដើមពី terminal ឧទាហរណ៍ `python3 greet.py Dara`។ ការ run នីមួយៗ គឺជា process ថ្មីមួយ ដែលចាប់ផ្ដើមដោយ memory ទទេ។ ពាក្យនៅក្រោយឈ្មោះ file មកដល់ក្នុង list `sys.argv` ជា text (`sys.argv[0]` គឺជាឈ្មោះ file)។ ក្នុង notebook អ្នកអាចចាប់ផ្ដើម program តាមរបៀបដូចគ្នា ដោយ `subprocess.run([sys.executable, "greet.py", "Dara"], ...)`: `sys.executable` គឺជា Python ដែល run notebook នេះ ហើយ exit code 0 មានន័យថាជោគជ័យ។

````python
import subprocess
import sys

program = '''import sys

def main():
    names = sys.argv[1:]          # the words after the file name, as text
    if len(names) == 0:
        print("Usage: python3 greet.py NAME ...")
        sys.exit(1)               # a non-zero exit code means "something went wrong"
    for name in names:
        print("Hello,", name)

if __name__ == "__main__":
    main()
'''
with open("greet.py", "w", encoding="utf-8") as file:
    file.write(program)

result = subprocess.run([sys.executable, "greet.py", "Dara", "Sophea"], capture_output=True, text=True)
print(result.stdout.strip())
print("Exit code:", result.returncode)

result = subprocess.run([sys.executable, "greet.py"], capture_output=True, text=True)
print(result.stdout.strip())
print("Exit code:", result.returncode)
````

## 17 · ការ run test ក្នុង process ថ្មី

notebook មួយចងចាំអ្វីៗទាំងអស់ដែលអ្នកបាន run សូម្បីតែ function ដែលអ្នកបានលុប ឬប្ដូរឈ្មោះរួចហើយ។ ដូច្នេះ test អាចជាប់នៅទីនេះ ប៉ុន្តែនៅតែធ្លាក់នៅលើកុំព្យូទ័រផ្សេង។ **process ថ្មី** ចាប់ផ្ដើមដោយមិនចងចាំអ្វីទាំងអស់ ដូច្នេះវាជាការពិនិត្យដ៏ស្មោះត្រង់: ក្នុង notebook សូមប្រើ **Runtime → Restart session and run all**។ សម្រាប់ program សូម run file test ពី terminal។ file test មួយ import code ដែលវា test មាន function `test_` ដែលប្រើ `assert` ហើយបញ្ចប់ដោយ exit code ដែលមិនមែនសូន្យ ពេល test មួយធ្លាក់។

````python
import subprocess
import sys

with open("shop_rules.py", "w", encoding="utf-8") as file:
    file.write('''FREE_DELIVERY_FROM_DOLLARS = 25

def free_delivery(order_dollars):
    return order_dollars >= FREE_DELIVERY_FROM_DOLLARS
''')

with open("test_shop_rules.py", "w", encoding="utf-8") as file:
    file.write('''from shop_rules import free_delivery

def test_normal():
    assert free_delivery(40) == True
    assert free_delivery(10) == False

def test_boundary():
    assert free_delivery(25) == True
    assert free_delivery(24.99) == False

if __name__ == "__main__":
    test_normal()
    test_boundary()
    print("All tests passed")
''')

result = subprocess.run([sys.executable, "test_shop_rules.py"], capture_output=True, text=True)
print((result.stdout + result.stderr).strip())
print("Exit code:", result.returncode)
````

````python
import subprocess
import sys

with open("ticket_rules.py", "w", encoding="utf-8") as file:
    file.write('''CHILD_UNDER_AGE = 12

def ticket_price(age):
    if age <= CHILD_UNDER_AGE:
        return 2
    return 5
''')

with open("test_ticket_rules.py", "w", encoding="utf-8") as file:
    file.write('''from ticket_rules import ticket_price

def test_normal():
    assert ticket_price(8) == 2
    assert ticket_price(30) == 5

# TODO 1: add test_boundary(): an 11-year-old pays 2, a 12-year-old pays 5; call it below

if __name__ == "__main__":
    test_normal()
    print("All tests passed")
''')

result = subprocess.run([sys.executable, "test_ticket_rules.py"], capture_output=True, text=True)
print((result.stdout + result.stderr).strip())
print("Exit code:", result.returncode)
# TODO 2: ពេល test ថ្មីរបស់អ្នកធ្លាក់ សូមកែ ticket_rules.py ខាងលើ ហើយ run cell ម្ដងទៀត
````
