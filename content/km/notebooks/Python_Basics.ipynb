## 00 · ការ run កូដ, comment, ការចូលបន្ទាត់ (indentation)

នៅពេលអ្នកចុច ▶ ឬ **Shift+Enter** Python នឹងដំណើរការបន្ទាត់នៃ cell ពីលើចុះក្រោម។ ប្រសិនបើអ្នកកែ cell មួយ គ្មានអ្វីប្រែប្រួលទេ រហូតដល់អ្នក run វាម្ដងទៀត។

អក្សរនៅក្រោយសញ្ញា `#` គឺជា **comment**។ Python មិនអើពើវាទេ ដូច្នេះវាជាកំណត់ចំណាំសម្រាប់មនុស្សអាន។

បន្ទាត់ដែលបញ្ចប់ដោយសញ្ញាចុចពីរ `:` ចាប់ផ្ដើម **block** មួយ។ បន្ទាត់ដែលស្ថិតនៅក្នុង block នោះ ត្រូវចូលបន្ទាត់ (indent) 4 ដកឃ្លា (spaces)។

**ប្រយ័ត្ន:** ការចូលបន្ទាត់ (indentation) កំណត់ថាបន្ទាត់ណាស្ថិតនៅក្នុង block ណា។ វាមិនមែនសម្រាប់លម្អទេ។

````python
battery_percent = 15          # comment មួយ: Python មិនអើពើផ្នែកនេះទេ
print("Battery:", battery_percent)
if battery_percent < 20:
    print("Low battery: please charge your phone")   # នៅក្នុង block របស់ if
print("Check finished")       # មិនបានចូលបន្ទាត់: ដំណើរការជានិច្ច
````

**លំហាត់។** សារ "Unplug when full" ត្រូវបង្ហាញតែនៅពេលទូរស័ព្ទកំពុងសាកថ្មប៉ុណ្ណោះ។ ចូរកែការចូលបន្ទាត់ (indentation) ឱ្យត្រូវ។ បន្ទាប់មក run cell ម្ដងជាមួយ `is_charging = True` និងម្ដងទៀតជាមួយ `False`។

````python
is_charging = False
if is_charging:
    print("Phone is charging")
print("Unplug when full")      # TODO: បន្ទាត់នេះត្រូវនៅក្នុង block របស់ if
print("Check finished")
````

<details><summary>មើលចម្លើយគំរូមួយ</summary>

ត្រូវចូលបន្ទាត់ `print("Unplug when full")` 4 ដកឃ្លា ដើម្បីឱ្យវាតម្រង់ជួរជាមួយបន្ទាត់ខាងលើ។ ពេល `is_charging` ជា `False` ឥឡូវនេះអ្នកឃើញតែ `Check finished` ប៉ុណ្ណោះ។ ពេលវាជា `True` អ្នកឃើញទាំងបីបន្ទាត់។

## 01 · តម្លៃ (value), type, ការបំប្លែង type

តម្លៃ (value) នីមួយៗមាន **type** មួយ៖ `int` (ចំនួនគត់), `float` (ចំនួនទសភាគ), `str` (អត្ថបទ ឬ text នៅក្នុងសញ្ញាសម្រង់) ឬ `bool` (`True` ឬ `False`)។ `name = value` រក្សាទុកតម្លៃមួយក្រោមឈ្មោះមួយ ហើយ `type(value)` បង្ហាញ type របស់វា។

អត្ថបទដែលមើលទៅដូចលេខ នៅតែជាអត្ថបទដដែល។ ដូច្នេះ ត្រូវបំប្លែងវាដោយ `int()` ឬ `float()` មុនពេលគណនា។ ប្រើ `str()` ដើម្បីបំប្លែងលេខទៅជាអត្ថបទ។

**ប្រយ័ត្ន:** `"False"` ក្នុងសញ្ញាសម្រង់ គឺជាអត្ថបទ មិនមែនជាតម្លៃ Boolean `False` ទេ។

````python
students_in_class = 32          # int: ចំនួនគត់
average_mark = 71.5             # float: ចំនួនទសភាគ
class_name = "MIS 101"          # str: អត្ថបទ ត្រូវនៅក្នុងសញ្ញាសម្រង់ជានិច្ច
exam_finished = True            # bool: True ឬ False
print(type(students_in_class), type(average_mark), type(class_name), type(exam_finished))

typed_age = "19"                # អត្ថបទ ឧទាហរណ៍ ដែលវាយបញ្ចូលក្នុង form
age = int(typed_age)            # បំប្លែងអត្ថបទទៅជាចំនួនគត់
print(age + 1)
print("Next year you will be " + str(age + 1))
print(bool("False"))            # True: អត្ថបទណាដែលមិនទទេ ត្រូវរាប់ថាជា True
````

## 02 · ការគណនា (arithmetic)

`+ - * /` ដំណើរការដូចក្នុងគណិតវិទ្យា ហើយ `/` តែងតែផ្ដល់ចំនួនទសភាគ (`float`)។

- `//` ចែក ហើយបោះសំណល់ចោល។
- `%` ផ្ដល់សំណល់។
- `**` គឺជាស្វ័យគុណ។

Python គណនា `*` និង `/` មុន `+` និង `-`។ ដូច្នេះ ចូរប្រើវង់ក្រចក ដើម្បីឱ្យលំដាប់នៃការគណនាច្បាស់លាស់។

**ប្រយ័ត្ន:** `"3" * 4` គឺ `"3333"` ព្រោះលេខ 3 ក្នុងសញ្ញាសម្រង់ គឺជាអត្ថបទ។

````python
print(2 + 3 * 4, (2 + 3) * 4)     # គុណមុន លើកលែងតែអ្នកប្រើវង់ក្រចក
print(7 / 2, 7 // 2, 7 % 2)       # ចែក, ចំនួនក្រុមពេញ, សំណល់
print(2 ** 3)                      # ស្វ័យគុណ: 2 x 2 x 2
print(round(10 / 3, 2))            # បង្គត់ត្រឹម 2 ខ្ទង់ទសភាគ
print(3 * 4, "3" * 4)              # លេខ ធៀបនឹងអត្ថបទ
````

**លំហាត់។** និស្សិត 50 នាក់ ត្រូវធ្វើការជាក្រុម ក្រុមនីមួយៗ 6 នាក់។ តើមានក្រុមពេញប៉ុន្មាន? ហើយតើនៅសល់និស្សិតប៉ុន្មាននាក់? ចូរប្រើ `//` និង `%`។

````python
class_size = 50
group_size = 6
# TODO: គណនាចំនួនក្រុមពេញ និងចំនួននិស្សិតដែលនៅសល់
full_groups = 0
students_left_over = 0
print(full_groups, "full groups,", students_left_over, "students left over")
````

## 03 · អត្ថបទ (text) និង f-string

អត្ថបទ (**string**) ត្រូវនៅក្នុងសញ្ញាសម្រង់។ method របស់ string ដូចជា `.strip()`, `.lower()`, `.upper()`, `.title()` និង `.replace()` បញ្ជូនត្រឡប់មកវិញ (return) នូវ string **ថ្មី** មួយ។ string ដើមមិនប្រែប្រួលទេ។

**f-string** មានអក្សរ `f` នៅមុនសញ្ញាសម្រង់ ហើយដាក់តម្លៃចូលក្នុង `{}`។ `{price:.2f}` បង្ហាញលេខដែលមានពីរខ្ទង់ទសភាគ។

**ប្រយ័ត្ន:** នៅក្នុង `{}` របស់ f-string ត្រូវប្រើសញ្ញាសម្រង់ប្រភេទផ្សេង ឧទាហរណ៍ `f"{student['name']}"`។

````python
typed_name = "  sok DARA  "
clean_name = typed_name.strip().title()     # លុបដកឃ្លាខាងក្រៅ រួចដាក់អក្សរធំនៅដើមពាក្យនីមួយៗ
print(clean_name)
print(len(typed_name), len(clean_name))     # 12 8: string ដើមមិនបានប្រែប្រួលទេ
print(clean_name.lower(), clean_name.replace("Sok", "Chan"))

item = "notebook"
price_dollars = 1.5
quantity = 3
print(f"{quantity} x {item} = {price_dollars * quantity:.2f} dollars")
````

## 06 · Loop និងការបូកសរុប

`for` loop ដំណើរការ block ដែលចូលបន្ទាត់របស់វា ម្ដងសម្រាប់ item នីមួយៗក្នុង list។

- ដើម្បីបូកសរុប ចូរបង្កើតចំនួនសរុប **មុន** loop ហើយបូកបន្ថែមវា **នៅក្នុង** loop ដោយប្រើ `+=`។
- ដើម្បីបូកសរុបតាមក្រុម ចូរប្រើ dictionary: `totals[key] = totals.get(key, 0) + amount`។

**ប្រយ័ត្ន:** ប្រសិនបើអ្នកសរសេរ `total = value` នៅក្នុង loop វានឹងសរសេរជាន់ចំនួនសរុបចាស់រាល់ដង មិនមែនបូកបន្ថែមទេ។

## 10 · តេស្ត (test) ដោយប្រើ `assert`

តេស្ត (test) មួយ ដំណើរការកូដរបស់អ្នកជាមួយធាតុចូល (input) មួយ ហើយប្រៀបធៀបលទ្ធផលជាមួយចម្លើយដែលអ្នកបានសម្រេច **មុន** ពេល run វា។ `assert condition` មិនធ្វើអ្វីទេ ប្រសិនបើលក្ខខណ្ឌជា `True` ប៉ុន្តែវានឹងឈប់ដោយ `AssertionError` ប្រសិនបើលក្ខខណ្ឌជា `False`។

ចូរសាកល្បងករណីបីប្រភេទ៖

- **ករណីធម្មតា** (normal): តម្លៃធម្មតាៗ។
- **ករណីព្រំដែន** (boundary): ត្រូវនឹងកម្រិតកំណត់ចំៗ និងនៅជិតបំផុតសងខាងវា។
- **ករណីចម្លែក** (odd): សូន្យ ទទេ អវិជ្ជមាន ឬមិនធម្មតា។

**ប្រយ័ត្ន:** ប្រសិនបើតេស្តមួយធ្លាក់ អ្នកត្រូវកែកូដ មិនមែនកែចម្លើយដែលរំពឹងទុកទេ លើកលែងតែចម្លើយដែលរំពឹងទុកនោះខុស។

````python
FREE_DELIVERY_FROM_DOLLARS = 25

def free_delivery(order_dollars):
    return order_dollars >= FREE_DELIVERY_FROM_DOLLARS

assert free_delivery(40) == True        # ធម្មតា
assert free_delivery(10) == False       # ធម្មតា
assert free_delivery(25) == True        # ព្រំដែន: ត្រូវនឹងកម្រិតចំៗ
assert free_delivery(24.99) == False    # ព្រំដែន: ទាបជាងកម្រិតបន្តិច
assert free_delivery(0) == False        # ចម្លែក: ការបញ្ជាទិញទទេ
print("All free_delivery tests passed")
````

## 12 · JSON និង JSON Lines

**JSON** គឺជាទម្រង់អត្ថបទ សម្រាប់ dictionary និង list ដែលស្ទើរតែគ្រប់ប្រព័ន្ធអាចអានបាន។

- `json.dump(data, file)` រក្សាទុកទិន្នន័យ Python ទៅក្នុង file ហើយ `json.load(file)` អានវាមកវិញ។
- `json.dumps` និង `json.loads` ធ្វើដូចគ្នា ប៉ុន្តែជាមួយ string។

**JSON Lines** (`.jsonl`) រក្សាទុក record JSON មួយក្នុងមួយបន្ទាត់។ ដូច្នេះ អ្នកអាចបន្ថែម record មួយនៅខាងចុង (mode `"a"`) ដោយមិនចាំបាច់សរសេរ file ទាំងមូលឡើងវិញ។ ហេតុនេះហើយ វាមានប្រយោជន៍សម្រាប់កំណត់ត្រាប្រវត្តិ (history log)។

**ប្រយ័ត្ន:** JSON សរសេរ `true`, `false` និង `null` នៅកន្លែងដែល Python សរសេរ `True`, `False` និង `None`។

````python
import json

phone = {"owner": "Dara", "battery_percent": 80, "charging": False, "apps": ["maps", "camera"]}
print(json.dumps(phone))                  # អត្ថបទ JSON: false ជំនួសឱ្យ False

with open("phone.json", "w", encoding="utf-8") as file:
    json.dump(phone, file, indent=2)      # រក្សាទុក dictionary ទាំងមូល
with open("phone.json", encoding="utf-8") as file:
    loaded = json.load(file)              # អានវាមកវិញ
print(loaded["apps"], loaded == phone)

with open("calls.jsonl", "w", encoding="utf-8") as file:    # JSON Lines: record មួយក្នុងមួយបន្ទាត់
    file.write(json.dumps({"to": "Sophea", "minutes": 3}) + "\n")
    file.write(json.dumps({"to": "Vanna", "minutes": 12}) + "\n")
with open("calls.jsonl", encoding="utf-8") as file:
    for line in file:
        print(json.loads(line))
````

## 13 · Class, object, method

**class** មួយ ពិពណ៌នាអំពីប្រភេទនៃរបស់មួយ ដូចជាទូរស័ព្ទ។ **object** មួយ គឺជាទូរស័ព្ទពិតមួយគ្រឿង ដែលបង្កើតចេញពី class នោះ។ object នីមួយៗរក្សាទិន្នន័យផ្ទាល់ខ្លួនរបស់វា។

`__init__` ដំណើរការនៅពេលអ្នកបង្កើត object មួយ ហើយរក្សាទុកទិន្នន័យរបស់វាជា **attribute** នៅលើ `self`។ **method** គឺជា function នៅក្នុង class។ `self` មានន័យថា "object នេះ"។

````python
class Phone:
    def __init__(self, owner, battery_percent):
        self.owner = owner
        self.battery_percent = battery_percent

    def use(self, minutes):
        """Each minute of use costs 1% battery, but never below 0."""
        self.battery_percent = max(0, self.battery_percent - minutes)

    def is_low(self):
        return self.battery_percent < 20

dara_phone = Phone("Dara", 50)
sophea_phone = Phone("Sophea", 90)
dara_phone.use(35)
print(dara_phone.owner, dara_phone.battery_percent, dara_phone.is_low())
print(sophea_phone.owner, sophea_phone.battery_percent, sophea_phone.is_low())
````

## 14 · ការ sort ដោយប្រើ key function

`sorted(items)` return list ថ្មីមួយដែលបាន sort រួច ហើយទុក list ដើមឱ្យនៅដដែល។ សម្រាប់ list នៃ dictionary ចូរផ្ដល់ **key function** មួយ៖ វាទទួល item មួយ ហើយ return តម្លៃដែលត្រូវប្រើសម្រាប់ sort។ បន្ថែម `reverse=True` ដើម្បីឱ្យតម្លៃធំបំផុតនៅមុនគេ។

````python
products = [
    {"name": "pen", "price": 0.5},
    {"name": "bag", "price": 12.0},
    {"name": "notebook", "price": 1.5},
]

def by_price(product):
    """Key function: take one product, return the value to sort by."""
    return product["price"]

print(sorted([3, 1, 2]))
for product in sorted(products, key=by_price):                    # ថោកបំផុតមុនគេ
    print(product["name"], product["price"])
print(sorted(products, key=by_price, reverse=True)[0]["name"])    # ថ្លៃបំផុត
````

ដូច្នេះ កូដនៅក្រោម `if __name__ == "__main__":` ដំណើរការតែពេល file ត្រូវបានចាប់ផ្ដើមជាកម្មវិធីប៉ុណ្ណោះ។ ការ import file នោះ (ឧទាហរណ៍ ដើម្បីសាកល្បង function របស់វា) មិនចាប់ផ្ដើមអ្វីទេ។

````python
import importlib
import math                                   # module មួយពី standard library
import os
import sys

print(math.sqrt(16), math.ceil(4.2))

shop_code = '''TAX_RATE = 0.10

def price_with_tax(price_dollars):
    return round(price_dollars * (1 + TAX_RATE), 2)

def main():
    print("Demo: 10 dollars with tax is", price_with_tax(10))

if __name__ == "__main__":
    main()
'''
with open("shop_tools.py", "w", encoding="utf-8") as file:
    file.write(shop_code)

# ការរៀបចំសម្រាប់ notebook: រក module ក្នុង folder នេះ ហើយ load កំណែថ្មីបំផុត។
# ក្នុងកម្មវិធី .py ធម្មតា អ្នកត្រូវការតែបន្ទាត់ import ប៉ុណ្ណោះ។
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())
importlib.invalidate_caches()
sys.modules.pop("shop_tools", None)

import shop_tools                             # main() មិនដំណើរការទេ: __name__ គឺ "shop_tools"
print(shop_tools.price_with_tax(20))
print(shop_tools.__name__)
````

## 18 · សន្លឹកជំនួយ terminal (cheat-sheet)

**terminal** គឺជាផ្ទាំងបង្អួចមួយ ដែលអ្នកវាយពាក្យបញ្ជា (command) ជំនួសឱ្យការចុច។ ចូរវាយ command ទាំងនេះក្នុង terminal មិនមែនក្នុង code cell ទេ ព្រោះវាមិនមែនជាកូដ Python ទេ។ នៅលើកុំព្យូទ័រផ្ទាល់ខ្លួនរបស់អ្នក សូមប្រើ **Terminal** (macOS ឬ Linux) ឬ **PowerShell** (Windows)។ ក្នុង Colab សូមប្រើប៊ូតុង **Terminal** នៅរបារខាងឆ្វេង។

| ខ្ញុំចង់… | macOS / Linux | Windows (PowerShell) |
|---|---|---|
| មើលថាខ្ញុំនៅក្នុង folder ណា | `pwd` | `pwd` |
| បង្ហាញបញ្ជី file នៅទីនេះ | `ls` | `ls` |
| ចូលទៅក្នុង folder មួយ | `cd foodbridge_kit` | `cd foodbridge_kit` |
| ឡើងទៅ folder ខាងលើមួយកម្រិត | `cd ..` | `cd ..` |
| ពិនិត្យថាខ្ញុំមាន Python កំណែណា | `python3 --version` | `py --version` |
| run កម្មវិធីមួយ | `python3 greet.py` | `py greet.py` |
| run កម្មវិធីមួយជាមួយ argument | `python3 greet.py Dara Sophea` | `py greet.py Dara Sophea` |
| run file តេស្តមួយ | `python3 test_shop_rules.py` | `py test_shop_rules.py` |
| run តេស្តទាំងអស់របស់ kit វគ្គសិក្សា | `python3 run_tests.py` | `py run_tests.py` |
| បញ្ឈប់កម្មវិធីដែលកំពុងដំណើរការ | **Ctrl+C** | **Ctrl+C** |
| ធ្វើ command មុនម្ដងទៀត | **↑** (ព្រួញឡើងលើ) | **↑** (ព្រួញឡើងលើ) |
| បំពេញឈ្មោះ file វែងៗ | **Tab** | **Tab** |

- ប្រសិនបើរកមិនឃើញ `python3` សូមសាកល្បង `python` ឬ `py -3`។
- Run កម្មវិធីពី folder ដែលមាន file នោះ ឬផ្ដល់ path របស់វា៖ `python3 programs/greet.py`។
- folder ដែលអ្នកកំពុងស្ថិតនៅ គឺសំខាន់៖ កម្មវិធីស្វែងរក file របស់វានៅក្នុង folder នោះ។
