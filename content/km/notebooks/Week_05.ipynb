```python
form = {"food": "Mango slices", "portions": "10"}   # គ្មាន key "donor"
form.get("donor", "")      # ផ្ដល់ "" ជំនួសឱ្យ KeyError
```

យើងឃើញថា តម្លៃនីមួយៗមកដល់ជាអត្ថបទ (text)។ បញ្ហាបន្ទាប់: អត្ថបទនោះផ្ទាល់ មិនស្អាតទេ — ដកឃ្លា និងអក្សរធំ ធ្វើឱ្យពាក្យតែមួយ មើលទៅដូចជាពាក្យពីរ។

### 2 · រៀបចំអត្ថបទឱ្យស្អាត: `strip()` និង `lower()`

**method** គឺជា function ដែលជារបស់តម្លៃមួយ។ អ្នកសរសេរវាបន្ទាប់ពីចំណុច (dot)។ `strip()` លុបដកឃ្លា (spaces) នៅចុងទាំងសងខាង។ `lower()` ប្ដូរអក្សរទាំងអស់ទៅជាអក្សរតូច។ ទាំងពីរបង្កើតអត្ថបទ **ថ្មី** មួយ។ អត្ថបទដើមមិនប្រែប្រួលទេ។

🤔 **ទាយមើល** លទ្ធផលនៃបន្ទាត់នីមួយៗ រួចចូរ Run cell។

````python
typed = "  Dara "                 # ឈ្មោះដែលវាយបញ្ចូលក្នុង login form
print(typed == "Dara")
print(typed.strip() == "Dara")
print(typed.strip().lower())
print(len(typed), len(typed.strip()))

CATEGORIES = ["bread", "rice meals", "fruit", "vegetables", "snacks"]
print(" Bread " in CATEGORIES)
print(" Bread ".strip().lower() in CATEGORIES)
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

ប្រសិនបើយើងមិនរៀបចំអត្ថបទឱ្យស្អាតទេ `" Bread "` និង `"bread"` ជាអត្ថបទពីរខុសគ្នា។ ប្រព័ន្ធនឹងគិតថា វាជាប្រភេទ (category) ពីរផ្សេងគ្នា។ បន្ទាប់ពីរៀបចំរួច ទាំងពីរក្លាយជា `"bread"` ហើយ `in` អាចរកឃើញវានៅក្នុង list។

method មួយទៀតសម្រាប់អត្ថបទ ដែលអ្នកនឹងជួបក្នុងសប្ដាហ៍នេះ៖ `", ".join(words)`។ វាភ្ជាប់ list នៃអត្ថបទ ឱ្យទៅជាអត្ថបទតែមួយ ដោយដាក់ `, ` នៅចន្លោះ។ ឧទាហរណ៍៖ `", ".join(["bread", "fruit"])` ផ្ដល់ `"bread, fruit"`។

យើងអាចរៀបចំអត្ថបទឱ្យស្អាត ដើម្បីឱ្យពាក្យដែលស្មើគ្នា ត្រូវគ្នា។ បញ្ហាបន្ទាប់: អត្ថបទស្អាត នៅតែជាអត្ថបទដដែល — `"30"` មិនទាន់ជាលេខ 30 ទេ។

### 3 · `int()` បម្លែងអត្ថបទទៅជាចំនួនគត់ ឬបរាជ័យ

អ្នកបានជួប `int()` រួចហើយ នៅសប្ដាហ៍ទី 1។ វាក៏មិនរាប់ដកឃ្លានៅចុងទាំងសងខាងដែរ។ ប៉ុន្តែ ប្រសិនបើអត្ថបទនោះមិនមែនជាចំនួនគត់ទេ វាធ្វើឱ្យកម្មវិធីកុំព្យូទ័រឈប់ ដោយមាន `ValueError`។

````python
print(int("30") + 4)
print(int(" 12 "))
print(int("-3"))
# print(int("many"))     # លុប # ចេញ ដើម្បីមើល error រួចដាក់ # នោះវិញ
````

🤔 តើតម្លៃណាខ្លះដែល `int()` អាចបម្លែងបាន៖ `"7"`, `"seven"`, `"7.5"`, `""`? ចូរសាកល្បងវានៅក្នុង cell ខាងលើ។

**ចម្លើយរបស់ខ្ញុំ:** …

សូមកត់សម្គាល់ថា `int("-3")` ដំណើរការបាន។ `-3` **ជា** ចំនួនគត់មែន ប៉ុន្តែវាមិនមែនជាចំនួនចំណែកដែលសមហេតុផលទេ។ ការបម្លែងគឺជាការពិនិត្យទីមួយ។ **ចន្លោះតម្លៃ (range)** គឺជាការពិនិត្យទីពីរ៖ តើវានៅចន្លោះដែនកំណត់ឬទេ?

យើងអាចបម្លែងអត្ថបទទៅជាលេខបាន នៅពេលអត្ថបទនោះត្រឹមត្រូវ។ បញ្ហាបន្ទាប់: ពេលខ្លះវាមិនត្រឹមត្រូវទេ — ហើយតម្លៃមិនត្រឹមត្រូវតែមួយ មិនគួរបញ្ឈប់កម្មវិធីកុំព្យូទ័រទាំងមូលទេ។

### 4 · `try` / `except ValueError`: ត្រៀមទុកសម្រាប់បញ្ហា

ប្រសិនបើអ្នក **រំពឹងទុក** ថានឹងមានបញ្ហា អ្នកអាចត្រៀមទុកសម្រាប់វាបាន។ Python ដំណើរការបន្ទាត់ដែលនៅក្រោម `try`។ ប្រសិនបើបន្ទាត់ណាមួយបង្ក (raise) `ValueError` Python នឹងលោតទៅ `except` ជំនួសឱ្យការឈប់។

````python
quantity_text = "two"           # វាយបញ្ចូលក្នុង order form របស់ហាងទូរស័ព្ទ
try:
    quantity = int(quantity_text)
    print("Order for", quantity, "phones")
except ValueError:
    print("Please type the quantity as a number, for example 2.")
````

🤔 ចូរប្ដូរ `"two"` ទៅជា `"2"` ហើយប៉ាន់ស្មានលទ្ធផល មុនពេលអ្នក Run។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

ទម្លាប់ល្អពីរ៖

- រក្សា `try` ឱ្យ **តូច**៖ ដាក់តែបន្ទាត់ដែលអាចបរាជ័យប៉ុណ្ណោះ (ការបម្លែង)។
- ចាប់ **តែ** `ValueError` ប៉ុណ្ណោះ ព្រោះវាជាបញ្ហាដែលអ្នករំពឹងទុក។ កំហុសផ្សេងទៀតក្នុងកូដរបស់អ្នក គួរតែនៅតែបង្ហាញ ដើម្បីឱ្យអ្នកអាចកែតម្រូវវាបាន។

យើងអាចឆ្លងកាត់តម្លៃមិនត្រឹមត្រូវមួយ ដោយមិនគាំង។ បញ្ហាបន្ទាប់: ជួរពិតមួយ អាចមានបញ្ហាច្រើនក្នុងពេលតែមួយ ហើយអ្នកបរិច្ចាគត្រូវដឹងបញ្ហាទាំងអស់នោះ ក្នុងពេលតែមួយ។

### 5 · list នៃសារ error

ចូរស្រមៃមើល form មួយ ដែលបង្ហាញតែបញ្ហាទីមួយប៉ុណ្ណោះ។ អ្នកបរិច្ចាគកែបញ្ហានោះ រួចផ្ញើម្ដងទៀត ហើយបែរជាទទួលបានបញ្ហាបន្ទាប់ទៀត។ នេះខាតពេល។ ទ្វារដែលល្អជាងនេះ ពិនិត្យ **គ្រប់យ៉ាង** ប្រមូលបញ្ហាទាំងអស់ទៅក្នុង list មួយ ហើយពន្យល់បញ្ហានីមួយៗជាពាក្យសាមញ្ញ។

````python
name = "  "
age_text = "seventeen"
errors = []

if name.strip() == "":
    errors.append("name is required.")
try:
    age = int(age_text)
except ValueError:
    errors.append("age must be a whole number, for example 19.")

print(errors)
print(len(errors), "problem(s)")
````

ចូរប្រៀបធៀបសាររបស់ Python ផ្ទាល់ `invalid literal for int() with base 10: 'many'` ជាមួយសារ `portions ត្រូវតែជាចំនួនគត់ ឧទាហរណ៍ 12។` អ្នកអានសារទាំងនេះគឺអ្នកបរិច្ចាគ មិនមែនអ្នកសរសេរកម្មវិធី (programmer) ទេ។ សារត្រូវប្រាប់ថា **អ្វី** ខុស និង **របៀប** កែវា។

យើងអាចប្រមូលបញ្ហាទាំងអស់ទៅក្នុង list មួយ។ បញ្ហាបន្ទាប់: អ្នកហៅ function ត្រូវការចម្លើយទាំងពីរក្នុងពេលតែមួយ — តម្លៃល្អ ឬ list នៃបញ្ហា។

### 6 · បញ្ជូនតម្លៃពីរត្រឡប់មកវិញ រួចបំបែកវា (unpack)

function មួយអាច return តម្លៃពីរក្នុងពេលតែមួយ៖ `return age, errors`។ អ្នកហៅ function **បំបែក (unpack)** តម្លៃទាំងនោះ ទៅក្នុងឈ្មោះពីរ៖ `age, errors = check_age("17")`។ តម្លៃទីមួយចូលទៅក្នុងឈ្មោះទីមួយ ហើយតម្លៃទីពីរចូលទៅក្នុងឈ្មោះទីពីរ។

````python
def check_age(age_text):
    """Return (age, errors). age is None when there is a problem."""
    try:
        age = int(age_text)
    except ValueError:
        return None, ["age must be a whole number, for example 19."]
    if age < 16:
        return None, ["volunteers must be 16 or older."]
    return age, []

age, errors = check_age("17")
print(age, errors)
````

🤔 **ទាយមើល** តើ `check_age("abc")` និង `check_age("12")` return អ្វី? បន្ទាប់មក ចូរបន្ថែមពីរបន្ទាត់ទៅក្នុង cell ខាងលើ ហើយពិនិត្យមើល។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

នេះគឺជាទម្រង់នៃផ្នែករបស់អ្នក ក្នុងសប្ដាហ៍នេះ៖ តម្លៃល្អមួយ ជាមួយ list ទទេ **ឬ** `None` ជាមួយ list នៃបញ្ហា។ មិនមានទាំងពីរក្នុងពេលតែមួយទេ។ `None` មានន័យថា "គ្មានសំណើណាមួយត្រូវបានបង្កើតទេ"។ ដូច្នេះ គ្មានអ្វីនៅក្នុងប្រព័ន្ធប្រែប្រួលឡើយ។ នេះហើយគឺផ្នែកទាំងបីនៃគោលការណ៍ទី 5៖ ពិនិត្យនៅមាត់ទ្វារ ពន្យល់ជាពាក្យសាមញ្ញ និងមិនប្ដូរអ្វីទាំងអស់។

វាគឺជាទម្រង់តម្លៃពីរពីផ្នែកទី 6 សម្រាប់ field តែមួយប៉ុណ្ណោះ — មិនមាន field ផ្សេង និងមិនមាន loop ពិនិត្យ field ដែលត្រូវបំពេញទេ។ ខាងក្រោមនេះ គឺជាការពិនិត្យនោះ:

````python
def clean_hours_left(text):
    """Return (hours, errors) for one hours_left text."""
    try:
        hours = int(text)
    except ValueError:
        return None, ["hours_left must be a whole number of hours."]
    if hours < 1:
        return None, ["too late to collect: hours_left must be at least 1."]
    return hours, []

print(clean_hours_left("0"))
````

**របៀបដែលវាដំណើរការ:** `int("0")` ជោគជ័យ ហើយផ្ដល់ `0` ដូច្នេះគ្មាន `ValueError` កើតឡើងទេ។ បន្ទាប់មក `0 < 1` គឺ `True` ដូច្នេះ function return `(None, ['too late to collect: hours_left must be at least 1.'])` — គ្មានសំណើ មានតែសារសាមញ្ញមួយ។ ជាមួយ `"3"` កូដដដែលនេះ return `(3, [])`។

នៅវគ្គទី 3 អ្នកនឹងសរសេរលំនាំដដែលនេះ សម្រាប់ជួរទាំងមូល នៅក្នុង `clean_offer`: field ប្រាំ ការពិនិត្យទាំងអស់ list នៃសារ error តែមួយ។

### ជំហានទី 2 · `clean_offer`

ផ្នែកភាគច្រើននៃ function នេះ ត្រូវបានសរសេររួចសម្រាប់អ្នក។ ចូរបំពេញ `TODO` ទាំងបី។ ចូរអាន comment ឱ្យបានល្អ៖ វាប្រាប់អ្នកយ៉ាងច្បាស់ថា ផ្នែកនីមួយៗត្រូវធ្វើអ្វី។

````python
def clean_offer(raw):
    """Check one raw row at the door.

    Return (offer, []) for a good row, or (None, errors) for a bad row.
    """
    errors = []

    # TODO 2: រៀបចំអត្ថបទឱ្យស្អាត។ បន្ថែម .strip() នៅបន្ទាត់ទាំងបួន ហើយបន្ថែម .lower() សម្រាប់ category ផង។
    donor = raw.get("donor", "")
    contact = raw.get("contact", "")
    food = raw.get("food", "")
    category = raw.get("category", "")

    # field ដែលត្រូវតែបំពេញ មិនអាចទទេបានទេ (សរសេររួចសម្រាប់អ្នក ដូច P3)។
    for field in REQUIRED_FIELDS:
        if raw.get(field, "").strip() == "":
            errors.append(field + " is required.")

    # លេខ: helper របស់អ្នកបម្លែងអត្ថបទទៅជា int ឬ None។
    portions = to_whole_number(raw.get("portions", ""))
    hours_left = to_whole_number(raw.get("hours_left", ""))

    # TODO 3: ពិនិត្យ range។ បន្ថែម elif មួយនៅក្រោម if នីមួយៗ:
    #   portions តិចជាង MIN_PORTIONS ឬលើស MAX_PORTIONS
    #       -> "portions must be between ... and ..."
    #   hours_left តិចជាង REJECT_IF_HOURS_LEFT_BELOW
    #       -> "too late to collect: ... hours left, the minimum is ..."
    if portions is None:
        errors.append("portions must be a whole number, for example 12.")
    if hours_left is None:
        errors.append("hours_left must be a whole number of hours.")

    # TODO 4: ប្រសិនបើ category មិននៅក្នុង CATEGORIES ចូរបន្ថែម
    #   "category must be one of: bread, rice meals, ..."  (ជំនួយ: ", ".join(CATEGORIES))

    if errors:
        return None, errors        # កុំប្ដូរអ្វីទាំងអស់: គ្មានសំណើសោះ
    offer = {"donor": donor, "contact": contact, "food": food, "category": category,
             "portions": portions, "hours_left": hours_left}
    return offer, []
````

ចូរសាកល្បងវាលើជួរពីរ៖ ជួរមួយរញ៉េរញ៉ៃ ប៉ុន្តែត្រឹមត្រូវ និងជួរមួយទៀតមានបញ្ហាបី។

````python
messy = {"donor": " Lotus Cafe ", "contact": "lotus@example.org", "food": "Steamed buns",
         "category": " Snacks ", "portions": " 15 ", "hours_left": "3"}
print(clean_offer(messy))

bad = {"donor": "Lotus Cafe", "food": "Soup", "category": "soup",
       "portions": "-3", "hours_left": "soon"}
print(clean_offer(bad))
````

នៅពេលអ្នកធ្វើរួច បន្ទាត់ទីមួយបង្ហាញ dictionary មួយ ដែលមាន `'category': 'snacks'` និង `'portions': 15` (គ្មានសញ្ញាសម្រង់ ព្រោះវាជាលេខ) រួចបន្ទាប់មកគឺ `[]`។ បន្ទាត់ទីពីរបង្ហាញ `None` និងសារ **បី**។

### ពិនិត្យវា

កម្មវិធីពិនិត្យ (checker) សាកល្បងជួរត្រឹមត្រូវមួយ ដែលមានដកឃ្លាលើស។ បន្ទាប់មក វាសាកល្បងតម្លៃមិនត្រឹមត្រូវ ម្ដងមួយៗ។ វាប្រើដែនកំណត់ **របស់អ្នក** ដូច្នេះវានៅតែដំណើរការបាន បន្ទាប់ពីអ្នកប្ដូរដែនកំណត់ទាំងនោះ។

````python
MY_SETTINGS = {
    "MIN_PORTIONS": MIN_PORTIONS,
    "MAX_PORTIONS": MAX_PORTIONS,
    "REJECT_IF_HOURS_LEFT_BELOW": REJECT_IF_HOURS_LEFT_BELOW,
    "CATEGORIES": CATEGORIES,
    "REQUIRED_FIELDS": REQUIRED_FIELDS,
}
bridge.check("clean_offer", clean_offer, MY_SETTINGS)
````

### ដាក់បញ្ចូលវាទៅក្នុងប្រព័ន្ធ

ឥឡូវនេះ ទ្វាររបស់អ្នកការពារប្រព័ន្ធពិត។ ចូរមើលជំហានទី 1 នៃថ្ងៃ៖ បន្ទាត់ 🚫 នីមួយៗ គឺជាជួរដែល `clean_offer` **របស់អ្នក** បដិសេធ ដោយប្រើសារ **របស់អ្នក**។

````python
my_system = bridge.System(MY_SETTINGS, hooks={"clean_offer": clean_offer})
my_system.run_day()
````

ការបដិសេធ មិនត្រូវប្ដូរអ្វីទាំងអស់។ ចូរសាកល្បងវា៖ បញ្ចូលសំណើមិនត្រឹមត្រូវមួយ ហើយប្រៀបធៀប record មុន និងក្រោយ។ បន្ទាប់មក ចូរបញ្ចូលសំណើដែលរញ៉េរញ៉ៃ ប៉ុន្តែត្រឹមត្រូវ ពីខាងលើ។

````python
before = my_system.records()
offer_id, errors = my_system.add_offer({"donor": "Hotel Kitchen", "contact": "hotel@example.org",
                                        "food": "Steamed rice", "category": "Rice Meals",
                                        "portions": "-3", "hours_left": "8"})
print("Bad offer →", offer_id, errors)
print("Records unchanged:", my_system.records() == before)

offer_id, errors = my_system.add_offer(messy)
print("Messy but good offer →", offer_id, errors)
for record in my_system.records():
    print("  ", record["id"], "·", record["food"], "·", record["category"], "·", record["portions"], "portions")
````

### 🧭 ការសម្រេចចិត្តរបស់អ្នក · តើទ្វារទទួលយកអ្វីខ្លះ?

អ្នកសម្របសម្រួលស្នើឱ្យអ្នកកំណត់ដែនកំណត់របស់ទ្វារ។ ដែនកំណត់នីមួយៗការពារមនុស្សមួយចំនួន ហើយបដិសេធមនុស្សមួយចំនួនទៀត។

- **`MIN_PORTIONS` / `MAX_PORTIONS`:** តើសំណើ 1 ចំណែក សមនឹងការធ្វើដំណើររបស់អ្នកស្ម័គ្រចិត្តម្នាក់ដែរឬទេ? តើមានអង្គការដៃគូណាមួយអាចទទួល 200 ចំណែកបានដែរឬទេ?
- **`REJECT_IF_HOURS_LEFT_BELOW`:** ប្រសិនបើនៅសល់តែ 1 ម៉ោង តើអ្នកស្ម័គ្រចិត្តនៅតែអាចទៅដល់ទាន់ដែរឬទេ?
- **`REQUIRED_FIELDS`:** តើ field ណាខ្លះដែលអ្នកបរិច្ចាគត្រូវបំពេញ? ឧទាហរណ៍៖ **តើ `contact` គួរជា field ដែលត្រូវបំពេញឬទេ?** ប្រសិនបើមានព័ត៌មានទំនាក់ទំនង ក្រុមការងារអាចរៀបចំការទៅយក និងទូរស័ព្ទទៅ នៅពេលមានបញ្ហា។ ប៉ុន្តែពេលនោះ ប្រព័ន្ធនឹងរក្សាទុកទិន្នន័យផ្ទាល់ខ្លួនច្រើនជាងមុន ដែលវាត្រូវការពារ (សប្ដាហ៍ទី 8)។ អ្នកបរិច្ចាគតូចៗខ្លះ អាចឈប់ផ្ដល់អាហារ។

1. ចូរប្ដូរតម្លៃថេរ (constant) នៅក្នុង **ដែនកំណត់របស់អ្នក** ខាងលើ។ អ្នកអាចរក្សាតម្លៃ default ដដែលបាន ប្រសិនបើអ្នកអាចពន្យល់ពីមូលហេតុ។
2. Run ម្ដងទៀត ចាប់ពី cell នោះ ចុះមកដល់ទីនេះ។ ការពិនិត្យនៅតែជាប់ ព្រោះវាប្រើដែនកំណត់របស់អ្នក។
3. ចូរ Run cell ខាងក្រោម ដើម្បីមើលថា ទ្វាររបស់អ្នកធ្វើអ្វីខ្លះជាមួយថ្ងៃគំរូ។

````python
for row in bridge.validation.read_offers_csv("foodbridge_kit/data/offers.csv"):
    offer, errors = clean_offer(row)
    if errors:
        print("🚫", row["food"], "→", " ".join(errors))
    else:
        print("✅", offer["food"], "·", offer["portions"], "portions ·", offer["hours_left"], "hours left")
````
