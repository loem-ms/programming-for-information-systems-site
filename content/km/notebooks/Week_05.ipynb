### 1 · តម្លៃនីមួយៗពី file គឺជាអត្ថបទ (text)

អ្នកបរិច្ចាគផ្ញើការផ្ដល់ជូននៅក្នុង file CSV: តារាងមួយដែលរក្សាទុកជាអត្ថបទធម្មតា។ cell ខាងក្រោមអាន file ថ្ងៃនេះ `offers.csv` ដោយប្រើរូបមន្តខ្លី 2 បន្ទាត់ (recipe)៖ `csv.DictReader` ប្ដូរជួរនីមួយៗទៅជា dictionary មួយ (ឈ្មោះជួរឈរ គឺជា key) ហើយ `list(...)` រក្សាជួរទាំងអស់ទុក។ ចូរប្រើវាតាមដែលវាមាន។ `with open` នឹងពន្យល់នៅសប្ដាហ៍ទី 7។

🤔 **ទាយមើល** បន្ទាត់ពីរចុងក្រោយដែលនឹង print មុនពេលអ្នក Run cell។

````python
import csv

with open("foodbridge_kit/data/offers.csv", encoding="utf-8") as file:   # រូបមន្ត (recipe)
    rows = list(csv.DictReader(file))

for row in rows:
    print(row)
first = rows[0]
print(first["portions"] + first["hours_left"])     # 🤔
print(type(first["portions"]))
````

អ្នកបានសួរជំនួយការ AI មួយថា: *"សូមសរសេរកូដ Python ដែលបន្ថែមជួរនៃ file របស់អ្នកបរិច្ចាគ ទៅក្នុងការផ្ដល់ជូនរបស់យើង។"* **Version A** គឺជាចម្លើយរបស់វា ជាមួយកំណត់ចំណាំថា *"វាមិនដែលគាំងទេ!"* **Version B** មកពីមិត្តរួមក្រុមម្នាក់។ ទាំងពីរដំណើរការលើជួរទាំង 9។ ចូរអានទាំងពីរ រួច Run cell នោះ។

````python
# Version A · ពីជំនួយការ AI: "វាមិនដែលគាំងទេ!"
def add_offer_a(raw, saved):
    offer = {"donor": raw["donor"], "food": raw["food"]}
    saved.append(offer)                        # រក្សាទុកភ្លាមៗ
    try:
        offer["portions"] = int(raw["portions"])
    except:
        offer["portions"] = 0                  # មិនគាំងដាច់ខាត
    if raw["category"] not in CATEGORIES:
        return "error"
    offer["category"] = raw["category"]
    return "OK"


# Version B · ពីមិត្តរួមក្រុមម្នាក់
def add_offer_b(raw, saved):
    errors = []
    donor = raw.get("donor", "").strip()
    category = raw.get("category", "").strip().lower()
    if donor == "":
        errors.append("donor is required.")
    if category not in CATEGORIES:
        errors.append("category must be one of: " + ", ".join(CATEGORIES) + ".")
    try:
        portions = int(raw.get("portions", ""))
        if portions < 1 or portions > 200:
            errors.append("portions must be between 1 and 200.")
    except ValueError:
        errors.append("portions must be a whole number.")
    if errors:
        return "refused: " + " ".join(errors)
    saved.append({"donor": donor, "food": raw.get("food", "").strip(),
                  "category": category, "portions": portions})
    return "OK"


saved_a = []
saved_b = []
for row in rows:
    print(row["food"], "→ A:", add_offer_a(row, saved_a), "| B:", add_offer_b(row, saved_b))
print()
print("A saved", len(saved_a), "offers:")
for offer in saved_a:
    print("  ", offer)
print("B saved", len(saved_b), "offers.")
````

### 📊 តើច្រកពិនិត្យរបស់អ្នកនឹងបដិសេធការផ្ដល់ជូនប៉ុន្មាន?

ដែនកំណត់មួយនៅតែជាការស្មាន រហូតដល់អ្នកសាកវាលើទិន្នន័យ: `semester.csv` មានការផ្ដល់ជូនទាំង 404 កាលពីឆមាសមុន។ `count_below` របស់អ្នករាប់ការផ្ដល់ជូនដែលតិចជាង `LIMIT`។ ក្រាហ្វបង្ហាញម៉ោងនៅសល់ នៅពេលការផ្ដល់ជូននីមួយៗមកដល់។ វាជា histogram (ក្រាហ្វរបារដែលរាប់តម្លៃក្នុងចន្លោះនីមួយៗ)៖ របារនីមួយៗ គឺជាចន្លោះ (bin) 1 ម៉ោង។ ចូរបំពេញចំណងជើង (title) និងស្លាក (label) នៃអ័ក្សទាំងពីរ **ជាភាសាអង់គ្លេស** រួចគូរដែនកំណត់របស់អ្នក។

*ហេតុអ្វីជាភាសាអង់គ្លេស? នៅក្នុង Colab ក្រាហ្វ matplotlib បង្ហាញអក្សរខ្មែរជាប្រអប់ទទេ (□) ដូច្នេះចំណងជើង ស្លាក និងលេខលើអ័ក្ស ត្រូវសរសេរជាភាសាអង់គ្លេស។*

````python
import csv

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:   # រូបមន្តដដែល (recipe)
    semester = list(csv.DictReader(file))

LIMIT = 3          # 🧭 REJECT_IF_HOURS_LEFT_BELOW មួយដែលអ្នកកំពុងគិត
refused = count_below(semester, "hours_left", LIMIT)       # ✍️ សាកសរសេរ របស់អ្នកពីវគ្គទី 1
if refused is None:
    print("Finish count_below in Session 1 to see the count.")
else:
    print("Limit", LIMIT, "→ the door refuses", refused, "of", len(semester), "offers",
          "(" + str(round(100 * refused / len(semester))) + "%).")

hours = []
for row in semester:
    hours.append(int(row["hours_left"]))
try:
    import matplotlib.pyplot as plt
    plt.figure(figsize=(8, 3))
    plt.hist(hours, bins=range(0, 68))
    plt.title("TODO: what does the chart show?")
    plt.xlabel("TODO")
    plt.ylabel("TODO")
    # TODO: គូរដែនកំណត់របស់អ្នក: plt.axvline(LIMIT, color="red")
    plt.show()
except ImportError:
    print("matplotlib is missing. Colab has it; on your own computer: pip install matplotlib")
````

ដែនកំណត់ 3 បដិសេធ 60 ក្នុងចំណោម 404 ការផ្ដល់ជូន (15%), ដែនកំណត់ 2 បដិសេធ 15 (4%), ហើយតម្លៃ default គឺ 1 មិនបដិសេធអ្វីទាំងអស់។ ការផ្ដល់ជូននីមួយៗដែលត្រូវបដិសេធ គឺជាអាហារដែលអ្នកបរិច្ចាគចង់ផ្ដល់ឱ្យ ហើយការផ្ដល់ជូននីមួយៗដែលត្រូវទទួលយក អាចជាអាហារដែលគ្មាននរណាទៅយកទាន់ពេល។

2. **តើប្រើ ID អ្វីខ្លះ?** ឧទាហរណ៍ លេខគណនី ឬអត្តលេខនិស្សិត (student ID)។ តើហេតុអ្វីបានជា record នីមួយៗត្រូវការ ID មួយ?
3. **តើមានអ្វីកើតឡើង ពេលមានការកែតម្រូវ?** តើតម្លៃចាស់ត្រូវបានរក្សាទុក ត្រូវបានជំនួស ឬត្រូវបានបង្ហាញថាជាការកែតម្រូវ? តើអ្នកដឹងដោយរបៀបណា?
4. **តើនរណាត្រូវការប្រវត្តិនេះ** ហើយហេតុអ្វី?

ចូរពិពណ៌នាតែប្រភេទទិន្នន័យប៉ុណ្ណោះ កុំចម្លងលេខពិត ឬព័ត៌មានផ្ទាល់ខ្លួនរបស់អ្នក។ ចូរប្រើប្រភពយ៉ាងហោចណាស់មួយ ហើយប្រាប់ឈ្មោះប្រភពនោះ។ ប្រសិនបើអ្នកប្រើជំនួយការ AI ចូរបង្ហាញកន្លែងមួយ ដែលអ្នកបានពិនិត្យ ឬកែតម្រូវវា។ ចូរយកឧទាហរណ៍របស់អ្នក មកចែករំលែកនៅ **ចាប់ផ្ដើមថ្នាក់ សប្ដាហ៍ទី 6**។ ដាក់ស្នើ (submit) តាម email ទៅកាន់ submit.progmis@gmail.com។ ថ្ងៃផុតកំណត់ និងឈ្មោះ file នឹងត្រូវប្រកាសនៅក្នុងថ្នាក់។

**សប្ដាហ៍ក្រោយ:** ការផ្ដល់ជូនដែលស្អាតហើយ នៅតែឆ្លងកាត់ដំណាក់កាលជាបន្តបន្ទាប់ (ថ្មី → បានអនុម័ត → បានទៅយក → បានប្រគល់) ហើយប្រព័ន្ធរបស់អ្នកត្រូវអនុញ្ញាតតែការប្ដូរដែលសមហេតុផលប៉ុណ្ណោះ។ អ្នកនឹងបន្ថែម `my_foodbridge/lifecycle.py`។
