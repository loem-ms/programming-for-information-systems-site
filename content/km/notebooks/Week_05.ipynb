### នៅចុងសប្ដាហ៍នេះ អ្នកនឹងអាច …

- **ពន្យល់** ថាហេតុអ្វីបានជាគ្រប់តម្លៃដែលមកពី file សុទ្ធតែជាអត្ថបទ (text) ហើយសម្អាតវាដោយប្រើ `strip()` និង `lower()`;
- **អាន** traceback មួយ ហើយ **ចាប់ (catch)** `ValueError` ដោយប្រើ `try` / `except`;
- **សរសេរ** ការពិនិត្យមួយ ដែលប្រមូលគ្រប់បញ្ហា ហើយ return តម្លៃពីរ៖ `(offer, [])` ឬ `(None, errors)`;
- **សម្រេចចិត្ត** ថាតើច្រកពិនិត្យរបស់អ្នកទទួលយកអ្វីខ្លះ ដោយមានលេខមួយពីទិន្នន័យឆមាសមុន ជាភស្តុតាង (evidence)។

### ដំណើររឿង

នៅសប្ដាហ៍ទី 4 អ្នកបានចាប់ផ្ដើម folder ប្រព័ន្ធផ្ទាល់ខ្លួនរបស់អ្នក គឺ `my_foodbridge/` ដែលមាន `rules.py` និង `records.py`។ function របស់អ្នករក្សាការសន្យារបស់វា នៅពេលធាតុចូលស្អាត។

ព្រឹកនេះ file របស់អ្នកបរិច្ចាគមកដល់ ប៉ុន្តែវាមិនស្អាតទេ។ ជួរមួយសរសេរថា ចំនួនចំណែកគឺ `many`។ ជួរមួយទៀតមានប្រភេទ (category) `soup` ដែលគ្មានអង្គការដៃគូណាទទួលយកទេ។ ជួរទីបីគ្មានឈ្មោះអ្នកបរិច្ចាគ ដូច្នេះគ្មាននរណាដឹងថា ត្រូវទៅយក mango slices (ចំណិតស្វាយ) នៅឯណាទេ។ សប្ដាហ៍នេះ អ្នកនឹងបន្ថែម file ទីបី គឺ `data_model.py` ដែលមាន `clean_offer`៖ ការពិនិត្យទិន្នន័យមុនទទួលចូលប្រព័ន្ធ។

## ចាប់ផ្ដើមថ្នាក់ · ចែករំលែកកិច្ចការស្រាវជ្រាវទី 3 (10 នាទី)

សម្រាប់កិច្ចការស្រាវជ្រាវទី 3 អ្នកបានអានអំពី **បញ្ហា (error) និង exception** (Python Basics និងជំពូកមួយពីសៀវភៅ) បានសួរ AI អំពីចំណុចដែលនៅមិនទាន់ច្បាស់ បានសាកល្បងការអះអាង (claim) មួយក្នុង Colab ហើយបានសរសេរមតិរបស់អ្នក លើសំណួរ *"តើកម្មវិធីកុំព្យូទ័រគួរគាំង ឬបន្តដំណើរការដោយស្មានយកចម្លើយមួយ?"*។ ចូរធ្វើការជាក្រុមបីនាក់៖

2. ចូរបង្ហាញការអះអាងដែលអ្នកបានសាកល្បង និងឧទាហរណ៍តូចរបស់អ្នក។ តើការអះអាងនោះត្រឹមត្រូវឬទេ?
3. **ជ្រើសរើសខាងមួយ។** គាំង ឬបន្តដំណើរការដោយស្មានយកចម្លើយ? ចូររកមនុស្សម្នាក់ក្នុងក្រុមរបស់អ្នក ដែលមានមតិខុសពីអ្នកខ្លាំងជាងគេ។ ម្នាក់ៗផ្ដល់មូលហេតុមួយ ក្នុងប្រយោគតែមួយ។

ចូររក្សាចម្លើយរបស់អ្នកចំពោះសំណួរទី 3 ទុក។ នៅវគ្គទី 1 អ្នកនឹងឃើញពីរបៀបដែល Python អនុញ្ញាតឱ្យអ្នកជ្រើសរើស៖ `try` / `except`។

## វគ្គទី 1 · ស្វែងយល់ (90 នាទី)

### គំនិតសំខាន់: គុណភាពទិន្នន័យ (data quality) នៅព្រំដែនរបស់ប្រព័ន្ធ

នៅព្រលានយន្តហោះ បុគ្គលិកពិនិត្យលិខិតឆ្លងដែនរបស់អ្នក **មុនពេល** អ្នកឡើងយន្តហោះ មិនមែនក្រោយពេលយន្តហោះហោះឡើងទេ។ ប្រព័ន្ធព័ត៌មានក៏មានច្រកដែរ៖ ទម្រង់បែបបទ (form) មួយ file មួយ ឬសារមួយពី app ផ្សេង។ ច្រក គឺជាកន្លែងដែលចំណាយតិចបំផុត ក្នុងការទប់ស្កាត់ទិន្នន័យខុស។ នៅពេលតម្លៃខុសមួយចូលដល់ខាងក្នុងហើយ វានឹងរាលដាលទៅក្នុងផលសរុប អេក្រង់ និងការសម្រេចចិត្ត ហើយមនុស្សនឹងធ្វើសកម្មភាពតាមវា។ មនុស្សវាយ `many` វាយដកឃ្លាលើស ឬទុក field ចោលទទេ។ ប្រព័ន្ធល្អរំពឹងទុករឿងនេះជាមុន។ វាពិនិត្យគ្រប់តម្លៃនៅច្រក ពន្យល់បញ្ហានីមួយៗជាពាក្យសាមញ្ញ ហើយមិនរក្សាទុកអ្វីទាំងអស់ រហូតដល់ជួរនោះត្រឹមត្រូវ។ នោះគឺជា **គោលការណ៍ទី 5: កុំទុកចិត្ត input។**

```text
  a form  ──┐
  a file  ──┼──►  🚪 the door: check every value  ──►  your records  ──►  totals, screens, decisions
  an app  ──┘            │
                         └──►  refused, with a reason the person can act on
```

**តើ Python អានវាយ៉ាងដូចម្ដេច?** រូបមន្តខ្លី 2 បន្ទាត់ (recipe) ដែលប្រើ `csv.DictReader` ប្ដូរបន្ទាត់នីមួយៗទៅជា dictionary មួយ ហើយបន្ទាត់ទីមួយនៃ file ផ្ដល់ key ទាំងអស់។ file រក្សាទុកតែតួអក្សរដែលមនុស្សបានវាយប៉ុណ្ណោះ។ វាមិនស្គាល់អ្វីដែលហៅថាលេខទេ។ ដូច្នេះ គ្រប់តម្លៃមកដល់ជា `str` សូម្បីតែ `30` ក៏ដោយ។

### 2 · method បង្កើតអត្ថបទថ្មី: `strip()` និង `lower()`

📖 **វាជាអ្វី?** **method** គឺជា function ដែលភ្ជាប់ជាមួយតម្លៃមួយ។ អ្នកសរសេរវានៅក្រោយសញ្ញាចុច (dot)៖ `typed.strip()`។ អត្ថបទ (text) មាន method ជាច្រើន។ `strip()` លុបដកឃ្លា (spaces) នៅចុងទាំងសងខាង។ `lower()` ប្ដូរគ្រប់អក្សរទៅជាអក្សរតូច។

**ហេតុអ្វីបានជាត្រូវការវា?** មនុស្សតែងវាយដកឃ្លាលើស ហើយលាយអក្សរធំ និងអក្សរតូចចូលគ្នា៖ `"Bread"`, `" bread"`, `"BREAD"`។ កុំព្យូទ័រប្រៀបធៀបអត្ថបទម្ដងមួយអក្សរៗ ដូច្នេះ `"Bread" == "bread"` គឺ `False`។ ត្រូវសម្អាតអត្ថបទជាមុនសិន រួចសឹមប្រៀបធៀប។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** អត្ថបទក្នុង Python **មិនដែលត្រូវបានកែប្រែដោយផ្ទាល់ (in place) ទេ**។ method មួយអានអត្ថបទនោះ ហើយបង្កើតអត្ថបទ **ថ្មី**៖ អត្ថបទដើមនៅដដែលទាំងស្រុង។ ដើម្បីរក្សាអត្ថបទថ្មីទុក ចូរដាក់ឈ្មោះឱ្យវា៖ `donor = donor.strip()`។ អ្នកអាចសរសេរ method តៗគ្នាជាខ្សែ (chain) បាន៖ ក្នុង `" Bread ".strip().lower()` សញ្ញាចុចនីមួយៗធ្វើការលើលទ្ធផលនៃជំហានមុន។

**តើអ្នកជួបវានៅកន្លែងណា?** form ចូលគណនី (login form) លុបដកឃ្លានៅជុំវិញអាសយដ្ឋាន email របស់អ្នក។ ប្រអប់ស្វែងរក (search box) មិនខ្វល់ថាអ្នកវាយអក្សរធំ ឬអក្សរតូចទេ។

| កូដ | លទ្ធផល (អត្ថបទថ្មី) | អត្ថបទដើម |
|---|---|---|
| `"  Dara ".strip()` | `"Dara"` | នៅតែជា `"  Dara "` |
| `"Rice Meals".lower()` | `"rice meals"` | នៅតែជា `"Rice Meals"` |
| `" Bread ".strip().lower()` | `"bread"` | នៅតែជា `" Bread "` |

### 3 · Exception: នៅពេល Python មិនអាចបន្តទៅមុខទៀតបាន

📖 **វាជាអ្វី?** `int()` ប្ដូរអត្ថបទទៅជាចំនួនគត់៖ `int("30")` ផ្ដល់ `30` ហើយ `int(" 12 ")` ផ្ដល់ `12` (វាមិនខ្វល់ពីដកឃ្លានៅចុងទាំងសងខាងទេ)។ នៅពេលអត្ថបទនោះមិនមែនជាចំនួនគត់ `int()` គ្មានចម្លើយត្រឹមត្រូវណាមួយដើម្បីផ្ដល់ឱ្យទេ។ ដូច្នេះ វា **បង្ក (raise) exception មួយ**៖ សញ្ញាមួយដែលប្រាប់ថាមានអ្វីខុស ដោយមាន type មួយ ដូចជា `ValueError` និងសារមួយ។ `int("-3")` ក៏ដំណើរការដែរ ព្រោះលេខអវិជ្ជមានក៏ជាចំនួនគត់ដែរ។ ចំណែកការពិនិត្យថា តើលេខមួយត្រូវនឹងវិធានរបស់អ្នកឬទេ (យ៉ាងហោចណាស់ 1000 រៀល ឬពី 1 ដល់ 200 ចំណែក) គឺជាជំហានមួយទៀតដាច់ដោយឡែក។

**ហេតុអ្វីបានជាត្រូវការវា?** ចម្លើយខុសដោយស្ងាត់ៗ ដូចជា 0 នឹងរបូតចូលទៅក្នុងផលសរុបរបស់អ្នក។ ការឈប់ភ្លាមៗ ដោយបង្ហាញសារយ៉ាងច្បាស់ មានសុវត្ថិភាពជាង៖ គ្រប់គ្នាមើលឃើញវា។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** exception ចាប់ផ្ដើមនៅបន្ទាត់ដែលបានបង្កវា។ Python បញ្ឈប់ function នោះភ្លាមៗ ហើយ exception **ធ្វើដំណើរឡើងទៅលើ**៖ ដំបូងវាត្រឡប់ទៅបន្ទាត់ដែលបានហៅ function នោះ បន្ទាប់មកទៅបន្ទាត់ដែលបានហៅ function ខាងក្រៅនោះទៀត ហើយបន្តបែបនេះ។ ប្រសិនបើគ្មានអ្វី **ចាប់ (catch)** វានៅតាមផ្លូវទេ កម្មវិធីនឹងឈប់ ហើយ Python print **traceback** មួយ៖ ផ្លូវដែល exception បានធ្វើដំណើរកាត់។

| `ValueError` | type ត្រឹមត្រូវ ប៉ុន្តែតម្លៃគ្មានន័យ | `int("many")` |
| `KeyError` | dictionary មួយគ្មាន key នោះទេ | `row["phone"]` |
| `NameError` | ឈ្មោះមួយមិនមានទេ (ជាញឹកញាប់ដោយសារវាយអក្សរខុស (typo)) | `print(totl)` |
| `TypeError` | type ខុស សម្រាប់ប្រមាណវិធី (operation) មួយ | `"30" + 5` |

🤔 ដំបូង ចូរ **ទាយមើល** លទ្ធផលជាមួយ `"18"`។ បន្ទាប់មក ចូរប្ដូរ `"18"` ទៅជា `"many"` Run cell ម្ដងទៀត ហើយអាន traceback ដោយប្រើតារាងខាងក្រោម (①②③ បង្ហាញផ្លូវរបស់វា)។ ក្រោយមក ចូរប្ដូរវាត្រឡប់ទៅ `"18"` វិញ។

````python
def read_portions(text):
    return int(text)                          # ① ជាមួយ "many" exception ចាប់ផ្ដើមនៅទីនេះ

def total_portions(texts):
    total = 0
    for text in texts:
        total = total + read_portions(text)   # ② ... ធ្វើដំណើរឡើងមកដល់ទីនេះ រួចចេញទៅក្រៅ
    return total

print(total_portions(["30", "18"]))           # ③ ... ហើយឡើងមកដល់ទីនេះ
````

### 4 · `try` / `except ValueError`: ចាប់វា ហើយសម្រេចចិត្ត

📖 **វាជាអ្វី?** `try` / `except` ចាប់ exception មួយដែលអ្នករំពឹងទុក ខណៈពេលដែលវាកំពុងធ្វើដំណើរឡើងទៅលើ ដើម្បីឱ្យកម្មវិធីរបស់អ្នកអាចសម្រេចចិត្តថាត្រូវធ្វើអ្វី ជំនួសឱ្យការឈប់។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** Python ដំណើរការបន្ទាត់ដែលនៅក្រោម `try` ម្ដងមួយៗ។ ប្រសិនបើបន្ទាត់ទាំងអស់ដំណើរការបាន វារំលង `except`។ ប្រសិនបើបន្ទាត់ណាមួយបង្ក `ValueError` Python ចាកចេញពីក្រុមបន្ទាត់ដែលចូលបន្ទាត់ (block) ក្រោម `try` ភ្លាមៗ៖ បន្ទាត់ដែលនៅបន្ទាប់ពីវា មិនដំណើរការទេ។ Python លោតទៅ `except ValueError:` ដំណើរការបន្ទាត់នៅទីនោះ ហើយបន្ទាប់មកបន្តទៅបន្ទាត់ដែលនៅក្រោយ `try` / `except` ទាំងមូល។ exception ដែលមាន type ផ្សេង មិនត្រូវបានចាប់ទេ៖ វានៅតែបន្តធ្វើដំណើរ។

**ហេតុអ្វីបានជាត្រូវការវា?** នៅក្នុងប្រព័ន្ធព័ត៌មាន ធាតុចូលខុស គឺជារឿងធម្មតា មិនមែនជារឿងកម្រទេ។ មនុស្សដែលបានវាយវា ត្រូវការសារច្បាស់លាស់មួយ ហើយមនុស្សផ្សេងទៀតទាំងអស់ ត្រូវការឱ្យកម្មវិធីនៅតែបន្តដំណើរការ។

**តើអ្នកជួបវានៅកន្លែងណា?** app បញ្ចូលលុយទូរស័ព្ទ (top-up) ដែលប្រាប់ថា *please type a number* (សូមវាយជាលេខ)។ web server (ម៉ាស៊ីនមេ) មួយ ដែលប្ដូរការស្នើសុំ (request) ខុសមួយ ទៅជាទំព័រ error មួយ ខណៈពេលដែលគេហទំព័រនៅតែដំណើរការសម្រាប់អ្នកផ្សេងទៀតទាំងអស់។

````python
quantity_text = "two"                  # វាយបញ្ចូលក្នុង form កុម្ម៉ង់ទំនិញរបស់ហាងទូរស័ព្ទ
try:
    quantity = int(quantity_text)
    print("Order for", quantity, "phones")
except ValueError:
    print("Please type the quantity as a number, for example 2.")
print("The shop is still open.")
````

#### តម្លៃពីរត្រឡប់មកវិញ ជា tuple មួយ

📖 ការពិនិត្យមួយ ច្រើនតែត្រូវបញ្ជូនចម្លើយពីរត្រឡប់មកវិញ៖ តម្លៃដែលស្អាត និង list នៃបញ្ហា។ `return price, errors` ខ្ចប់តម្លៃទាំងពីរ ទៅក្នុង **tuple** មួយ៖ ក្រុមតម្លៃមួយដែលមិនប្រែប្រួល នៅក្នុងវង់ក្រចក ដូចជា `(2500, [])`។ កូដដែលហៅ function **បំបែក (unpack)** tuple នោះ ទៅជាឈ្មោះពីរ តាមលំដាប់ដដែល៖ `price, errors = read_price("2500")`។ ការផ្ទេរប្រាក់តាមធនាគារ ក៏ដំណើរការដូចគ្នាដែរ៖ លទ្ធផលត្រឡប់មកវិញជា "រួចរាល់" ឬ "បានបដិសេធ៖ មិនមានលុយគ្រប់គ្រាន់"។

![return offer, errors បញ្ជូន tuple មួយត្រឡប់មកវិញ ហើយកូដដែលហៅ function បំបែកវាទៅជាឈ្មោះពីរ](../../assets/diagrams/km/tuple_return.svg)

ទម្រង់នៃច្រកពិនិត្យសប្ដាហ៍នេះ៖ តម្លៃល្អមួយ ជាមួយ list ទទេ **ឬ** `None` ជាមួយបញ្ហាទាំងឡាយ។ មិនដែលមានទាំងពីរក្នុងពេលតែមួយទេ។ `None` មានន័យថា កូដដែលហៅ function មិនរក្សាទុកអ្វីទេ៖ **កុំប្ដូរអ្វីទាំងអស់ នៅពេលធាតុចូលមិនត្រឹមត្រូវ**។

| ធាតុចូល … | function return |
|---|---|
| ល្អ | `(value, [])` |
| មិនល្អ | `(None, ["a plain message", …])` |

🤔 **ទាយមើល** បន្ទាត់ទាំងពីរដែលនឹងត្រូវ print។ តើវង់ក្រចកនៅក្នុងបន្ទាត់ទីមួយ ប្រាប់អ្នកពីអ្វី?

````python
def read_price(text):
    try:
        return int(text), []
    except ValueError:
        return None, ["price must be a whole number of riel, for example 2500."]

print(read_price("2500"))
price, errors = read_price("2.5 dollars")      # បំបែក tuple ទៅជាឈ្មោះពីរ
print(price, errors)
````

អ្នកបានសួរជំនួយការ AI (AI assistant) មួយថា៖ *"សូមសរសេរកូដ Python ដែលបន្ថែមជួរនៃ file របស់អ្នកបរិច្ចាគ ទៅក្នុងការផ្ដល់ជូនរបស់យើង។"* **Version A** គឺជាចម្លើយរបស់វា ជាមួយកំណត់ចំណាំថា *"វាមិនដែលគាំងទេ!"* **Version B** មកពីមិត្តរួមក្រុមម្នាក់។ ទាំងពីរដំណើរការលើជួរទាំង 9។ ចូរអានទាំងពីរ រួច Run cell នោះ។

````python
# Version A · ពីជំនួយការ AI: "វាមិនដែលគាំងទេ!"
def add_offer_a(raw, saved):
    offer = {"donor": raw["donor"], "food": raw["food"]}
    saved.append(offer)                        # រក្សាទុកភ្លាមៗ
    try:
        offer["portions"] = int(raw["portions"])
    except:
        offer["portions"] = 0                  # កុំឱ្យគាំងដាច់ខាត
    if raw["category"] not in CATEGORIES:
        return "error"
    offer["category"] = raw["category"]
    return "OK"


# Version B · ពីមិត្តរួមក្រុមម្នាក់
def add_offer_b(raw, saved):
    errors = []
    donor = raw.get("donor", "").strip()
    category = raw.get("category", "").strip().lower()
    if donor == "":
        errors.append("donor is required.")
    if category not in CATEGORIES:
        errors.append("category must be one of: " + ", ".join(CATEGORIES) + ".")
    try:
        portions = int(raw.get("portions", ""))
        if portions < 1 or portions > 200:
            errors.append("portions must be between 1 and 200.")
    except ValueError:
        errors.append("portions must be a whole number.")
    if len(errors) > 0:
        return "refused: " + " ".join(errors)
    saved.append({"donor": donor, "food": raw.get("food", "").strip(),
                  "category": category, "portions": portions})
    return "OK"


saved_a = []
saved_b = []
for row in rows:
    print(row["food"], "→ A:", add_offer_a(row, saved_a), "| B:", add_offer_b(row, saved_b))
print()
print("A saved", len(saved_a), "offers:")
for offer in saved_a:
    print("  ", offer)
print("B saved", len(saved_b), "offers.")
````

- TODO 1 គឺល្បែងផ្គុំកូដ (puzzle) ពីលំហាត់ Ladder 1, TODO 2 គឺ Ladder 2 ហើយ TODO 3 គឺ tuple មួយ ដូចក្នុងសាកសរសេរទី 4។

```python
MAX_PORTIONS = 200

def clean_offer(raw):

    errors = []

    # ជំហានទី 1 · field ដែលត្រូវតែបំពេញ: សារមួយ សម្រាប់ field ទទេនីមួយៗ (សរសេររួចសម្រាប់អ្នក)
    for field in REQUIRED_FIELDS:
        if raw.get(field, "").strip() == "":
            errors.append(field + " is required.")

    # ជំហានទី 2 · សម្អាតអត្ថបទ: strip គ្រប់ field អត្ថបទ ប្ដូរ category ជាអក្សរតូច (សរសេររួចសម្រាប់អ្នក)
    offer = {}
    for field in ["donor", "contact", "food", "category"]:
        offer[field] = raw.get(field, "").strip()
    offer["category"] = offer["category"].lower()

    # ជំហានទី 4 សម្រាប់ category: តើវាស្ថិតនៅក្នុង CATEGORIES ឬទេ? (សរសេររួចសម្រាប់អ្នក: សាកសរសេរទី 4 របស់អ្នក)
    if offer["category"] != "" and offer["category"] not in CATEGORIES:
        errors.append("category must be one of: " + ", ".join(CATEGORIES) + ".")

    # TODO 1 · ជំហានទី 3 និង 4 សម្រាប់ portions (Ladder 1):
    #   ក្នុង try:  offer["portions"] = int(raw.get("portions", ""))
    #                នៅពេលវាតិចជាង MIN_PORTIONS ឬលើស MAX_PORTIONS ចូរបន្ថែម
    #                "portions must be between " + str(MIN_PORTIONS) + " and " + str(MAX_PORTIONS) + "."
    #   except ValueError:  បន្ថែម "portions must be a whole number."

    # TODO 2 · ជំហានទី 3 និង 4 សម្រាប់ hours_left (Ladder 2):
    #   ក្នុង try:  offer["hours_left"] = int(raw.get("hours_left", ""))
    #                នៅពេលវាតិចជាង REJECT_IF_HOURS_LEFT_BELOW ចូរបន្ថែម "hours left is too short to collect it safely."
    #   except ValueError:  បន្ថែម "hours left must be a whole number."

    # TODO 3 · ជំហានទី 5: ប្រសិនបើ len(errors) > 0 ចូរ return None, errors; បើមិនដូច្នោះទេ ចូរ return offer, []
    return None, ["not done yet"]
```

ដែនកំណត់នីមួយៗ ការពារមនុស្សមួយចំនួន ហើយបដិសេធមនុស្សមួយចំនួនទៀត។ ដែនកំណត់មួយ នៅតែជាការស្មាន រហូតដល់អ្នកសាកល្បងវាលើទិន្នន័យ។ `semester.csv` ផ្ទុកការផ្ដល់ជូនទាំង 404 កាលពីឆមាសមុន។ cell ដែលរៀបចំរួចខាងក្រោម រាប់ថា ប្រសិនបើប្រើដែនកំណត់នីមួយៗ តើវា **នឹង** បដិសេធការផ្ដល់ជូនប៉ុន្មាន ហើយគូរក្រាហ្វបង្ហាញពេលដែលការផ្ដល់ជូនទាំងនោះមកដល់។ ចូរ Run វា ហើយអានវា៖ មិនមានអ្វីត្រូវសរសេរទេ។

របារនីមួយៗនៃក្រាហ្វ រាប់ចំនួនការផ្ដល់ជូន ដែលមកដល់ ដោយមានម៉ោងនៅសល់ស្មើនឹងលេខនៅក្រោមរបារនោះ។ ការផ្ដល់ជូនទាំងអស់ដែលនៅខាង **ឆ្វេង** នៃដែនកំណត់មួយ នឹងត្រូវបានបដិសេធ។

````python
# @title Last semester's offers: what would each limit refuse? { display-mode: "form" }
# cell ដែលរៀបចំរួច: វាអាន semester.csv print តារាងតូចពីរ ហើយគូរក្រាហ្វមួយ។ គ្រាន់តែ Run វា។
import csv
import matplotlib.pyplot as plt

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    semester = list(csv.DictReader(file))
hours = []
portions = []
for row in semester:
    hours.append(int(row["hours_left"]))
    portions.append(int(row["portions"]))

print("REJECT_IF_HOURS_LEFT_BELOW → offers the door would refuse (of " + str(len(semester)) + ")")
for limit in [1, 2, 3, 4, 5]:
    refused = 0
    for value in hours:
        if value < limit:
            refused = refused + 1
    print("   " + str(limit) + " → " + str(refused) + " (" + str(round(100 * refused / len(semester))) + "%)")
print()
print("MAX_PORTIONS → offers the door would refuse (of " + str(len(semester)) + ")")
for limit in [35, 50, 100, 200]:
    refused = 0
    for value in portions:
        if value > limit:
            refused = refused + 1
    print("   " + str(limit) + " → " + str(refused))
print("   (the largest offer last semester: " + str(max(portions)) + " portions)")

plt.figure(figsize=(8, 3))
plt.hist(hours, bins=range(0, 68))
plt.title("Hours left when each offer arrived (last semester, 404 offers)")
plt.xlabel("Hours left")
plt.ylabel("Number of offers")
plt.show()
````
