## វគ្គទី 1 · រៀន: តម្លៃ ឈ្មោះ និងអត្ថបទ (30 នាទី)

កម្មវិធីកុំព្យូទ័រ (program) នីមួយៗធ្វើការជាមួយ **តម្លៃ (value)**៖ លេខ អត្ថបទ (text) និងពិត/មិនពិត (true/false)។ Python រក្សាទុកតម្លៃនីមួយៗជាមួយ **type** មួយ។ type នោះកំណត់ថាតើអ្នកអាចធ្វើអ្វីខ្លះជាមួយតម្លៃនោះ។

### 1 · តម្លៃ (value) និង type

````python
print(30)
print(2.5)
print("Bread rolls")
print(True)
print(type(30), type(2.5), type("Bread rolls"), type(True))
````

- `int` ជាចំនួនគត់, `float` ជាលេខដែលមានចំណុចទសភាគ, `str` ជាអត្ថបទ (text), ហើយ `bool` គឺ `True` ឬ `False`។
- អត្ថបទត្រូវមានសញ្ញាសម្រង់ (quotes) ជានិច្ច។ `"30"` ជាអត្ថបទដែលមើលទៅដូចលេខ ចំណែក `30` គឺជាលេខ។

### 2 · ឈ្មោះ (variable)

**ឈ្មោះ** មួយចង្អុលទៅតម្លៃមួយ។ សញ្ញា `=` មានន័យថា *"ឱ្យឈ្មោះនេះចង្អុលទៅតម្លៃនេះ"*។ វាមិនមែនជាសញ្ញា `=` (ស្មើ) ក្នុងគណិតវិទ្យាទេ។

````python
portions = 30
hours_left = 5
food = "Bread rolls"
print(food, portions, hours_left)
````

អ្នកអាចឱ្យឈ្មោះមួយចង្អុលទៅតម្លៃថ្មីបាន។ ពេលនោះ តម្លៃចាស់ត្រូវបានបំភ្លេចចោល។

🤔 **ទាយមើល** ថាតើ cell នេះនឹង print អ្វី រួចសូម Run វា។

````python
portions = 30
portions = portions + 12
print(portions)
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **ហេតុអ្វី:** …

### 3 · ការគណនា (arithmetic)

| សញ្ញាប្រមាណវិធី (operator) | អត្ថន័យ | `30 ? 4` |
|---|---|---|
| `+` `-` `*` | បូក ដក គុណ | `34`, `26`, `120` |
| `/` | ចែក (ផ្ដល់ float ជានិច្ច) | `7.5` |
| `//` | ចែកយកតែចំនួនគត់ | `7` |
| `%` | សំណល់ | `2` |

````python
portions = 30
volunteers = 4
print(portions / volunteers)
print(portions // volunteers)
print(portions % volunteers)
````

🤔 អ្នកស្ម័គ្រចិត្តមិនអាចដឹកកន្លះចំណែកបានទេ។ ក្នុងចំណោមចម្លើយទាំងបី តើមួយណាមានប្រយោជន៍សម្រាប់អ្នកសម្របសម្រួល ហើយតើមួយណាអាចធ្វើឱ្យយល់ច្រឡំ? ចូរសរសេរចម្លើយមួយប្រយោគ។

### 4 · អត្ថបទ និងលេខ មិនលាយបញ្ចូលគ្នាដោយអចេតនាទេ

សញ្ញា `+` បូកលេខ ប៉ុន្តែវា **តភ្ជាប់** អត្ថបទ។ Python មិនព្រមតភ្ជាប់អត្ថបទជាមួយលេខទេ ព្រោះវាមិនដឹងថាអ្នកចង់បានអ្វី។

````python
print(3 + 4)
print("3" + "4")
print(int("30") + 4)      # int() ប្ដូរអត្ថបទ ទៅជាចំនួនគត់
print("Portions: " + str(30))   # str() ប្ដូរលេខ ទៅជាអត្ថបទ
````

### 5 · f-string: អត្ថបទដែលមានតម្លៃនៅខាងក្នុង

ចូរដាក់អក្សរ `f` នៅមុខសញ្ញាសម្រង់ ហើយសរសេរឈ្មោះនៅក្នុង `{ }`។ Python នឹងបំពេញតម្លៃទាំងនោះឱ្យអ្នក។

````python
food = "Bread rolls"
portions = 30
hours_left = 5
print(f"{food}: {portions} portions, {hours_left} hours left")
````
