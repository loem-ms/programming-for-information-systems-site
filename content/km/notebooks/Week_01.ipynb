កម្មវិធីនីមួយៗធ្វើការជាមួយ **តម្លៃ (value)**: លេខ អត្ថបទ (text) និងពិត/មិនពិត (true/false)។ Python រក្សាទុកតម្លៃនីមួយៗជាមួយ **type** (ប្រភេទទិន្នន័យ) មួយ ហើយ type នោះកំណត់ថា តើអ្នកអាចធ្វើអ្វីខ្លះជាមួយតម្លៃនោះ។

````python
print(30)
print(2.5)
print("Bread rolls")
print(True)
print(type(30), type(2.5), type("Bread rolls"), type(True))
````

- `int` ជាចំនួនគត់, `float` ជាលេខដែលមានចំណុចទសភាគ, `str` ជាអត្ថបទ (text), ហើយ `bool` គឺ `True` ឬ `False`។
- អត្ថបទត្រូវមានសញ្ញាសម្រង់ (quotes) ជានិច្ច។ `"30"` ជាអត្ថបទដែលមើលទៅដូចលេខ ចំណែក `30` គឺជាលេខ។

### ✍️ សាកសរសេរទី 2 · តាមដានកូដលើក្រដាសជាមុនសិន

**ការតាមដានកូដ (trace)** មានន័យថា អានកូដម្ដងមួយបន្ទាត់ ដូចដែលកុំព្យូទ័រធ្វើ ហើយសរសេរតម្លៃនីមួយៗទុក។ ចូរអាន cell នេះ ប៉ុន្តែ **កុំទាន់ Run វា**។ នៅលើក្រដាស ចូររកតម្លៃចុងក្រោយរបស់ `card_balance_riel` និងរបស់ `total_riel`។ សរសេរចម្លើយទាំងពីររបស់អ្នក ក្នុងបន្ទាត់ពីរដំបូង រួច Run។

````python
my_card_balance_riel = 0     # TODO: ចម្លើយរបស់អ្នកសម្រាប់ card_balance_riel
my_total_riel = 0            # TODO: ចម្លើយរបស់អ្នកសម្រាប់ total_riel

card_balance_riel = 10000                      # កាតអាហារដ្ឋានរបស់និស្សិត
card_balance_riel = card_balance_riel - 4000   # អាហារថ្ងៃត្រង់
card_balance_riel = card_balance_riel + 2000   # បញ្ចូលលុយបន្ថែម

fare_riel = 1500                               # សំបុត្រឡានក្រុងមួយ
trips = 2
total_riel = fare_riel * trips
fare_riel = 2000                               # តម្លៃឡើង ក្រោយពេលអ្នកបានបង់រួច

print("✅ card_balance_riel" if my_card_balance_riel == card_balance_riel else f"❌ card_balance_riel ends at {card_balance_riel}: go line by line, right side first")
print("✅ total_riel" if my_total_riel == total_riel else f"❌ total_riel is {total_riel}: it was worked out before fare_riel changed")
````

ការតាមដានទីពីរ គឺជាអន្ទាក់ដែលគេជួបញឹកញាប់។ `total_riel` ត្រូវបានគណនា **តែម្ដង** ជាមួយថ្លៃចាស់។ ការប្ដូរ `fare_riel` នៅពេលក្រោយ មិនប្ដូរវាទេ។ ឈ្មោះមួយចង្អុលទៅតម្លៃមួយ មិនមែនទៅរូបមន្តទេ។

### 3 · ការគណនា (arithmetic): តើចម្លើយណាមានប្រយោជន៍?

| សញ្ញាប្រមាណវិធី (operator) | អត្ថន័យ | `30 ? 4` |
|---|---|---|
| `+` `-` `*` | បូក ដក គុណ | `34`, `26`, `120` |
| `/` | ចែក (ផ្ដល់ float ជានិច្ច) | `7.5` |
| `//` | ចែកយកតែចំនួនគត់ | `7` |
| `%` | សំណល់ (អ្វីដែលនៅសល់) | `2` |

````python
portions = 30
volunteers = 4
print(portions / volunteers)
print(portions // volunteers)
print(portions % volunteers)
````

🤔 អ្នកស្ម័គ្រចិត្តមិនអាចដឹកកន្លះចំណែកបានទេ។ ក្នុងចំណោមចម្លើយទាំងបី តើមួយណាមានប្រយោជន៍សម្រាប់អ្នកសម្របសម្រួល ហើយតើមួយណាអាចធ្វើឱ្យយល់ច្រឡំ? ចូរសរសេរមួយប្រយោគ។

### ✍️ សាកសរសេរទី 3 · ប្រអប់ spring rolls

អាហារដ្ឋានវេចខ្ចប់ spring rolls ចំនួន 50 ដាក់ក្នុងប្រអប់ ប្រអប់មួយដាក់បាន 6។ តើមានប្រអប់ **ពេញ** ប៉ុន្មាន ហើយនៅសល់ spring rolls ប៉ុន្មាន? ចូរគណនាដោយប្រើឈ្មោះ និង `//` ហើយនិង `%` មិនមែនដោយប្រើលេខដែលអ្នកគិតក្នុងក្បាលទេ។

````python
spring_rolls = 50
rolls_per_box = 6

full_boxes = 0     # TODO: ប្រើ //
rolls_left = 0     # TODO: ប្រើ %

print(f"{full_boxes} full boxes, {rolls_left} rolls left over")
print("✅ full_boxes" if full_boxes == spring_rolls // rolls_per_box else "❌ full_boxes: how many whole boxes? (//)")
print("✅ rolls_left" if rolls_left == spring_rolls % rolls_per_box else "❌ rolls_left: what is left after the full boxes? (%)")
````

លទ្ធផលរំពឹងទុក: `8 full boxes, 2 rolls left over`។ 🤔 ឥឡូវនេះ ចូរប្ដូរ `spring_rolls` ទៅជា 60 ហើយទាយបន្ទាត់ថ្មី មុនពេលអ្នក Run។ ប្រសិនបើអ្នកបានវាយ `8` ជំនួសឱ្យការគណនា ការពិនិត្យនឹងចាប់បាន។

### 4 · អត្ថបទ និងលេខ មិនលាយបញ្ចូលគ្នាដោយចៃដន្យទេ

សញ្ញា `+` បូកលេខ ប៉ុន្តែវា **តភ្ជាប់** អត្ថបទ។ Python មិនព្រមតភ្ជាប់អត្ថបទជាមួយលេខទេ ព្រោះវាមិនអាចទាយថាអ្នកចង់បានអ្វី។ `int()` ប្ដូរអត្ថបទទៅជាចំនួនគត់ ហើយ `str()` ប្ដូរលេខទៅជាអត្ថបទ។

### 5 · f-string: អត្ថបទដែលមានតម្លៃ និងឯកតា

ការតភ្ជាប់ដោយប្រើ `+` និង `str()` ធ្វើឱ្យបន្ទាត់វែង។ ចូរដាក់អក្សរ `f` នៅមុខសញ្ញាសម្រង់ ហើយសរសេរឈ្មោះនៅក្នុង `{ }`: Python នឹងបំពេញតម្លៃឱ្យអ្នក។ ចូរសរសេរ **ឯកតានៅក្រោយលេខនីមួយៗ** នៅក្នុងអត្ថបទ: `40 portions` មិនមែនត្រឹមតែ `40` ទេ។

### ✍️ សាកសរសេរទី 5 · បន្ទាត់សម្រាប់អេក្រង់ចំណតឡានក្រុង

អេក្រង់នៅចំណតឡានក្រុងមួយ បង្ហាញមួយបន្ទាត់សម្រាប់ឡានក្រុងនីមួយៗ។ ចូរបង្កើតបន្ទាត់នេះដោយប្រើ f-string និងឈ្មោះទាំងបី មិនមែនលេខទេ:

`Phnom Penh to Kampot: 12 seats left, leaves in 40 minutes`

````python
route = "Phnom Penh to Kampot"
seats_left = 12
minutes_to_departure = 40

bus_line = "?"     # TODO: f-string ដែលមានឈ្មោះទាំងបី និងឯកតារបស់វា

print(bus_line)
expected = route + ": " + str(seats_left) + " seats left, leaves in " + str(minutes_to_departure) + " minutes"
print("✅ bus_line" if bus_line == expected else "❌ compare your line with the example above, letter by letter")
````

#### 🧩 លំហាត់តាមកម្រិត (Ladder 2) · សរសេរពី cell ទទេ: លេខដែលមកដល់ជាអត្ថបទ

Riverside Restaurant ទូរស័ព្ទមកផ្ដល់ស៊ុបមី (Noodle soup)។ ទម្រង់រក្សាទុកចំណែកជា **អត្ថបទ**: `"25"`។ បន្ទាប់មក ភោជនីយដ្ឋានបន្ថែម 5 ចំណែកទៀត។ ចូរសរសេរកូដដែលគណនាផលបូកពិតប្រាកដជា **លេខ** ហើយបង្កើតបន្ទាត់ `Noodle soup: 30 portions` ដោយប្រើ f-string។ លើកនេះគ្មានជំនួយទេ។

````python
food = "Noodle soup"
portions_text = "25"     # ពីទម្រង់: ជាអត្ថបទ
extra_portions = 5

total_portions = 0       # កូដរបស់អ្នក
line = "?"               # កូដរបស់អ្នក

print(line)
print("✅ total_portions" if total_portions == int(portions_text) + extra_portions else "❌ total_portions: turn the text into a number first, then add")
print("✅ line" if line == "Noodle soup: 30 portions" else "❌ line should read exactly: Noodle soup: 30 portions")
````


### ពិនិត្យបន្ទាត់របស់អ្នក

ការពិនិត្យ (check) គឺជាកូដតូចមួយ ដែលសាកល្បងការសន្យាមួយ។ ចូរ Run វា។ សញ្ញា ❌ នីមួយៗប្រាប់អ្នកថាខ្វះអ្វី។ អ្នកមិនទាន់ចាំបាច់យល់កូដរបស់ការពិនិត្យនេះនៅឡើយទេ។
