### គំនិតសំខាន់: module និង contract

ឆ្នាំងសាកទូរស័ព្ទ ធ្វើការសន្យាមួយ។ ភ្លើងពីព្រីជញ្ជាំងចូល ភ្លើងដែលមានសុវត្ថិភាពសម្រាប់ទូរស័ព្ទរបស់អ្នកចេញ ហើយវាមិនដែលធ្វើឱ្យទូរស័ព្ទខូចទេ។ អ្នកមិនចាំបាច់ដឹងថា មានអ្វីនៅខាងក្នុងទេ។ អ្នកទុកចិត្តវា ព្រោះវារក្សាការសន្យាដដែលរាល់ពេល ហើយព្រោះមាននរណាម្នាក់បាន test វារួចហើយ។

function ក៏ដូចគ្នាដែរ។ **contract** (កិច្ចសន្យា) របស់វា និយាយបីយ៉ាង:

- **Input:** អ្វីដែលអ្នកត្រូវតែផ្ដល់ឱ្យវា។ `find_offer` ត្រូវការ list នៃ record និង ID មួយ។
- **Output (លទ្ធផល):** អ្វីដែលវា return។ `find_offer` return record នោះ ឬ `None` ពេលគ្មាន ID បែបនោះ។
- **អ្វីដែលវាមិនដែលធ្វើ។** `find_offer` មិនដែលប្ដូរ list ដែលអ្នកផ្ដល់ឱ្យវា ហើយមិនដែល print ទេ។

**module** គឺជា file មួយ ដែលមាន function ជាច្រើនដែលមាន contract។ module `bridge.records` របស់ kit គឺជាឧទាហរណ៍មួយ: `total_portions` និង `find_offer` ផ្ទាល់ខ្លួនរបស់ប្រព័ន្ធ ស្ថិតនៅទីនោះ។ ពេលគ្រប់ផ្នែករក្សា contract របស់វា អ្នកអាចបង្កើតប្រព័ន្ធមួយពីផ្នែកៗ ហើយជំនួសផ្នែកមួយ (ដោយ function ផ្ទាល់ខ្លួនរបស់អ្នក ឬ function ដែល AI សរសេរ) ដោយមិនធ្វើឱ្យផ្នែកផ្សេងទៀតខូច។ **test** គឺជារបៀបដែលអ្នក *បង្ហាញ* ថាផ្នែកមួយរក្សា contract របស់វា។

### 1 · Parameter និង argument

**Parameter** គឺជាឈ្មោះនៅក្នុងបន្ទាត់ `def`: គឺជា input របស់ function។ **Argument** គឺជា value ដែលអ្នកផ្ដល់ឱ្យ ពេលអ្នកហៅវា។ Python ផ្គូផ្គងវាតាមទីតាំង: argument ទីមួយទៅ parameter ទីមួយ។ អ្នកក៏អាចផ្ដល់វាតាមឈ្មោះបានដែរ។

🤔 **ទាយមើល** បន្ទាត់ចុងក្រោយ មុនពេលអ្នក run cell។

````python
def price_after_discount(price_riel, discount_percent):
    return price_riel - price_riel * discount_percent // 100

print(price_after_discount(10000, 10))
print(price_after_discount(discount_percent=25, price_riel=4000))   # តាមឈ្មោះ: លំដាប់មិនសំខាន់ទេ
print(price_after_discount(10, 10000))                              # 🤔 argument ត្រូវបានដាក់ប្ដូរកន្លែងគ្នា
````

**ការទាយ:** …  **លទ្ធផល:** …

តម្លៃ −990 រៀល គឺគ្មានន័យទេ ប៉ុន្តែ Python មិនដឹងរឿងនោះទេ។ មានតែមនុស្ស ឬ test ប៉ុណ្ណោះ ដែលកត់សម្គាល់ឃើញ។ ឈ្មោះ parameter ដែលច្បាស់លាស់ (មានឯកតា) ធ្វើឱ្យងាយមើលឃើញ argument ដែលដាក់ប្ដូរកន្លែងគ្នា។

### 2 · `return` ឬ `print`?

function ទាំងពីរខាងក្រោម បូកតម្លៃដដែល។ មានតែមួយប៉ុណ្ណោះ ដែលផ្ដល់ចម្លើយត្រឡប់មកវិញ។

````python
def total_printed(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    print(total)

def total_returned(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    return total

printed_result = total_printed([4000, 1500])
returned_result = total_returned([4000, 1500])
print("printed_result is", printed_result)
print("returned_result is", returned_result)
````

- `print` **បង្ហាញ** value មួយដល់មនុស្ស។ បន្ទាប់មក វាបាត់ទៅ។
- `return` **ផ្ដល់** value ត្រឡប់ទៅ code ដែលបានហៅ function នោះ។ code នោះអាចបូកវា ប្រៀបធៀបវា test វា ឬដាក់វាក្នុងរបាយការណ៍។

ប្រព័ន្ធមិនដែលអានអេក្រង់ទេ។ function ដែលគ្រាន់តែ print គឺគ្មានប្រយោជន៍សម្រាប់ផ្នែកផ្សេងទៀតនៃប្រព័ន្ធ ហើយមិនអាច test បានទេ។

### 3 · `None` មានន័យថា "គ្មានចម្លើយ"

function ដែលបញ្ចប់ដោយគ្មាន `return` ផ្ដល់ `None` ត្រឡប់មកវិញ។ នោះហើយជាអ្វីដែលបានកើតឡើងចំពោះ `printed_result`។ ពេលខ្លះ `None` គឺ **ដោយចេតនា**: `find_offer` return `None` ដើម្បីប្រាប់ថា "រកមិនឃើញ"។ ពិនិត្យវាដោយ `is None`។

````python
prices_by_item = {"notebook": 4000, "pen": 1500}

def find_price(item_name):
    return prices_by_item.get(item_name)

price = find_price("ruler")
if price is None:
    print("Not sold here")
else:
    print(price, "riel")
````

`None` ដោយចេតនា គឺជាផ្នែកមួយនៃ contract ដូច្នេះត្រូវសរសេរវាទុក។ `None` ដោយចៃដន្យ (ភ្លេចសរសេរ `return`) គឺជា bug។

### 4 · Docstring: សរសេរការសន្យាទុក

**docstring** គឺជា text ក្នុងសញ្ញាសម្រង់បីជាន់ (triple quotes) នៅបន្ទាត់ដំបូងក្នុង function មួយ។ វាប្រាប់ពី contract។ `help(...)` បង្ហាញវា។ function របស់ kit ក៏មាន docstring ដែរ។

````python
def average_mark(marks):
    """Return the average of a list of marks.

    Returns 0 for an empty list. Does not change the list.
    """
    if len(marks) == 0:
        return 0
    total = 0
    for mark in marks:
        total += mark
    return total / len(marks)

help(average_mark)
help(bridge.records.find_offer)
````

### 5 · Function ដែលហៅ function

បង្កើត function ធំៗ ពី function តូចៗ។ function តូចនីមួយៗធ្វើ **ការងារមួយ** ហើយអាច test ដាច់ដោយឡែកបាន។

````python
URGENT_HOURS = 3

def is_urgent(hours_left):
    return hours_left <= URGENT_HOURS

def urgent_portions(records):
    """Return the total portions of urgent records. Does not change records."""
    total = 0
    for record in records:
        if is_urgent(record["hours_left"]):
            total += record["portions"]
    return total

print(urgent_portions(bridge.sample_records()))
````

`is_urgent` សម្រេច; `urgent_portions` បូកសរុប។ បើអ្នកសម្របសម្រួលប្ដូរអត្ថន័យនៃ "បន្ទាន់" អ្នកកែតែកន្លែងមួយប៉ុណ្ណោះ។

### 6 · `assert`: ការពិនិត្យដែលឈប់ ហើយស្រែកប្រាប់

`assert` ពិនិត្យថា អ្វីមួយគឺ `True`។ បើវាពិត គ្មានអ្វីកើតឡើងទេ។ បើមិនពិត Python ឈប់ដោយ `AssertionError`។ ការសន្យាដែលមិនត្រូវបានរក្សា មិនគួររំលងទៅដោយស្ងាត់ៗទេ។

````python
assert price_after_discount(10000, 10) == 9000
assert average_mark([]) == 0
print("Both asserts passed")
# assert average_mark([50, 70]) == 50    # លុប # ចេញ ដើម្បីមើល assert ដែលបរាជ័យ រួចដាក់វាវិញ
````

### 7 · Test function: ករណីធម្មតា boundary និងទទេ

**test function** មានឈ្មោះចាប់ផ្ដើមដោយ `test_` ហើយវាប្រើ `assert` ដើម្បីពិនិត្យករណីមួយ។ test ល្អ គ្របដណ្ដប់ករណីបីប្រភេទ:

| ប្រភេទករណី | សំណួរ | សម្រាប់ `average_mark` |
|---|---|---|
| **normal** (ធម្មតា) | តើវាដំណើរការ សម្រាប់ input ធម្មតាទេ? | `[60, 80]` → 70 |
| **boundary** (ព្រំដែន) | តើវាដំណើរការនៅត្រង់គែមទេ? | `[0, 100]` → 50; `[100]` → 100 |
| **empty ឬ odd** (ទទេ ឬចម្លែក) | input ដែលបាត់ ទទេ ឬមិនធម្មតា? | `[]` → 0 |

cell នេះក៏កំណត់ `run_tests` ដែលជា test runner តូចមួយ។ អ្នកមិនចាំបាច់សរសេរវាទេ; គ្រាន់តែអានវា។ វាហៅ test នីមួយៗ ហើយ print ✅ ពេលគ្មាន `assert` ណាមួយបរាជ័យ ឬ ❌ ជាមួយមូលហេតុ។ (`try`/`except` អនុញ្ញាតឱ្យ notebook បន្តទៅមុខ ក្រោយពេលមានការបរាជ័យ; អ្នកនឹងរៀនវានៅសប្ដាហ៍ទី 5។)

````python
def run_tests(tests):
    """Run each test function. Print ✅ or ❌ for each one; never stop the notebook."""
    passed = 0
    for test in tests:
        try:
            test()
            print("✅", test.__name__)          # function នីមួយៗ ស្គាល់ឈ្មោះខ្លួនឯង
            passed += 1
        except AssertionError as problem:
            print("❌", test.__name__, "→", problem)
        except Exception as problem:
            print("❌", test.__name__, "→ error:", type(problem).__name__, problem)
    print(passed, "of", len(tests), "tests passed")

def test_average_normal():
    assert average_mark([60, 80]) == 70

def test_average_boundary():
    assert average_mark([0, 100]) == 50
    assert average_mark([100]) == 100

def test_average_empty():
    assert average_mark([]) == 0

run_tests([test_average_normal, test_average_boundary, test_average_empty])
````

សូមកត់សម្គាល់ `run_tests([test_average_normal, ...])`: អ្នកផ្ដល់ **function** ដោយគ្មាន `( )`។ runner ជាអ្នកហៅវា។

### កិច្ចការស្រាវជ្រាវទី 3 · ការពិនិត្យ input (5%)

*ដាក់ឱ្យធ្វើក្រោយ midterm។* ហេតុអ្វីបានជាប្រព័ន្ធពិនិត្យ input **មុន** ពេលរក្សាទុកវា? ជ្រើសរើស form មួយដែលអ្នកធ្លាប់ប្រើ: form ចុះឈ្មោះនៅសាកលវិទ្យាល័យ ការកក់សំបុត្រឡានក្រុង ឬសំបុត្រផ្សេងៗ ការបញ្ចូលលុយទូរស័ព្ទ ការទូទាត់ប្រាក់ពេលទិញទំនិញអនឡាញ ឬការចុះឈ្មោះចូលក្លឹប។

សម្រាប់ **field ពីរ** នៃ form នោះ ចូរផ្ដល់:

1. **value ដែលអនុញ្ញាត** (ឧទាហរណ៍ "ចំនួនគត់ពី 1 ដល់ 10");
2. ឧទាហរណ៍ **មិនត្រឹមត្រូវ** មួយ;
3. **សារ** ដែល form បង្ហាញ ឬសារដែលវា *គួរតែ* បង្ហាញ;
4. អ្វីដែលត្រូវតែ **នៅដដែល** ពេល input មិនត្រឹមត្រូវ។
