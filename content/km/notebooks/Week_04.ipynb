## វគ្គទី 1 · ស្វែងយល់: function ដែលអាចទុកចិត្តបាន (90 នាទី)

### គំនិតសំខាន់: module និងកិច្ចសន្យា (contract)

ឆ្នាំងសាកទូរស័ព្ទ មានការសន្យាមួយ។ ចរន្តភ្លើងពីព្រីចូលទៅ ហើយចរន្តដែលមានសុវត្ថិភាពសម្រាប់ទូរស័ព្ទរបស់អ្នក ចេញមកវិញ។ វាមិនដែលធ្វើឱ្យទូរស័ព្ទខូចឡើយ។ អ្នកមិនចាំបាច់ដឹងថាមានអ្វីនៅខាងក្នុងទេ។ អ្នកទុកចិត្តវា ព្រោះវារក្សាការសន្យាដដែលរាល់ពេល ហើយព្រោះមាននរណាម្នាក់បានសាកល្បង (test) វារួចហើយ។

function ក៏ដូចគ្នាដែរ។ **កិច្ចសន្យា (contract)** របស់វា បញ្ជាក់រឿងបីយ៉ាង៖

- **ធាតុចូល (input):** អ្វីដែលអ្នកត្រូវតែផ្ដល់ឱ្យវា។ `find_offer` ត្រូវការ list នៃ record និង ID មួយ។
- **លទ្ធផល (output):** អ្វីដែលវា return។ `find_offer` return record នោះ ឬ `None` នៅពេលគ្មាន ID នោះ។
- **អ្វីដែលវាមិនដែលធ្វើ។** `find_offer` មិនដែលប្ដូរ list ដែលអ្នកផ្ដល់ឱ្យវាទេ ហើយក៏មិនដែល print អ្វីដែរ។

**module** គឺជា file មួយដែលផ្ទុក function ជាច្រើន ហើយ function នីមួយៗមានកិច្ចសន្យារបស់វា។ នៅពេលគ្រប់ផ្នែករក្សាកិច្ចសន្យារបស់ខ្លួន អ្នកអាចផ្គុំប្រព័ន្ធមួយពីផ្នែកៗបាន ហើយអាចជំនួសផ្នែកមួយ (ដោយ function ផ្ទាល់ខ្លួនរបស់អ្នក ឬ function ដែល AI សរសេរ) ដោយមិនធ្វើឱ្យផ្នែកផ្សេងទៀតខូច។ **តេស្ត (test)** គឺជារបៀបដែលអ្នក *បង្ហាញ* ថាផ្នែកមួយរក្សាកិច្ចសន្យារបស់វា។

សប្ដាហ៍នេះ អ្នកនឹងហាត់ទម្លាប់ប្រាំយ៉ាង នៃការគិតបែបអ្នកសរសេរកម្មវិធី (program thinking)៖

- **ដាក់ឈ្មោះ** ឱ្យរបស់នានា ដើម្បីឱ្យអត្ថន័យរបស់វាច្បាស់លាស់។
- **បំបែកបញ្ហាជាជំហានតូចៗ (decompose)**។
- **តាមដានកូដដោយដៃ (trace)** ដោយដាក់ម្រាមដៃលើកូដ។
- **សាកល្បងនៅចំណុចព្រំដែន**។
- **អានសារ error** ចាប់ពីបន្ទាត់ចុងក្រោយ។

### 1 · ការហៅ (call) function មួយ ផ្ដល់តម្លៃឱ្យ parameter នីមួយៗ

📖 **វាជាអ្វី?** ឈ្មោះនៅក្នុងបន្ទាត់ `def` គឺជា **parameter** ពោលគឺធាតុចូល (input) របស់ function។ តម្លៃដែលអ្នកផ្ដល់ឱ្យ នៅពេលអ្នកហៅ function នោះ គឺជា **argument**។

**ហេតុអ្វីបានជាត្រូវការវា?** អ្នកសរសេរជំហាននានាតែម្ដង ដោយប្រើឈ្មោះជំនួសឱ្យលេខ។ បន្ទាប់មក អ្នកអាចដំណើរការជំហានទាំងនោះ សម្រាប់តម្លៃណាក៏បាន៖ គ្រប់វិក្កយបត្រ គ្រប់ការផ្ដល់ជូន (offer)។

**តើ Python ដំណើរការការហៅ function យ៉ាងដូចម្ដេច?** នៅពេល Python ជួប `price_after_discount(10000, 10)` វា៖

1. **ភ្ជាប់ (bind)** ឈ្មោះ parameter នីមួយៗ ទៅនឹងតម្លៃរបស់ argument មួយ៖ argument ទីមួយទៅ parameter ទីមួយ argument ទីពីរទៅ parameter ទីពីរ។ (អ្នកក៏អាចផ្គូផ្គងវា **តាមឈ្មោះ** បានដែរ៖ `discount_percent=25`។ ពេលនោះ លំដាប់មិនសំខាន់ទេ។)
2. ដំណើរការបន្ទាត់នានានៅខាងក្នុង function។ ឈ្មោះដែលបង្កើតនៅខាងក្នុងវា ដូចជា `discount_riel` គឺជាឈ្មោះ **local** (ឈ្មោះដែលមានតែនៅខាងក្នុង function)៖ វាមានតែក្នុងអំឡុងពេលនៃការហៅនេះប៉ុណ្ណោះ។
3. នៅពេលដល់ `return` វាបញ្ជូនតម្លៃត្រឡប់ទៅបន្ទាត់ដែលបានហៅវាវិញ។ ការហៅបញ្ចប់ ហើយ parameter និងឈ្មោះ local ទាំងអស់ **បាត់ទៅវិញ**។

អេក្រង់របស់អ្នកសម្របសម្រួល គួរបង្ហាញ *"Send a volunteer now"* (បញ្ជូនអ្នកស្ម័គ្រចិត្តម្នាក់ឥឡូវនេះ) នៅពេលមានការផ្ដល់ជូនបន្ទាន់យ៉ាងហោចណាស់មួយ។ function នេះរាប់ការផ្ដល់ជូនបន្ទាន់ ប៉ុន្តែគ្មានអ្វីអាចប្រើចម្លើយរបស់វាបានទេ។ ចូរ Run cell ហើយអានព័ត៌មានជំនួយ 💡។ បន្ទាប់មក ចូរកែតម្រូវ function នេះ។

````python
URGENT_HOURS = 3

def count_urgent(hours_left_list):
    count = 0
    for hours_left in hours_left_list:
        if hours_left <= URGENT_HOURS:
            count += 1
    print(count)

bridge.check("count_urgent", count_urgent, examples=[
    ([2, 5, 3, 20], 2),
    ([12, 24], 0),
    ([], 0),
])
````

### 5 · ពី flowchart ទៅជា function

📖 **វាជាអ្វី?** **flowchart (គំនូសតាងលំហូរ)** គឺជារូបភាពនៃវិធានមួយ។ វាប្រើរាងបីប្រភេទ៖ រាងពងក្រពើ សម្រាប់ចំណុចចាប់ផ្ដើម **រាងពេជ្រ (diamond)** សម្រាប់សំណួរ មែន/មិនមែន (yes/no) និង **ប្រអប់ (box)** សម្រាប់ចំណុចបញ្ចប់។

**ហេតុអ្វីបានជាត្រូវការវា?** flowchart គឺជាកន្លែងដែលចំណាយតិចបំផុត ក្នុងការរកឃើញកំហុស។ អ្នកសម្របសម្រួលដែលមិនចេះអាន Python នៅតែអាចពិនិត្យវាបាន មុនពេលនរណាម្នាក់សរសេរកូដ។

**តើវាក្លាយជា Python យ៉ាងដូចម្ដេច?** **រាងពេជ្រ (diamond)** នីមួយៗ ក្លាយជា `if` មួយ។ **ប្រអប់លទ្ធផលបញ្ចប់ (end box)** នីមួយៗ ក្លាយជា `return` មួយ។ Python សួរសំណួរ `if` ពីលើចុះក្រោម ហើយ `return` ទីមួយដែលដំណើរការ នឹងបញ្ចប់ការហៅ ដូច្នេះសំណួរនៅខាងក្រោមត្រូវបានរំលង។ នេះហើយជាមូលហេតុដែលលំដាប់នៃរាងពេជ្រ មានសារៈសំខាន់។

### 7 · Test function: តេស្តនីមួយៗ ពិនិត្យករណីមួយប្រភេទ

📖 **វាជាអ្វី?** **function សម្រាប់តេស្ត (test function)** គឺជា function ដែលមានឈ្មោះចាប់ផ្ដើមដោយ `test_`។ វាមិនទទួលធាតុចូលទេ ហើយវាប្រើ `assert` ដើម្បីពិនិត្យ **ករណីមួយប្រភេទ**៖

| ប្រភេទ | សំណួរ | សម្រាប់ `total_price` |
|---|---|---|
| **ធម្មតា** (normal) | តើវាដំណើរការត្រឹមត្រូវ សម្រាប់ធាតុចូលធម្មតាដែរឬទេ? | `[4000, 1500]` → 5500 |

**ហេតុអ្វីបានជាត្រូវការវា?** អ្នកអាចដំណើរការតេស្តរបស់អ្នកម្ដងទៀត ក្រោយការកែប្រែនីមួយៗ ក្នុងរយៈពេលតែមួយវិនាទី។ នៅពេលតេស្តមួយធ្លាក់ ឈ្មោះរបស់វាប្រាប់អ្នកថា ករណីប្រភេទណាដែលខូច។ ហើយតេស្តមួយត្រូវតែអាច **ធ្លាក់** បាន៖ តេស្តដែលគ្មាន `assert` មិនពិនិត្យអ្វីសោះ។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `def test_…` គ្រាន់តែកំណត់ (define) តេស្តប៉ុណ្ណោះ៖ មិនទាន់មានអ្វីត្រូវបានពិនិត្យនៅឡើយទេ។ **កម្មវិធីដំណើរការតេស្ត (test runner)** ហៅតេស្តនីមួយៗតាមវេន។ នៅពេល `assert` មួយធ្លាក់ runner កត់ ❌ ហើយបន្តទៅតេស្តបន្ទាប់ ដូច្នេះតេស្តមួយដែលធ្លាក់ មិនលាក់តេស្តផ្សេងទៀតទេ។

**តើអ្នកជួបវានៅកន្លែងណា?** ក្រុមអ្នកបង្កើតកម្មវិធី (software team) គ្រប់ក្រុម រក្សាទុក file តេស្តមួយនៅក្បែរកូដរបស់ខ្លួន។ អ្នកនឹងសរសេរ file របស់អ្នក គឺ `tests.py` នៅសប្ដាហ៍ទី 10 ហើយ midterm នឹងឱ្យអ្នកសរសេរតេស្តដែលចាប់កំហុស (bug) បាន។

cell បន្ទាប់ គឺជា test runner តូចមួយដែលរៀបចំរួចហើយ ឈ្មោះ `run_tests`៖ វាហៅតេស្តនីមួយៗ ហើយ print ✅ ឬ ❌។ ចូរ Run វាម្ដង។ អ្នកគ្រាន់តែប្រើវា អ្នកមិនចាំបាច់អានវាទេ។

👀 **ឧទាហរណ៍។** cell ដែលនៅបន្ទាប់ពី runner មានតេស្តបី សម្រាប់ `total_price`។ 🤔 **ទាយមើល** ថាតើតេស្តទាំងនោះជាប់ប៉ុន្មាន។

````python
# @title A tiny test runner (run this cell) { display-mode: "form" }
def run_tests(tests):
    """Call each test function. Print ✅ or ❌ for each one; never stop the notebook."""
    passed = 0
    for test in tests:
        name = getattr(test, "__name__", repr(test))
        if not callable(test):
            print("❌", name, "→ not a test function: give its name without ( )")
            continue
        try:
            test()
            print("✅", name)
            passed += 1
        except AssertionError as problem:
            print("❌", name, "→", str(problem) or "an assert failed")
        except Exception as problem:
            print("❌", name, "→ error:", type(problem).__name__, problem)
    print(passed, "of", len(tests), "tests passed")
````

````python
def test_total_normal():
    assert total_price([4000, 1500]) == 5500

def test_total_boundary():
    assert total_price([4000]) == 4000

def test_total_empty():
    assert total_price([]) == 0

run_tests([test_total_normal, test_total_boundary, test_total_empty])
````

#### ⚖️ ល្អ ឬមានហានិភ័យ?

មិត្តរួមថ្នាក់ម្នាក់ បានសុំជំនួយការ AI (AI assistant) ឱ្យសរសេរ *"function ដែលប្រាប់ចំណែកដែលបានប្រគល់ ហើយសម្អាតការផ្ដល់ជូនដែលបានបញ្ចប់ ព្រមទាំងមានតេស្តមួយ។"* វាបានសរសេរ **Version A**។ **Version B** គឺជាការសរសេរឡើងវិញ។ យើងបានបន្ថែម `_a` និង `_b` ទៅឈ្មោះ ដើម្បីឱ្យទាំងពីរអាចនៅក្នុង notebook នេះជាមួយគ្នាបាន។ សូម Run cell ទាំងពីរ៖ គ្រប់តេស្តសុទ្ធតែជាប់។

````python
def delivered_portions_a(records):
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
            records.remove(record)          # សម្អាតការផ្ដល់ជូនដែលបានបញ្ចប់
    print("Delivered:", total, "portions")

def test_delivered_a():
    delivered_portions_a(bridge.sample_records())

run_tests([test_delivered_a])
````

````python
def delivered_portions_b(records):
    """Return the total portions with status "delivered". Does not change records."""
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
    return total

def test_delivered_normal():
    assert delivered_portions_b(bridge.sample_records()) == 18

def test_delivered_does_not_change_records():
    records = bridge.sample_records()
    delivered_portions_b(records)
    assert len(records) == 4

run_tests([test_delivered_normal, test_delivered_does_not_change_records])
````

> ❌ *"ចូរសរសេរ function មួយសម្រាប់ចំណែកដែលមិនទាន់បញ្ចប់ (open portions)។"*
>
> ✅ *"ចូរសរសេរ function Python 3 មួយឈ្មោះ `open_portions(records)`។ `records` គឺជា list នៃ dictionary ដែលមាន key `"status"` និង `"portions"` (ជា int)។ ចូរ return ចំណែកសរុបនៃ record ដែលមាន status `"new"`, `"approved"` ឬ `"picked_up"`។ ចូរ return 0 សម្រាប់ list ទទេ។ កុំ print។ កុំប្ដូរ list។ ប្រើតែ Python ធម្មតា គ្មាន library។ ឧទាហរណ៍៖ record ពីរ គឺ 30 ចំណែក picked_up និង 18 ចំណែក delivered → 30។"*

ការស្នើសុំទីពីរ គឺជាប័ណ្ណការងារសប្ដាហ៍នេះ ដែលសរសេរឱ្យពេញលេញ៖ ឈ្មោះ ធាតុចូល លទ្ធផល ករណីចម្លែក អ្វីដែលវាមិនត្រូវធ្វើជាដាច់ខាត និងឧទាហរណ៍មួយ។

**2 · ពិនិត្យអ្វីដែល AI ផ្ដល់មកវិញ ដូចជាមនុស្សដែលអ្នកមិនស្គាល់ ជាអ្នកសរសេរវា។** ចូរអានគ្រប់បន្ទាត់។ ចូរ Run ការពិនិត្យ និងតេស្តផ្ទាល់ខ្លួនរបស់អ្នក។ ចូរសួរថា៖ *តើមានអ្វីកើតឡើងជាមួយ list ទទេ? នៅចំណុចព្រំដែន?* ប្រសិនបើអ្នកមិនអាចពន្យល់បន្ទាត់ណាមួយបានទេ អ្នកមិនអាចប្រើវាបានឡើយ។

✍️ **ពិនិត្យចម្លើយរបស់ AI។** ជំនួយការ AI មួយ បានឆ្លើយការស្នើសុំមិនច្បាស់លាស់ខាងលើ ដោយ function ខាងក្រោម។ ចូរ Run ការពិនិត្យ។ ចូររកបន្ទាត់ដែលបំពានកិច្ចសន្យា ហើយកែវា។

````python
def ai_open_portions(records):
    # សរសេរដោយជំនួយការ AI
    total = 0
    for record in records:
        if record["status"] != "rejected":
            total += record["portions"]
    return total

bridge.check("open_portions", ai_open_portions)
````

2. **សួរ AI អំពីចំណុចដែលនៅមិនទាន់ច្បាស់។** ចូរសួរសំណួរបន្ត (follow-up question) យ៉ាងហោចណាស់ពីរ៖ មួយសុំឧទាហរណ៍ពីជីវិតប្រចាំថ្ងៃនៅកម្ពុជា មួយទៀតសួរថា *"តើមានអ្វីអាចនឹងខុសឆ្គង?"*។ ចូររក្សាទុកបន្ទាត់សំខាន់ៗ 6–10 បន្ទាត់ នៃការសន្ទនានោះ។

4. **មតិរបស់អ្នក។** ចូរសរសេរប្រហែលកន្លះទំព័រ (≈ 150–250 ពាក្យជាភាសាអង់គ្លេស) លើសំណួរ *"តើកម្មវិធីកុំព្យូទ័រគួរគាំង ឬបន្តដំណើរការដោយស្មានយកចម្លើយមួយ?"* ចូរជ្រើសរើសជំហរមួយ ហើយផ្ដល់មូលហេតុ ដោយយកឧទាហរណ៍ពីករណី FoodBridge (ឧទាហរណ៍ ការផ្ដល់ជូនមួយដែលចំនួនចំណែកសរសេរថា `many`)។
5. **ប្រភព និងភាពស្មោះត្រង់។** ចូររាយផ្នែក និងជំពូកដែលអ្នកបានអាន និងឧបករណ៍ AI ដែលអ្នកបានប្រើ។
