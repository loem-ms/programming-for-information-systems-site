### 1 · Parameter និង argument

**Parameter** គឺជាឈ្មោះនៅក្នុងបន្ទាត់ `def` ពោលគឺជាធាតុចូល (input) របស់ function។ **Argument** គឺជាតម្លៃដែលអ្នកផ្ដល់ឱ្យ នៅពេលអ្នកហៅ function នោះ។ Python ផ្គូផ្គងវាតាមទីតាំង: argument ទីមួយទៅឱ្យ parameter ទីមួយ។ អ្នកក៏អាចផ្ដល់វាតាមឈ្មោះបានដែរ។

🤔 **ទាយមើល** បន្ទាត់ចុងក្រោយ មុនពេលអ្នក Run cell។

````python
def price_after_discount(price_riel, discount_percent):
    return price_riel - price_riel * discount_percent // 100

print(price_after_discount(10000, 10))
print(price_after_discount(discount_percent=25, price_riel=4000))   # តាមឈ្មោះ: លំដាប់មិនសំខាន់ទេ
print(price_after_discount(10, 10000))                              # 🤔 argument ត្រូវបានដាក់ប្ដូរកន្លែងគ្នា
````

Parameter អនុញ្ញាតឱ្យអ្នកផ្ដល់ធាតុចូល (input) ឱ្យ function មួយ។ បញ្ហាបន្ទាប់: function មួយអាចគណនាចម្លើយត្រូវ ប៉ុន្តែមិនបញ្ជូនអ្វីត្រឡប់មកវិញសោះ។

### 2 · `return` ឬ `print`?

function ទាំងពីរខាងក្រោម បូកតម្លៃទំនិញ (price) ដដែល។ ប៉ុន្តែមានតែមួយប៉ុណ្ណោះ ដែលបញ្ជូនចម្លើយត្រឡប់មកវិញ (return)។

````python
def total_printed(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    print(total)

def total_returned(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    return total

printed_result = total_printed([4000, 1500])
returned_result = total_returned([4000, 1500])
print("printed_result is", printed_result)
print("returned_result is", returned_result)
````

- `print` **បង្ហាញ** តម្លៃមួយដល់មនុស្ស។ បន្ទាប់មក តម្លៃនោះបាត់ទៅ។
- `return` **បញ្ជូន** តម្លៃនោះត្រឡប់ទៅកូដដែលបានហៅ function នោះវិញ។ កូដនោះអាចយកវាទៅបូក ប្រៀបធៀប តេស្ត ឬដាក់ក្នុងរបាយការណ៍។

ប្រព័ន្ធមិនដែលអានអេក្រង់ទេ។ function ដែលគ្រាន់តែ print គឺគ្មានប្រយោជន៍សម្រាប់ផ្នែកផ្សេងទៀតនៃប្រព័ន្ធ ហើយមិនអាចតេស្តបានទេ។

តម្លៃដែលបាន return អាចឱ្យកូដផ្សេងទៀតប្រើបាន។ បញ្ហាបន្ទាប់: ពេលខ្លះ ចម្លើយដ៏ស្មោះត្រង់ គឺគ្មានចម្លើយ។

### 3 · `None` មានន័យថា "គ្មានចម្លើយ"

function ដែលបញ្ចប់ដោយគ្មាន `return` នឹងផ្ដល់ `None` ត្រឡប់មកវិញ។ នោះហើយជាអ្វីដែលបានកើតឡើងចំពោះ `printed_result`។ ពេលខ្លះ `None` គឺ **ដោយចេតនា**: `find_offer` return `None` ដើម្បីប្រាប់ថា "រកមិនឃើញ"។ ចូរពិនិត្យវាដោយប្រើ `is None`។

````python
prices_by_item = {"notebook": 4000, "pen": 1500}

def find_price(item_name):
    return prices_by_item.get(item_name)

price = find_price("ruler")
if price is None:
    print("Not sold here")
else:
    print(price, "riel")
````

### 4 · Docstring: សរសេរការសន្យាទុក

**docstring** គឺជាអត្ថបទក្នុងសញ្ញាសម្រង់បីជាន់ (triple quotes) នៅបន្ទាត់ដំបូងក្នុង function មួយ។ វាប្រាប់ពីកិច្ចសន្យា (contract)។ `help(...)` បង្ហាញវា។ function របស់ kit ក៏មាន docstring ដែរ។

````python
def average_mark(marks):
    """Return the average of a list of marks.

    Returns 0 for an empty list. Does not change the list.
    """
    if len(marks) == 0:
        return 0
    total = 0
    for mark in marks:
        total += mark
    return total / len(marks)

help(average_mark)
help(bridge.records.find_offer)
````

docstring សរសេរការសន្យាទុក។ បញ្ហាបន្ទាប់: function ធំមួយដែលមានការងារច្រើន ពិបាកតេស្ត ដូច្នេះ ចូរបង្កើតវាពី function តូចៗ។

### 5 · Function ដែលហៅ function ផ្សេងទៀត

ចូរបង្កើត function ធំៗ ពី function តូចៗ។ function តូចនីមួយៗធ្វើ **ការងារមួយ** ហើយអាចតេស្តដាច់ដោយឡែកពីគ្នាបាន។

````python
URGENT_HOURS = 3

def is_urgent(hours_left):
    return hours_left <= URGENT_HOURS

def urgent_portions(records):
    """Return the total portions of urgent records. Does not change records."""
    total = 0
    for record in records:
        if is_urgent(record["hours_left"]):
            total += record["portions"]
    return total

print(urgent_portions(bridge.sample_records()))
````

`is_urgent` ជាអ្នកសម្រេច ចំណែក `urgent_portions` ជាអ្នកបូកសរុប។ ប្រសិនបើអ្នកសម្របសម្រួលប្ដូរអត្ថន័យនៃ "បន្ទាន់" អ្នកកែតែកន្លែងមួយប៉ុណ្ណោះ។

### 7 · Test function: ករណីធម្មតា ករណីព្រំដែន និងករណីទទេ

**function សម្រាប់តេស្ត (test function)** មានឈ្មោះចាប់ផ្ដើមដោយ `test_` ហើយវាប្រើ `assert` ដើម្បីពិនិត្យករណីមួយ។ តេស្តល្អ គ្របដណ្ដប់ករណីបីប្រភេទ៖

| ប្រភេទករណី | សំណួរ | សម្រាប់ `average_mark` |
|---|---|---|
| **ធម្មតា** (normal) | តើវាដំណើរការត្រឹមត្រូវ សម្រាប់ធាតុចូលធម្មតាដែរឬទេ? | `[60, 80]` → 70 |
| **ករណីព្រំដែន** (edge case) | តើវាដំណើរការត្រឹមត្រូវ នៅចំណុចព្រំដែនដែរឬទេ? | `[0, 100]` → 50; `[100]` → 100 |
| **ទទេ ឬចម្លែក** (empty or odd) | ចុះធាតុចូលដែលបាត់ ទទេ ឬមិនធម្មតាវិញ? | `[]` → 0 |

cell នេះក៏កំណត់ `run_tests` ដែលជាកម្មវិធីដំណើរការតេស្ត (test runner) តូចមួយ។ អ្នកមិនចាំបាច់សរសេរវាទេ គ្រាន់តែអានវាបានហើយ។ វាហៅតេស្តនីមួយៗ ហើយ print ✅ នៅពេលគ្មាន `assert` ណាមួយបរាជ័យ ឬ print ❌ ជាមួយមូលហេតុ។ (`try`/`except` អនុញ្ញាតឱ្យ notebook បន្តទៅមុខ ក្រោយពេលមានការបរាជ័យ។ អ្នកនឹងរៀនវានៅសប្ដាហ៍ទី 5។)

````python
def run_tests(tests):
    """Run each test function. Print ✅ or ❌ for each one; never stop the notebook."""
    passed = 0
    for test in tests:
        try:
            test()
            print("✅", test.__name__)          # function នីមួយៗ ស្គាល់ឈ្មោះខ្លួនឯង
            passed += 1
        except AssertionError as problem:
            print("❌", test.__name__, "→", problem)
        except Exception as problem:
            print("❌", test.__name__, "→ error:", type(problem).__name__, problem)
    print(passed, "of", len(tests), "tests passed")

def test_average_normal():
    assert average_mark([60, 80]) == 70

def test_average_boundary():
    assert average_mark([0, 100]) == 50
    assert average_mark([100]) == 100

def test_average_empty():
    assert average_mark([]) == 0

run_tests([test_average_normal, test_average_boundary, test_average_empty])
````

សូមកត់សម្គាល់ `run_tests([test_average_normal, ...])`: អ្នកផ្ដល់ **function** ដោយគ្មាន `( )`។ runner ជាអ្នកហៅ function ទាំងនោះ។

### ឧទាហរណ៍គំរូ · ចំណែកដែលបានប្រគល់តែប៉ុណ្ណោះ

មុនពេលអ្នកបង្កើត ចូរមើលលំនាំនេះម្ដងសិន លើសំណួរតូចជាង។ អ្នកសម្របសម្រួលសួរថា: *«តើមានប៉ុន្មានចំណែក ដែលបានប្រគល់រួច នៅព្រឹកនេះ?»*

យើងឆ្លើយជាបួនជំហាន៖

1. ចាប់ផ្ដើមផលបូកនៅ 0។
2. loop លើ record ទាំងអស់។
3. បូកចំណែក **តែនៅពេល** status របស់ record គឺ `"delivered"`។
4. បញ្ជូនផលបូកត្រឡប់មកវិញ (return) ក្រោយពេល loop ចប់។

វាគឺជាលំនាំ accumulator ពីសប្ដាហ៍ទី 3 ដោយមាន `if` មួយនៅខាងក្នុង — ហើយវា return ចម្លើយរបស់វា ជំនួសឱ្យការ print វា។ ខាងក្រោមនេះ គឺជា record បីនៅព្រឹកនេះ:

````python
morning_records = [
    {"id": "M01", "food": "Bread rolls", "status": "delivered", "portions": 18},
    {"id": "M02", "food": "Bananas", "status": "approved", "portions": 40},
    {"id": "M03", "food": "Fried rice boxes", "status": "delivered", "portions": 7},
]

def delivered_portions(records):
    """Return the total portions of records with status "delivered". Does not change records."""
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
    return total

print(delivered_portions(morning_records))
````

**របៀបដែលវាដំណើរការ:** loop ទៅមើល record ទាំងបី ប៉ុន្តែ `if` អនុញ្ញាតឱ្យតែ record ពីរប៉ុណ្ណោះ ទៅដល់បន្ទាត់ `+=`: 18 + 7 = 25។ record របស់ Bananas ត្រូវបានទៅមើល ពិនិត្យ រួចរំលងចោល — វាត្រូវបានអនុម័ត ប៉ុន្តែមិនទាន់បានប្រគល់ទេ។

នៅវគ្គទី 2 អ្នកនឹងសរសេរលំនាំដដែលនេះ សម្រាប់ `open_portions`: នៅទីនោះ `if` ពិនិត្យ list នៃ status បី ជំនួសឱ្យ status តែមួយ ហើយតេស្តរបស់អ្នកបញ្ជាក់ថា function នោះរក្សាការសន្យារបស់វា។

ឥឡូវនេះ អ្នកអាចសរសេរ function មួយ ហើយតេស្តវាបាន។ បញ្ហាបន្ទាប់: វាយតម្លៃ function របស់អ្នកដទៃ — ហើយបង្កើត function មួយ ដែលសេចក្ដីសង្ខេបរបស់អ្នកសម្របសម្រួលពឹងផ្អែកលើវា។

### ⚖️ ល្អ ឬមានហានិភ័យ?

មិត្តរួមថ្នាក់ម្នាក់ បានសុំជំនួយការ AI មួយ ឱ្យសរសេរ *"function ដែលផ្ដល់ចំណែកដែលបានប្រគល់ ហើយសម្អាតសំណើដែលបានបញ្ចប់ ព្រមទាំងមានតេស្តមួយ។"* វាបានសរសេរ **Version A**។ **Version B** គឺជាការសរសេរឡើងវិញ។ យើងបានបន្ថែម `_a` និង `_b` ទៅឈ្មោះ ដើម្បីឱ្យទាំងពីរអាចនៅក្នុង notebook នេះជាមួយគ្នាបាន។ សូម Run cell ទាំងពីរ៖ គ្រប់តេស្តសុទ្ធតែជាប់។

````python
def delivered_portions_a(records):
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
            records.remove(record)          # សម្អាតសំណើដែលបានបញ្ចប់
    print("Delivered:", total, "portions")

def test_delivered_a():
    delivered_portions_a(bridge.sample_records())

run_tests([test_delivered_a])
````

````python
def delivered_portions_b(records):
    """Return the total portions with status "delivered". Does not change records."""
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
    return total

def test_delivered_normal():
    assert delivered_portions_b(bridge.sample_records()) == 18

def test_delivered_two_in_a_row():
    two = [{"status": "delivered", "portions": 10}, {"status": "delivered", "portions": 5}]
    assert delivered_portions_b(two) == 15

def test_delivered_empty():
    assert delivered_portions_b([]) == 0

def test_delivered_keeps_the_list():
    records = bridge.sample_records()
    delivered_portions_b(records)
    assert records == bridge.sample_records()

run_tests([test_delivered_normal, test_delivered_two_in_a_row, test_delivered_empty,
           test_delivered_keeps_the_list])
````
