### ដំណើររឿង

## វគ្គទី 1 · ស្វែងយល់: ការផ្ទៀងផ្ទាត់ (90 នាទី)

### គំនិតសំខាន់: ការផ្ទៀងផ្ទាត់ (verification)

រៀងរាល់ព្រឹក មុនពេលឡានក្រុងដំបូងចេញពីស្ថានីយ ជាងម៉ាស៊ីនម្នាក់ដើរពិនិត្យជុំវិញវា ជាមួយបញ្ជីត្រួតពិនិត្យ (checklist) មួយ៖ ហ្វ្រាំង ភ្លើង កង់ ទ្វារ។ នាងគូសធីកប្រអប់នីមួយៗ ហើយចុះហត្ថលេខាលើសន្លឹកនោះ។ ការនិយាយថា «ម្សិលមិញវាដំណើរការល្អ» មិនគ្រប់គ្រាន់ទេ ព្រោះសុវត្ថិភាពរបស់មនុស្ស ពឹងផ្អែកលើឡានក្រុងនោះ។

ប្រព័ន្ធដែលមនុស្សពឹងផ្អែកលើ ក៏ត្រូវការរបស់បីយ៉ាងដូចគ្នាដែរ៖ **បញ្ជីត្រួតពិនិត្យដែលនរណាក៏អាចដំណើរការម្ដងទៀតបាន** (តេស្ត) **ការពិនិត្យដោយចាប់ផ្ដើមសាជាថ្មីពីដើម** (clean run) និង **ភ្នែកមួយគូទៀត** (ការត្រួតពិនិត្យកូដ ឬ code review)។ កូដដែល AI សរសេរ ត្រូវការរបស់ទាំងបីនេះជាងគេ៖ វាច្រើនតែមើលទៅត្រឹមត្រូវ ហើយ AI អាចនិយាយថា «ខ្ញុំបានសាកល្បងហើយ» ទោះបីគ្មាននរណាបានសាកល្បងក៏ដោយ។

ក្នុងសប្ដាហ៍នេះ អ្នកនឹងប្រើទម្លាប់បីយ៉ាង នៃការគិតបែបអ្នកសរសេរកម្មវិធី (program thinking) ដោយចេតនា៖ **តាមដានកូដដោយដៃ (trace)** (គណនាចម្លើយ មុនពេលអ្នកដំណើរការកូដ) **សាកល្បងនៅចំណុចព្រំដែន** (ចំណុចព្រំដែននៃគ្រប់វិធាន និងករណីទទេ ឬករណីចម្លែក) និង **អានសារ error** ចាប់ពីបន្ទាត់ចុងក្រោយ។

### 1 · តេស្ត គឺជាកូដដែលពិនិត្យកូដ

📖 **វាជាអ្វី?** **តេស្ត (test)** គឺជា function តូចមួយ ដែលពិនិត្យ function មួយទៀត។ វាហៅផ្នែកនោះជាមួយធាតុចូល (input) មួយ ហើយប្រៀបធៀបចម្លើយ ជាមួយចម្លើយដែលរំពឹងទុក ដោយប្រើ `assert` (សប្ដាហ៍ទី 4)។ អ្នកបានសរសេរករណីតេស្ត តាំងពីសប្ដាហ៍ទី 4 មក។ ក្នុងសប្ដាហ៍នេះ ករណីតេស្តទាំងនោះនឹងមានកន្លែងផ្ទាល់ខ្លួនរបស់វា។

**ហេតុអ្វីបានជាត្រូវការវា?** អ្នកមិនអាចមើលគ្រប់ចម្លើយដោយភ្នែក ក្រោយការកែប្រែនីមួយៗបានទេ។ តេស្តពិនិត្យក្នុងរយៈពេលមួយវិនាទី រាល់ពេល ហើយមិនដែលហត់ទេ។ វាក៏ប្រាប់អ្នកបន្ទាប់ផងដែរថា ផ្នែកនោះសន្យាអ្វីខ្លះ។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `def test_…` គ្រាន់តែកំណត់ (define) តេស្តប៉ុណ្ណោះ។ **កម្មវិធីដំណើរការតេស្ត (runner)** ជាអ្នកហៅវា។ នៅខាងក្នុងតេស្ត `assert` ជាប់ដោយស្ងាត់ៗ នៅពេលការប្រៀបធៀបជា `True` ហើយធ្លាក់ ដោយប្រាប់ឱ្យដឹងយ៉ាងច្បាស់តាម `AssertionError` នៅពេលវាជា `False`។ runner ចាប់ (catch) error នោះ កត់ថា "fail" (ធ្លាក់) ហើយបន្តទៅតេស្តបន្ទាប់។

| ជំហាន | អ្វីដែលកើតឡើង | ឧទាហរណ៍ |
|---|---|---|
| 1 · ហៅ (call) | តេស្តហៅផ្នែកនោះ ជាមួយធាតុចូលមួយ | `urgency(3)` |
| 2 · ប្រៀបធៀប | `assert` ប្រៀបធៀបចម្លើយ ជាមួយចម្លើយដែលរំពឹងទុក | `== "urgent"` |
| 3 · ជាប់ ឬធ្លាក់ (verdict) | `True` → គ្មានអ្វីកើតឡើងទេ (ជាប់) · `False` → `AssertionError` (ធ្លាក់) | ✅ ឬ ❌ |

**តេស្តមួយត្រូវតែអាចធ្លាក់បាន។** ការជាប់លើកូដត្រឹមត្រូវ គឺជាភស្តុតាងតែពាក់កណ្ដាលប៉ុណ្ណោះ។ តេស្តដែលមានប្រយោជន៍ ក៏ត្រូវ **ធ្លាក់** លើកំណែ (version) ដែលមានកំហុស (bug) ដែរ។ ដូច្នេះ ក្នុងសប្ដាហ៍នេះ អ្នកនឹងសាកល្បងតេស្តនីមួយៗ លើកំណែជាច្រើននៃ function មួយ៖ កំណែមួយត្រឹមត្រូវ និងកំណែផ្សេងទៀត ដែលមានកំហុសដាក់ដោយចេតនា។

**តើអ្នកជួបវានៅកន្លែងណា?** កំណែថ្មី (update) នីមួយៗរបស់ app នៅលើទូរស័ព្ទរបស់អ្នក បានជាប់តេស្តរាប់ពាន់រួចហើយ។ ធនាគារ និងហាង app (app store) ដំណើរការតេស្តទាំងនោះដោយស្វ័យប្រវត្តិ មុនពេលចេញផ្សាយការកែប្រែណាមួយ។

cell បន្ទាប់ត្រូវបានរៀបចំរួចហើយ។ វាមាន `verdict(test, version)`៖ ចូរផ្ដល់តេស្តមួយ និងកំណែមួយនៃ function ឱ្យវា ហើយវានឹង return `"pass"` (ជាប់) ឬ `"fail"` (ធ្លាក់)។ ការគាំង (crash) រាប់ថាជា `"fail"`។ វាបញ្ជូនកំណែនោះ ទៅឱ្យតេស្ត ជា **ធាតុចូល**៖ function ក៏ជាតម្លៃមួយដែរ ដូច្នេះអ្នកអាចផ្ដល់វាទៅឱ្យ function មួយទៀតបាន ដូចដែលអ្នកបានផ្ដល់ `hours_left_of` ទៅឱ្យ `sorted` នៅសប្ដាហ៍ទី 8។ cell នោះក៏រៀបចំកំណែត្រឹមត្រូវ និងកំណែដែលមានកំហុស ដែលអ្នកប្រើក្នុងវគ្គទី 1 ផងដែរ។ អ្នកមិនចាំបាច់អានវាជាមុនទេ ព្រោះតេស្តរបស់អ្នក មកពីកិច្ចសន្យា។

````python
# @title Prepared: verdict() and the versions to test (run this cell) { display-mode: "form" }
def verdict(test, version):
    """Run one test on one version of a function. Return "pass" or "fail"."""
    try:
        test(version)
    except Exception:            # assert មួយធ្លាក់ ឬកំណែនោះគាំង
        return "fail"
    return "pass"


def urgency_correct(hours_left):
    if hours_left <= 3:
        return "urgent"
    if hours_left <= 12:
        return "soon"
    return "later"


def urgency_bug_3(hours_left):                # 3 ម៉ោង លែងជា urgent ទៀតហើយ
    if hours_left < 3:
        return "urgent"
    if hours_left <= 12:
        return "soon"
    return "later"


def urgency_bug_12(hours_left):               # 12 ម៉ោង លែងជា soon ទៀតហើយ
    if hours_left <= 3:
        return "urgent"
    if hours_left < 12:
        return "soon"
    return "later"


def urgency_bug_zero(hours_left):             # ភ្លេចករណីនៅសល់ 0 ម៉ោង
    if hours_left >= 1 and hours_left <= 3:
        return "urgent"
    if hours_left <= 12:
        return "soon"
    return "later"


def label_correct(portions, approved):
    if not approved:
        return "hold"
    if portions >= 20:
        return "team"
    return "standard"


def label_bug_approval(portions, approved):   # ភ្លេចការអនុម័ត
    if portions >= 20:
        return "team"
    return "standard"


def label_bug_edge(portions, approved):       # > ជំនួសឱ្យ >=
    if not approved:
        return "hold"
    if portions > 20:
        return "team"
    return "standard"


def top_lowest(marks):                        # ពិន្ទុទាបបំផុតបី
    return sorted(marks)[:3]


def rate_correct(part, whole):
    if whole == 0:
        return 0.0
    return round(100 * part / whole, 1)


def rate_bug_fraction(part, whole):           # ភ្លេច 100: ជាប្រភាគ មិនមែនជាភាគរយទេ
    if whole == 0:
        return 0.0
    return round(part / whole, 1)


print("Ready: verdict() and", 10, "versions to test.")
````

👀 **ឧទាហរណ៍។** តេស្តពីរ សម្រាប់ `urgency`។ នៅខាងក្នុងតេស្ត `urgency` គឺជា parameter មួយ៖ វាមានន័យថា *កំណែដែលយើងបញ្ជូនចូលទៅ*។ 🤔 **ទាយមើល** លទ្ធផលទាំងបួន (ជាប់ ឬធ្លាក់) មុនពេលអ្នក Run cell។

````python
def test_one_hour(urgency):
    assert urgency(1) == "urgent"

def test_three_hours(urgency):
    assert urgency(3) == "urgent"

print("1 hour: ", verdict(test_one_hour, urgency_correct), verdict(test_one_hour, urgency_bug_3))
print("3 hours:", verdict(test_three_hours, urgency_correct), verdict(test_three_hours, urgency_bug_3))
````

### ✍️ សាកសរសេរទី 1 · `test_label`

កិច្ចសន្យាពីសប្ដាហ៍ទី 2 ដោយប្រើចំនួនអប្បបរមាសម្រាប់ក្រុម តាមតម្លៃ default គឺ 20៖ មិនទាន់អនុម័ត → `"hold"`; 20 ចំណែក ឬច្រើនជាងនេះ → `"team"`; ក្រៅពីនេះ → `"standard"`។ ចូរសរសេរ `test_label` ដើម្បីឱ្យវា **ជាប់** លើ `label_correct` ហើយ **ធ្លាក់** លើកំណែដែលមានកំហុសទាំងពីរ។ ឧទាហរណ៍នីមួយៗខាងក្រោម គឺ *(តេស្ត និងកំណែមួយ, លទ្ធផលដែលអ្នករំពឹងទុក)*។ មុនពេលអ្នកសរសេរតេស្តរួច cell នោះបង្ហាញ 1/3 ព្រោះតេស្តដែលគ្មាន `assert` មិនអាចចាប់កំហុសណាមួយបានទេ។

````python
def test_label(pickup_label):
    # ✍️ ចូរប្ដូរបន្ទាត់នីមួយៗទៅជា assert មួយ ឧទាហរណ៍  assert pickup_label(19, True) == "standard"
    # 19 ចំណែក អនុម័តហើយ       → "standard"   (តិចជាងចំនួនអប្បបរមាសម្រាប់ក្រុម 1)
    # 20 ចំណែក អនុម័តហើយ       → "team"       (ស្មើចំនួនអប្បបរមាសម្រាប់ក្រុមគត់)
    # 20 ចំណែក មិនទាន់អនុម័ត    → "hold"
    pass   # ជំនួសបន្ទាត់នេះ ដោយ assert បីរបស់អ្នក


bridge.check("verdict", verdict, examples=[
    ((test_label, label_correct), "pass"),
    ((test_label, label_bug_approval), "fail"),
    ((test_label, label_bug_edge), "fail"),
])
````

### 2 · ចម្លើយដែលរំពឹងទុក មកពីកិច្ចសន្យា

📖 **វាជាអ្វី?** តម្លៃនៅខាងក្រោយ `==` ក្នុងតេស្តមួយ គឺជា **ចម្លើយដែលរំពឹងទុក**។ អ្នកសរសេរវាពី **កិច្ចសន្យា (contract)**៖ ប័ណ្ណការងារ (ticket) វិធាន docstring ឬការសម្រេចចិត្តរបស់អ្នក។

**ហេតុអ្វីវាសំខាន់?** Python គ្រាន់តែប្រៀបធៀបតម្លៃពីរប៉ុណ្ណោះ។ វាមិនអាចដឹងថាតម្លៃមួយណាត្រឹមត្រូវទេ។ ប្រសិនបើអ្នក run កូដជាមុនសិន ហើយចម្លងអ្វីដែលវា print ទៅដាក់ក្នុងតេស្ត នោះតេស្តនឹងការពារអ្វីក៏ដោយដែលកូដធ្វើ **រួមទាំងកំហុសរបស់វាផង**។ ជំនួយការ AI (AI assistant) ដែលសរសេរតេស្តតាមរបៀបនេះ នឹងរាយការណ៍ថា «តេស្តទាំងអស់ជាប់» សម្រាប់កូដដែលខូច។

**តើត្រូវធ្វើយ៉ាងដូចម្ដេច?** ចូរតាមដានកូដដោយដៃ (trace) ជាមុនសិន៖ គណនាចម្លើយពីកិច្ចសន្យា សរសេរវាទុក ហើយទើប run កូដ។ ប្រសិនបើលទ្ធផលទាំងពីរមិនដូចគ្នា មានមួយក្នុងចំណោមវាខុស។ ចូរអានកិច្ចសន្យាម្ដងទៀត ដើម្បីសម្រេចថាមួយណាខុស។

**តើអ្នកជួបវានៅកន្លែងណា?** ក្នុងក្រុមការងារពិត លទ្ធផលដែលរំពឹងទុក ច្រើនតែត្រូវបានសរសេរពីតម្រូវការ (requirements) មុនពេលមានកូដទៀតផង។ ក្រុមខ្លះថែមទាំងសរសេរតេស្តមុន ហើយសរសេរកូដក្រោយ។

👀 **ឧទាហរណ៍។** សេចក្ដីព្រាង (draft) នៃ `open_portions` នេះមានកំហុស។ កិច្ចសន្យា (សប្ដាហ៍ទី 4)៖ បូកចំណែកនៃការផ្ដល់ជូន ដែលមាន status new, approved ឬ picked_up។ 🤔 ជាមុនសិន ចូរគណនា **ដោយដៃ** ថាវាគួរ return អ្វី សម្រាប់ record គំរូទាំង 4 (F001 picked_up 30 · F002 delivered 18 · F003 rejected 40 · F004 approved 12)។ បន្ទាប់មក ចូរ **ទាយមើល** ថាសេចក្ដីព្រាងនោះ print អ្វី។

````python
def open_portions_draft(records):        # សេចក្ដីព្រាងដែលមានកំហុស
    total = 0
    for record in records:
        total = total + record["portions"]
    return total

print(open_portions_draft(bridge.sample_records()))     # 🤔
````

**គណនាដោយដៃ (ពីកិច្ចសន្យា):** …  **រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **មូលហេតុ:** …

🐞 **កំហុសដែលគេជួបញឹកញាប់ · តម្លៃរំពឹងទុក ដែលចម្លងពីលទ្ធផល (output)។** សេចក្ដីព្រាងនោះ print `100`។ តេស្តដែលសរសេរពីលទ្ធផលនោះ គឺ `assert open_portions_draft(records) == 100` នឹង **ជាប់** ហើយឥឡូវវាការពារកំហុសនោះ៖ ក្ដារព័ត៌មាននឹងសន្យាថាមាន 100 ចំណែក ខណៈដែលនៅសល់តែ 42 ចំណែកប៉ុណ្ណោះ ដែលមិនទាន់បញ្ចប់ (open)។ កិច្ចសន្យាផ្ដល់ `30 + 12 = 42` (picked up និង approved; delivered និង rejected មិនរាប់ទេ)។ របៀបកែ៖ ចូរសរសេរ `== 42` ដែលអ្នកបានគណនាដោយដៃ។ តេស្តនោះធ្លាក់ ហើយនោះជាដំណឹងល្អ៖ វាបានរកឃើញកំហុស។

### ✍️ សាកសរសេរទី 2 · `test_urgency`

`urgency` មានចំណុចព្រំដែន **ពីរ**៖ 3 ម៉ោង (urgent ឬ soon) និង 12 ម៉ោង (soon ឬ later)។ ករណីពីរត្រូវបានផ្ដល់ឱ្យរួចហើយ។ ចូរបន្ថែមករណីរបស់អ្នកផ្ទាល់ ពីកិច្ចសន្យា **មុនពេល** អ្នក Run cell៖ ករណីព្រំដែនមួយនៅ 12 ម៉ោង និងករណីចម្លែកមួយ (ម៉ោងនៅសល់តិចបំផុត៖ 0)។ តេស្តរបស់អ្នកត្រូវតែជាប់លើ `urgency_correct` ហើយធ្លាក់លើកំណែដែលមានកំហុសទាំងបី។

````python
def test_urgency(urgency):
    assert urgency(3) == "urgent"     # ចំណុចព្រំដែននៅ 3 ម៉ោង
    assert urgency(4) == "soon"       # លើសវាមួយម៉ោង
    # ✍️ បន្ថែមករណីរបស់អ្នក: ករណីព្រំដែនមួយ និងករណីចម្លែកមួយ ឧទាហរណ៍:
    # assert urgency(12) == ...       # ចំណុចព្រំដែននៅ 12 ម៉ោង
    # assert urgency(0) == ...        # ចម្លែក: គ្មានពេលនៅសល់សោះ


bridge.check("verdict", verdict, examples=[
    ((test_urgency, urgency_correct), "pass"),
    ((test_urgency, urgency_bug_3), "fail"),
    ((test_urgency, urgency_bug_12), "fail"),
    ((test_urgency, urgency_bug_zero), "fail"),
])
````

### 3 · ពិនិត្យអ្វីដែលមិនត្រូវប្រែប្រួល

📖 **វាជាអ្វី?** កិច្ចសន្យាមួយ ក៏ប្រាប់ផងដែរថា function មួយ **មិនត្រូវ** ប្ដូរអ្វីខ្លះ។ `most_urgent_first` នៃសប្ដាហ៍ទី 8 return list **ថ្មី** មួយ ហើយទុក list របស់អ្នកឱ្យនៅដដែល។ `append_event` នៃសប្ដាហ៍ទី 7 បន្ថែមបន្ទាត់មួយ ហើយមិនដែលសរសេរជាន់លើបន្ទាត់ប្រវត្តិមុនៗទេ។ ការប្រែប្រួលដែល function មួយធ្វើ នៅខាងក្រៅចម្លើយរបស់វា ត្រូវបានគេហៅថា **ផលចំហៀង (side effect)**។

**ហេតុអ្វីវាសំខាន់?** ផ្នែកជាច្រើនប្រើ list តែមួយរួមគ្នា (សប្ដាហ៍ទី 4៖ parameter គឺជាឈ្មោះទីពីរ សម្រាប់តម្លៃដដែល)។ ប្រសិនបើរបាយការណ៍មួយតម្រៀប (sort) list ផ្ទាល់របស់អ្នកសម្របសម្រួល អេក្រង់បន្ទាប់នឹងបង្ហាញលំដាប់ខុសពីមុន ហើយពេលរក្សាទុក (save) កូដអាចសរសេរលំដាប់នោះចូលទៅក្នុងថាស (disk)។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `sorted(marks)` អាន list ហើយបង្កើត list **ថ្មី** មួយ។ `marks.sort()` គឺជា method មួយរបស់ list៖ វារៀបលំដាប់ list **នោះផ្ទាល់** ឡើងវិញ ហើយបញ្ជូន `None` មកវិញ។

| | `sorted(marks)` | `marks.sort()` |
|---|---|---|
| បញ្ជូនមកវិញ | list ថ្មីមួយ ដែលរៀបតាមលំដាប់ | `None` |
| list `marks` ក្រោយការហៅ | នៅដដែល | ប្រែប្រួល៖ ឥឡូវរៀបតាមលំដាប់ |
| ប្រើវានៅពេល | អ្នកត្រូវការលំដាប់មួយ សម្រាប់អេក្រង់ ឬរបាយការណ៍ | list នោះជារបស់អ្នក ហើយអ្នកចង់ឱ្យវាប្រែប្រួល |

**តើត្រូវសាកល្បងវាយ៉ាងដូចម្ដេច?** ចូរបង្កើតធាតុចូលដោយខ្លួនឯង ហៅ function រួច assert ថាធាតុចូលនោះ នៅតែដូចអ្វីដែលអ្នកបានបង្កើតពិតប្រាកដ។

**តើអ្នកជួបវានៅកន្លែងណា?** នៅពេលអ្នកតម្រៀបតារាងមួយ នៅលើគេហទំព័រទិញទំនិញ ទិដ្ឋភាព (view) របស់អ្នកប្រែប្រួល ប៉ុន្តែ record ផ្ទាល់របស់ហាង មិនប្រែប្រួលទេ។

👀 **ឧទាហរណ៍។** កំណែពីរនៃ "ពិន្ទុខ្ពស់បំផុតបី" (`reverse=True` ដាក់ពិន្ទុខ្ពស់បំផុតមុនគេ សប្ដាហ៍ទី 8)។ 🤔 **ទាយមើល** បន្ទាត់ទាំងពីរ ជាពិសេស `my_marks` នៅបន្ទាត់ចុងក្រោយ។

````python
def top_three_new(marks):
    return sorted(marks, reverse=True)[:3]

def top_three_in_place(marks):        # តម្រៀប list ដែលគេផ្ដល់ឱ្យវាផ្ទាល់
    marks.sort(reverse=True)
    return marks[:3]

my_marks = [5, 9, 7, 10]
print(top_three_new(my_marks), my_marks)
print(top_three_in_place(my_marks), my_marks)     # 🤔
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **មូលហេតុ:** …

🐞 **កំហុសដែលគេជួបញឹកញាប់ · យកអ្វីដែល `.sort()` បញ្ជូនមកវិញ មកដាក់ក្នុង variable។** `top = marks.sort()` បន្ទាប់មក `top[:3]` ឈប់ដោយ៖

```text
TypeError: 'NoneType' object is not subscriptable
```

ចូរអានសារ error ចាប់ពីបន្ទាត់ចុងក្រោយ៖ `'NoneType'` ប្រាប់ថាតម្លៃនោះគឺ `None` ហើយ "not subscriptable" ប្រាប់ថា អ្នកមិនអាចយក `[...]` ពីវាបានទេ។ `.sort()` បានប្ដូរ `marks` ហើយបញ្ជូន `None` មកវិញ។ របៀបកែ៖ `sorted(marks, reverse=True)[:3]` ដែលបញ្ជូន list ថ្មីមួយមកវិញ ហើយទុក `marks` ឱ្យនៅដដែល។

### ✍️ សាកសរសេរទី 3 · `test_top_three`

កិច្ចសន្យា៖ `top_three(marks)` return ពិន្ទុខ្ពស់បំផុតបី ពីខ្ពស់ទៅទាប **ហើយទុក list ដែលគេផ្ដល់ឱ្យវា ឱ្យនៅដដែល**។ ករណីមួយត្រូវបានផ្ដល់ឱ្យរួចហើយ។ ចូរបន្ថែមអ្វីដែលមិនត្រូវប្រែប្រួល និងករណីចម្លែកមួយ។ តេស្តរបស់អ្នកត្រូវតែជាប់លើ `top_three_new` ហើយធ្លាក់លើ `top_three_in_place` និងលើ `top_lowest`។

````python
def test_top_three(top_three):
    marks = [5, 9, 7, 10]
    assert top_three(marks) == [10, 9, 7]
    # ✍️ បន្ថែមករណីរបស់អ្នក: អ្វីដែលមិនត្រូវប្រែប្រួល និងករណីចម្លែកមួយ ឧទាហរណ៍:
    # assert marks == ...                # list ដែលខ្ញុំផ្ដល់ឱ្យ ក្រោយការហៅ
    # assert top_three([...]) == ...     # ចម្លែក: ពិន្ទុតិចជាងបី


bridge.check("verdict", verdict, examples=[
    ((test_top_three, top_three_new), "pass"),
    ((test_top_three, top_three_in_place), "fail"),
    ((test_top_three, top_lowest), "fail"),
])
````

📖 **វាជាអ្វី?** កូដពីជំនួយការ AI ប្រៀបដូចជាកូដដែលសរសេរដោយមនុស្សមិនស្គាល់មុខម្នាក់។ វាច្រើនតែស្អាត និងមើលទៅមានទំនុកចិត្ត។ AI ថែមទាំងអាចនិយាយថា «ខ្ញុំបានសាកល្បងហើយ» ទោះបីគ្មាននរណាបានសាកល្បងក៏ដោយ។

**ហេតុអ្វីវាសំខាន់?** អ្នកទទួលខុសត្រូវចំពោះកូដដែលអ្នកដំណើរការសម្រាប់មនុស្ស មិនថានរណាជាអ្នកសរសេរវាទេ (សប្ដាហ៍ទី 4)។ អ្នកអាចទទួលខុសត្រូវបាន តែចំពោះកូដដែលអ្នកបានពិនិត្យរួចប៉ុណ្ណោះ។

**តើត្រូវពិនិត្យវាយ៉ាងដូចម្ដេច? ទម្លាប់បីយ៉ាង។**

1. **អានគ្រប់បន្ទាត់។** ចូរតាមដានឧទាហរណ៍មួយដោយដៃ (trace)។ សម្រាប់ប្រយោគនីមួយៗនៃការស្នើសុំរបស់អ្នក ចូររកបន្ទាត់ដែលធ្វើការងារនោះ។ ប្រយោគណាដែលគ្មានបន្ទាត់កូដ គឺជាកំហុស (bug)។
2. **ដំណើរការករណីតេស្តផ្ទាល់ខ្លួនរបស់អ្នក** ដែលសរសេរពីកិច្ចសន្យា មុនពេលអ្នកមើលលទ្ធផល (ចំណុចទី 2)។
3. **សួរអំពីចំណុចព្រំដែន៖** សូន្យ list ទទេ ចំណុចព្រំដែនគត់ និងពាក្យមួយនៅកន្លែងដែលគួរតែជាលេខ។

**តើអ្នកជួបវានៅកន្លែងណា?** ក្នុងក្រុមអ្នកបង្កើតកម្មវិធី (software team) នេះហៅថា **ការត្រួតពិនិត្យកូដ (code review)**៖ គ្មានការកែប្រែណាមួយត្រូវបានបញ្ចូលទេ រហូតដល់មនុស្សទីពីរម្នាក់បានអានវា ហើយបានដំណើរការតេស្ត។

👀 **ឧទាហរណ៍។** នៅសប្ដាហ៍ទី 9 ផ្នែក B អ្នកបានអាន function `rate(part, whole)`។ កិច្ចសន្យារបស់វាមានបីប្រយោគ៖ *ចូរ return ផ្នែក (part) ជាភាគរយនៃទាំងមូល (whole)។ ចូរបង្គត់ឱ្យនៅសល់ខ្ទង់ទសភាគ 1 ខ្ទង់។ ចូរ return 0.0 នៅពេលទាំងមូលស្មើ 0។* មិត្តរួមថ្នាក់ម្នាក់បានផ្ដល់កិច្ចសន្យានោះឱ្យ AI ហើយសុំឱ្យវាសរសេរ function នោះ។ AI បានឆ្លើយមកវិញដោយកូដខាងក្រោម ហើយនិយាយថា *"I tested it, it works ✅"* (ខ្ញុំបានសាកល្បងហើយ វាដំណើរការ)។ 🤔 **អានគ្រប់បន្ទាត់** រួច **ទាយមើល** លទ្ធផលទាំងបី។ ចូរ Run cell។ បន្ទាប់មក ចូរលុបសញ្ញា `#` នៅបន្ទាត់ចុងក្រោយ Run ម្ដងទៀត អានសារ error ចាប់ពីបន្ទាត់ចុងក្រោយ រួចដាក់ `#` វិញ។

````python
def rate_ai(part, whole):
    # សរសេរដោយជំនួយការ AI: "I tested it, it works ✅"
    return round(100 * part / whole, 1)

print(rate_ai(3, 20))
print(rate_ai(20, 20))
# print(rate_ai(0, 0))      # ប្រភេទមួយ ដែលមិនទាន់មានការផ្ដល់ជូនសោះ
````

### ✍️ សាកសរសេរទី 4 · `test_rate`

ចូរសរសេរតេស្ត ដែល AI មិនដែលបានដំណើរការ។ ករណីធម្មតាមួយត្រូវបានផ្ដល់ឱ្យរួចហើយ។ ចូរបន្ថែមករណីព្រំដែនមួយ (ក្រុមទាំងមូល៖ 20 ក្នុងចំណោម 20) និងករណីចម្លែកមួយ (ក្រុមទទេ៖ 0 ក្នុងចំណោម 0)។ ចូរយកចម្លើយដែលរំពឹងទុកពីកិច្ចសន្យា មិនមែនពីការ run កូដទេ។ តេស្តរបស់អ្នកត្រូវតែជាប់លើ `rate_correct` ហើយធ្លាក់លើ `rate_ai` និងលើ `rate_bug_fraction`។

````python
def test_rate(rate):
    assert rate(3, 20) == 15.0        # ធម្មតា: 3 ក្នុងចំណោម 20 គឺ 15 %
    # ✍️ បន្ថែមករណីរបស់អ្នក: ករណីព្រំដែនមួយ និងករណីចម្លែកមួយ ឧទាហរណ៍:
    # assert rate(20, 20) == ...      # ក្រុមទាំងមូល
    # assert rate(0, 0) == ...        # ក្រុមទទេ: តើកិច្ចសន្យានិយាយថាដូចម្ដេច?


bridge.check("verdict", verdict, examples=[
    ((test_rate, rate_correct), "pass"),
    ((test_rate, rate_ai), "fail"),
    ((test_rate, rate_bug_fraction), "fail"),
])
````

**Lab 1 · ពិនិត្យថាលេខសមហេតុផល (sanity check)។** ក្រាហ្វមួយអាចមើលទៅល្អ ប៉ុន្តែនៅតែខុស។ មុនពេលអ្នកទុកចិត្តលេខមួយ ចូរពិនិត្យអ្វីដែល **ត្រូវតែ** ពិតអំពីវា៖ ផ្នែកទាំងអស់បូកគ្នា ស្មើនឹងទាំងមូល អត្រាមួយនៅចន្លោះ 0 % និង 100 % ហើយក្រុមទទេមិនធ្វើឱ្យកូដគាំងទេ។ cell ដែលរៀបចំរួចខាងក្រោម ពិនិត្យការផ្ដល់ជូនទាំង 404 នៃឆមាស ដោយប្រើ `rate_correct` (ប្រសិនបើ runtime ទើបភ្ជាប់ឡើងវិញ ចូរ Run cell ដែលរៀបចំរួចនៃចំណុចទី 1 ជាមុនសិន)។ 🤔 **ទាយមើល:** តើការពិនិត្យណានឹងឈប់ ប្រសិនបើកូដអាន file រំលងជួរចុងក្រោយ? តើការពិនិត្យណានឹងឈប់ ប្រសិនបើ `rate` ចែកបញ្ច្រាស (ទាំងមូល ÷ ផ្នែក)?

````python
# @title Lab 1 · Sanity checks on the semester (run this cell) { display-mode: "form" }
import csv

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    offers = list(csv.DictReader(file))


def count_by(rows, field):                       # ពីសប្ដាហ៍ទី 6
    counts = {}
    for row in rows:
        counts[row[field]] = counts.get(row[field], 0) + 1
    return counts


statuses = count_by(offers, "status")
parts_total = 0
for status in statuses:
    parts_total = parts_total + statuses[status]
assert parts_total == len(offers) == 404, "the statuses must add up to all 404 offers"
print("✅ 1 · The statuses add up to the whole:", statuses, "→", parts_total, "of 404 offers")

rejected = []
for offer in offers:
    if offer["status"] == "rejected":
        rejected.append(offer)
rejections = count_by(rejected, "category")
by_category = count_by(offers, "category")
print("✅ 2 · Every rejection rate is between 0 and 100 %:")
for category in by_category:
    category_rate = rate_correct(rejections.get(category, 0), by_category[category])
    assert 0 <= category_rate <= 100, category + ": a rate must stay between 0 and 100 %"
    print("     ", category, rejections.get(category, 0), "of", by_category[category], "→", category_rate, "%")

assert rate_correct(0, 0) == 0.0, "an empty group must give 0.0"
print("✅ 3 · An empty group gives 0.0, not a crash")
````

### ⚖️ ល្អ ឬមានហានិភ័យ?

អ្នកបានសួរជំនួយការ AI មួយថា៖ *"Write tests for my FoodBridge rules."* (សូមសរសេរតេស្តសម្រាប់វិធាន FoodBridge របស់ខ្ញុំ)។ វាឆ្លើយមកវិញជាមួយ **Version A** ហើយនិយាយថា៖ *"All 3 tests pass ✅."* (តេស្តទាំង 3 ជាប់)។ cell ខាងក្រោមផ្ដល់ផ្នែក **ខូច** បី ទៅឱ្យ Version A ហើយដំណើរការតេស្តរបស់វា។ ចូរអានលទ្ធផល។

````python
# ផ្នែកខូចបី
def pickup_label(portions, approved):     # ភ្លេចការអនុម័ត
    if portions >= 20:
        return "team"
    return "standard"


def urgency(hours_left):                  # 3 ម៉ោង លែងជា urgent ទៀតហើយ
    if hours_left < 3:
        return "urgent"
    return "later"


def open_portions(records):               # រាប់អាហារដែលបានប្រគល់រួចផងដែរ
    total = 0
    for record in records:
        total = total + record["portions"]
    return total


# Version A · តេស្តពីជំនួយការ AI
def test_pickup_label():
    assert pickup_label(30, True) in ["hold", "team", "standard"]


def test_urgency():
    print("urgency(2) is", urgency(2))


def test_open_portions():
    records = [{"portions": 10, "status": "approved"}]
    assert open_portions(records) == open_portions(records)


test_pickup_label()
test_urgency()
test_open_portions()
print("Version A: all 3 tests passed, and all 3 parts are broken.")
````
