### 1 · សំណួររំលឹក: តេស្តពីសប្ដាហ៍ទី 4 (10 នាទី)

1. `assert 5 > 3` និង `assert 3 > 5, "too small"`: តើមានអ្វីកើតឡើងចំពោះបន្ទាត់នីមួយៗ?

3. ករណីធម្មតា ចំណុចព្រំដែន ឬករណីចម្លែក? `urgency(20)` · `urgency(3)` · `total_portions([])`
4. វិធានចែងថា *3 ម៉ោង ឬតិចជាងនេះ គឺ urgent*។ កំហុសមួយសរសេរ `<` ជំនួសឱ្យ `<=`។ តើតេស្តមួយណាចាប់វាបាន: `urgency(1) == "urgent"` ឬ `urgency(3) == "urgent"`?

<details><summary>ចម្លើយ 1–3</summary>

1. បន្ទាត់ទីមួយមិនធ្វើអ្វីទេ (វាពិត)។ បន្ទាត់ទីពីរបញ្ឈប់ ដោយ `AssertionError: too small`។
2. `test_`។ វាជាប់ នៅពេលគ្មាន `assert` ណាមួយបញ្ឈប់វា។
3. ករណីធម្មតា (តម្លៃប្រចាំថ្ងៃ) · ចំណុចព្រំដែន (ត្រង់កន្លែងដែលវិធានប្ដូរចម្លើយ) · ករណីចម្លែក (list ទទេ)។

</details>

តេស្តមួយមានប្រយោជន៍ លុះត្រាតែវា **អាចធ្លាក់បាន**: លើកំណែមួយដែលមានកំហុស។ ដើម្បីមើលរឿងនេះ យើងឱ្យ function ដែលត្រូវតេស្ត ទៅតេស្តជា **ធាតុចូល** របស់វា (ដូចដែលអ្នកបានឱ្យ function key មួយទៅ `sorted` នៅសប្ដាហ៍ទី 8)។ `verdict` Run តេស្តមួយ លើកំណែមួយ ហើយ return `"pass"` ឬ `"fail"`; ការគាំង (crash) រាប់ថាជា `"fail"`។

🤔 **ទាយមើល:** តើលទ្ធផលទាំងបួននឹងជាអ្វី?

````python
def verdict(test, version):
    """Run one test on one version of a function. Return "pass" or "fail"."""
    try:
        test(version)
    except Exception:            # assert មួយធ្លាក់ ឬកូដគាំង
        return "fail"
    return "pass"


def urgency_correct(hours_left):
    if hours_left <= 3:
        return "urgent"
    if hours_left <= 12:
        return "soon"
    return "later"


def urgency_bug_3(hours_left):                # 3 ម៉ោង លែងជា urgent ទៀតហើយ
    if hours_left < 3:
        return "urgent"
    if hours_left <= 12:
        return "soon"
    return "later"


def test_one_hour(urgency):
    assert urgency(1) == "urgent"


def test_three_hours(urgency):
    assert urgency(3) == "urgent"


print("1 hour: ", verdict(test_one_hour, urgency_correct), "·", verdict(test_one_hour, urgency_bug_3))
print("3 hours:", verdict(test_three_hours, urgency_correct), "·", verdict(test_three_hours, urgency_bug_3))
````

### Lab 1 · ការពិនិត្យសមហេតុផល (sanity check) លើលេខ (10 នាទី)

ក្រាហ្វមួយអាចមើលទៅល្អ ប៉ុន្តែនៅតែខុស។ មុនពេលអ្នកទុកចិត្តលេខមួយ ចូរពិនិត្យអ្វីដែល **ត្រូវតែ** ពិតអំពីវា: ផ្នែកទាំងអស់បូកគ្នាស្មើនឹងទាំងមូល អត្រាមួយនៅចន្លោះ 0 % និង 100 % ហើយក្រុមទទេមិនធ្វើឱ្យកូដគាំងទេ (គ្មានការចែកនឹង 0)។ `rate` ពីសប្ដាហ៍ទី 9 របស់អ្នក គឺជារបស់ទីមួយដែលត្រូវតេស្ត។ កំណែខាងក្រោមមានកំហុស ដែលការពិនិត្យទាំងនេះចាប់បាន។

````python
def rate_correct(part, whole):
    if whole == 0:
        return 0.0
    return round(100 * part / whole, 1)


def rate_bug_zero(part, whole):          # គាំង នៅពេលក្រុមទទេ
    return round(100 * part / whole, 1)


def rate_bug_fraction(part, whole):      # ភ្លេច 100: ជាប្រភាគ មិនមែនជាភាគរយទេ
    if whole == 0:
        return 0.0
    return round(part / whole, 1)


def rate_bug_flipped(part, whole):       # ចែកបញ្ច្រាស: ច្រើនជាង 100 %
    if part == 0:
        return 0.0
    return round(100 * whole / part, 1)
````

````python
import csv

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    offers = list(csv.DictReader(file))


def count_by(rows, field):                       # ពីសប្ដាហ៍ទី 6
    counts = {}
    for row in rows:
        counts[row[field]] = counts.get(row[field], 0) + 1
    return counts


statuses = count_by(offers, "status")
parts_total = 0
for status in statuses:
    parts_total = parts_total + statuses[status]
assert parts_total == len(offers) == 404
print("✅ 1 · The statuses add up to the whole:", parts_total, "of 404 offers")

rejected = []
for offer in offers:
    if offer["status"] == "rejected":
        rejected.append(offer)
rejections = count_by(rejected, "category")
by_category = count_by(offers, "category")
for category in by_category:
    assert 0 <= rate_correct(rejections.get(category, 0), by_category[category]) <= 100
print("✅ 2 · Every rejection rate is between 0 and 100 %")

assert rate_correct(0, 0) == 0.0
print("✅ 3 · An empty group gives 0.0, not a crash")
````

    """Return an HTML page (a str): the open offers, most urgent first."""
    rows = ""
    for record in most_urgent_first(records):
        if record["status"] in ["new", "approved", "picked_up"]:

````python
# ផ្នែកខូចបី
def pickup_label(portions, approved):     # ភ្លេចការអនុម័ត
    if portions >= 20:
        return "team"
    return "standard"


def urgency(hours_left):                  # 3 ម៉ោង លែងបន្ទាន់ទៀតហើយ
    if hours_left < 3:
        return "urgent"
    return "later"


def open_portions(records):               # រាប់អាហារដែលបានប្រគល់រួចផងដែរ
    total = 0
    for record in records:
        total = total + record["portions"]
    return total


# Version A · តេស្តពីជំនួយការ AI
def test_pickup_label():
    assert pickup_label(30, True) in ["hold", "team", "standard"]


def test_urgency():
    print("urgency(2) is", urgency(2))


def test_open_portions():
    records = [{"portions": 10, "status": "approved"}]
    assert open_portions(records) == open_portions(records)


test_pickup_label()
test_urgency()
test_open_portions()
print("Version A: all 3 tests passed, and all 3 parts are broken.")
````

### ⭐⭐ ក្រោយពេលដាក់ស្នើ: តេស្តមួយទៀត

ពេលរង់ចាំវេនរបស់អ្នកនៅតុគ្រូ ចូរសាកលំហាត់តាមកម្រិត (Ladder 2) (វាមិនប្ដូរ folder របស់អ្នកទេ)។ កំណែខាងក្រោម គឺ `open_portions` ដែលត្រឹមត្រូវមួយ និងកំណែបីដែលមានកំហុស។

````python
def open_correct(records):
    total = 0
    for record in records:
        if record["status"] in ["new", "approved", "picked_up"]:
            total = total + record["portions"]
    return total


def open_bug_all(records):              # រាប់អាហារដែលបានប្រគល់រួចផងដែរ
    total = 0
    for record in records:
        total = total + record["portions"]
    return total


def open_bug_picked_up(records):        # ភ្លេច picked_up
    total = 0
    for record in records:
        if record["status"] in ["new", "approved"]:
            total = total + record["portions"]
    return total


def open_bug_empty(records):            # list ទទេ ផ្ដល់ None ជំនួសឱ្យ 0
    if records == []:
        return None
    return open_correct(records)
````

1. **ទៅកាន់ខ្លួនអ្នកនៅសប្ដាហ៍ទី 1:** នៅសប្ដាហ៍ទី 1 តើអ្នកបានឆ្លើយសំណួរ «តើប្រព័ន្ធព័ត៌មាន (information system) ជាអ្វី?» យ៉ាងដូចម្ដេច? ឥឡូវនេះ តើអ្នកនឹងឆ្លើយក្នុងមួយប្រយោគយ៉ាងដូចម្ដេច? ចូរប្រាប់ពីគោលការណ៍មួយ ដែលបានប្ដូរគំនិតរបស់អ្នក។
2. **ការសម្រេចចិត្តដែលអ្នកមិនសូវប្រាកដជាងគេ:** តើជាការសម្រេចចិត្តមួយណា ក្នុងចំណោមការសម្រេចចិត្តទាំងបីរបស់អ្នក? តើទិន្នន័យអ្វី ឬសំឡេងរបស់នរណា ដែលអាចប្ដូរគំនិតរបស់អ្នកបាន?
3. **ពេលមួយដែលអ្នកបានពិនិត្យ មុននឹងទុកចិត្ត:** កូដរបស់ AI កូដផ្ទាល់ខ្លួនរបស់អ្នក ឬក្រាហ្វមួយដែលមើលទៅត្រឹមត្រូវ ប៉ុន្តែខុស។ តើអ្នកកត់សម្គាល់ឃើញយ៉ាងដូចម្ដេច? លើកក្រោយ តើអ្នកនឹងពិនិត្យអ្វីមុនគេ?
4. **ជំហានបន្ទាប់របស់អ្នក:** ចូរជ្រើសរើសប្រព័ន្ធមួយនៅជុំវិញខ្លួនអ្នក (ការចុះឈ្មោះចូលរៀន កម្មវិធីធនាគារ កម្មវិធីហៅឡាន)។ ក្នុងពីរបន្ទាត់: តើនរណារងផលប៉ះពាល់ពីវា ហើយតើអ្នកនឹងពិនិត្យយ៉ាងដូចម្ដេចថា វាដំណើរការត្រឹមត្រូវ?
