### 1 · test គឺជា function ដែលពិនិត្យការសន្យាមួយ

អ្នកបានជួប `assert` នៅសប្ដាហ៍ទី 4។ **test function** ដាក់ការសន្យាមួយ ក្នុង function ដែលឈ្មោះចាប់ផ្ដើមដោយ `test_`។ ឈ្មោះប្រាប់ពីការសន្យា។ សារនៅក្រោយសញ្ញាក្បៀស ប្រាប់ពីអ្វីដែលគួរតែពិត។ test ដែលជាប់ មិន print អ្វីទាំងអស់។ test ដែលធ្លាក់ raise `AssertionError`។

````python
def discount(price):
    if price >= 100:
        return 10
    return 0

def test_discount_at_the_boundary():
    assert discount(100) == 10, "a price of exactly 100 gets the discount"

def test_no_discount_below_the_boundary():
    assert discount(99) == 0, "a price below 100 gets no discount"

test_discount_at_the_boundary()
test_no_discount_below_the_boundary()
print("Both tests passed (a passing test prints nothing).")
````

🤔 **ទាយមើល:** ប្ដូរ `>=` ទៅ `>` ក្នុង `discount` ហើយ run cell ម្ដងទៀត។ តើ test មួយណាធ្លាក់ ហើយបន្ទាត់ចុងក្រោយនៃ error និយាយអ្វី? បន្ទាប់មក ប្ដូរវាត្រឡប់ទៅ `>=` វិញ ដើម្បីឱ្យ **Runtime → Restart session and run all** នៅតែដំណើរការ។

### 2 · Run test ច្រើន ដោយមិនឈប់

test ដែលធ្លាក់មួយ មិនគួរលាក់ test ផ្សេងទៀតទេ។ runner តូចនេះហៅ test នីមួយៗ ចាប់ការធ្លាក់ណាមួយដោយ `try`/`except` (សប្ដាហ៍ទី 5) ហើយ print ✅ ឬ ❌។ list នៃ function ដំណើរការដូច hooks ដែរ: អ្នកសរសេរឈ្មោះ ដោយគ្មាន `( )`។ `test.__name__` គឺជាឈ្មោះរបស់ function។

````python
def run_tests(tests):
    passed = 0
    for test in tests:
        try:
            test()
            print("✅", test.__name__)
            passed = passed + 1
        except Exception as error:
            print("❌", test.__name__, "→", type(error).__name__ + ":", error)
    print(passed, "of", len(tests), "tests passed")

run_tests([test_discount_at_the_boundary, test_no_discount_below_the_boundary])
````

`run_tests.py` របស់ kit ធ្វើដូចគ្នានេះ សម្រាប់ files។ Final project របស់អ្នកមាន function `run_tests` ដូចគ្នានេះ (ផ្នែកទី 5)។ វា run គ្រប់ test ដែលអ្នកដាក់ក្នុង list របស់វា។

### 3 · ការអាន output របស់ test

Run test របស់ kit ក្នុង process ថ្មីមួយ ដូចនៅសប្ដាហ៍ទី 9។ អាន **បន្ទាត់ចុងក្រោយ** មុនគេ: វារាប់ចំនួន test។

````python
import subprocess
import sys

kit_tests = subprocess.run([sys.executable, "run_tests.py"], cwd="foodbridge_kit",
                           capture_output=True, text=True)
print(kit_tests.stdout.splitlines()[-1])
````

ពេល test មួយធ្លាក់ output មើលទៅដូចនេះ (ពីប្រព័ន្ធរបស់មិត្តរួមថ្នាក់ម្នាក់):

```text

❌ test_25_portions_need_a_team → AssertionError: 25 portions is my team minimum, so it needs a team
✅ test_offer_without_contact_is_refused
2 of 3 tests passed
```

អានវាតាមលំដាប់នេះ:

1. **បន្ទាត់ចុងក្រោយ:** តើធ្លាក់ប៉ុន្មាន?
2. **ឈ្មោះរបស់ test ដែលធ្លាក់:** តើការសន្យាណាដែលខូច?
3. **សារ:** តើអ្វីគួរតែពិត?
4. **បន្ទាប់មក សម្រេចចិត្ត:** តើ code ខុស test ខុស ឬការសម្រេចចិត្តមួយបានប្ដូរ?

🤔 ក្នុង output ខាងលើ តើអ្វីជារឿង **ទីមួយ** ដែលអ្នកនឹងពិនិត្យ? (ជំនួយ: តើលេខ 25 ស្ថិតនៅកន្លែងណា?)

### 4 · clean run: Runtime → Restart session and run all

- **Colab:** Runtime → Restart session and run all។ **Jupyter:** Kernel → Restart & Run All។
- គ្រប់ cell ត្រូវតែ run ពីលើចុះក្រោម ដោយគ្មាន error។ បន្ទាប់មក អាន output ម្ដងទៀត។

អ្នកបានឃើញនៅសប្ដាហ៍ទី 9 ថា process ថ្មីមិនដឹងអ្វីពី notebook របស់អ្នកទេ។ 🤔 **ទាយមើល** ថានឹងមានអ្វីកើតឡើងនៅទីនេះ រួច run។

### 6 · ការពិនិត្យ code ដែល AI សរសេរ

ជំនួយការ AI សរសេរ code ដែល **run បាន** ហើយ **មើលទៅ** ដូចជាបានបញ្ចប់។ run បាន មិនមែនមានន័យថាត្រឹមត្រូវទេ។ ការពិនិត្យ (review) មានប្រាំជំហាន:

1. **អានការស្នើសុំ។** តើ code គួរសន្យាអ្វី?
2. **អាន code ជាមួយបញ្ជីត្រួតពិនិត្យ។** កត់បញ្ហានីមួយៗ និងបន្ទាត់របស់វា។
3. **test edge case:** ករណីធម្មតា ករណី boundary និងករណីចម្លែក (ទទេ សូន្យ ឬបាត់)។
4. **ភ្ជាប់វាចូល ហើយតាមដាន ⚠️។** fallback មានន័យថា code របស់អ្នក *មិន* ត្រូវបានប្រើទេ។
5. **សម្រេចចិត្ត:** ទទួលយក កែ ឬបដិសេធ ហើយសរសេរមូលហេតុ។

🤔 ជំនួយការ AI បានសរសេរ function នេះ។ **ទាយមើល** ថានឹងមានអ្វីកើតឡើង សម្រាប់និស្សិតដែលមិនទាន់មានពិន្ទុសោះ។ លុប `#` រួច run អាន error បន្ទាប់មកដាក់ `#` មកវិញ។

````python
def average_mark(marks):          # សរសេរដោយជំនួយការ AI
    return sum(marks) / len(marks)

print(average_mark([60, 70, 80]))
# print(average_mark([]))
````

helper បី ធ្វើឱ្យ test ខ្លី។ `ALL_SETTINGS` គឺជា setting ទាំងអស់: value របស់អ្នក ពេលអ្នកបានប្ដូរវា ហើយ default សម្រាប់ setting ផ្សេងទៀត។

````python
ALL_SETTINGS = bridge.System(MY_SETTINGS).settings  # បង្ហាញ ⚠️ សម្រាប់ setting ដែលសរសេរខុស


def fresh_system():
    """Return a new system with my settings and hooks, after the sample day."""
    system = bridge.System(MY_SETTINGS, hooks=MY_HOOKS, quiet=True)
    system.run_day(show=False)
    return system


def offer_row(system, portions):
    """Return a raw offer row that passes every input check (portions may be changed)."""
    return {"donor": "Test Donor", "contact": "test@example.org", "food": "Test rice",
            "category": system.settings["CATEGORIES"][0], "portions": str(portions),
            "hours_left": "8"}


def approved_offer(system, portions):
    """Add a test offer, approve it for the first partner, and return the Offer."""
    offer_id, errors = system.add_offer(offer_row(system, portions))
    assert offer_id is not None, "the test offer was refused: " + "; ".join(errors)
    system.move(offer_id, "approved", partner=system.settings["PARTNERS"][0])
    return system.get(offer_id)


changed = bridge.settings.changed_settings(ALL_SETTINGS)
print("Settings that differ from the defaults:", list(changed))
print("Your hooks:", list(MY_HOOKS))
````

### ⭐ P1 · អាន output ហើយរក bug

មិត្តរួមថ្នាក់ម្នាក់បានសរសេរ `too_late` និង test បី។ Run cell ហើយអាន output តាមលំដាប់ដែលបានរៀនក្នុងវគ្គទី 1។ រក bug កែវា (តួអក្សរតែមួយ) ហើយ run ម្ដងទៀត រហូតដល់ test ទាំងបីជាប់។

````python
REJECT_IF_HOURS_LEFT_BELOW = 1

def too_late(hours_left):
    """Return True when there is not enough time left to collect the food."""
    return hours_left <= REJECT_IF_HOURS_LEFT_BELOW

def test_zero_hours_is_too_late():
    assert too_late(0), "0 hours left: too late to collect"

def test_exactly_the_minimum_is_accepted():
    assert not too_late(1), "1 hour left is exactly the minimum, so the offer is accepted"

def test_plenty_of_time_is_accepted():
    assert not too_late(10), "10 hours left: plenty of time"

run_tests([test_zero_hours_is_too_late, test_exactly_the_minimum_is_accepted,
           test_plenty_of_time_is_accepted])
````
