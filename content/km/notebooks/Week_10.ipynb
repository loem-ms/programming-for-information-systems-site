🤔 **ទាយមើល:** តើបន្ទាត់ចុងក្រោយ print អ្វី?

### 2 · តេស្តមួយ គឺជា function ដែលអាចបរាជ័យបាន

**តេស្ត (test)** គឺជា function មួយ ដែលឈ្មោះចាប់ផ្ដើមដោយ `test_`។ វាពិនិត្យ **ការសន្យាមួយ** ដោយប្រើបន្ទាត់ `assert` មួយចំនួន។ វាជាប់ (pass) នៅពេលគ្មានអ្វីកើតឡើង ហើយវាធ្លាក់ (fail) នៅពេល `assert` មួយបញ្ឈប់វា។ តេស្តមួយមានប្រយោជន៍ លុះត្រាតែវា **អាចធ្លាក់បាន**: តេស្តដែលជាប់ ទោះបីកូដខូចក៏ដោយ មិនការពារនរណាម្នាក់ទេ។

ដើម្បីមើលថាតេស្តរបស់អ្នកអាចធ្លាក់បានឬទេ យើងប្រគល់ function ដែលត្រូវតេស្ត ទៅឱ្យវាជា **ធាតុចូល (input)** (ដូចដែលអ្នកបានឱ្យ key function ទៅ `sorted` នៅសប្ដាហ៍ទី 8)។ បន្ទាប់មក យើងសាកតេស្តនោះលើកំណែ (version) ដែលត្រឹមត្រូវ និងលើកំណែដែលមានកំហុស (bug)។ `verdict` ធ្វើការនេះ ហើយ return `"pass"` ឬ `"fail"`។ ការគាំង (crash) ក៏រាប់ជា `"fail"` ដែរ។

🤔 **ទាយមើល:** តើលទ្ធផលទាំងពីរនឹងជាអ្វី? តើតេស្តខ្លីនេះ ចាប់កំហុសនៅក្នុង `fare_bug_12` បានឬទេ?

```python
def verdict(test, version):
    """Run one test on one version of a function. Return "pass" or "fail"."""
    try:
        test(version)
    except Exception:            # assert មួយធ្លាក់ ឬកូដគាំង
        return "fail"
    return "pass"


def fare_correct(age):
    if age < 0:
        return None
    if age < 12:
        return 1000
    return 2000


def fare_bug_12(age):            # កុមារអាយុ 12 ឆ្នាំ បង់តម្លៃកុមារ
    if age < 0:
        return None
    if age <= 12:
        return 1000
    return 2000


def fare_bug_negative(age):      # ភ្លេចថា អាយុក្រោម 0 គឺជាកំហុស
    if age < 12:
        return 1000
    return 2000


def test_fare_short(bus_fare):
    assert bus_fare(30) == 2000


print(verdict(test_fare_short, fare_correct))
print(verdict(test_fare_short, fare_bug_12))       # 🤔
```

```python
def average_correct(marks):
    if marks == []:
        return 0
    return round(sum(marks) / len(marks), 1)


def average_bug_empty(marks):        # គាំងពេល list ទទេ
    return round(sum(marks) / len(marks), 1)


def average_bug_whole(marks):        # // បោះចោលខ្ទង់ទសភាគ
    if marks == []:
        return 0
    return sum(marks) // len(marks)


print(average_correct([8, 9]))
print(average_bug_whole([8, 9]))     # 🤔
```

### ការកែប្រែផ្ទាល់ (live change)

ក្រោយការបង្ហាញរបស់អ្នក គ្រូនឹងសុំឱ្យអ្នកប្ដូររឿងមួយ ឧទាហរណ៍ *«លាក់ប្រភេទ (category) ពីសាធារណៈ»* ឬ *«ក្រុមមួយ ចាប់ពី 25 ចំណែក»*។ អ្នកត្រូវ៖

- និយាយថាការសម្រេចចិត្តនោះនៅកន្លែងណា
- **ទាយមើល** ថាអ្វីនឹងប្រែប្រួល
- ធ្វើការប្ដូរនៅក្នុង file នោះ ហើយ Run cell

ចូរហាត់ម្ដងឥឡូវនេះ។ **សូមចាំ:** ចូរប្ដូរការកែប្រែផ្ទាល់ត្រឡប់មកវិញ មុនពេលអ្នកបង្កើត ZIP (ឬបង្កើត ZIP សម្រាប់ដាក់ស្នើ មុនពេលធ្វើការកែប្រែផ្ទាល់) ដើម្បីកុំឱ្យអ្នកដាក់ស្នើតេស្តដែលធ្លាក់។

### ដាក់ស្នើ (submit)

1. ចូរប្ដូរការកែប្រែផ្ទាល់ត្រឡប់មកវិញ (លុះត្រាតែអ្នកបានបង្កើត ZIP មុនពេលនោះរួចហើយ) រួចធ្វើ clean run ចុងក្រោយ។ ត្រូវប្រាកដថា `README.md` បានបំពេញគ្រប់គ្រាន់។
2. ចូរបង្កើត file ZIP មួយនៃ folder របស់អ្នក។ នៅក្នុង Google Drive: ចុចកណ្ដុរស្ដាំលើ `my_foodbridge` → **Download** (Drive បង្កើត ZIP ឱ្យ)។ ឬ Run cell ខាងក្រោម (វាក៏ធានាថា `saved/offers.json` មាននៅទីនោះផងដែរ) រួចនៅក្នុងផ្ទាំង **Files** ចុចកណ្ដុរស្ដាំលើ `my_foodbridge_submission.zip` → **Download**។
3. ចូរផ្ញើ ZIP តាមអ៊ីមែលទៅ `submit.progmis@gmail.com`។ ថ្ងៃផុតកំណត់ និងឈ្មោះ file នឹងប្រកាសនៅក្នុងថ្នាក់។

```python
import json
import os
import shutil

os.makedirs("my_foodbridge/saved", exist_ok=True)
if not os.path.exists("my_foodbridge/saved/offers.json"):
    with open("my_foodbridge/saved/offers.json", "w", encoding="utf-8") as f:
        json.dump(bridge.sample_records(), f, indent=2)

shutil.make_archive("my_foodbridge_submission", "zip", ".", "my_foodbridge")
print("Made my_foodbridge_submission.zip: download it from the Files panel.")
```
