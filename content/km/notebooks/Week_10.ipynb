### 1 · តេស្ត (test) គឺជា function ដែលពិនិត្យការសន្យាមួយ

អ្នកបានស្គាល់ `assert` រួចហើយនៅសប្ដាហ៍ទី 4។ **តេស្ត** (test function) ដាក់ការសន្យាមួយនៅក្នុង function ដែលឈ្មោះរបស់វាចាប់ផ្ដើមដោយ `test_`។ ឈ្មោះនោះប្រាប់ពីការសន្យា។ សារដែលនៅក្រោយសញ្ញាក្បៀស ប្រាប់ថាអ្វីដែលត្រូវពិត។ តេស្តដែលជាប់ មិន print អ្វីទាំងអស់។ តេស្តដែលធ្លាក់ នឹង raise `AssertionError`។

````python
def discount(price):
    if price >= 100:
        return 10
    return 0

def test_discount_at_the_boundary():
    assert discount(100) == 10, "a price of exactly 100 gets the discount"

def test_no_discount_below_the_boundary():
    assert discount(99) == 0, "a price below 100 gets no discount"

test_discount_at_the_boundary()
test_no_discount_below_the_boundary()
print("Both tests passed (a passing test prints nothing).")
````

🤔 **ទាយមើល:** ចូរប្ដូរ `>=` ទៅជា `>` ក្នុង `discount` ហើយ Run cell ម្ដងទៀត។ តើតេស្តមួយណាធ្លាក់ ហើយតើបន្ទាត់ចុងក្រោយនៃសារ error សរសេរថាអ្វី? បន្ទាប់មក ចូរប្ដូរវាត្រឡប់ទៅ `>=` វិញ ដើម្បីឱ្យ **Runtime → Restart session and run all** នៅតែដំណើរការបាន។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **ហេតុអ្វី:** …

យើងអាចសរសេរតេស្តមួយ សម្រាប់ការសន្យាមួយ។ បញ្ហាបន្ទាប់: តេស្តមួយដែលធ្លាក់ មិនគួរលាក់តេស្តផ្សេងទៀតទាំងអស់ទេ។

### 2 · ដំណើរការតេស្តច្រើន ដោយមិនឈប់

តេស្តមួយដែលធ្លាក់ មិនគួរលាក់តេស្តផ្សេងទៀតទេ។ កម្មវិធីតូចនេះ (runner) ហៅតេស្តនីមួយៗ ចាប់កំហុសណាមួយដោយប្រើ `try`/`except` (សប្ដាហ៍ទី 5) ហើយ print ✅ ឬ ❌។ list នៃ function ប្រើដូច hooks ដែរ៖ អ្នកសរសេរតែឈ្មោះ ដោយគ្មាន `( )`។ `test.__name__` គឺជាឈ្មោះរបស់ function។

````python
def run_tests(tests):
    passed = 0
    for test in tests:
        try:
            test()
            print("✅", test.__name__)
            passed = passed + 1
        except Exception as error:
            print("❌", test.__name__, "→", type(error).__name__ + ":", error)
    print(passed, "of", len(tests), "tests passed")

run_tests([test_discount_at_the_boundary, test_no_discount_below_the_boundary])
````

`run_tests.py` របស់ kit ធ្វើការងារដូចគ្នានេះ សម្រាប់ file។ Final project របស់អ្នកក៏មាន function `run_tests` ដូចគ្នានេះដែរ (ផ្នែកទី 5)។ វាដំណើរការគ្រប់តេស្តដែលអ្នកដាក់ក្នុង list របស់វា។

យើងអាចដំណើរការតេស្តច្រើន ហើយឃើញលទ្ធផលនីមួយៗ។ បញ្ហាបន្ទាប់: អានលទ្ធផលឱ្យបានល្អគ្រប់គ្រាន់ ដើម្បីរកកំហុស (bug) ពិតប្រាកដ។

### 3 · ការអានលទ្ធផល (output) របស់តេស្ត

````python
import subprocess
import sys

kit_tests = subprocess.run([sys.executable, "run_tests.py"], cwd="foodbridge_kit",
                           capture_output=True, text=True)
print(kit_tests.stdout.splitlines()[-1])
````

### 4 · clean run: Runtime → Restart session and run all

notebook ចងចាំគ្រប់ cell ដែលអ្នកបាន Run សូម្បីតែ cell ដែលអ្នកបានកែ ឬលុបនៅពេលក្រោយក៏ដោយ។ ដូច្នេះ notebook អាចមើលទៅល្អនៅលើអេក្រង់របស់អ្នក ប៉ុន្តែនៅតែអាចដំណើរការមិនបាន នៅពេលអ្នកផ្សេងបើកវា។ **clean run** ចាប់ផ្ដើមពីអង្គចងចាំទទេ៖

- **Colab:** Runtime → Restart session and run all។ **Jupyter:** Kernel → Restart & Run All។
- គ្រប់ cell ត្រូវតែដំណើរការពីលើចុះក្រោម ដោយគ្មានបញ្ហា (error)។ បន្ទាប់មក ចូរអានលទ្ធផល (output) ម្ដងទៀត។

អ្នកបានឃើញនៅសប្ដាហ៍ទី 9 រួចហើយថា process ថ្មីមិនដឹងអ្វីទាំងអស់ពី notebook របស់អ្នក។ 🤔 **ទាយមើល៖** តើនឹងមានអ្វីកើតឡើងនៅទីនេះ? បន្ទាប់មក ចូរ Run។

````python
total_today = 48       # ស្រមៃថា អ្នកបានវាយបន្ទាត់នេះក្នុង cell មួយ ហើយក្រោយមកបានលុប cell នោះ

with open("tomorrow.py", "w") as file:
    file.write("print('Portions picked up today:', total_today)\n")

fresh = subprocess.run([sys.executable, "tomorrow.py"], capture_output=True, text=True)
print("Return code:", fresh.returncode)
print("Last line of the error:", fresh.stderr.splitlines()[-1])
````

1. **ចូរអានការស្នើសុំ។** តើកូដគួរសន្យាអ្វី?
2. **ចូរអានកូដ ដោយប្រើបញ្ជីត្រួតពិនិត្យ។** កត់ត្រាបញ្ហានីមួយៗ និងបន្ទាត់ដែលមានបញ្ហា។
3. **ចូរសាកល្បងករណីនៅចំណុចព្រំដែន៖** ករណីធម្មតាមួយ (normal case) ករណីព្រំដែនមួយ (edge case) និងករណីចម្លែកមួយ (ទទេ សូន្យ ឬបាត់)។
4. **ចូរដាក់បញ្ចូល (plug in) វា ហើយមើលរក ⚠️។** ប្រសិនបើមាន fallback មានន័យថាកូដរបស់អ្នក *មិន* ត្រូវបានប្រើទេ។
5. **ចូរសម្រេចចិត្ត៖** ទទួលយក កែសម្រួល ឬបដិសេធ ហើយសរសេរមូលហេតុឱ្យច្បាស់។

🤔 ជំនួយការ AI បានសរសេរ function នេះ។ **ទាយមើល៖** តើនឹងមានអ្វីកើតឡើង សម្រាប់និស្សិតដែលមិនទាន់មានពិន្ទុសោះ? ចូរលុប `#` រួច Run ហើយអានសារ error។ បន្ទាប់មក ចូរដាក់ `#` ត្រឡប់មកវិញ។

````python
def average_mark(marks):          # សរសេរដោយជំនួយការ AI
    return sum(marks) / len(marks)

print(average_mark([60, 70, 80]))
# print(average_mark([]))
````

យើងឆ្លើយជាបីជំហាន៖

1. ចូរសរសេរវិធានតូចនោះដោយឡែក ដោយដាក់ចំនួនអប្បបរមាជា setting ដែលមានឈ្មោះ។
2. ចូរសរសេរ function តេស្តមួយ សម្រាប់ការសន្យាមួយ ដោយ function នីមួយៗមាន `assert` មួយ និងសារមួយ។
3. ចូរហៅតេស្តទាំងនោះ: តេស្តដែលជាប់ មិន print អ្វីទេ ដូច្នេះ `print` ចុងក្រោយមួយ ប្រាប់ថាតេស្តទាំងពីរជាប់។

វាគឺជាលំនាំតេស្តពីផ្នែកទី 1 ដោយដក argument `approved` ចេញ: វិធាននេះមាន input តែមួយ ដូច្នេះមានតែចំណុចព្រំដែនទេ ដែលត្រូវបង្ហាញ។

````python
TEAM_MINIMUM = 25

def needs_team(portions):
    return portions >= TEAM_MINIMUM

def test_exactly_the_minimum_needs_a_team():
    assert needs_team(25) == True, "25 portions is the minimum, so it needs a team"

def test_one_below_the_minimum_does_not():
    assert needs_team(24) == False, "24 portions is below the minimum, so no team"

test_exactly_the_minimum_needs_a_team()
test_one_below_the_minimum_does_not()
print("Both tests passed")
````

**របៀបដែលវាដំណើរការ:** `needs_team(25)` ផ្ដល់ `True` ព្រោះ 25 >= 25 ដូច្នេះ `assert` ទីមួយជាប់ ហើយមិន print អ្វីទេ។ `needs_team(24)` ផ្ដល់ `False` ដូច្នេះ `assert` ទីពីរក៏ជាប់ដែរ។ `print` ចុងក្រោយដំណើរការ លុះត្រាតែ assert ទាំងពីរជាប់: ប្រសិនបើវិធានប្រើ `>` ជំនួសឱ្យ `>=` `assert` ទីមួយនឹងបញ្ឈប់ cell ដោយសាររបស់វា ហើយ `print` នឹងមិនដំណើរការឡើយ។

````python
ALL_SETTINGS = bridge.System(MY_SETTINGS).settings  # បង្ហាញ ⚠️ សម្រាប់ setting ដែលសរសេរខុស


def fresh_system():
    """Return a new system with my settings and hooks, after the sample day."""
    system = bridge.System(MY_SETTINGS, hooks=MY_HOOKS, quiet=True)
    system.run_day(show=False)
    return system


def offer_row(system, portions):
    """Return a raw offer row that passes every input check (portions may be changed)."""
    return {"donor": "Test Donor", "contact": "test@example.org", "food": "Test rice",
            "category": system.settings["CATEGORIES"][0], "portions": str(portions),
            "hours_left": "8"}


def approved_offer(system, portions):
    """Add a test offer, approve it for the first partner, and return the Offer."""
    offer_id, errors = system.add_offer(offer_row(system, portions))
    assert offer_id is not None, "the test offer was refused: " + "; ".join(errors)
    system.move(offer_id, "approved", partner=system.settings["PARTNERS"][0])
    return system.get(offer_id)


changed = bridge.settings.changed_settings(ALL_SETTINGS)
print("Settings that differ from the defaults:", list(changed))
print("Your hooks:", list(MY_HOOKS))
````

````python
REJECT_IF_HOURS_LEFT_BELOW = 1

def too_late(hours_left):
    """Return True when there is not enough time left to collect the food."""
    return hours_left <= REJECT_IF_HOURS_LEFT_BELOW

def test_zero_hours_is_too_late():
    assert too_late(0), "0 hours left: too late to collect"

def test_exactly_the_minimum_is_accepted():
    assert not too_late(1), "1 hour left is exactly the minimum, so the offer is accepted"

def test_plenty_of_time_is_accepted():
    assert not too_late(10), "10 hours left: plenty of time"

run_tests([test_zero_hours_is_too_late, test_exactly_the_minimum_is_accepted,
           test_plenty_of_time_is_accepted])
````

ឥឡូវ ចូរសាកល្បងការអះអាងរបស់ AI ថា *"វាដំណើរការបាន"*។ ចូរដាក់បញ្ចូល (plug in) Version A ទៅក្នុងប្រព័ន្ធរបស់អ្នក ហើយអាន **គ្រប់** បន្ទាត់នៃលទ្ធផល។

````python
def ai_pickup_label(portions, approved):
    """Improved pickup label."""
    if portions >= 25:
        print("team")
    else:
        print("standard")

ai_system = bridge.System(MY_SETTINGS, hooks={"pickup_label": ai_pickup_label})
ai_system.run_day(show=False)
print(ai_system.report())
print("Fell back:", ai_system.fallbacks)
````

- **គោលការណ៍ទី 6 · មើលរំលងការអនុម័ត។** `approved` មិនដែលត្រូវបានអានទេ។ សំណើថ្មីដែលអ្នកសម្របសម្រួលមិនទាន់អនុម័ត នឹងទទួលបាន label `team` ជំនួសឱ្យ `hold`។ អ្នកស្ម័គ្រចិត្តអាចបើកឡានទៅយកអាហារ ដែលក្រោយមកត្រូវបានបដិសេធ ដូចជាចេកដែលគ្មានកន្លែងរក្សាទុកត្រជាក់ (F003)។ ប្រសិនបើ settings របស់អ្នកទទួលយក noodle soup (គុយទាវ ដែលជាសំណើថ្មីមួយ) អ្នកអាចឃើញបញ្ហានេះក្នុងលទ្ធផលខាងលើ៖ AI បាន print `team` សម្រាប់ `pickup_label(25, False)`។

#### ⭐ version ដែលអ្នកបានពិនិត្យ

ចូរកែ function របស់ AI ដើម្បីឱ្យវាជាប់ការពិនិត្យ៖

- return ចម្លើយ ជំនួសឱ្យការ print។
- ត្រូវពិនិត្យការអនុម័ត (`approved`)។
- អានចំនួនអប្បបរមាសម្រាប់ក្រុម ពី `ALL_SETTINGS` ជំនួសឱ្យការសរសេរលេខ 25។

````python
def reviewed_pickup_label(portions, approved):
    """Improved pickup label."""
    if portions >= 25:
        print("team")
    else:
        print("standard")

bridge.check("pickup_label", reviewed_pickup_label, MY_SETTINGS)
````
