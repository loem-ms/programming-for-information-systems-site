### 1 · សរសេរ និងអាន text file

`open` ភ្ជាប់កម្មវិធីរបស់អ្នកទៅនឹង file មួយ។ `with` បិទ file ឱ្យអ្នក នៅចុងបញ្ចប់នៃ block សូម្បីតែនៅពេលមានអ្វីមួយខុសក៏ដោយ។ `"w"` មានន័យថា *write* (សរសេរ)។ `for` loop លើ file មួយ ផ្ដល់ឱ្យអ្នកម្ដងមួយបន្ទាត់។

🤔 **ទាយមើល:** តើ loop នឹង print ប៉ុន្មានបន្ទាត់? ហើយតើ `line.strip()` ដកអ្វីចេញ? បន្ទាប់មក ចូរ run cell នេះ។

````python
with open("shopping.txt", "w", encoding="utf-8") as file:
    file.write("rice\n")
    file.write("eggs\n")

with open("shopping.txt", encoding="utf-8") as file:
    text = file.read()
print(text)

with open("shopping.txt", encoding="utf-8") as file:
    for line in file:
        print("-", line.strip())
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

- `"\n"` បញ្ចប់បន្ទាត់មួយ។ `write` មិនបន្ថែមវាឱ្យអ្នកទេ ហើយ `strip()` ដកវាចេញ នៅពេលអ្នកអាន។
- `encoding="utf-8"` អនុញ្ញាតឱ្យ file ផ្ទុកភាសាណាក៏បាន រួមទាំងភាសាខ្មែរ។
- ប្រសិនបើគ្មាន `"w"` ទេ `open` អាន file តែប៉ុណ្ណោះ។

នៅក្នុង Colab សូមចុចលើរូប folder នៅខាងឆ្វេង ដើម្បីមើល `shopping.txt`។ **File នៅក្នុង folder របស់ Colab នឹងបាត់ នៅពេល runtime reset។** ដើម្បីរក្សា file ទុក សូម download វា (right-click → Download) ឬរក្សាទុកវានៅក្នុង Google Drive របស់អ្នក។

### 2 · `"w"` ធ្វើឱ្យ file ទទេ, `"a"` បន្ថែមនៅខាងចុង

🤔 **ទាយមើល:** តើផ្នែកចុងក្រោយនឹង print អ្វី? សូមចាំថា៖ ឥឡូវនេះ `shopping.txt` មាន *rice* និង *eggs* នៅក្នុងវា។

````python
with open("shopping.txt", "w", encoding="utf-8") as file:
    file.write("bananas\n")

with open("shopping.txt", "a", encoding="utf-8") as file:
    file.write("bread\n")

with open("shopping.txt", encoding="utf-8") as file:
    print(file.read())
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **ហេតុអ្វី:** …

`"w"` ធ្វើឱ្យ file ទទេ **ភ្លាមៗនៅពេលវាបើក** គឺមុនពេលអ្នកសរសេរអ្វីមួយចូល។ Rice និង eggs បាត់អស់ ហើយ Python មិនព្រមានអ្នកអ្វីសោះ។ `"a"` (*append*) រក្សាអ្វីៗដែលមានស្រាប់ទាំងអស់ ហើយបន្ថែមនៅខាងចុង។ ការបើក file ដោយប្រើ mode ខុស គឺជាមូលហេតុមួយដែលកើតឡើងញឹកញាប់បំផុត ដែលធ្វើឱ្យកម្មវិធីកុំព្យូទ័របាត់បង់ទិន្នន័យ។

### 4 · JSON: list និង dictionary ជា text

File មួយអាចផ្ទុកបានតែ text ប៉ុណ្ណោះ។ **JSON** គឺជាវិធីស្ដង់ដារមួយ សម្រាប់សរសេរ list, dictionary, លេខ, text និង `True`/`False` ជា text។ មនុស្សអាចអានវាបាន ហើយភាសាសរសេរកម្មវិធីស្ទើរតែទាំងអស់ អាចអាន (load) វាបាន។

`json.dumps` បម្លែងទិន្នន័យទៅជា JSON text។ `json.loads` បម្លែង JSON text ត្រឡប់ទៅជាទិន្នន័យវិញ។ (អក្សរ `s` មានន័យថា *string*។)

````python
import json

phone = {"owner": "Dara", "credit_usd": 5, "plans": ["data", "calls"], "active": True}
text = json.dumps(phone)
print(text)
print(type(text))

back = json.loads(text)
print(back["credit_usd"] + 1)
````

### 5 · `json.dump` និង `json.load` ជាមួយ file

នៅពេលគ្មានអក្សរ `s` ទេ function ដដែលទាំងនេះធ្វើការដោយផ្ទាល់ជាមួយ file។

````python
classes = [{"code": "MIS101", "students": 42}, {"code": "ACC102", "students": 35}]

with open("classes.json", "w", encoding="utf-8") as file:
    json.dump(classes, file, indent=2)

with open("classes.json", encoding="utf-8") as file:
    loaded = json.load(file)

print(loaded[1]["students"])
print(loaded == classes)
````

`indent=2` ដាក់ item នីមួយៗនៅលើបន្ទាត់ដាច់ដោយឡែក ដូច្នេះមនុស្សក៏អាចអាន file នេះបានដែរ។ ចូរបើក `classes.json` នៅក្នុង file panel ហើយមើលវា។

### 6 · JSON Lines: ប្រវត្តិដែលយើងគ្រាន់តែបន្ថែមចូល

ប្រវត្តិមានទម្រង់ខុសពីនេះ។ យើងមិនដែលសរសេរវាឡើងវិញទេ គឺយើង **បន្ថែមមួយបន្ទាត់ សម្រាប់ព្រឹត្តិការណ៍នីមួយៗ**។ បន្ទាត់នីមួយៗគឺជា JSON object តូចមួយ។ របៀបនេះហៅថា **JSON Lines** ហើយឈ្មោះ file ប្រភេទនេះបញ្ចប់ដោយ `.jsonl`។

````python
def log_event(path, event):
    with open(path, "a", encoding="utf-8") as file:     # "a": បន្ថែម មិនដែលធ្វើឱ្យទទេ
        file.write(json.dumps(event) + "\n")

log_event("phone_history.jsonl", {"time": "09:00", "who": "Dara", "event": "top up", "usd": 5})
log_event("phone_history.jsonl", {"time": "09:30", "who": "Dara", "event": "bought data", "usd": 3})

with open("phone_history.jsonl", encoding="utf-8") as file:
    for line in file:
        event = json.loads(line)
        print(event["time"], event["who"], event["event"], event["usd"])
````

### ផ្នែកទី 3 · ដាក់បញ្ចូល (plug in): រក្សាទុក និងអាន (load) ដោយប្រើ function របស់អ្នក

ដំបូង សូមមើល settings របស់អ្នកសម្រាប់សប្ដាហ៍នេះ។ អ្នកនឹងសម្រេចចិត្តលើ `LOG_EVENTS` នៅចុងវគ្គនេះ។ បន្ទាប់មក cell បន្ទាប់នឹង៖

- ដំណើរការថ្ងៃគំរូ។
- រក្សាទុកសំណើ ដោយប្រើ `to_record` **របស់អ្នក** និងការរក្សាទុកដោយសុវត្ថិភាព (safe save)។
- អានសំណើមកវិញ ដោយប្រើ `from_record` **របស់អ្នក**។

````python
LOG_EVENTS = ["added", "input_rejected", "moved"]   # 🧭 ការសម្រេចចិត្តរបស់អ្នក (មើលចុងវគ្គនេះ)
MY_SETTINGS = {"LOG_EVENTS": LOG_EVENTS}
````

````python
my_system = bridge.System(MY_SETTINGS)
my_system.run_day(show=False)

# 1. រក្សាទុក: Offer objects -> records (to_record របស់អ្នក) -> JSON file (safe save)
records_to_save = []
for offer in my_system.offers:
    records_to_save.append(to_record(offer))
bridge.storage.save_records(records_to_save, "my_offers.json")
print("Saved", len(records_to_save), "records to my_offers.json")

# 2. អានមកវិញ (load): JSON file -> records -> Offer objects (from_record របស់អ្នក)
loaded_offers = []
for record in bridge.storage.load_records("my_offers.json"):
    loaded_offers.append(from_record(record))
print("Loaded", len(loaded_offers), "offers")
````

````python
# តើមានអ្វីប្រែប្រួលនៅតាមផ្លូវទេ? ប្រៀបធៀបសំណើដែលបានអានមកវិញ ជាមួយ records ផ្ទាល់របស់ប្រព័ន្ធ។
records_after = []
for offer in loaded_offers:
    if offer is None:
        print("from_record returned None: it is not finished yet")
    else:
        print(offer.describe(), "· partner:", offer.partner)
        records_after.append(offer.to_record())

if records_after == my_system.records():
    print("✅ Every offer came back exactly the same.")
else:
    print("❌ Something was lost or changed between saving and loading.")
````

ចូរបើក `my_offers.json` នៅក្នុង file panel។ វាជា text ធម្មតា ដែលមនុស្សអាចអានបាន ហើយវានឹងនៅតែមាន ទោះបីកម្មវិធីឈប់ដំណើរការក៏ដោយ (រហូតដល់ Colab reset)។
