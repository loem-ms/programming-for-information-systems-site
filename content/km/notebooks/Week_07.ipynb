### នៅចុងសប្ដាហ៍នេះ អ្នកនឹងអាច …

- **ពន្យល់** ពីភាពខុសគ្នារវាងអង្គចងចាំ (memory) និងថាស (disk) ហើយថាតើ file មួយ និង path របស់វាជាអ្វី;
- **សរសេរ និងអាន** file ដោយប្រើ `open` និង `with` ហើយ **ជ្រើសរើស** `"w"` ឬ `"a"` ដើម្បីកុំឱ្យបាត់បង់ទិន្នន័យ;
- **រក្សាទុក** record ជា JSON **រក្សា** ប្រវត្តិជា JSON Lines ហើយ **ពន្យល់** ថាហេតុអ្វីបានជាការរក្សាទុកដោយសុវត្ថិភាព សរសេរ file បណ្ដោះអាសន្នមួយជាមុនសិន;
- **សម្រេចចិត្ត** ថាតើប្រវត្តិរបស់អ្នករក្សាទុកព្រឹត្តិការណ៍ប្រភេទណាខ្លះ ដោយយកលេខមួយពីប្រវត្តិដែលអ្នកបានរក្សាទុកផ្ទាល់ ធ្វើជាភស្តុតាង (evidence)។

### ដំណើររឿង

folder របស់អ្នកមាន file បួន៖ `rules.py` និង `records.py` (សប្ដាហ៍ទី 4), `data_model.py` ដែលជាការពិនិត្យទិន្នន័យមុនទទួលចូលប្រព័ន្ធ (សប្ដាហ៍ទី 5) និង `lifecycle.py` ដែលអនុញ្ញាតតែការប្ដូរស្ថានភាពដែលសមហេតុផលប៉ុណ្ណោះ (សប្ដាហ៍ទី 6)។ **កូដ** របស់អ្នកមានសុវត្ថិភាព៖ វាស្ថិតនៅក្នុង Google Drive របស់អ្នក។

ប៉ុន្តែ **ទិន្នន័យ** របស់អ្នកមិនមានសុវត្ថិភាពទេ។ ម្សិលមិញ នៅម៉ោង 4 រសៀល Colab បានចាប់ផ្ដើមឡើងវិញ ហើយទិន្នន័យពេញមួយថ្ងៃបាត់អស់៖ ការផ្ដល់ជូន (offer) ការអនុម័ត និងការប្រគល់នីមួយៗ ស្ថិតនៅតែក្នុងអង្គចងចាំប៉ុណ្ណោះ។ បន្ទាប់មក Sunrise Bakery ទូរស័ព្ទមកសួរថា៖ *«តើនំបុ័ង (bread rolls) 30 ចំណែករបស់យើង ទៅណាហើយ?»* គ្មាន file សម្រាប់បើកមើល ហើយក៏គ្មានប្រវត្តិសម្រាប់អានដែរ។ នៅសប្ដាហ៍នេះ អ្នកនឹងបន្ថែម `storage.py` ដើម្បីឱ្យប្រព័ន្ធរបស់អ្នកចងចាំបាន។

## ចាប់ផ្ដើមថ្នាក់ · ចែករំលែកកិច្ចការស្រាវជ្រាវទី 5 (10 នាទី)

ក្នុងកិច្ចការស្រាវជ្រាវទី 5 អ្នកបានអានអំពីអ្វីដែលកើតឡើងចំពោះទិន្នន័យ នៅពេលដាច់ភ្លើង (file, មូលដ្ឋានទិន្នន័យ (database), backup (ទិន្នន័យបម្រុងទុក)) បានសួរ AI អំពីចំណុចដែលនៅមិនទាន់ច្បាស់ បានពិនិត្យការអះអាង (claim) មួយនៅក្នុង Colab ហើយបានសរសេរមតិរបស់អ្នកលើសំណួរ *«តើមានពេលណាដែលប្រព័ន្ធគួរលុបទិន្នន័យចោលពិតប្រាកដដែរឬទេ?»* ចូរប្រៀបធៀបជាមួយមិត្តរួមថ្នាក់ពីរនាក់៖

1. ចូរប្រាប់រឿងមួយ ដែលសៀវភៅ និង AI ពន្យល់ខុសគ្នា ឬកន្លែងដែល AI ឆ្លើយមិនច្បាស់ ឬឆ្លើយខុស។ តើអ្នកបានសាកល្បងការអះអាងមួយណានៅក្នុង Colab ហើយតើវាត្រឹមត្រូវឬទេ?
2. តើមានពេលណាដែលប្រព័ន្ធគួរលុបទិន្នន័យចោលពិតប្រាកដដែរឬទេ? ចូរប្រាប់ជំហររបស់អ្នក និងមូលហេតុ ក្នុងពីរប្រយោគ។ បន្ទាប់មក ចូរស្ដាប់ឱ្យបានមូលហេតុល្អមួយ ពីភាគីម្ខាងទៀត។
3. នៅសប្ដាហ៍នេះ ប្រវត្តិរបស់អ្នកគ្រាន់តែកើនឡើងប៉ុណ្ណោះ៖ `append_event` បន្ថែមមួយបន្ទាត់ ហើយមិនដែលសរសេរបន្ទាត់មុនៗឡើងវិញទេ ចំណែកការរក្សាទុកដោយសុវត្ថិភាព (safe save) មិនដែលបន្សល់ file ដែលសរសេរបានពាក់កណ្ដាលទេ។ តើគំនិតទាំងពីរនេះ ស្របនឹងជំហររបស់អ្នកក្នុងសំណួរទី 2 ដែរឬទេ? តើមានអ្វីដែល FoodBridge នៅតែគួរលុបចោលនៅថ្ងៃណាមួយដែរឬទេ ឧទាហរណ៍ ព័ត៌មានទំនាក់ទំនងឯកជនរបស់អ្នកបរិច្ចាគ?

## វគ្គទី 1 · ស្វែងយល់: អង្គចងចាំ file និងប្រវត្តិ (90 នាទី)

### គំនិតសំខាន់: អង្គចងចាំភ្លេច ប៉ុន្តែ record មិនត្រូវភ្លេចទេ

សូមគិតអំពីហាងតូចមួយ។ នៅពេលថ្ងៃ ម្ចាស់ហាងកត់ត្រាការលក់នៅលើក្ដារស ដែលត្រូវលុបឱ្យស្អាតនៅពេលយប់។ ម្ចាស់ហាងដែលប្រុងប្រយ័ត្ន ក៏មាន **សៀវភៅកត់ត្រាការលក់** មួយដែរ៖ ការលក់នីមួយៗ ត្រូវបានសរសេរជាបន្ទាត់ថ្មីមួយដោយទឹកខ្មៅ ព្រមទាំងម៉ោង។ កំហុសមួយត្រូវបានកែ ដោយសរសេរបន្ទាត់ថ្មីមួយទៀត ("សងលុយវិញ, ម៉ោង 2 រសៀល") មិនមែនដោយហែកទំព័រចោលទេ។

អង្គចងចាំ (memory) របស់កម្មវិធីកុំព្យូទ័រដែលកំពុងដំណើរការ គឺប្រៀបដូចជាក្ដារសនោះ៖ list និង dictionary របស់អ្នក បាត់អស់ នៅពេលកម្មវិធីឈប់ ឬនៅពេល Colab ចាប់ផ្ដើមឡើងវិញ។ ប្រព័ន្ធព័ត៌មានមួយ ត្រូវការរបស់ពីរយ៉ាងទៀត៖

- **ទិន្នន័យនៅតែមាន (persistence):** record ត្រូវបានរក្សាទុកក្នុង file មួយ ដូច្នេះនៅថ្ងៃស្អែក វានៅតែមាន។
- **ប្រវត្តិសម្រាប់ត្រួតពិនិត្យ (audit history):** បញ្ជីនៃព្រឹត្តិការណ៍ (event) ដែលប្រាប់ថា **តើមានអ្វីបានកើតឡើង និងនៅពេលណា** (ក្នុងប្រព័ន្ធពិត វាក៏ប្រាប់ផងដែរថា **នរណា** ជាអ្នកធ្វើ)។ យើងគ្រាន់តែបន្ថែមទៅក្នុងវាប៉ុណ្ណោះ៖ វាជាប្រវត្តិ **បានតែបន្ថែម (append-only)** មិនកែ និងមិនលុបបន្ទាត់ចាស់ទេ។ ដូច្នេះ នៅពេលអ្នកបរិច្ចាគម្នាក់សួរថា *«តើនំបុ័ងរបស់ខ្ញុំ ទៅណាហើយ?»* នឹងមាននរណាម្នាក់អាចឆ្លើយបាន។

ការគិតបែបអ្នកសរសេរកម្មវិធី (program thinking) នៅសប្ដាហ៍នេះ៖ **តាមដានកូដដោយដៃ (trace)** ដើម្បីដឹងថា file មួយផ្ទុកអ្វីខ្លះ ក្រោយជំហាននីមួយៗ **សាកល្បងនៅចំណុចព្រំដែន** (មិនទាន់មាន file, ប្រវត្តិទទេ) និង **អានសារ error** ចាប់ពីបន្ទាត់ចុងក្រោយ។

### 1 · អង្គចងចាំ (memory) និងថាស (disk): file មួយ និង path របស់វា

📖 **វាជាអ្វី?** **អង្គចងចាំ (memory)** ផ្ទុកតម្លៃនានារបស់កម្មវិធីដែលកំពុងដំណើរការ៖ គ្រប់ឈ្មោះ គ្រប់ list និងគ្រប់ dictionary ដែលអ្នកបង្កើត។ **ថាស (disk)** (ថាសរឹង (hard disk), SSD, Google Drive) ផ្ទុក **file**។ file គឺជាទិន្នន័យមួយដុំដែលមានឈ្មោះ នៅលើថាស ហើយច្រើនតែជាអត្ថបទ (text)។ **path** (ទីតាំងរបស់ file) ប្រាប់ថា file មួយនៅឯណា៖ folder ដែលវានៅក្នុងនោះ បន្ទាប់មកឈ្មោះរបស់វា ដោយភ្ជាប់គ្នាដោយសញ្ញា `/`។ `my_foodbridge/demo_day/offers.json` គឺជា file `offers.json` នៅក្នុង folder `demo_day` ដែលនៅក្នុង folder `my_foodbridge` របស់អ្នក។ path ដែលគ្មាន folder ដូចជា `shopping.txt` មានន័យថា "នៅក្នុង folder បច្ចុប្បន្ន (current folder)"។ នៅក្នុង Colab folder នោះគឺជាកន្លែងធ្វើការបណ្ដោះអាសន្នរបស់ runtime (`/content`) ដែលនឹងបាត់ទៅវិញ នៅពេល runtime ត្រូវបានលុប។ ចំណែក notebook របស់អ្នកផ្ទាល់ ត្រូវបានរក្សាទុកក្នុង Drive ហើយ folder `my_foodbridge/` របស់អ្នកក៏ដូច្នោះដែរ។

**ហេតុអ្វីបានជាត្រូវការវា?** អង្គចងចាំដំណើរការលឿន ប៉ុន្តែវាភ្លេច៖ នៅពេលកម្មវិធីឈប់ ឬ runtime ចាប់ផ្ដើមឡើងវិញ អង្គចងចាំនឹងទទេ។ file នៅលើថាស នៅតែមាន រហូតដល់មាននរណាម្នាក់លុបវា ដូច្នេះអ្វីៗដែលអ្នកត្រូវការនៅថ្ងៃស្អែក ត្រូវតែសរសេរទុកនៅលើថាស។

|---|---|---|

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `open(path)` ផ្ដល់ឱ្យអ្នកនូវ **file object** មួយ៖ ការតភ្ជាប់ទៅកាន់ file ដែលអ្នកប្រើសម្រាប់អាន ឬសរសេរ។ ប្រសិនបើគ្មាន mode (របៀបបើក file) ទេ `open` គ្រាន់តែអាន file ប៉ុណ្ណោះ។ `open(path, "w", encoding="utf-8")` បើក file សម្រាប់ **សរសេរ** (ចំណុចទី 2 និយាយលម្អិតបន្ថែម) ហើយ `encoding="utf-8"` អនុញ្ញាតឱ្យ file ផ្ទុកភាសាណាក៏បាន រួមទាំងភាសាខ្មែរ។ `file.write(text)` ដាក់អត្ថបទនោះចូលក្នុង file ដូចដែលវាមានពិតប្រាកដ ហើយមិនបន្ថែមអ្វីទាំងអស់៖ បន្ទាត់មួយបញ្ចប់ តែនៅកន្លែងដែលអ្នកសរសេរ `"\n"` ប៉ុណ្ណោះ ដែលជាសញ្ញាចុះបន្ទាត់ (line break)។ `with open(...) as file:` បិទការតភ្ជាប់នៅចុងបញ្ចប់នៃក្រុមបន្ទាត់ដែលចូលបន្ទាត់ (block) របស់វា **ទោះបីមានអ្វីមួយខុសនៅពាក់កណ្ដាលទីក៏ដោយ**។ ការបិទនេះសំខាន់ ព្រោះវាជាពេលដែល Python សរសេរអត្ថបទចុងក្រោយ ចូលទៅក្នុងថាសឱ្យចប់សព្វគ្រប់។ `for` loop លើ file មួយ ផ្ដល់ឱ្យអ្នកម្ដងមួយបន្ទាត់ ហើយបន្ទាត់នីមួយៗនៅតែបញ្ចប់ដោយ `"\n"`។

**តើអ្នកជួបវានៅកន្លែងណា?** ទូរស័ព្ទរបស់អ្នកនៅតែរក្សារូបថតរបស់អ្នក នៅពេលវាបិទ ព្រោះរូបថតទាំងនោះជា file។ ហ្គេមមួយ "រក្សាទុក (save)" កម្រិតដែលអ្នកលេងដល់ ទៅក្នុង file មួយ។ ទម្រង់បែបបទ (form) លើ web ដែលភ្លេចអ្វីដែលអ្នកបានវាយ ក្រោយពេលអ្នកបើកទំព័រឡើងវិញ (reload) បានរក្សាវាទុកតែក្នុងអង្គចងចាំប៉ុណ្ណោះ។

👀 **ឧទាហរណ៍។** 🤔 **ទាយមើល** ថាតើ loop នឹង print ប៉ុន្មានបន្ទាត់ ហើយ `line.strip()` ដកអ្វីចេញ។

````python
with open("shopping.txt", "w", encoding="utf-8") as file:
    file.write("rice\n")
    file.write("eggs\n")

with open("shopping.txt", encoding="utf-8") as file:
    for line in file:
        print("-", line.strip())
````

### 2 · `"w"` ធ្វើឱ្យ file ទទេ, `"a"` បន្ថែមនៅខាងចុង

📖 **វាជាអ្វី?** argument ទីពីររបស់ `open` គឺជា **mode** (របៀបបើក file)៖ អ្វីដែលអ្នកចង់ធ្វើជាមួយ file នោះ។

| mode | file មិនទាន់មាន | file មានរួចហើយ | ប្រើវាសម្រាប់ |
|---|---|---|---|
| គ្មាន mode (ឬ `"r"`) | `FileNotFoundError` | អានវាពីដើម | ការអាន |

### 3 · JSON: record ជាអត្ថបទ និង "តើ file មានឬទេ?"

📖 **វាជាអ្វី?** **file អត្ថបទ (text file)** ផ្ទុកបានតែអត្ថបទប៉ុណ្ណោះ មិនមែន list និង dictionary របស់ Python ទេ។ record របស់អ្នកគឺជា list និង dictionary ដូច្នេះវាត្រូវតែប្ដូរទៅជាអត្ថបទជាមុនសិន។ **JSON** គឺជាវិធីស្ដង់ដារមួយ សម្រាប់សរសេរវាជាអត្ថបទ។ វាមើលទៅស្ទើរតែដូច Python៖

| Python | អត្ថបទ JSON |
|---|---|
| `{"id": "F001", "portions": 30}` | `{"id": "F001", "portions": 30}` |
| `True`, `False`, `None` | `true`, `false`, `null` |
| list នៃ dictionary `[{...}, {...}]` | list នៃ object `[{...}, {...}]` |

**ហេតុអ្វីបានជាត្រូវការវា?** នៅសប្ដាហ៍ទី 5 គ្រប់តម្លៃដែលមកពី file CSV គឺជាអត្ថបទ ដូច្នេះអ្នកត្រូវការ `int()`។ JSON រក្សាលេខឱ្យនៅជាលេខ ហើយវាអាចផ្ទុក list នៅខាងក្នុង dictionary បាន។ មនុស្សអាចអានវាបាន ហើយភាសាសរសេរកម្មវិធីស្ទើរតែទាំងអស់ អាចទាញយកវាមកប្រើ (load) បាន។ (notebook នេះក៏ជា JSON ដែរ៖ file `.ipynb` គឺជាអត្ថបទ JSON។)

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `json.dump(data, file)` ប្ដូរ list និង dictionary របស់អ្នកទៅជាអត្ថបទ JSON ហើយសរសេរវាចូលទៅក្នុង file ដែលកំពុងបើក។ `json.load(file)` អានអត្ថបទនោះ ហើយបង្កើត list និង dictionary **ថ្មី** នៅក្នុងអង្គចងចាំ។ `indent=2` ដាក់ item នីមួយៗនៅលើបន្ទាត់ដាច់ដោយឡែក ដូច្នេះមនុស្សអាចអាន file នោះបាន។ `json.dumps(data)` និង `json.loads(text)` (ដែលមានអក្សរ `s` នៅខាងចុង) ធ្វើការងារដូចគ្នា ប៉ុន្តែជាមួយ **string** ជំនួសឱ្យ file។ អ្នកត្រូវការវា នៅពេលអ្នកសរសេរម្ដងមួយបន្ទាត់។

មុនពេលអ្នកអាន file មួយ ចូរសួរជាមុនថា តើវាមានឬទេ៖ `os.path.exists(path)` ឆ្លើយ `True` ឬ `False`។ នៅថ្ងៃដំបូងនៃប្រព័ន្ធថ្មីមួយ មិនទាន់មាន file នៅឡើយទេ ហើយនោះជារឿងធម្មតា មិនមែនជាបញ្ហា (error) ទេ។ លំនាំ (pattern) ដែលអ្នកនឹងប្រើពេញមួយសប្ដាហ៍៖ **ប្រសិនបើ file មាន ត្រូវអានវា។ ប្រសិនបើមិនមាន ត្រូវចាប់ផ្ដើមជាមួយ list ទទេ។**

**តើអ្នកជួបវានៅកន្លែងណា?** នៅពេល app អាកាសធាតុមួយ សួរ server រកការព្យាករណ៍អាកាសធាតុថ្ងៃនេះ ចម្លើយត្រឡប់មកវិញជាអត្ថបទ JSON។

👀 **ឧទាហរណ៍។** 🤔 **ទាយមើល** តម្លៃទាំងពីរដែលនឹងត្រូវ print៖ តើ `loaded[1]["students"] + 1` ជាលេខ ឬជា error (អត្ថបទ + លេខ)? តើ `diary.json` មានឬទេ?

````python
import json
import os

classes = [{"code": "MIS101", "students": 42}, {"code": "ACC102", "students": 35}]
with open("classes.json", "w", encoding="utf-8") as file:
    json.dump(classes, file, indent=2)
with open("classes.json", encoding="utf-8") as file:
    loaded = json.load(file)
print(loaded[1]["students"] + 1, os.path.exists("diary.json"))
````

### 4 · JSON Lines: ប្រវត្តិដែលអ្នកគ្រាន់តែបន្ថែមចូល

| | `offers.json` (record) | `history.jsonl` (ប្រវត្តិ) |
|---|---|---|

### ផ្នែក B · មើលពីខាងក្នុង (under the hood) (≈ 45 នាទី)

#### 🧑‍🏫 តើ `storage.py` ដំណើរការយ៉ាងដូចម្ដេច

file របស់អ្នកមាន function បួន ហើយ **ប្រព័ន្ធ** ជាអ្នកហៅវា។ នេះគឺជាលំហូរទិន្នន័យ (data flow) នៅថ្ងៃគំរូ៖

```text
 the system (memory)                              my_foodbridge/demo_day/ (disk)
 an offer is added or moved ── save_records ──▶  offers.json.tmp ── os.replace ──▶ offers.json   (replaced)
 an event of a kind in LOG_EVENTS ── append_event ───────────────────────────────▶ history.jsonl (+1 line)

 after a restart:   load_records ◀── offers.json        read_events ◀── history.jsonl
```

**ការរក្សាទុកដោយសុវត្ថិភាព (safe save)។** តើមានអ្វីកើតឡើង ប្រសិនបើភ្លើងដាច់ ខណៈពេលកំពុងសរសេរ `offers.json`? ជាមួយ `open("offers.json", "w")` file ត្រូវបានធ្វើឱ្យទទេជាមុនសិន។ ក្នុងរយៈពេលមួយភ្លែត ច្បាប់ចម្លង **តែមួយគត់** នៃគ្រប់ record ត្រូវបានសរសេរបានត្រឹមពាក់កណ្ដាល។ ប្រសិនបើភ្លើងដាច់នៅពេលនោះ នៅព្រឹកបន្ទាប់ `json.load` នឹងជួបអត្ថបទដែលខូច (`JSONDecodeError` ពីវគ្គទី 1) ហើយប្រព័ន្ធមិនអាចចាប់ផ្ដើមបានទេ។ ការផ្ដល់ជូនទាំងអស់បាត់បង់ មិនមែនត្រឹមតែការផ្លាស់ប្ដូរចុងក្រោយប៉ុណ្ណោះទេ។

ការរក្សាទុកដោយសុវត្ថិភាព មិនដែលសរសេរចូលទៅក្នុង file ពិតដោយផ្ទាល់ទេ៖

1. សរសេរ list ថ្មីទាំងមូល ចូលទៅក្នុង **file បណ្ដោះអាសន្ន (temporary file)** មួយ ឈ្មោះ `offers.json.tmp` នៅក្នុង folder ដដែល។
2. ប្ដូរវាចូលជំនួស ដោយប្រើ `os.replace("offers.json.tmp", "offers.json")`។

`os.replace` ធ្វើឱ្យឈ្មោះ `offers.json` ចង្អុលទៅ file ថ្មី **ក្នុងមួយជំហាន**។ គ្មានពេលណាមួយដែល `offers.json` ជាចាស់ពាក់កណ្ដាល ថ្មីពាក់កណ្ដាលទេ។ ដូច្នេះ ការដាច់ភ្លើងអាចប៉ះពាល់បានតែ file បណ្ដោះអាសន្នប៉ុណ្ណោះ។ ប្រសិនបើវាឈប់ក្នុងអំឡុងជំហានទី 1 `offers.json` នៅតែជា file ពេញលេញចុងក្រោយ៖ យ៉ាងអាក្រក់បំផុត គឺបាត់បង់តែការផ្លាស់ប្ដូរចុងក្រោយប៉ុណ្ណោះ មិនដែលបាត់អស់ទាំងអស់ទេ។ file បណ្ដោះអាសន្ននៅក្នុង folder ដដែល ព្រោះការប្ដូរចូលជំនួសក្នុងមួយជំហាននេះ ដំណើរការបានតែនៅលើថាសតែមួយប៉ុណ្ណោះ។

#### កូដចាប់ផ្ដើមនៃវគ្គទី 3 ម្ដងមួយបន្ទាត់

នៅវគ្គទី 3 `save_records` ត្រូវបានផ្ដល់ឱ្យអ្នករួច ពេញលេញ។ ចូរអានវាម្ដងមួយបន្ទាត់៖

```python
def save_records(records, path):
    """Save every record as JSON. Write a temporary file first, then swap it in."""
    temp_path = path + ".tmp"                                # offers.json → offers.json.tmp
    with open(temp_path, "w", encoding="utf-8") as file:     # "w" មានសុវត្ថិភាពនៅទីនេះ: វាធ្វើឱ្យទទេតែ file បណ្ដោះអាសន្នប៉ុណ្ណោះ
        json.dump(records, file, indent=2)
    os.replace(temp_path, path)                              # ប្ដូរវាចូលជំនួស ក្នុងមួយជំហាន
```

អ្នកសរសេរ function បីទៀត។ នេះគឺជាកិច្ចសន្យា (contract) របស់វា និងមូលហេតុដែលវាត្រូវបានកំណត់បែបនេះ៖

| Function | អ្វីដែលវាទទួល | អ្វីដែលវា return | ហេតុអ្វីបានជាកិច្ចសន្យាបែបនេះ |
|---|---|---|---|
| `save_records(records, path)` | list នៃ record និងកន្លែងដែលត្រូវរក្សាទុកវា | គ្មានអ្វីទេ (`None`)៖ file គឺជាលទ្ធផល | list ទាំងមូលត្រូវបានរក្សាទុករាល់ពេល ព្រោះ record ប្រាប់ពីអ្វីដែលពិត **នៅពេលនេះ** |
| `load_records(path)` | កន្លែងដែលត្រូវរក | list ដែលបានរក្សាទុក ឬ `[]` នៅពេលគ្មាន file | `[]` មិនមែន `None` ទេ៖ នៅថ្ងៃដំបូង ប្រព័ន្ធគ្រាន់តែចាប់ផ្ដើមដោយគ្មានការផ្ដល់ជូន ហើយ loop លើ `[]` មិនធ្វើអ្វីទាំងអស់ |
| `append_event(path, event)` | កន្លែងដែលប្រវត្តិនៅ និងព្រឹត្តិការណ៍មួយ (dictionary មួយ) | គ្មានអ្វីទេ៖ file មានបន្ទាត់មួយបន្ថែមទៀត | `"a"` ដើម្បីកុំឱ្យបន្ទាត់មុនៗត្រូវបានសរសេរឡើងវិញ។ វាសរសេរ **គ្រប់** ព្រឹត្តិការណ៍ដែលគេផ្ដល់ឱ្យវា៖ ប្រព័ន្ធជាអ្នកអាន `LOG_EVENTS` ហើយសម្រេចថាត្រូវផ្ញើព្រឹត្តិការណ៍ណាខ្លះ |
| `read_events(path)` | កន្លែងដែលប្រវត្តិនៅ | គ្រប់ព្រឹត្តិការណ៍ ដោយចាស់បំផុតនៅមុនគេ ឬ `[]` នៅពេលគ្មាន file | វារំលងបន្ទាត់ទទេ ដូច្នេះបន្ទាត់ទទេតែមួយ មិនអាចធ្វើឱ្យប្រព័ន្ធឈប់បានទេ |

🔭 **ក្នុងការងារពិត។** មូលដ្ឋានទិន្នន័យ (database) រក្សាទុកតាមរបៀបដូចគ្នា ដោយប្រើ **ប្រតិបត្តិការ (transaction)**៖ ការផ្លាស់ប្ដូរមួយត្រូវបានរក្សាទុកទាំងអស់ ឬមិនរក្សាទុកសោះ មិនដែលពាក់កណ្ដាលទេ ដែលជាគំនិតដូចគ្នានឹង file បណ្ដោះអាសន្ន និង `os.replace` របស់អ្នក។ ប្រព័ន្ធពិតក៏រក្សា **កំណត់ត្រាសម្រាប់ត្រួតពិនិត្យ (audit log)** ដែលប្រាប់ថា នរណាបានប្ដូរអ្វី និងនៅពេលណា ព្រមទាំងរក្សា **backup** នៅកន្លែងផ្សេងមួយទៀត ដូច្នេះថាសខូចតែមួយ មិនអាចធ្វើឱ្យបាត់បង់អ្វីៗទាំងអស់បានទេ។

#### ⚖️ ល្អ ឬមានហានិភ័យ?

អ្នកបានសុំជំនួយការ AI (AI assistant) មួយថា៖ *"ចូរសរសេរ function មួយ ដែលរក្សាទុកការផ្ដល់ជូន និងព្រឹត្តិការណ៍របស់ថ្ងៃនេះ នៅចុងថ្ងៃ។ ចូររក្សា file ឱ្យតូច។"* វាបានផ្ដល់ឱ្យអ្នកនូវ **Version A**។ **Version B** ប្រើផ្នែកនានាដែលអ្នកបានរៀននៅសប្ដាហ៍នេះ។ នៅថ្ងៃធម្មតាដែលគ្មានបញ្ហា ទាំងពីរដំណើរការដោយគ្មាន error។

**Version A** (ពីជំនួយការ AI)

```python
def save_day(records, todays_events):
    with open("offers.json", "w") as file:
        json.dump(records, file)
    # រក្សាប្រវត្តិឱ្យតូច: យកតែព្រឹត្តិការណ៍ថ្ងៃនេះ
    with open("history.jsonl", "w") as file:
        for event in todays_events:
            file.write(json.dumps(event) + "\n")
```

**Version B**

```python
def save_day(records, todays_events):
    temp_path = "offers.json.tmp"
    with open(temp_path, "w", encoding="utf-8") as file:   # 1. សរសេរទិន្នន័យថ្មីនៅទីនេះ
        json.dump(records, file, indent=2)
    os.replace(temp_path, "offers.json")                    # 2. ប្ដូរវាចូលជំនួស ក្នុងមួយជំហាន
    with open("history.jsonl", "a", encoding="utf-8") as file:
        for event in todays_events:
            file.write(json.dumps(event) + "\n")            # បន្ថែមនៅខាងចុង; មិនដែលសរសេរឡើងវិញ
```

**ភស្តុតាង (evidence) របស់អ្នក។** cell ដែលរៀបចំរួចខាងក្រោម អាន `my_foodbridge/demo_day/history.jsonl` ដែលជាប្រវត្តិដែល `bridge.show("history")` ទើបតែរក្សាទុក ហើយរាប់ព្រឹត្តិការណ៍តាមប្រភេទនីមួយៗ ដោយប្រើលំនាំដូចគ្នានឹង `count_events` របស់អ្នក។ អ្នកគ្រាន់តែអានលទ្ធផលរបស់វាប៉ុណ្ណោះ។

````python
# @title Count the events in the saved demo day { display-mode: "form" }
# អាន my_foodbridge/demo_day/history.jsonl ហើយរាប់ព្រឹត្តិការណ៍តាមប្រភេទនីមួយៗ ដូច count_events។
import json
import os

history_path = "my_foodbridge/demo_day/history.jsonl"
demo_events = []
if os.path.exists(history_path):
    with open(history_path, encoding="utf-8") as file:
        for line in file:
            if line.strip() != "":
                demo_events.append(json.loads(line))
event_counts = {}
for event in demo_events:
    event_counts[event["event"]] = event_counts.get(event["event"], 0) + 1
print(history_path, "holds", len(demo_events), "events:")
for kind in ["added", "input_rejected", "moved"]:
    print("  ", kind, "→", event_counts.get(kind, "not logged"))
````

3. **ប្រៀបធៀប និងពិនិត្យ។** ចូរសាកល្បង function របស់ AI ជាមួយករណីផ្ទាល់ខ្លួនរបស់អ្នក (ករណីធម្មតា ករណីព្រំដែន ករណីចម្លែក) នៅក្នុង Colab ហើយតាមដានវា (trace) នៅក្នុង Python Tutor។ តើ AI ត្រឹមត្រូវ មិនច្បាស់ ឬខុស នៅកន្លែងណាខ្លះ?
4. **មតិរបស់អ្នក។** ចូរសរសេរប្រហែលកន្លះទំព័រ (≈ 150–250 ពាក្យជាភាសាអង់គ្លេស) លើសំណួរ *"តើនរណាជាអ្នកទទួលខុសត្រូវ នៅពេលកូដដែល AI សរសេរ ធ្វើឱ្យនរណាម្នាក់រងផលប៉ះពាល់?"* ចូរជ្រើសរើសជំហរមួយ ហើយផ្ដល់មូលហេតុ ដោយយកឧទាហរណ៍ពីករណីនៃសប្ដាហ៍នេះ ឧទាហរណ៍ function រក្សាទុករបស់ AI ដែលបានបើកប្រវត្តិដោយ `"w"`។
5. **ប្រភព និងភាពស្មោះត្រង់។** ចូររាយផ្នែក និងជំពូកដែលអ្នកបានអាន និងឧបករណ៍ AI ដែលអ្នកបានប្រើ។
