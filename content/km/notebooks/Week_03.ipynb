## វគ្គទី 1 · រៀន និងសាកល្បង: record ច្រើន (75 នាទី)

### គំនិតសំខាន់: record, ID និងគំរូទិន្នន័យ (data model)

ចូរគិតអំពីបញ្ជីឈ្មោះនិស្សិតក្នុងថ្នាក់។ និស្សិតម្នាក់ៗ ត្រូវបានកត់ទុកក្នុង **ជួរដេកមួយ** ហើយគ្រប់ជួរដេកមានជួរឈរដូចគ្នា: ID និស្សិត (អត្តលេខ) ឈ្មោះ និងថ្នាក់។ និស្សិតពីរនាក់អាចមានឈ្មោះដូចគ្នា ប៉ុន្តែមិនដែលមាន ID និស្សិតដូចគ្នាទេ។ ការិយាល័យរកព័ត៌មានរបស់អ្នក តាម ID។

ប្រព័ន្ធរបស់យើងក៏ដំណើរការដូចគ្នាដែរ។ សំណើនីមួយៗ គឺជា **record** មួយ: ព័ត៌មានអំពីសំណើពិតមួយ។ គ្រប់ record មាន **field** ដូចគ្នា និង **ID** មួយដែលមិនស្ទួនគ្នា ដូចជា `"F001"`។ បញ្ជី field ទាំងនោះ ហៅថា **គំរូទិន្នន័យ (data model)**។ ការជ្រើសរើស field គឺជាការសម្រេចចិត្តផ្នែករចនា។

| `status`, `partner` | `"approved"`, `"Hope Children's Centre"` | វានៅដំណាក់កាលណា និងនរណាជាអ្នកទទួល |

- **តើហេតុអ្វីបានជាប្រើ ID មិនប្រើឈ្មោះអាហារ?** ហាងនំពីរ អាចផ្ដល់ "Bread rolls" ដូចគ្នា ហើយអ្នកបរិច្ចាគអាចកែឈ្មោះអាហារ។ ID មិនដែលស្ទួនគ្នា ហើយមិនដែលប្រែប្រួលទេ។

### 2 · dictionary ដាក់ឈ្មោះឱ្យតម្លៃនីមួយៗ

**dictionary** រក្សាទុកទិន្នន័យជាគូ: **key** (ឈ្មោះ) និង **តម្លៃ (value)**។ អ្នករកតម្លៃមួយតាម key របស់វា មិនមែនតាមទីតាំងទេ។

🤔 **ទាយមើល** លទ្ធផលរបស់បន្ទាត់ពីរចុងក្រោយ។

````python
item = {"name": "Notebook", "price_riel": 4000}
print(item["price_riel"])
item["price_riel"] = 3500        # ប្ដូរតម្លៃមួយ
item["colour"] = "blue"          # បន្ថែម key ថ្មីមួយ
print(item)
print(item.get("size"))
print(item.get("size", "one size"))
````

### 4 · `for`: ធ្វើរឿងដដែល សម្រាប់គ្រប់ record

**loop** (ការធ្វើម្ដងហើយម្ដងទៀត) ដំណើរការបន្ទាត់កូដដែលចូលបន្ទាត់ (indent) ម្ដង សម្រាប់ item នីមួយៗ។ ឈ្មោះដែលនៅបន្ទាប់ពី `for` សំដៅទៅលើ item ដែល loop កំពុងមើលនៅពេលនោះ។ ចូរប្រើឈ្មោះ **ពហុវចនៈ** (plural) សម្រាប់ list (`offers`) និងឈ្មោះ **ឯកវចនៈ** (singular) សម្រាប់ item មួយ (`offer`)។

### 5 · ការបូកបន្តបន្ទាប់ (accumulator): ចាប់ផ្ដើម បូក រួចប្រើ

អ្នកសម្របសម្រួលត្រូវការ **តួលេខតែមួយ** សម្រាប់ពេញមួយថ្ងៃ ដូច្នេះ យើងរក្សាផលបូកដែលកើនឡើងជាបន្តបន្ទាប់ (running total) ខណៈពេល loop កំពុងដំណើរការ។ លំនាំនេះ ហៅថា **ការបូកបន្តបន្ទាប់ (accumulator)**។

```{iframe} https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/km/anim_for_loop.html
:width: 100%
:title: for loop និងការបូកបន្តបន្ទាប់ (accumulator)

[`for` loop និងការបូកបន្តបន្ទាប់ (accumulator)](https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/km/anim_for_loop.html)
```

````python
prices_riel = [4000, 1500, 2000]
total_riel = 0                    # 1. ចាប់ផ្ដើមមុន loop
for price_riel in prices_riel:
    total_riel += price_riel      # 2. បូក item នីមួយៗ
print(total_riel)                 # 3. ប្រើលទ្ធផល ក្រោយ loop
````

`total_riel += price_riel` មានន័យថា `total_riel = total_riel + price_riel`។ ជំហានទាំងបីដដែលនេះ អាចធ្វើបានច្រើនជាងការបូក៖

| ការងារ | ចាប់ផ្ដើម | ក្នុង loop |
|---|---|---|
| **បូក** (add) | `total = 0` | `total += value` |
| **រាប់** (count) | `count = 0` | `count += 1` |
| **ច្រោះ** (filter) | ដូចខាងលើ | បន្ទាត់ដដែល ក្រោម `if` មួយ |
| **ប្រមូល** (collect) | `found = []` | `found.append(value)` |

🤔 **ទាយមើល:** តើកូដនឹង print អ្វី ប្រសិនបើ `total_riel = 0` ត្រូវបានផ្លាស់ទៅ **ក្នុង** loop (នៅជាប់ពីលើបន្ទាត់ `+=`)? ចូរសាកល្បងវា រួចផ្លាស់វាមកកន្លែងដើមវិញ។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **មូលហេតុ:** …

### 6 · ស្វែងរក record មួយ តាម ID របស់វា

ការបូកសរុប មើល record ទាំងអស់។ ការស្វែងរក គឺរកមើល record **តែមួយ**។ នៅពេល ID ត្រូវគ្នា ចូរ **បញ្ជូនត្រឡប់មកវិញ (return)** record នោះ: `return` បញ្ឈប់ loop ភ្លាមៗ។ មានតែ **ក្រោយ** ពេល loop បានពិនិត្យគ្រប់ record រួចហើយប៉ុណ្ណោះ ទើបអ្នកដឹងថាវាមិនមាន; ពេលនោះ ចូរ return `None`។

````python
shop_items = [
    {"code": "N01", "name": "Notebook", "price_riel": 4000},
    {"code": "P02", "name": "Pen", "price_riel": 1500},
]

def find_item(items, wanted_code):
    for item in items:
        if item["code"] == wanted_code:
            return item          # រកឃើញ: ឈប់ ហើយបញ្ជូនវាត្រឡប់ទៅវិញ
    return None                  # បានពិនិត្យគ្រប់ item: រកមិនឃើញ

print(find_item(shop_items, "P02"))
print(find_item(shop_items, "X99"))
````

`return None` តម្រង់ជួរជាមួយ `for` មិនមែនជាមួយ `if` ទេ: វាដំណើរការតែនៅពេល loop បានបញ្ចប់ ដោយរកមិនឃើញអ្វីសោះ។

### 🧭 ការសម្រេចចិត្តរបស់អ្នក · តើយើងទទួលយកប្រភេទណាខ្លះ? (`CATEGORIES`)

````python
def describe_day(title, day):
    numbers = day.numbers()
    print(title)
    print("  accepted:", numbers["offers"], "offers,", numbers["portions_offered"], "portions offered")
    print("  refused at the door:", numbers["inputs_refused"])

describe_day("Default: " + ", ".join(bridge.DEFAULTS["CATEGORIES"]), day)

soup_day = bridge.System({"CATEGORIES": bridge.DEFAULTS["CATEGORIES"] + ["soup"]})   # + ភ្ជាប់ list ពីរ
soup_day.run_day(show=False)
describe_day("With soup added", soup_day)
print()
print(soup_day.report())
````
