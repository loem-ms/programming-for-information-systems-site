ផ្នែកនីមួយៗដែលអ្នកបានបង្កើត ស្ថិតនៅក្នុង module ដាច់ដោយឡែករបស់វា។ file `README.md` របស់ kit មានផែនទីមួយ។ module នីមួយៗក៏ចាប់ផ្ដើមដោយការពិពណ៌នាខ្លីមួយ ដែលហៅថា **docstring**។ Python រក្សាការពិពណ៌នានោះទុកក្នុង `__doc__` ហើយរក្សាឈ្មោះរបស់ module ទុកក្នុង `__name__`។ ចូរ Run cell ដើម្បីអានទាំងពីរ។

````python
with open("foodbridge_kit/README.md") as file:
    readme = file.read()

start = readme.find("## Map of the parts")   # ទីតាំងដែលផែនទីចាប់ផ្ដើម
print(readme[start:])                       # [start:] មានន័យថា: ពីទីនោះ រហូតដល់ចុងបញ្ចប់

for module in [bridge.rules, bridge.records, bridge.validation, bridge.models, bridge.storage, bridge.reports]:
    first_line = module.__doc__.splitlines()[0]
    print(module.__name__, "·", first_line)
````

### 4 · របៀបដែល `System` ភ្ជាប់ផ្នែកនានា

![hook អនុញ្ញាតឱ្យ function របស់អ្នកជំនួសវិធាន default; បើមានបញ្ហា ប្រព័ន្ធត្រឡប់ទៅប្រើ default វិញ](../../assets/diagrams/km/hooks.svg)

`bridge.System(MY_SETTINGS, hooks=...)` ធ្វើការងារបីយ៉ាង៖

1. វាបង្កើត **settings ពេញលេញ**: តម្លៃដើម បូកនឹងការផ្លាស់ប្ដូររបស់អ្នក។ ប្រសិនបើឈ្មោះណាមួយសរសេរខុស វានឹងរំលងឈ្មោះនោះ ហើយបង្ហាញការព្រមាន ⚠️។
2. វារៀបចំ **default function** មួយ សម្រាប់ hook នីមួយៗ រួចដាក់បញ្ចូល (plug in) function របស់អ្នក។
3. រាល់ពេលដែលវាត្រូវការចម្លើយ វា **សួរ hook**។ ប្រសិនបើ function របស់អ្នកគាំង (crash) ឬផ្ដល់ចម្លើយចម្លែក ប្រព័ន្ធនឹងប្រាប់មូលហេតុ (តែម្ដង) ហើយប្រើ default ជំនួសវិញ ដូច្នេះថ្ងៃនោះនៅតែបន្តបាន។

🤔 ប្រព័ន្ធនេះមាន setting មួយដែលសរសេរខុស និង hook មួយដែល "ស្រែក" ជាអក្សរធំ។ **ទាយមើល** តើ pickup label ទាំងពីរនឹងចេញជាអ្វី? បន្ទាប់មក Run។

````python
def shouting_label(portions, approved):
    return "TEAM!"            # មិនមែនជា "hold", "team" ឬ "standard" ទេ

test_system = bridge.System({"TEAM_PICKUP_MIN_PORTION": 25}, hooks={"pickup_label": shouting_label})
test_system.run_day(show=False)
for offer_id in ["F001", "F004"]:
    offer = test_system.get(offer_id)
    print(offer_id, offer.portions, "portions →", test_system.label(offer))
````

### 5 · ការដំណើរការ file `.py` ក្នុង process ថ្មីមួយ

notebook ចងចាំអ្វីៗទាំងអស់ដែលអ្នកបាន Run។ ចំណែកកម្មវិធីកុំព្យូទ័រ (program) ពិតប្រាកដ ចាប់ផ្ដើម **ពីសូន្យ** (fresh)៖ វាដឹងតែអ្វីដែលមាននៅក្នុង file របស់វាប៉ុណ្ណោះ។ module `subprocess` របស់ Python អាចចាប់ផ្ដើម Python ថ្មីមួយ ហើយដំណើរការ file មួយ ដូចដែល terminal ធ្វើដែរ។

````python
import subprocess
import sys

with open("hello.py", "w") as file:
    file.write('print("Hello from hello.py")\n')
    file.write("print(2 + 3)\n")

result = subprocess.run([sys.executable, "hello.py"], capture_output=True, text=True)
print("It printed:")
print(result.stdout)
print("Return code:", result.returncode)
````

- `sys.executable` គឺជា Python ដែលកំពុងដំណើរការ notebook នេះ។ យើងឱ្យវាដំណើរការ `hello.py`។
- `capture_output=True` ប្រមូលអ្វីៗដែលកម្មវិធីកុំព្យូទ័រ print។ `text=True` ផ្ដល់លទ្ធផលនោះមកយើងជាអត្ថបទ (text)។
- `result.stdout` គឺជាលទ្ធផល (output) ធម្មតា។ `result.stderr` ផ្ទុកសារ error។ return code `0` មានន័យថា *គ្មានបញ្ហា (error) ទេ*។

🤔 notebook ស្គាល់ `secret` ខាងក្រោម។ តើ process ថ្មីស្គាល់វាដែរឬទេ? **ទាយមើល** រួច Run។

````python
secret = "only in this notebook"

with open("ask.py", "w") as file:
    file.write("print(secret)\n")

result = subprocess.run([sys.executable, "ask.py"], capture_output=True, text=True)
print("Return code:", result.returncode)
print("Last line of the error:", result.stderr.splitlines()[-1])
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **ហេតុអ្វី:** …

process ថ្មីគឺស្មោះត្រង់៖ ប្រសិនបើកម្មវិធីកុំព្យូទ័ររបស់អ្នកត្រូវការអ្វីមួយ អ្វីនោះត្រូវតែមាននៅក្នុងកម្មវិធីនោះ។ នៅសប្ដាហ៍ក្រោយ អ្នកនឹងប្រើគំនិតនេះ ដើម្បីពិនិត្យថា notebook ទាំងមូលរបស់អ្នកអាចដំណើរការបានពីសូន្យ (clean start)។

process ថ្មីមួយបង្ហាញយ៉ាងច្បាស់ថា តើកម្មវិធីកុំព្យូទ័រមួយផ្ទុកអ្វីខ្លះ។ បញ្ហាបន្ទាប់: file មួយ ជារឿយៗត្រូវធ្វើជาทាំងកម្មវិធីកុំព្យូទ័រ និង module ក្នុងពេលតែមួយ។

### 6 · ការ import file មួយ ឬការដំណើរការវា: `if __name__ == "__main__":`

Python ផ្ដល់ `__name__` មួយឱ្យ module នីមួយៗ។ ពេលអ្នក **ដំណើរការ** (run) file មួយ `__name__` របស់វាគឺ `"__main__"`។ ពេល file ផ្សេងទៀត **import** វា `__name__` របស់វាគឺជាឈ្មោះ module របស់វា។ notebook ខ្លួនវាក៏ដំណើរការដែរ ដូច្នេះ `__name__` របស់វាក៏ជា `"__main__"` (សូមមើលបន្ទាត់ទីមួយនៃ cell)។

បន្ទាត់ `if __name__ == "__main__":` មានន័យថា *"ធ្វើការនេះ តែពេលខ្ញុំត្រូវបានដំណើរការផ្ទាល់ប៉ុណ្ណោះ មិនមែនពេលខ្ញុំត្រូវបាន import ទេ"*។ វាអនុញ្ញាតឱ្យ file មួយ ជាទាំង **កម្មវិធីកុំព្យូទ័រ** (program) និង **module**។ ក្នុងឧទាហរណ៍នេះ `shop.py` មាន function មួយដែលមានប្រយោជន៍ និង demo តូចមួយ។ `main.py` import `shop` ដើម្បីប្រើ function នោះ។

````python
print("This notebook says: my __name__ is", __name__)

shop_code = '''
TAX_RATE = 0.10

def price_with_tax(price):
    return round(price * (1 + TAX_RATE), 2)

print("shop.py says: my __name__ is", __name__)

if __name__ == "__main__":
    print("Demo: 100 with tax is", price_with_tax(100))
'''

main_code = '''
import shop

print("main.py says: my __name__ is", __name__)
print("main.py uses shop:", shop.price_with_tax(50))
'''

with open("shop.py", "w") as file:
    file.write(shop_code)
with open("main.py", "w") as file:
    file.write(main_code)

print(subprocess.run([sys.executable, "shop.py"], capture_output=True, text=True).stdout)
````

យើងឆ្លើយជាបួនជំហាន៖

1. ចាប់ផ្ដើមពីច្បាប់ចម្លងមួយនៃ settings default។
2. ដាក់ការផ្លាស់ប្ដូររបស់ថ្ងៃនេះ ពីលើច្បាប់ចម្លងនោះ។
3. ទុកឱ្យវិធានអានលេខ ពី settings ដែលបានបញ្ចប់ — មិនដែលអានពីលេខដែលសរសេរនៅក្នុងវិធានឡើយ។
4. print label ដែលវិធានផ្ដល់ឱ្យ។

វាគឺជាផ្នែក settings ពាក់កណ្ដាលនៃផ្នែកទី 3 ដោយគ្មាន `System` និងគ្មាន hooks។ ខាងក្រោមនេះ គឺជា settings default ការផ្លាស់ប្ដូររបស់ថ្ងៃនេះ និងវិធានរបស់សប្ដាហ៍ទី 2 ដែលអានពី settings:

````python
defaults = {"TEAM_PICKUP_MIN_PORTIONS": 20, "URGENT_HOURS": 3}
changes = {"TEAM_PICKUP_MIN_PORTIONS": 25}
settings = dict(defaults)
settings.update(changes)

def pickup_label(portions, approved, settings):
    if not approved:
        return "hold"
    if portions >= settings["TEAM_PICKUP_MIN_PORTIONS"]:
        return "team"
    return "standard"

print(pickup_label(22, True, settings))
````

**របៀបដែលវាដំណើរការ:** `dict(defaults)` ចម្លង settings default ដូច្នេះ settings ដើមនៅតែជា 20 និង 3 ដដែល។ `settings.update(changes)` សរសេរ 25 ពីលើចំនួនអប្បបរមាសម្រាប់ក្រុម របស់ច្បាប់ចម្លង។ វិធានប្រៀបធៀប 22 ទៅនឹងលេខដែលវារកឃើញក្នុង settings: 22 >= 25 គឺ `False` ដូច្នេះវា print `standard`។ ប្រសិនបើប្រើតម្លៃដើម (default) 20 នោះ 22 ចំណែកដដែលនេះ នឹងជា `"team"` — ពេលអ្នកប្ដូរ setting វិធានដដែលនឹងឆ្លើយខុសគ្នា។

នៅវគ្គទី 3 ជំហានទី 1 `bridge.System` ធ្វើបែបនេះសម្រាប់គ្រប់ setting ទាំងអស់ក្នុងពេលតែមួយ ហើយដាក់បញ្ចូល (plug in) hooks របស់អ្នកផងដែរ។

### ⭐ P2 · អានគ្រឿងម៉ាស៊ីន

នេះជាផ្នែកខ្លីមួយនៃ `system.py` (បានកាត់ឱ្យខ្លីបន្តិច)។ ចូរអានវា ហើយឆ្លើយសំណួរខាងក្រោម។ អ្នកមិនចាំបាច់សរសេរកូដបែបនេះទេ។ អ្នកគ្រាន់តែត្រូវអានវាប៉ុណ្ណោះ។

```python
def label(self, offer):
    approved = offer.status != "new" or not self.settings["REQUIRE_APPROVAL"]
    return self._call("pickup_label", offer.portions, approved)

def _call(self, name, *args):                 # *args = ធាតុចូល (input) ទាំងអស់ ដែលបញ្ជូនបន្ត
    function, default = self.hooks[name], self.defaults[name]
    try:
        result = function(*args)
        problem = result_problem(name, result)   # None ពេលចម្លើយមើលទៅត្រឹមត្រូវ
    except Exception as error:
        problem = "it raised " + type(error).__name__ + ": " + str(error)
    if problem is None:
        return result
    if name not in self.fallbacks:
        print("⚠️ Your " + name + " did not work: " + problem)
    self.fallbacks.setdefault(name, problem)   # ចងចាំបញ្ហាទីមួយ
    return default(*args)
```

1. `label` គណនា `approved` ដោយខ្លួនឯង។ តើ `approved` មានតម្លៃអ្វី សម្រាប់ការផ្ដល់ជូន (offer) ដែលមាន status `"new"`? ចុះតើវាមានតម្លៃអ្វី ពេល `REQUIRE_APPROVAL` គឺ `False`?
2. តើបញ្ហា **ពីរប្រភេទ** អ្វីខ្លះ ដែលធ្វើឱ្យ `_call` ប្រើវិធាន default?
3. ឧបមាថា hook របស់អ្នកបរាជ័យ សម្រាប់សំណើ **ពីរ** ផ្សេងគ្នា។ តើអ្នកឃើញការព្រមាន ⚠️ ប៉ុន្មានដង? តើបញ្ហាត្រូវបានចងចាំទុកនៅកន្លែងណា?

````python
def capital_label(portions, approved):
    if not approved:
        return "Hold"
    if portions >= 20:
        return "Team"
    return "Standard"

capital_system = bridge.System(hooks={"pickup_label": capital_label})
capital_system.run_day(show=False)
for offer_id in ["F001", "F004"]:
    offer = capital_system.get(offer_id)
    print(offer_id, offer.portions, "portions →", capital_system.label(offer))
print("Remembered:", capital_system.fallbacks)
````

AI ក៏បានសរសេរ `web_row` សម្រាប់ web page ដែរ។ វាដំណើរការបាន ប៉ុន្តែមានការសម្រេចចិត្តពីរ ត្រូវបានកប់នៅខាងក្នុង។ ចូរកែវាដើម្បីឱ្យ៖

- ស្លាកបន្ទាន់ (urgent badge) ប្រើ `settings["URGENT_HOURS"]` ជំនួសឱ្យ `3`។
- បង្ហាញតែ field ដែលមានក្នុង `settings["PUBLIC_FIELDS"]` ប៉ុណ្ណោះ តាមលំដាប់នោះ។

check cell ខាងក្រោមសាកល្បងជាមួយ **settings ពីរផ្សេងគ្នា**។ កូដដែលធ្វើតាម settings នឹងជាប់ការពិនិត្យទាំងពីរ។

````python
def web_row(record, settings):
    # សរសេរដោយជំនួយការ AI។ មានការសម្រេចចិត្តពីរ កប់នៅទីនេះ។
    if record["hours_left"] <= 3:
        badge = "URGENT"
    else:
        badge = ""
    shown = {}
    for field in record:
        if field != "contact":
            shown[field] = record[field]
    return badge, shown
````

### ជំហានទី 2 · ដាក់បញ្ចូល (plug in) function ផ្ទាល់ខ្លួនរបស់អ្នក (ស្រេចចិត្ត)

អ្នកអាចចម្លងដាក់ (paste) function ដែលអ្នកបានសរសេរនៅសប្ដាហ៍មុនៗ។ ប្រសិនបើ function មួយប្រើ helper (ដូចជា `to_whole_number` ពីសប្ដាហ៍ទី 5) ឬ dictionary (ដូចជា `ALLOWED_MOVES` ពីសប្ដាហ៍ទី 6) សូម paste វាផងដែរ។ បន្ទាប់មក ចូរលុប `#` នៅពីមុខបន្ទាត់របស់ function នោះ ក្នុង `MY_HOOKS`។ ប្រសិនបើអ្នកមិន paste អ្វីទាំងអស់ ប្រព័ន្ធនឹងប្រើវិធាន default របស់វា។ នោះមិនអីទេ សម្រាប់កម្រិតទី 1។

**ចូរលុបបន្ទាត់ `NAME = value` ណាមួយដែលអ្នក paste ចូល** ដូចជា `TEAM_PICKUP_MIN_PORTIONS = 20` ពីសប្ដាហ៍ទី 2 ឬ `REQUIRE_APPROVAL = True` ពីសប្ដាហ៍ទី 6។ ឥឡូវនេះ តម្លៃទាំងនោះមកពី `MY_SETTINGS` (cell ខាងលើ)។ បន្ទាត់ដែល paste ចូល នឹងប្ដូរតម្លៃនោះត្រឡប់ទៅវិញដោយស្ងាត់ៗ ហើយ function របស់អ្នក និង settings របស់អ្នក នឹងមិនស្របគ្នា។

````python
# Paste function របស់អ្នក នៅក្រោមបន្ទាត់នេះ (ស្រេចចិត្ត)។
#   សប្ដាហ៍ទី 2: pickup_label(portions, approved) និង urgency(hours_left)
#   សប្ដាហ៍ទី 5: clean_offer(raw)
#   សប្ដាហ៍ទី 6: can_move(old_status, new_status)
#   សប្ដាហ៍ទី 8: public_row(record)


MY_HOOKS = {
    # "pickup_label": pickup_label,
    # "urgency": urgency,
    # "clean_offer": clean_offer,
    # "can_move": can_move,
    # "public_row": public_row,
}

for name in MY_HOOKS:
    bridge.check(name, MY_HOOKS[name], MY_SETTINGS)
if not MY_HOOKS:
    print("No hooks plugged in: the system uses its default rules.")
````

### ជំហានទី 3 · ផ្គុំប្រព័ន្ធរបស់អ្នក ហើយផ្ទៀងផ្ទាត់វា

kit ដែលបានទាញយក (download) រក្សាការសម្រេចចិត្តរបស់អ្នកទុកក្នុង `my_settings.py` ដាច់ដោយឡែកពីគ្រឿងម៉ាស៊ីន។ ចូរសរសេរ settings របស់អ្នកចូលក្នុង file នោះ ហើយដំណើរការ menu អក្សរ (text menu) របស់ kit ជាកម្មវិធីកុំព្យូទ័រដាច់ដោយឡែកមួយ។ `input=` វាយចម្លើយ menu ជំនួសអ្នក (`1` = ទិដ្ឋភាពអ្នកសម្របសម្រួល, `q` = ចាកចេញ)។ `repr()` សរសេរតម្លៃមួយ តាមរបៀបដែលកូដ Python សរសេរវា។

````python
with open("foodbridge_kit/my_settings.py", "w") as file:
    for name in MY_SETTINGS:
        if MY_SETTINGS[name] != bridge.DEFAULTS[name]:      # សរសេរតែការផ្លាស់ប្ដូររបស់អ្នក
            file.write(name + " = " + repr(MY_SETTINGS[name]) + "\n")

menu_run = subprocess.run([sys.executable, "run_menu.py", "--fresh"], cwd="foodbridge_kit",
                          input="1\nq\n", capture_output=True, text=True)
print(menu_run.stdout)
````
