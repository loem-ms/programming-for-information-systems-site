````python
import subprocess
import sys


def run_file(name):
    """Run a .py file in a new Python and show what it printed."""
    result = subprocess.run([sys.executable, name], capture_output=True, text=True)
    print(result.stdout + result.stderr)


run_file("prices.py")
````

- [ ] ខ្ញុំបានគូរក្រាហ្វខ្សែ និង histogram ហើយខ្ញុំអាចប្រាប់ថាហេតុអ្វីក្រាហ្វ A បំភាន់។
- [ ] `app.py` របស់ខ្ញុំជាប់ការពិនិត្យ `bridge.show("app")` បង្ហាញ menu របស់ខ្ញុំ ហើយខ្ញុំបានបើកប្រព័ន្ធរបស់ខ្ញុំដោយ `bridge.start_web(folder="my_foodbridge")`។
- [ ] ខ្ញុំបានធ្វើការសម្រេចចិត្តមួយ ដោយផ្អែកលើទិន្នន័យ ហើយបានសរសេរ decision log។
- [ ] ខ្ញុំបានចាប់ផ្ដើមផែនការ Final project របស់ខ្ញុំ (ការសម្រេចចិត្តបី រួចរាល់មុនសប្ដាហ៍ទី 10)។
