## វគ្គទី 1 · រៀន និងសាកល្បង (90 នាទី)

### គំនិតសំខាន់: entity, status និង workflow

សូមគិតពីកញ្ចប់អីវ៉ាន់មួយ ដែលអ្នកកុម្ម៉ង់តាមអនឡាញ: *ordered (បានកុម្ម៉ង់) → shipped (បានផ្ញើចេញ) → delivered (បានប្រគល់)*។ កញ្ចប់អីវ៉ាន់គឺជា **entity** (ធាតុទិន្នន័យសំខាន់មួយដែលប្រព័ន្ធតាមដាន)។ **status** (ស្ថានភាព) ប្រាប់ថាវាកំពុងស្ថិតនៅដំណាក់កាលណា។ ចំណែក **workflow** (លំហូរការងារ) គឺជាសំណុំនៃការប្ដូរស្ថានភាពដែលសមហេតុផល។ កញ្ចប់អីវ៉ាន់មិនអាចប្រគល់បាន មុនពេលវាត្រូវបានផ្ញើចេញទេ ហើយកញ្ចប់ដែលបានប្រគល់រួច ក៏មិនត្រឡប់ទៅជា "ordered" វិញដែរ។

សំណើមួយក៏មានដំណើរស្រដៀងគ្នានេះដែរ៖

![lifecycle របស់សំណើ: new, approved, picked_up, delivered ឬ rejected ឬ cancelled](../../assets/diagrams/km/lifecycle.svg)

មនុស្សជាអ្នកធ្វើការប្ដូរទាំងនេះ។ **អ្នកសម្របសម្រួល** អនុម័ត ឬបដិសេធសំណើថ្មី ហើយជ្រើសរើសអង្គការដៃគូ ឯអ្នកស្ម័គ្រចិត្តទៅយក និងប្រគល់អាហារ។ កម្មវិធីកុំព្យូទ័រ **កត់ត្រា** ការសម្រេចចិត្ត។ ចំណែកការវិនិច្ឆ័យថា អាហារមានសុវត្ថិភាពឬអត់ នៅតែជាការសម្រេចចិត្តរបស់មនុស្ស។ rejected (បានបដិសេធ), cancelled (បានលុបចោល) និង delivered (បានប្រគល់) គឺជា **ស្ថានភាពចុងក្រោយ (end state)**: មិនអាចប្ដូរចេញពីស្ថានភាពទាំងនេះបានទៀតទេ។

### ⭐⭐ 8 · អាន class មួយ: `Offer` របស់ kit

```python
class Offer:
    """One food offer from a donor. Change its status only with move_to()."""

    def __init__(self, offer_id, donor, food, category, portions, hours_left, contact=""):
        self.id = offer_id
        self.donor = donor
        self.contact = contact
        self.food = food
        self.category = category
        self.portions = portions
        self.hours_left = hours_left
        self.status = "new"
        self.partner = ""

    def move_to(self, new_status, is_allowed):
        """Change status only if is_allowed(old, new) is True. Return True or False."""
        if not is_allowed(self.status, new_status):
            return False
        self.status = new_status
        return True
```

`__init__` ដំណើរការ នៅពេលសំណើថ្មីមួយត្រូវបានបង្កើត។ `self` មានន័យថា "សំណើនេះ"។ `move_to` គឺជា **method**។ 🤔 ចូរអាន class នេះ ហើយឆ្លើយសំណួរខាងក្រោម៖

### ⚖️ ល្អ ឬមានហានិភ័យ?

អ្នកបានសួរជំនួយការ AI មួយថា: *"អ្នកស្ម័គ្រចិត្តត្រូវការប៊ូតុង ដើម្បីធ្វើបច្ចុប្បន្នភាព (update) សំណើ ហើយអ្នកសម្របសម្រួលចង់បានវិធីកែកំហុស។"* វាបានសរសេរ **Version A**។ **Version B** គឺជាកំណែដែលមិត្តរួមក្រុមម្នាក់សរសេរឡើងវិញ។ ទាំងពីរដំណើរការបាន។ ចូរអានវានៅក្នុង cell ខាងក្រោម រួច Run វា។ ទាំងពីរសាកល្បងចុចប៊ូតុងលើ F004: ការីរបស់ Lotus Cafe ចំនួន 12 ចំណែក ដែលបានអនុម័តសម្រាប់ Old Market Elders' Group។

````python
# Version A · ពីជំនួយការ AI
def update_status(record, typed_status):
    record["status"] = typed_status.strip().lower()   # អ្វីក៏ដោយដែលអ្នកស្ម័គ្រចិត្តវាយបញ្ចូល

def mark_delivered(record):
    record["status"] = "delivered"                    # ចុចម្ដង ពី status ណាក៏បាន

def fix_mistake(record):
    record["status"] = "new"                          # reset: ចាប់ផ្ដើមសាជាថ្មី


# Version B · ពីមិត្តរួមក្រុម
def move(record, new_status, is_allowed):
    old_status = record["status"]
    if is_allowed(old_status, new_status):
        record["status"] = new_status
        return "done: " + record["id"] + " " + old_status + " -> " + new_status
    return "refused: " + record["id"] + " cannot move from " + old_status + " to " + new_status


DEMO_MOVES = {
    "new": ["approved", "rejected"],
    "approved": ["picked_up", "cancelled"],
    "picked_up": ["delivered"],
    "delivered": [],
    "rejected": [],
    "cancelled": [],
}


def lifecycle_rule(old_status, new_status):
    return new_status in DEMO_MOVES.get(old_status, [])


curry_a = bridge.sample_records()[3]      # F004 ដែលបានអនុម័ត
curry_b = bridge.sample_records()[3]
print("Version A")
mark_delivered(curry_a)
print("  1.", curry_a["id"], curry_a["status"])
fix_mistake(curry_a)
print("  2.", curry_a["id"], curry_a["status"])
update_status(curry_a, "Deliverd ")
print("  3.", curry_a["id"], curry_a["status"])
print("Version B")
print("  1.", move(curry_b, "delivered", lifecycle_rule))
print("  2.", move(curry_b, "new", lifecycle_rule))
print("  3.", move(curry_b, "picked_up", lifecycle_rule))
print("  4.", move(curry_b, "delivered", lifecycle_rule))
````
