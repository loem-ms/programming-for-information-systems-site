សូមគិតពីកញ្ចប់អីវ៉ាន់មួយដែលអ្នកកុម្ម៉ង់តាមអនឡាញ។ ប្រព័ន្ធរបស់ហាងតាមដានកញ្ចប់នោះ៖ *បានកុម្ម៉ង់ → បានផ្ញើចេញ → កំពុងដឹកមក → បានប្រគល់*។

សំណើមួយក៏មានដំណើរស្រដៀងគ្នានេះដែរ៖

### 1 · class គឺជាប្លង់ ហើយ object គឺជារបស់មួយ

រហូតមកដល់ពេលនេះ សំណើនីមួយៗគឺជា dictionary មួយ។ **class** ទៅមុខមួយជំហានទៀត: វារក្សាទុកទិន្នន័យ **និង** សកម្មភាពដែលពាក់ព័ន្ធនឹងទិន្នន័យនោះ នៅជាមួយគ្នា។ class គឺជាប្លង់ (blueprint)។ **object** នីមួយៗដែលបង្កើតចេញពីប្លង់នោះ គឺជារបស់ពិតមួយ។

````python
class Parcel:
    def __init__(self, code, address):
        self.code = code
        self.address = address
        self.status = "ordered"

    def describe(self):
        return f"{self.code} to {self.address}: {self.status}"

parcel_a = Parcel("P01", "Battambang")
parcel_b = Parcel("P02", "Siem Reap")
print(parcel_a.describe())
print(parcel_b.describe())
````

- `class Parcel:` ចាប់ផ្ដើមប្លង់។ ឈ្មោះ class ចាប់ផ្ដើមដោយអក្សរធំ។
- `__init__` (មានសញ្ញា underscore ពីរនៅសងខាង) ដំណើរការដោយស្វ័យប្រវត្តិ នៅពេលអ្នកបង្កើត object ដោយ `Parcel("P01", "Battambang")`។ វាកំណត់តម្លៃចាប់ផ្ដើម។
- `self` មានន័យថា **object នេះ**។ `self.status = "ordered"` ផ្ដល់ status ផ្ទាល់ខ្លួនដល់កញ្ចប់នេះ។
- `describe` គឺជា **method**: function ដែលជារបស់ class។ អ្នកហៅវាដោយប្រើចំណុច: `parcel_a.describe()`។ Python បំពេញ `self` ឱ្យអ្នកដោយស្វ័យប្រវត្តិ។

យើងអាចបង្កើត object មួយពីប្លង់ (blueprint) ដោយកំណត់តម្លៃចាប់ផ្ដើមរបស់វា។ បញ្ហាបន្ទាប់: តើការប្ដូរ object មួយ ធ្វើឱ្យ object ផ្សេងទៀតប្ដូរដែរឬទេ?

### ឧទាហរណ៍គំរូ · កញ្ចប់អីវ៉ាន់មួយ ការប្ដូរមួយ

មុនពេលអ្នកបង្កើត ចូរមើលលំនាំនេះម្ដងសិន លើដំណើរតូចជាង។ ហាងសួរថា: *«តើកញ្ចប់អីវ៉ាន់នេះ អាចប្រគល់បានហើយឬនៅ? វាទើបតែត្រូវបានកុម្ម៉ង់នៅព្រឹកនេះទេ។»*

យើងឆ្លើយជាបីជំហាន៖

1. រកមើល status ចាស់ `"ordered"` នៅក្នុង dictionary នៃការប្ដូរ។
2. យក list នៃ status ដែលវាអាចប្ដូរទៅបាន។
3. ឆ្លើយ `True` តែនៅពេល `"delivered"` នៅក្នុង list នោះ។

វាគឺជា dictionary នៃការប្ដូរដែលត្រូវបានអនុញ្ញាត ពីផ្នែកទី 6 ដែលពិនិត្យដោយ `in`។ ខាងក្រោមនេះ គឺជា lifecycle តូចរបស់កញ្ចប់អីវ៉ាន់:

````python
PARCEL_MOVES = {"ordered": ["shipped"], "shipped": ["delivered"], "delivered": []}

def can_parcel_move(old_status, new_status):
    return new_status in PARCEL_MOVES[old_status]

print(can_parcel_move("ordered", "delivered"))
````

**របៀបដែលវាដំណើរការ:** `PARCEL_MOVES["ordered"]` គឺ `["shipped"]` ហើយ `"delivered"` មិននៅក្នុង list នោះទេ ដូច្នេះ function បញ្ជូន `False` មកវិញ (return)។ កញ្ចប់អីវ៉ាន់ត្រូវផ្ញើចេញ (shipped) ជាមុនសិន។ មានតែក្រោយពេលនោះទេ ទើបវាអាចប្រគល់បាន។

### ⭐ P1 · អាន class មួយ ហើយប៉ាន់ស្មានលទ្ធផល

ក្រុមហ៊ុនឡានក្រុងមួយប្រើ class នេះ។ ចូរអានវា រួចសរសេរការប៉ាន់ស្មានរបស់អ្នកសម្រាប់បន្ទាត់ `print` ទាំងបួន ហើយទើប Run។

````python
class BusTicket:
    def __init__(self, passenger, price_riel):
        self.passenger = passenger
        self.price_riel = price_riel
        self.used = False

    def use(self):
        """A ticket can be used only once. Return True if this use is allowed."""
        if self.used:
            return False
        self.used = True
        return True

ticket = BusTicket("Sokha", 2000)
other = BusTicket("Dara", 1500)
print(ticket.passenger, ticket.used)
print(ticket.use())
print(ticket.use())
print(ticket.used, other.used)
````

## វគ្គទី 3 · បង្កើតផ្នែករបស់អ្នក: វិធាន lifecycle (60–80 នាទី)

នេះគឺជា **ការសម្រេចចិត្តមួយ ដែលត្រូវតែស៊ីគ្នានៅកន្លែងពីរ**: setting (`MY_SETTINGS`) និង dictionary របស់អ្នក។ ប្រសិនបើវាមិនស៊ីគ្នាទេ ប្រព័ន្ធនឹងនិយាយមួយបែប តែធ្វើមួយបែបទៀត។ TODO 2 បង្កើត dictionary **ចេញពី** setting ដូច្នេះទាំងពីរមិនអាចខុសគ្នាបានឡើយ។

ឥឡូវនេះ រាល់ការប្ដូរក្នុងប្រព័ន្ធពិត ត្រូវតែឆ្លងកាត់វិធាន **របស់អ្នក**។ ជាដំបូង សូមមើលថ្ងៃគំរូ៖

ឥឡូវនេះ ចូរសាកល្បងការប្ដូរមួយចំនួនដោយខ្លួនឯង រួមទាំងកំហុសបី៖ ការប្រគល់មុនពេលទៅយក ការ reset អាហារដែលបានប្រគល់រួច និងការលុបចោលការបដិសេធ (undo)។

````python
attempts = [
    ("F004", "delivered", ""),                        # ប្រគល់មុនពេលទៅយក
    ("F004", "picked_up", ""),
    ("F004", "delivered", ""),
    ("F002", "new", ""),                              # reset អាហារដែលបានប្រគល់រួច
    ("F003", "approved", "Hope Children's Centre"),   # លុបចោលការបដិសេធ (undo)
]
for offer_id, new_status, partner in attempts:
    ok, message = my_system.move(offer_id, new_status, partner=partner)
    if ok:
        print("✅", message)
    else:
        print("🚫", message)
````
