### 2 · ការតម្រៀប record ដោយប្រើ key function ដែលមានឈ្មោះ

Python មិនអាចដឹងដោយខ្លួនឯងថា ត្រូវតម្រៀប dictionary តាមអ្វីទេ៖ តាមឈ្មោះ? ឬតាមពិន្ទុ? អ្នកត្រូវប្រាប់វាដោយប្រើ **key function** មួយ។ Function នេះទទួល item មួយ ហើយ return តម្លៃដែលត្រូវប្រើសម្រាប់តម្រៀប។

🤔 **ទាយមើល:** តើឈ្មោះទាំងនេះនឹងចេញតាមលំដាប់ណា?

````python
students = [{"name": "Sokha", "mark": 72},
            {"name": "Dara", "mark": 90},
            {"name": "Vanna", "mark": 61}]


def by_mark(student):
    return student["mark"]


for student in sorted(students, key=by_mark):
    print(student["name"], student["mark"])
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

ចូរសរសេរ `key=by_mark` **ដោយគ្មាន** `( )`។ អ្នកផ្ដល់ function ខ្លួនវាផ្ទាល់ទៅឱ្យ `sorted` ហើយ `sorted` ហៅវាម្ដង សម្រាប់និស្សិតម្នាក់ៗ។ ឈ្មោះល្អ ប្រាប់ពីលំដាប់៖ `by_mark`, `hours_left_of`។

### 3 · ការជ្រើសរើស field: បញ្ជីអនុញ្ញាត (allowed list)

ទិដ្ឋភាពសាធារណៈ គឺជា **ច្បាប់ចម្លងតូចជាង** នៃ record មួយ។ វិធីដែលមានសុវត្ថិភាពក្នុងការបង្កើតវា៖ ចាប់ផ្ដើមពី dictionary ទទេមួយ ហើយចម្លងតែ field ដែលមាននៅក្នុង **បញ្ជីអនុញ្ញាត (allowed list)** ប៉ុណ្ណោះ។ (លេខទូរស័ព្ទនេះជាលេខប្រឌិត។)

````python
student = {"name": "Dara", "phone": "012 000 111", "mark": 90, "class": "MIS1"}
NOTICE_BOARD_FIELDS = ["name", "class"]


def notice_board_row(student):
    row = {}
    for field in NOTICE_BOARD_FIELDS:
        row[field] = student.get(field, "")
    return row


print(notice_board_row(student))

student["home_address"] = "Street 271"      # នរណាម្នាក់បន្ថែម field ឯកជនថ្មីមួយ នៅពេលក្រោយ
print(notice_board_row(student))
````

### 5 · ចំនួន (count) ឬផលបូក (sum)? តើអ្នកបរិច្ចាគណាសំខាន់ជាងគេ?

````python
import csv
import matplotlib.pyplot as plt

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    offers = list(csv.DictReader(file))


def count_by(rows, field):                       # ពីសប្ដាហ៍ទី 6
    counts = {}
    for row in rows:
        counts[row[field]] = counts.get(row[field], 0) + 1
    return counts


offers_by_donor = count_by(offers, "donor")
portions_by_donor = total_by(offers, "donor", "portions")      # សាកសរសេររបស់អ្នក
if portions_by_donor == {}:
    print("🔧 total_by is not finished yet: model numbers in use")
    portions_by_donor = {"Lotus Cafe": 766, "Sunrise Bakery": 1721, "Campus Canteen": 1780,
                         "Riverside Restaurant": 1292, "Night Market Vendor": 674,
                         "Green Market Stall": 1547, "Student Club Bake Sale": 379, "Hotel Kitchen": 827}
donors = list(offers_by_donor.keys())
offer_counts = []
portion_totals = []
for donor in donors:
    offer_counts.append(offers_by_donor[donor])
    portion_totals.append(portions_by_donor.get(donor, 0))
print(len(offers), "offers from", len(donors), "donors")
````

```python
def by_food_name(record):
    return record["food"]

def dashboard(records):
    offered = 0
    picked_up = 0
    delivered = 0
    for record in records:
        offered += record["portions"]
        if record["status"] == "picked_up":
            picked_up += record["portions"]
        if record["status"] == "delivered":
            delivered += record["portions"]
    lines = [f"Meals served: {offered + picked_up + delivered}"]
    for record in sorted(records, key=by_food_name):
        lines.append(f"{record['food']:<26} {record['portions']:>4}  {record['contact']}")
    return "\n".join(lines)
```

**ជំហានទី 2 · សរសេរទំព័រ។** ចូរបំពេញ `TODO` ទាំងបី រួច Run cell។ វារក្សាទុកទំព័រនៅក្នុង folder របស់អ្នក ហើយបង្ហាញវា។ ដើម្បីរក្សាច្បាប់ចម្លងមួយ៖ ផ្ទាំង **Files** → `my_foodbridge` → ចុចកណ្ដុរស្ដាំលើ `public_board.html` → **Download**។

````python
header = ""
for field in PUBLIC_FIELDS:
    header = header + "<th>" + html.escape(field) + "</th>"

rows = ""
for record in most_urgent_first(bridge.sample_records()):
    if record["status"] not in []:             # TODO 1: តើ status ណាខ្លះដែលអាហារនឹងមិនមក? ចូរដាក់វាក្នុង list
        row = public_row(record)
        cells = ""
        for field in PUBLIC_FIELDS:
            value = str(row.get(field, ""))    # TODO 2: ធ្វើឱ្យវាមានសុវត្ថិភាពសម្រាប់ទំព័រវេប: html.escape(...)
            cells = cells + "<td>" + value + "</td>"
        rows = rows + ""                       # TODO 3: បន្ថែមជួរដេកមួយក្នុងតារាង: "<tr>" + cells + "</tr>\n"

page = ("<meta charset='utf-8'>\n<h3>Campus FoodBridge · Public board</h3>\n"
        "<table border='1'>\n<tr>" + header + "</tr>\n" + rows + "</table>\n")
with open("my_foodbridge/public_board.html", "w", encoding="utf-8") as file:
    file.write(page)
print("Saved my_foodbridge/public_board.html with", rows.count("<tr>"), "offers")
if "@" in page:
    print("🔓 A contact is on your public page. Check PUBLIC_FIELDS.")
try:
    from IPython.display import HTML, display
    display(HTML(page))                    # 👀 Colab និង Jupyter គូរទំព័ររបស់អ្នកនៅទីនេះ
except ImportError:
    print("Open the file in a browser to see it.")
````
