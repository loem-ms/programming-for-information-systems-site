### 2 · ការតម្រៀប record ដោយប្រើ key function ដែលមានឈ្មោះ

Python មិនអាចដឹងដោយខ្លួនឯងថា ត្រូវតម្រៀប dictionary តាមអ្វីទេ៖ តាមឈ្មោះ? ឬតាមពិន្ទុ? អ្នកត្រូវប្រាប់វាដោយប្រើ **key function** មួយ។ Function នេះទទួល item មួយ ហើយ return តម្លៃដែលត្រូវប្រើសម្រាប់តម្រៀប។

````python
students = [{"name": "Sokha", "mark": 72},
            {"name": "Dara", "mark": 90},
            {"name": "Vanna", "mark": 61}]

def by_mark(student):
    return student["mark"]

for student in sorted(students, key=by_mark):
    print(student["name"], student["mark"])
````

ចូរសរសេរ `key=by_mark` **ដោយគ្មាន** `( )`។ អ្នកផ្ដល់ function ខ្លួនវាផ្ទាល់ទៅឱ្យ `sorted` ហើយ `sorted` ហៅវាម្ដង សម្រាប់និស្សិតម្នាក់ៗ។ ឈ្មោះល្អ ប្រាប់ពីលំដាប់នៃការតម្រៀប៖ `by_mark`, `most_urgent_first`។

🤔 **ទាយមើល:** តើនរណានឹងនៅមុនគេ ប្រសិនបើអ្នកបន្ថែម `reverse=True`? ចូរកែ cell ខាងលើ ហើយពិនិត្យមើល។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

យើងអាចតម្រៀប record សម្រាប់អ្នកអានដែលត្រូវការវាបាន។ បញ្ហាបន្ទាប់: record ដែលបានតម្រៀបរួច នៅតែមានគ្រប់ field រួមទាំង field ឯកជនផង។

### 3 · ការជ្រើសរើស field ពី dictionary

ទិដ្ឋភាពសាធារណៈ គឺជា **ច្បាប់ចម្លងតូចជាង** នៃ record មួយ។ វិធីដែលមានសុវត្ថិភាពក្នុងការបង្កើតវា គឺចាប់ផ្ដើមពីទទេ ហើយចម្លងតែ field ដែលមាននៅក្នុង **បញ្ជីអនុញ្ញាត (allowed list)** ប៉ុណ្ណោះ។ (លេខទូរស័ព្ទនេះជាលេខប្រឌិត។)

🤔 **ទាយមើល:** បន្ទាត់ពីរចុងក្រោយបន្ថែម field ឯកជនថ្មីមួយ។ តើក្ដារព័ត៌មាន (notice board) នឹងបង្ហាញវាដែរឬទេ?

````python
student = {"name": "Dara", "phone": "012 000 111", "mark": 90, "class": "MIS1"}
NOTICE_BOARD_FIELDS = ["name", "class"]

def notice_board_row(student):
    row = {}
    for field in NOTICE_BOARD_FIELDS:
        row[field] = student[field]
    return row

print(notice_board_row(student))

student["home_address"] = "Street 271"      # នរណាម្នាក់បន្ថែម field ឯកជនថ្មីមួយ នៅពេលក្រោយ
print(notice_board_row(student))
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …

វិធីម្យ៉ាងទៀត គឺចម្លងអ្វីៗទាំងអស់ រួចលុប field ឯកជនចេញ។ វិធីនេះមានហានិភ័យ។ ប្រសិនបើនរណាម្នាក់បន្ថែម field ឯកជនថ្មីនៅពេលក្រោយ field នោះនឹងបង្ហាញនៅលើក្ដារ រហូតដល់មាននរណាម្នាក់កត់សម្គាល់ឃើញ។ ផ្ទុយទៅវិញ ជាមួយបញ្ជីអនុញ្ញាត field ថ្មីនឹងនៅតែលាក់ រហូតដល់មនុស្សម្នាក់ **សម្រេចចិត្ត** បង្ហាញវា។

យើងអាចចម្លងតែ field ដែលអ្នកអានអាចមើលឃើញបាន។ បញ្ហាបន្ទាប់: បញ្ជីដែលមានសុវត្ថិភាព នៅតែពិបាកអាន ប្រសិនបើលេខរបស់វាមិនតម្រង់ជួរគ្នា។

### 4 · ការតម្រង់ជួរ column ដោយ f-string

នៅខាងក្នុង `{ }` សញ្ញា colon កំណត់ទទឹង៖ `<` ដាក់តម្លៃនៅខាងឆ្វេង ហើយ `>` ដាក់តម្លៃនៅខាងស្ដាំ។ ចូរដាក់ text នៅខាងឆ្វេង និងលេខនៅខាងស្ដាំ ដើម្បីឱ្យខ្ទង់លេខតម្រង់ជួរគ្នា។

````python
food = "Bread rolls"
portions = 30
print(f"[{food:<12}]")       # text នៅខាងឆ្វេង ទទឹង 12 តួអក្សរ
print(f"[{portions:>4}]")    # លេខនៅខាងស្ដាំ ទទឹង 4 តួអក្សរ

print(f"{food:<12} {portions:>4} portions")
food = "Rice"
portions = 3
print(f"{food:<12} {portions:>4} portions")
````

យើងអាចតម្រង់ជួរ field ដែលមានសុវត្ថិភាព ដើម្បីឱ្យអ្នកអានអាចមើលរំលងបានយ៉ាងរហ័ស។ បញ្ហាបន្ទាប់: អេក្រង់ក៏ត្រូវឆ្លើយនឹងជម្រើសដែលមនុស្សម្នាក់ធ្វើដែរ។

### 5 · Menu function ដែល return text

Menu មានការងារពីរ៖ **សួរ** អ្នកប្រើថាគេចង់បានអ្វី និង **សម្រេច** ថាត្រូវបង្ហាញអ្វី។ ចូរទុកការងារទាំងពីរនេះឱ្យដាច់ពីគ្នា។ ផ្នែកដែលសម្រេច គឺជា function ធម្មតាមួយ ដែល **return** text។ ធ្វើបែបនេះ អ្នកអាចសាកល្បង (test) វាបាន ដោយមិនចាំបាច់វាយអក្សរ។ Function ដដែលនេះ ក៏អាចប្រើបានសម្រាប់ text menu, ការពិនិត្យ (check) ឬទំព័រ web មួយដែរ។

````python
def library_answer(choice):
    if choice == "1":
        return "Open 8:00-17:00, Monday to Saturday."
    if choice == "2":
        return "You can borrow 3 books for 14 days."
    if choice == "q":
        return "Goodbye!"
    return "Unknown choice " + repr(choice) + ". Please type 1, 2 or q."

print(library_answer("1"))
print(library_answer("7"))
````

`repr(choice)` បង្ហាញ text នៅក្នុងសញ្ញាសម្រង់ ដូច្នេះសូម្បីតែចម្លើយទទេ ឬដកឃ្លា (space) មួយ ក៏អាចមើលឃើញដែរ។

🤔 តើ `library_answer("7")` នឹង return អ្វី ប្រសិនបើគ្មានបន្ទាត់ `return` ចុងក្រោយ? ហើយតើអ្នកប្រើនឹងឃើញអ្វី?

**ចម្លើយរបស់ខ្ញុំ:** …

យើងអាចបញ្ជូន text សម្រាប់ជម្រើស menu មួយត្រឡប់មកវិញបាន។ បញ្ហាបន្ទាប់: ការប្រើការ sort និង column ដែលតម្រង់ជួររួមគ្នា នៅលើ record ពិតរបស់ FoodBridge។

### ឧទាហរណ៍គំរូ · បន្ទាត់សាធារណៈខាងលើគេ

យើងឆ្លើយជាបីជំហាន៖

1. តម្រៀប (sort) record ទាំងអស់ដោយ key function ដោយយកម៉ោងនៅសល់តិចបំផុតនៅមុនគេ។
2. យក record ទីមួយ។
3. បង្កើតមួយបន្ទាត់ពី record នោះដោយ f-string ដោយអានតែ field ដែលអ្នកអានសាធារណៈត្រូវការ។

វាគឺជា key function ពីផ្នែកទី 2 និង f-string ដែលតម្រង់ជួរពីផ្នែកទី 4 ហើយបន្ទាត់នោះមិនដែលអាន `contact` ដែលជា field ឯកជនពីផ្នែកទី 3 ទេ។ ខាងក្រោមនេះ គឺជាសំណើបី៖

````python
board_records = [
    {"id": "F201", "food": "Bread rolls", "portions": 30, "hours_left": 5, "contact": "sunrise@example.org"},
    {"id": "F202", "food": "Fried rice boxes", "portions": 18, "hours_left": 2, "contact": "canteen@example.org"},
    {"id": "F203", "food": "Vegetable curry", "portions": 12, "hours_left": 4, "contact": "lotus@example.org"},
]

def by_hours_left(record):
    return record["hours_left"]

def top_public_line(records):
    first = sorted(records, key=by_hours_left)[0]
    return f"{first['food']:<20} {first['portions']:>3} portions, {first['hours_left']} hours left"

print(top_public_line(board_records))
````
