````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាពន្លា (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.7"
_KIT_SHA = "937b9aba05e2ee76cde15efd202da219a1a1cb876f5968d83cdf3142ded860ff"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
bridge.folder.connect()          # រក្សាទុក folder my_foodbridge/ របស់អ្នក (ក្នុង Google Drive នៅក្នុង Colab)
````

````python
def total_printed(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    print(total)

def total_returned(prices_riel):
    total = 0
    for price_riel in prices_riel:
        total += price_riel
    return total

printed_result = total_printed([4000, 1500])
returned_result = total_returned([4000, 1500])
print("printed_result is", printed_result)
print("returned_result is", returned_result)
````

- `print` **បង្ហាញ** តម្លៃមួយដល់មនុស្ស។ បន្ទាប់មក តម្លៃនោះបាត់ទៅ។
- `return` **បញ្ជូន** តម្លៃនោះត្រឡប់ទៅកូដដែលបានហៅ function វិញ។ កូដនោះអាចយកវាទៅបូក ប្រៀបធៀប តេស្ត ឬបង្ហាញលើអេក្រង់។

ប្រព័ន្ធមិនដែលអានអេក្រង់ទេ។ function ដែលគ្រាន់តែ print គឺគ្មានប្រយោជន៍សម្រាប់ផ្នែកផ្សេងទៀតនៃប្រព័ន្ធ ហើយមិនអាចតេស្តបានទេ។

### 5 · ពី flowchart ទៅជា function

មុនពេលអ្នកសរសេរវិធានមួយ ចូរគូរវាជាមុនសិន។ **flowchart (គំនូសតាងលំហូរ)** ប្រើប្រអប់បីទម្រង់៖ រាងពងក្រពើសម្រាប់ចំណុចចាប់ផ្ដើម ប្រអប់រាងពេជ្រ (diamond) សម្រាប់សំណួរលក្ខខណ្ឌ បាទ/ទេ (yes/no) និងប្រអប់ចតុកោណសម្រាប់លទ្ធផលបញ្ចប់។ នេះគឺជាវិធានទៅយកអាហារ (pickup) ពីសប្ដាហ៍ទី 2 របស់អ្នក៖

Flowchart of pickup_label: not approved gives hold; otherwise team at or above the team minimum, else standard

**ប្រអប់រាងពេជ្រ (លក្ខខណ្ឌ)** នីមួយៗ ក្លាយជា `if` មួយ។ **ប្រអប់លទ្ធផលបញ្ចប់** នីមួយៗ ក្លាយជា `return` មួយ។ ប្រសិនបើអ្នកដើរតាមព្រួញនីមួយៗម្ដង អ្នកបានតេស្តគ្រប់ផ្លូវទាំងអស់។

````python
TEAM_PICKUP_MIN_PORTIONS = 20

def pickup_label(portions, approved):
    if not approved:                              # ប្រអប់លក្ខខណ្ឌ (រាងពេជ្រ) ទី 1
        return "hold"
    if portions >= TEAM_PICKUP_MIN_PORTIONS:      # ប្រអប់លក្ខខណ្ឌ (រាងពេជ្រ) ទី 2
        return "team"
    return "standard"

print(pickup_label(30, False), pickup_label(20, True), pickup_label(19, True))
````

| **ធម្មតា** (normal) | តើវាដំណើរការត្រឹមត្រូវ សម្រាប់ធាតុចូលធម្មតាដែរឬទេ? | `[4000, 1500]` → 5500 |
| **ព្រំដែន** (boundary) | តើវាដំណើរការត្រឹមត្រូវ នៅចំណុចព្រំដែនដែរឬទេ? | `[4000]` → 4000 |
| **ទទេ ឬចម្លែក** (empty or odd) | ចុះធាតុចូលដែលបាត់ ទទេ ឬមិនធម្មតាវិញ? | `[]` → 0 |

cell នេះក៏បង្កើត `run_tests` ដែលជាកម្មវិធីដំណើរការតេស្ត (test runner) តូចមួយ។ អ្នកមិនចាំបាច់សរសេរវាទេ គ្រាន់តែអានវាបានហើយ។ វាហៅតេស្តនីមួយៗ ហើយ print ✅ នៅពេលគ្មាន `assert` ណាមួយមិនជាប់ ឬ print ❌ ជាមួយមូលហេតុ។

````python
def run_tests(tests):
    """Run each test function. Print ✅ or ❌ for each one; never stop the notebook."""
    passed = 0
    for test in tests:
        try:
            test()
            print("✅", test.__name__)
            passed += 1
        except AssertionError as problem:
            print("❌", test.__name__, "→", problem)
        except Exception as problem:
            print("❌", test.__name__, "→ error:", type(problem).__name__, problem)
    print(passed, "of", len(tests), "tests passed")

def test_total_normal():
    assert total_price([4000, 1500]) == 5500

def test_total_boundary():
    assert total_price([4000]) == 4000

def test_total_empty():
    assert total_price([]) == 0

run_tests([test_total_normal, test_total_boundary, test_total_empty])
````

សូមកត់សម្គាល់ `run_tests([test_total_normal, ...])`៖ អ្នកផ្ដល់ **function** ដោយគ្មាន `( )`។ runner ជាអ្នកហៅ function ទាំងនោះ។

មិត្តរួមថ្នាក់ម្នាក់ បានសុំជំនួយការ AI (AI assistant) ឱ្យសរសេរ *"function ដែលផ្ដល់ចំណែកដែលបានប្រគល់ ហើយសម្អាតការផ្ដល់ជូនដែលបានបញ្ចប់ ព្រមទាំងមានតេស្តមួយ។"* វាបានសរសេរ **Version A**។ **Version B** គឺជាការសរសេរឡើងវិញ។ យើងបានបន្ថែម `_a` និង `_b` ទៅឈ្មោះ ដើម្បីឱ្យទាំងពីរអាចនៅក្នុង notebook នេះជាមួយគ្នាបាន។ សូម Run cell ទាំងពីរ៖ គ្រប់តេស្តសុទ្ធតែជាប់។

````python
def delivered_portions_a(records):
    total = 0
    for record in records:
        if record["status"] == "delivered":
            total += record["portions"]
            records.remove(record)          # សម្អាតការផ្ដល់ជូនដែលបានបញ្ចប់
    print("Delivered:", total, "portions")

def test_delivered_a():
    delivered_portions_a(bridge.sample_records())

run_tests([test_delivered_a])
````

### ⭐⭐ ស្រេចចិត្ត · ពិនិត្យចម្លើយរបស់ AI

ជំនួយការ AI មួយ បានឆ្លើយសំណើមិនច្បាស់លាស់មួយ ដោយ function ខាងក្រោម។ ភារកិច្ចរបស់អ្នក៖

- ចូរ Run ការពិនិត្យ។
- ចូររកបន្ទាត់ដែលបំពានកិច្ចសន្យា ហើយកែវា។

````python
def ai_open_portions(records):
    # សរសេរដោយជំនួយការ AI
    total = 0
    for record in records:
        if record["status"] != "rejected":
            total += record["portions"]
    return total

bridge.check("open_portions", ai_open_portions)
````

## វគ្គទី 3 · សេចក្ដីណែនាំអំពី Midterm

វគ្គទី 3 គឺជាការប្រឡង midterm ធ្វើម្នាក់ឯង៖ **"បញ្ជរបណ្ណាល័យសាកលវិទ្យាល័យ (Campus Library Desk)"** រយៈពេល **60 នាទី** (+ 30 នាទី រៀបចំ ដាក់ស្នើ និងពិភាក្សាចម្លើយ)។ Notebook នឹងត្រូវដោះសោក្នុងថ្នាក់ដោយប្រើ **លេខកូដ PIN** ពីម៉ឺនុយ **Midterm**។

- **ការងារម្នាក់ឯង។** អនុញ្ញាត៖ កំណត់ចំណាំផ្ទាល់ខ្លួនរបស់អ្នក notebook ប្រចាំសប្ដាហ៍ និង notebook Python Basics។ មិនអនុញ្ញាត៖ ឧបករណ៍ AI ការផ្ញើសារ ឬការចែករំលែកកូដ។
- **ផ្នែកទាំង 4 (100 ពិន្ទុ)៖** 1 អាន និងដេញកូដដោយដៃ (trace) (24) · 2 បំប្លែង flowchart → function វិធាន + ផលប៉ះពាល់នៃការប្ដូរវិធាន (24) · 3 ធ្វើការជាមួយ records → dictionary សរុបលទ្ធផល (28) · 4 ត្រួតពិនិត្យ សរសេរ test និងកែសម្រួលកូដ AI + របាយការណ៍ផលប៉ះពាល់ (24) · ចុងបញ្ចប់៖ **បន្ទាត់កូដសំខាន់របស់ខ្ញុំ (My key line)** (ចាំបាច់) + ការ run កូដសាជាថ្មីទាំងអស់ពីដើម (clean run)។
- បន្ទាប់មក៖ ពិភាក្សាចម្លើយផ្ទាល់ និង **ការពិនិត្យផ្ទាល់មាត់ខ្លី 2–3 នាទី** លើកូដផ្ទាល់ខ្លួនរបស់អ្នក។

**របៀបត្រៀមខ្លួន៖** ចូរធ្វើ cell 🤔 ទាយមើល នៃសប្ដាហ៍ទី 1–4 ម្ដងទៀត ហើយពន្យល់ចម្លើយនីមួយៗឱ្យឮៗ; បំប្លែង flowchart → `if/elif/else` ដោយប្រើតម្លៃថេរមានឈ្មោះ (named constants); loop លើ records ដោយមិនកែធាតុចូល; សរសេរតេស្ត `assert` ដែលចាប់កំហុស (bug); ហើយប្រើ **Runtime → Restart session and run all** លើ notebook នេះ។

### កិច្ចការស្រាវជ្រាវទី 3 · ការពិនិត្យ input (5%)

*ដាក់ឱ្យធ្វើក្រោយ midterm។* តើហេតុអ្វីបានជាប្រព័ន្ធពិនិត្យធាតុចូល (input) **មុន** ពេលរក្សាទុកវា? ចូរជ្រើសរើសទម្រង់បែបបទ (form) មួយដែលអ្នកធ្លាប់ប្រើ ឧទាហរណ៍៖ form ចុះឈ្មោះនៅសាកលវិទ្យាល័យ ការកក់សំបុត្រឡានក្រុង ឬសំបុត្រផ្សេងៗ ការបញ្ចូលលុយទូរស័ព្ទ ការទូទាត់ប្រាក់ពេលទិញទំនិញអនឡាញ ឬការចុះឈ្មោះចូលក្លឹប។

សម្រាប់ **field ពីរ** នៃ form នោះ ចូរផ្ដល់៖

1. **តម្លៃដែលអនុញ្ញាត** (ឧទាហរណ៍ "ចំនួនគត់ពី 1 ដល់ 10");
2. ឧទាហរណ៍ **មិនត្រឹមត្រូវ** មួយ;
3. **សារ** ដែល form បង្ហាញ ឬសារដែលវា *គួរតែ* បង្ហាញ;
4. អ្វីដែលត្រូវតែ **នៅដដែល** នៅពេលធាតុចូលមិនត្រឹមត្រូវ។
