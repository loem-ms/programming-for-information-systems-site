- បន្ទាប់ពី midterm អ្នកនឹងមាន **ការពិនិត្យផ្ទាល់មាត់ 3 នាទី**: គ្រូនឹងចង្អុលទៅ code មួយបន្ទាត់របស់អ្នក និងតេស្តមួយរបស់អ្នក ហើយអ្នកត្រូវពន្យល់វា។ កូដដែលអ្នកមិនអាចពន្យល់បាន មិនត្រូវបានរាប់ថាជាការងាររបស់អ្នកទេ។

**ចាប់ផ្ដើម:** ចុចលើពាក្យ Copy to Drive (ឬ ចូលទៅកាន់ File រួចចុច Save a copy in Drive) បន្ទាប់មក Run setup cell។

```python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.3"
_KIT_SHA = "689b26d4950abba56ebe2affd1b5fc02e3165a03f712d0b3be8ba9c1432cea43"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
```

```python
def total(records):
    count = 0
    for record in records:
        count = count + record["portions"]
    print(count)

result = total([{"portions": 10}, {"portions": 5}])
print("Result:", result)
```

- [ ] ផ្នែក A: ការពន្យល់ចំនួនប្រាំ ដែលការពន្យល់នីមួយៗមាន *ហេតុអ្វី*។
- [ ] ផ្នែក B និង C: បាន Run ការពិនិត្យរួច (ទោះបីជា ✅ ឬមិនជាប់ក៏ដោយ សូមបង្ហាញការព្យាយាមដ៏ល្អបំផុតរបស់អ្នក)។
- [ ] ផ្នែក D: លទ្ធផលរបស់ `catch_bugs` បង្ហាញឱ្យឃើញច្បាស់។
- [ ] ផ្នែក E: តារាងមានបញ្ហាយ៉ាងហោចណាស់បី។
- [ ] ផ្នែក F: បំពេញកំណត់ត្រាការសម្រេចចិត្តរួច ហើយអ្នកបានប្រើ **Runtime → Restart session and run all** មុនពេលរក្សាទុក។

ចូរទាញយក notebook របស់អ្នក (**File → Download → .ipynb**) ហើយដាក់ស្នើតាមការណែនាំរបស់គ្រូ។ បន្ទាប់មក ចូរត្រៀមខ្លួនសម្រាប់ការពិនិត្យផ្ទាល់មាត់ 3 នាទី៖ អ្នកត្រូវត្រៀមពន្យល់បន្ទាត់ណាមួយនៃ `plan_day` របស់អ្នក និងតេស្តមួយរបស់អ្នក។

**អ្វីដែលអ្នកអាចធ្វើបានឥឡូវនេះ:** អានកូដ និងពន្យល់បានថា ហេតុអ្វីវាមានឥរិយាបថដូច្នោះ សរសេរវិធានតូចមួយដែលបញ្ជូនចម្លើយមួយត្រឡប់មកវិញ (return) សង្ខេប record ពេញមួយថ្ងៃដោយមិនប្ដូរវា សរសេរតេស្ត (test) ដែលចាប់កំហុស (bug) បាន វាយតម្លៃកូដដែល AI សរសេរ ដោយធៀបនឹងគោលការណ៍ទាំងបួន និងការពារការសម្រេចចិត្តមួយជាមួយ clean run។

**បញ្ហារបស់សប្ដាហ៍ក្រោយ:** ថ្ងៃនេះ function របស់អ្នកទទួលបាន record ស្អាត និងរួចរាល់។ នៅសប្ដាហ៍ទី 5 តើនឹងមានអ្វីកើតឡើង នៅពេលធាតុចូល (input) មកដល់ជាអត្ថបទរញ៉េរញ៉ៃ — មានដកឃ្លាលើស ពាក្យជំនួសលេខ និង field ដែលបាត់ — ដែលអ្នកមិនអាចទុកចិត្តបាន?
