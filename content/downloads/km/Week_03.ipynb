## មុនពេលចាប់ផ្ដើម

1. ចុចលើពាក្យ Copy to Drive (ឬ ចូលទៅកាន់ File រួចចុច Save a copy in Drive) រួចធ្វើការក្នុងច្បាប់ចម្លងរបស់អ្នក។
2. Run setup cell។ Run វាម្ដងទៀត ក្រោយពេល runtime restart។
3. ⭐ = កម្រិតមូលដ្ឋាន សម្រាប់គ្រប់គ្នា, ⭐⭐ = សម្រាប់សាកល្បងបន្ថែម។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.7"
_KIT_SHA = "f2b097068a1ca7d5251963015748526b066b60e0498b0930830bfcb445c723af"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

### គំនិតសំខាន់: record, ID និង data model

ចូរគិតអំពីបញ្ជីឈ្មោះនិស្សិតក្នុងថ្នាក់។ និស្សិតម្នាក់ៗ គឺជា **ជួរដេកមួយ**។ គ្រប់ជួរដេកមានជួរឈរដូចគ្នា: ID និស្សិត (អត្តលេខ) ឈ្មោះ និងថ្នាក់។ និស្សិតពីរនាក់អាចមានឈ្មោះដូចគ្នា ប៉ុន្តែមិនដែលមាន ID និស្សិតដូចគ្នាទេ។ នៅពេលការិយាល័យត្រូវការពិន្ទុរបស់អ្នក គេស្វែងរកអ្នកតាម ID មិនមែនតាមឈ្មោះទេ។

ប្រព័ន្ធរបស់យើងក៏ដំណើរការដូចគ្នាដែរ។ សំណើនីមួយៗ គឺជា **record** មួយ៖ ព័ត៌មានមួយក្រុមអំពីសំណើពិតមួយ។ គ្រប់ record មាន **field** ដូចគ្នា និង **ID** មួយដែលមិនស្ទួនគ្នា ដូចជា `"F001"`។ បញ្ជី field ទាំងនោះ ហៅថា **គំរូទិន្នន័យ (data model)**។ ការជ្រើសរើស field គឺជាការសម្រេចចិត្តផ្នែករចនា: តើយើងរក្សាទុកព័ត៌មានណាខ្លះ ហើយតើយើងទុកចោលព័ត៌មានណាខ្លះ?

### 1 · list រក្សាទុកតម្លៃច្រើន តាមលំដាប់

- វង់ក្រចកជ្រុង `[ ]` បង្កើត list។ item នីមួយៗ ខណ្ឌចែកដោយសញ្ញាក្បៀស។
- `volunteers[0]` គឺជា item **ទីមួយ**។ លេខទីតាំងនេះ ហៅថា **index**។
- `len(...)` រាប់ចំនួន item។ `.append(...)` បន្ថែម item មួយនៅខាងចុង។ `in` ផ្ដល់ `True` ឬ `False`។

🤔 **ទាយមើល** បន្ទាត់នីមួយៗ មុនពេលអ្នក Run cell។ តើ `volunteers[3]` នឹងជាអ្វី ក្រោយពេល `append`?

### 2 · dictionary ដាក់ឈ្មោះឱ្យតម្លៃនីមួយៗ

**dictionary** រក្សាទុកទិន្នន័យជាគូ: **key** (ឈ្មោះ) និង **តម្លៃ (value)**។ អ្នករកតម្លៃមួយតាម key របស់វា មិនមែនតាមទីតាំងទេ។

````python
item = {"name": "Notebook", "price_riel": 4000, "in_stock": 25}
print(item["price_riel"])
item["price_riel"] = 3500        # ប្ដូរតម្លៃមួយ
item["colour"] = "blue"          # បន្ថែម key ថ្មីមួយ
print(item)
print(list(item.keys()))
print(list(item.values()))
print(item.get("size"))
print(item.get("size", "one size"))
````

- វង់ក្រចកកោង `{ }` បង្កើត dictionary។ គូនីមួយៗសរសេរជា `key: value`។
- key ត្រូវតែដូចគ្នាបេះបិទ ដូចការប្រៀបធៀបអត្ថបទ (text) នៅសប្ដាហ៍ទី 2: `"Price_riel"` គឺជា key ផ្សេង។

ចុះប្រសិនបើ key នោះមិនមានវិញ? `item["size"]` ឈប់ដំណើរការ ហើយបង្ហាញ `KeyError`។ `.get(...)` គឺជាវិធីសួរដោយគួរសម: វាផ្ដល់ `None` (ឬតម្លៃដើម (default value) ដែលអ្នកជ្រើសរើស) ជំនួសឱ្យបញ្ហា (error)។

🤔 **ទាយមើល:** តើបន្ទាត់ពីរនៅចុងបញ្ចប់នៃ cell ខាងលើ print អ្វី?

### 3 · record គឺជា dictionary ហើយមួយថ្ងៃគឺជា list នៃ record

សំណើមួយ គឺជា dictionary មួយ។ សំណើច្រើន គឺជា **list នៃ dictionary**។ នេះជាទម្រង់ដែលប្រព័ន្ធព័ត៌មានស្ទើរតែទាំងអស់ប្រើ ចាប់ពីបញ្ជីឈ្មោះនិស្សិត រហូតដល់របាយការណ៍គណនីធនាគារ (bank statement)។

### 4 · `for`: ធ្វើរឿងដដែល សម្រាប់គ្រប់ item

**loop** ដំណើរការបន្ទាត់ដែលចូលបន្ទាត់ (indented) ម្ដងសម្រាប់ item នីមួយៗក្នុង list។ ឈ្មោះដែលនៅបន្ទាប់ពី `for` គឺជាឈ្មោះរបស់ item បច្ចុប្បន្ន។

loop ទីពីរ គឺជាគំនិតដដែល ប៉ុន្តែប្រើជាមួយ record។ ទម្លាប់ល្អមួយគឺ៖ ប្រើឈ្មោះ **ពហុវចនៈ** (plural) សម្រាប់ list (`offers`) និងឈ្មោះ **ឯកវចនៈ** (singular) សម្រាប់ item មួយ (`offer`)។

````python
prices_riel = [4000, 1500, 2000]
for price_riel in prices_riel:
    print("Price:", price_riel, "riel")
print("Done")

for offer in offers:
    print(offer["id"], "·", offer["food"], "·", offer["portions"], "portions")
````

loop អាចទៅមើល record នីមួយៗ។ ប៉ុន្តែអ្នកសម្របសម្រួលមិនត្រូវការការទៅមើលទេ — គាត់ត្រូវការលេខតែមួយ សម្រាប់ពេញមួយថ្ងៃ។ ដូច្នេះ យើងរក្សាផលបូកសរុបមួយ ខណៈពេល loop កំពុងដំណើរការ។

### 5 · accumulator: ចាប់ផ្ដើមពីសូន្យ រួចបូក item នីមួយៗ

ដើម្បីបូក list មួយ អ្នកត្រូវរក្សាផលបូកដែលកើនឡើងជាបន្តបន្ទាប់ (running total)។ លំនាំនេះ ហៅថា **accumulator** (លំនាំប្រមូលផលបូក)។

````python
prices_riel = [4000, 1500, 2000]
total_riel = 0                    # 1. ចាប់ផ្ដើមមុន loop
for price_riel in prices_riel:
    total_riel += price_riel      # 2. បូក item នីមួយៗ
print(total_riel)                 # 3. ប្រើលទ្ធផល ក្រោយ loop
````

- `total_riel += price_riel` មានន័យថា `total_riel = total_riel + price_riel`។
- លំនាំដដែលនេះ ក៏អាចប្រើដើម្បី **រាប់** បានដែរ: ចាប់ផ្ដើមដោយ `count = 0` រួច `count += 1` សម្រាប់ item នីមួយៗដែលអ្នកចង់រាប់។

🤔 **ទាយមើល:** តើនឹង print អ្វី ប្រសិនបើ `total_riel = 0` ត្រូវបានផ្លាស់ទៅ **ក្នុង** loop (ចូលបន្ទាត់ ហើយនៅពីលើបន្ទាត់ `+=` បន្តិច)? ចូរសាកល្បងវា រួចផ្លាស់វាមកកន្លែងដើមវិញ។

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **មូលហេតុ:** …

ការបូកសរុប មើល record ទាំងអស់បញ្ចូលគ្នា។ ការស្វែងរក ធ្វើផ្ទុយពីនេះ: វារក record តែមួយ ហើយឈប់ភ្លាមនៅពេលរកឃើញ។

### 6 · ស្វែងរក record មួយ តាម ID របស់វា

ដើម្បីរក record មួយ អ្នកត្រូវមើល record ម្ដងមួយៗ។ នៅពេល ID ត្រូវគ្នា ចូរ **បញ្ជូនត្រឡប់មកវិញ (return)** record នោះ។ `return` បញ្ឈប់ loop ភ្លាមៗ។ មានតែ **ក្រោយ** ពេល loop បានពិនិត្យគ្រប់ record រួចហើយប៉ុណ្ណោះ ទើបអ្នកដឹងថាវាមិនមាន។ ពេលនោះ ចូរ return `None`។

````python
shop_items = [
    {"code": "N01", "name": "Notebook", "price_riel": 4000},
    {"code": "P02", "name": "Pen", "price_riel": 1500},
    {"code": "R03", "name": "Ruler", "price_riel": 2000},
]

def find_item(items, wanted_code):
    for item in items:
        if item["code"] == wanted_code:
            return item          # រកឃើញ: ឈប់ ហើយបញ្ជូនវាត្រឡប់ទៅវិញ
    return None                  # បានពិនិត្យគ្រប់ item: រកមិនឃើញ

print(find_item(shop_items, "P02"))
print(find_item(shop_items, "X99"))
````

សូមមើលការចូលបន្ទាត់ (indentation) របស់ `return None`។ វាតម្រង់ជួរជាមួយ `for` មិនមែនជាមួយ `if` ទេ។ វាដំណើរការតែនៅពេល loop បានបញ្ចប់ ដោយរកមិនឃើញអ្វីសោះ។

### ឧទាហរណ៍គំរូ · ចំណែក bread តែប៉ុណ្ណោះ

មុនពេលអ្នកបង្កើត ចូរមើលលំនាំនេះម្ដងសិន លើសំណួរតូចជាង។ អ្នកសម្របសម្រួលសួរថា: *«តើយើងមានចំណែក bread ប៉ុន្មាន នៅព្រឹកនេះ?»*

យើងឆ្លើយជាបួនជំហាន៖

1. ចាប់ផ្ដើមផលបូកនៅ 0។
2. loop លើ record ទាំងអស់។
3. បូកចំណែក **តែនៅពេល** category របស់ record គឺ `"bread"`។
4. បញ្ជូនផលបូកត្រឡប់មកវិញ (return) ក្រោយពេល loop ចប់។

វាគឺជាលំនាំ accumulator ពីផ្នែកទី 5 ដោយមាន `if` មួយនៅខាងក្នុង។ ខាងក្រោមនេះ គឺជាសំណើបីនៅព្រឹកនេះ:

````python
morning_offers = [
    {"id": "M01", "food": "Bread rolls", "category": "bread", "portions": 30},
    {"id": "M02", "food": "Bananas", "category": "fruit", "portions": 40},
    {"id": "M03", "food": "Baguettes", "category": "bread", "portions": 12},
]

def bread_portions(records):
    total = 0
    for record in records:
        if record["category"] == "bread":
            total += record["portions"]
    return total

print(bread_portions(morning_offers))
````

**របៀបដែលវាដំណើរការ:** loop ទៅមើល record ទាំងបី ប៉ុន្តែ `if` អនុញ្ញាតឱ្យតែ record ពីរប៉ុណ្ណោះ ទៅដល់បន្ទាត់ `+=`: 30 + 12 = 42។ record របស់ Bananas ត្រូវបានទៅមើល ពិនិត្យ រួចរំលងចោល។

នៅវគ្គទី 3 ផ្នែកទី 1 អ្នកនឹងសរសេរលំនាំដដែលនេះ ដោយគ្មាន `if` ទាល់តែសោះ: សម្រាប់ `total_portions` គ្រប់ record សុទ្ធតែត្រូវរាប់។

### ⭐ P3 · កែតម្រូវការស្វែងរក

ជំនួយការ AI មួយបានសរសេរកូដស្វែងរកនេះ។ វារក F101 ឃើញ ប៉ុន្តែវាប្រាប់ថា F103 មិនមានទេ។ ភារកិច្ចរបស់អ្នក៖

- ស្វែងរកកំហុស (bug)។
- កែតម្រូវវា។

````python
def find_by_id(records, offer_id):
    for record in records:
        if record["id"] == offer_id:
            return record
        else:
            return None

print(find_by_id(evening_offers, "F101"))
print(find_by_id(evening_offers, "F103"))   # គួររក Chicken rice ឃើញ
print(find_by_id(evening_offers, "F999"))   # គួរតែជា None
````

### 🧭 ការសម្រេចចិត្តរបស់អ្នក · តើយើងទទួលយកប្រភេទណាខ្លះ?

setting `CATEGORIES` គឺជាបញ្ជីប្រភេទអាហារ ដែលក្រុមទទួលយក។ សំណើថ្មីនីមួយៗ ត្រូវបានពិនិត្យនៅមាត់ទ្វារ: ប្រសិនបើប្រភេទរបស់វាមិននៅក្នុងបញ្ជីទេ សំណើនោះនឹងត្រូវបានបដិសេធ។ សូមមើលថា ពាក្យតែមួយអាចធ្វើឱ្យប្រែប្រួលអ្វីខ្លះ:

````python
def describe_day(title, system):
    numbers = system.numbers()
    print(title)
    print("  accepted:", numbers["offers"], "offers,", numbers["portions_offered"], "portions offered")
    print("  refused at the door:", numbers["inputs_refused"])

print("Default categories:", bridge.DEFAULTS["CATEGORIES"])
default_day = bridge.System()
default_day.run_day(show=False)
describe_day("Before (default categories)", default_day)

# + ភ្ជាប់ list ពីរ: default categories រួចបន្ទាប់មក "soup"
soup_day = bridge.System({"CATEGORIES": bridge.DEFAULTS["CATEGORIES"] + ["soup"]})
soup_day.run_day(show=False)
describe_day("After (soup added)", soup_day)
print()
print(soup_day.report())
````

ការបន្ថែម `"soup"` អនុញ្ញាតឱ្យស៊ុបមីរបស់ Riverside Restaurant ចូលបាន: មានចំណែក 25 បន្ថែមទៀត ហើយវាឡើងទៅ **កំពូល** នៃបញ្ជីរបស់អ្នកសម្របសម្រួល ព្រោះវានៅសល់តែ 3 ម៉ោងប៉ុណ្ណោះ។ តើក្រុមត្រៀមខ្លួនសម្រាប់រឿងនោះហើយឬនៅ? ស៊ុបត្រូវការប្រអប់ដែលមិនកំពប់ និងអង្គការដៃគូដែលអាចកម្ដៅវាឡើងវិញបាន។ ការដកប្រភេទមួយចេញ មានលទ្ធផលផ្ទុយពីនេះ: សំណើទាំងនោះត្រូវបានបដិសេធនៅមាត់ទ្វារ ហើយត្រូវប្រាប់អ្នកបរិច្ចាគពីមូលហេតុ។
