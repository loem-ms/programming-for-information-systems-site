2. Run setup cell។ Run វាម្ដងទៀត ក្រោយពេល runtime restart។
3. ⭐ = មូលដ្ឋាន សម្រាប់គ្រប់គ្នា, ⭐⭐ = បន្ថែម (ស្រេចចិត្ត)។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ Food Rescue
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.2"
_KIT_SHA = "87d59ca87437a3c6cc62441c2e95bfb0b252ea6b51ab2ccb7b2875f022775c3a"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("food_rescue_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("food_rescue_kit/my_settings.py", "food_rescue_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("food_rescue_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "rescue" or m.startswith("rescue.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import rescue as fr
fr.set_language("km")         # screens and checks in Khmer
print("✅ ប្រព័ន្ធ Food Rescue រួចរាល់ហើយ (version " + fr.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

### គំនិតសំខាន់: record, ID និង data model

គិតអំពីបញ្ជីឈ្មោះនិស្សិតក្នុងថ្នាក់។ និស្សិតម្នាក់ៗ គឺជា **ជួរដេកមួយ**។ គ្រប់ជួរដេកមានជួរឈរដូចគ្នា: ID និស្សិត (អត្តលេខ) ឈ្មោះ និងថ្នាក់។ និស្សិតពីរនាក់អាចមានឈ្មោះដូចគ្នា ប៉ុន្តែមិនដែលមាន ID និស្សិតដូចគ្នាទេ។ ពេលការិយាល័យត្រូវការពិន្ទុរបស់អ្នក គេរកអ្នកតាម ID មិនមែនតាមឈ្មោះទេ។

ប្រព័ន្ធរបស់យើងក៏ដូចគ្នាដែរ។ សំណើនីមួយៗ គឺជា **record** មួយ: ព័ត៌មានមួយក្រុម អំពីសំណើពិតមួយ។ គ្រប់ record មាន **field** ដូចគ្នា និង **ID** មួយដែលមិនស្ទួន ដូចជា `"F001"`។ បញ្ជី field ទាំងនោះ ហៅថា **data model** (គំរូទិន្នន័យ)។ ការជ្រើសរើសវា គឺជាការសម្រេចចិត្តផ្នែកការរចនា: តើយើងរក្សាព័ត៌មានណាខ្លះ ហើយទុកចោលព័ត៌មានណាខ្លះ?

- **តើអ្វីដែលយើងចេតនាមិនដាក់?** គ្មាន field "ចំណែកសរុបថ្ងៃនេះ" ទេ។ ផលបូកសរុបត្រូវបាន **គណនា** ពី record ទាំងឡាយ ពេលអ្នកត្រូវការវា។ បើអ្នករក្សាទុកវាផងដែរ អ្នកនឹងមានព័ត៌មានដដែលពីរដង ហើយថ្ងៃណាមួយ ព័ត៌មានទាំងពីរកន្លែងនឹងខុសគ្នា។

### 1 · list រក្សា value ច្រើន តាមលំដាប់

- វង់ក្រចកជ្រុង `[ ]` បង្កើត list។ item នីមួយៗ ខណ្ឌចែកដោយសញ្ញាក្បៀស។
- `volunteers[0]` គឺជា item **ទីមួយ**។ លេខទីតាំងនេះ ហៅថា **index**។
- `len(...)` រាប់ចំនួន item។ `.append(...)` បន្ថែម item មួយនៅខាងចុង។ `in` ផ្ដល់ `True` ឬ `False`។

🤔 **ទាយមើល** បន្ទាត់នីមួយៗ មុនពេលអ្នក run cell។ ក្រោយ `append` តើ `volunteers[3]` នឹងជាអ្វី?

````python
volunteers = ["Dara", "Sophea", "Vanna"]
print(volunteers[0])          # ទីតាំងចាប់ផ្ដើមពី 0
print(len(volunteers))        # មាន item ប៉ុន្មាន
volunteers.append("Kosal")    # បន្ថែមមួយនៅខាងចុង
print(volunteers)
print(volunteers[3])
print("Sophea" in volunteers) # តើវានៅក្នុង list ឬទេ?
# print(volunteers[4])        # លុប # ចេញ ដើម្បីមើល error រួចដាក់វាវិញ
````

### 2 · dictionary ដាក់ឈ្មោះឱ្យ value នីមួយៗ

**dictionary** រក្សាទុកទិន្នន័យជាគូ: **key** (ឈ្មោះ) និង **value**។ អ្នករក value តាម key របស់វា មិនមែនតាមទីតាំងទេ។

````python
item = {"name": "Notebook", "price_riel": 4000, "in_stock": 25}
print(item["price_riel"])
item["price_riel"] = 3500        # ប្ដូរ value មួយ
item["colour"] = "blue"          # បន្ថែម key ថ្មីមួយ
print(item)
print(list(item.keys()))
print(list(item.values()))
print(item.get("size"))
print(item.get("size", "one size"))
````

- វង់ក្រចកកោង `{ }` បង្កើត dictionary។ គូនីមួយៗសរសេរជា `key: value`។
- key ត្រូវតែដូចគ្នាបេះបិទ ដូចការប្រៀបធៀប text នៅសប្ដាហ៍ទី 2: `"Price_riel"` គឺជា key ផ្សេង។

ចុះបើ key នោះមិនមាន? `item["size"]` ឈប់ដោយ `KeyError`។ `.get(...)` គឺជាវិធីសួរដោយគួរសម: វាផ្ដល់ `None` (ឬ default ដែលអ្នកជ្រើសរើស) ជំនួសឱ្យ error។

### 4 · `for`: ធ្វើរឿងដដែល សម្រាប់គ្រប់ item

**loop** run បន្ទាត់ដែលដកចូល (indented) ម្ដងហើយម្ដងទៀត: ម្ដង សម្រាប់ item នីមួយៗក្នុង list។ ឈ្មោះនៅបន្ទាប់ពី `for` ចង្អុលទៅ item បច្ចុប្បន្ន។

loop ទីពីរ គឺជាគំនិតដដែល ប៉ុន្តែជាមួយ record។ ទម្លាប់ល្អមួយ: ឈ្មោះ **ពហុវចនៈ** (plural) សម្រាប់ list (`offers`) និងឈ្មោះ **ឯកវចនៈ** (singular) សម្រាប់ item មួយ (`offer`)។

````python
prices_riel = [4000, 1500, 2000]
for price_riel in prices_riel:
    print("Price:", price_riel, "riel")
print("Done")

for offer in offers:
    print(offer["id"], "·", offer["food"], "·", offer["portions"], "portions")
````

### 5 · accumulator: ចាប់ផ្ដើមពីសូន្យ បូក item នីមួយៗ

ដើម្បីបូក list មួយ ត្រូវរក្សាផលបូកដែលកើនឡើងជាបន្តបន្ទាប់ (running total)។ លំនាំនេះ ហៅថា **accumulator**។

````python
prices_riel = [4000, 1500, 2000]
total_riel = 0                    # 1. ចាប់ផ្ដើមមុន loop
for price_riel in prices_riel:
    total_riel += price_riel      # 2. បូក item នីមួយៗ
print(total_riel)                 # 3. ប្រើលទ្ធផល ក្រោយ loop
````

- `total_riel += price_riel` មានន័យថា `total_riel = total_riel + price_riel`។
- លំនាំដដែលនេះ ក៏ **រាប់** បានដែរ: ចាប់ផ្ដើម `count = 0` រួច `count += 1` សម្រាប់ item នីមួយៗដែលអ្នកចង់រាប់។

🤔 **ទាយមើល:** តើនឹង print អ្វី បើ `total_riel = 0` ត្រូវបានផ្លាស់ទៅ **ក្នុង** loop (ដកចូល នៅពីលើបន្ទាត់ `+=`)? សាកល្បងវា រួចផ្លាស់វាមកកន្លែងដើមវិញ។

**ការទាយ:** …  **លទ្ធផល:** …  **មូលហេតុ:** …

### 6 · ស្វែងរក record មួយ តាម ID របស់វា

ដើម្បីរក record មួយ ត្រូវមើល record ម្ដងមួយៗ។ ពេល ID ត្រូវគ្នា ចូរ **return** វា: `return` បញ្ឈប់ loop ភ្លាមៗ។ មានតែ **ក្រោយ** ពេល loop បានពិនិត្យគ្រប់ record ហើយ ទើបអ្នកដឹងថាវាមិនមាន; ពេលនោះ return `None`។

````python
shop_items = [
    {"code": "N01", "name": "Notebook", "price_riel": 4000},
    {"code": "P02", "name": "Pen", "price_riel": 1500},
    {"code": "R03", "name": "Ruler", "price_riel": 2000},
]

def find_item(items, wanted_code):
    for item in items:
        if item["code"] == wanted_code:
            return item          # រកឃើញ: ឈប់ ហើយបញ្ជូនវាត្រឡប់ទៅវិញ
    return None                  # បានពិនិត្យគ្រប់ item: រកមិនឃើញ

print(find_item(shop_items, "P02"))
print(find_item(shop_items, "X99"))
````

សូមមើលការដកចូល (indentation) របស់ `return None`។ វាតម្រង់ជួរជាមួយ `for` មិនមែនជាមួយ `if` ទេ។ វា run តែពេល loop បានបញ្ចប់ ដោយរកមិនឃើញអ្វីសោះ។

### ⭐ P3 · ជួសជុលការស្វែងរក

````python
def find_by_id(records, offer_id):
    for record in records:
        if record["id"] == offer_id:
            return record
        else:
            return None

print(find_by_id(evening_offers, "F101"))
print(find_by_id(evening_offers, "F103"))   # គួររក Chicken rice ឃើញ
print(find_by_id(evening_offers, "F999"))   # គួរតែជា None
````

````python
# Version B (មិត្តរួមថ្នាក់)
offer_records = [
    {"id": "F001", "donor": "Sunrise Bakery", "food": "Bread rolls", "portions": 30},
    {"id": "F002", "donor": "Green Market Stall", "food": "Bananas", "portions": 40},
]
offer_records.append({"id": "F003", "donor": "Golden Oven Bakery", "food": "Bread rolls", "portions": 12})

total = 0
for record in offer_records:
    total += record["portions"]
print(len(offer_records), "offers,", total, "portions")
````

### 🧭 ការសម្រេចចិត្តរបស់អ្នក · តើយើងទទួលយកប្រភេទណាខ្លះ?

setting `CATEGORIES` គឺជាបញ្ជីប្រភេទអាហារ ដែលក្រុមទទួលយក។ សំណើថ្មីនីមួយៗ ត្រូវបានពិនិត្យនៅមាត់ទ្វារ: បើប្រភេទរបស់វាមិននៅក្នុងបញ្ជី សំណើនោះត្រូវបានបដិសេធ។ មើលថា ពាក្យតែមួយ ប្រែប្រួលអ្វីខ្លះ:

````python
def describe_day(title, system):
    numbers = system.numbers()
    print(title)
    print("  accepted:", numbers["offers"], "offers,", numbers["portions_offered"], "portions offered")
    print("  refused at the door:", numbers["inputs_refused"])

print("Default categories:", fr.DEFAULTS["CATEGORIES"])
default_day = fr.System()
default_day.run_day(show=False)
describe_day("Before (default categories)", default_day)

# + ភ្ជាប់ list ពីរ: default categories រួចបន្ទាប់មក "soup"
soup_day = fr.System({"CATEGORIES": fr.DEFAULTS["CATEGORIES"] + ["soup"]})
soup_day.run_day(show=False)
describe_day("After (soup added)", soup_day)
print()
print(soup_day.report())
````
