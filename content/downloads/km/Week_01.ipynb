## មុនពេលចាប់ផ្ដើម

1. **រក្សាទុកច្បាប់ចម្លងរបស់អ្នកផ្ទាល់ជាមុនសិន។** នៅក្នុង Colab សូមចុចលើពាក្យ Copy to Drive (ឬ ចូលទៅកាន់ File រួចចុច Save a copy in Drive)។ ធ្វើការក្នុងច្បាប់ចម្លងនោះ មិនមែនក្នុងច្បាប់ដើមនេះទេ។
2. **Run setup cell ខាងក្រោម** (ចុច ▶)។ Run វាម្ដងទៀត រាល់ពេលដែល Colab ប្រាប់ថា runtime បាន restart។
3. ធ្វើការពីលើចុះក្រោម។ ផ្នែកនីមួយៗប្រាប់ថាគួរប្រើពេលប៉ុន្មាន។

កម្រិត: **⭐ មូលដ្ឋាន** សម្រាប់គ្រប់គ្នា។ **⭐⭐ បន្ថែម** គឺស្រេចចិត្ត សម្រាប់ពេលអ្នកធ្វើរួចមុនម៉ោង។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.3"
_KIT_SHA = "9a7f0e70ab94bb1480312c1ccc2cd08ea21038ab8fa51cc0adbfe520961d0844"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

## វគ្គទី 1 · រៀន: value ឈ្មោះ និង text (30 នាទី)

Program គ្រប់មួយធ្វើការជាមួយ **value**: លេខ text និង ពិត/មិនពិត (true/false)។ Python រក្សា value នីមួយៗជាមួយ **type** មួយ ហើយ type នោះកំណត់ថាអ្នកអាចធ្វើអ្វីខ្លះជាមួយវា។

### 1 · Value និង type

````python
print(30)
print(2.5)
print("Bread rolls")
print(True)
print(type(30), type(2.5), type("Bread rolls"), type(True))
````

- `int` ជាចំនួនគត់ `float` ជាលេខដែលមានចំណុចទសភាគ `str` ជា text ហើយ `bool` គឺ `True` ឬ `False`។
- Text តែងតែមានសញ្ញាសម្រង់ (quotes)។ `"30"` ជា text ដែលមើលទៅដូចលេខ; `30` ជាលេខ។

### 2 · ឈ្មោះ (variable)

**ឈ្មោះ** មួយចង្អុលទៅ value មួយ។ `=` មានន័យថា *"ឱ្យឈ្មោះនេះចង្អុលទៅ value នេះ"*។ វាមិនមែនជា `=` ក្នុងគណិតវិទ្យាទេ។

````python
portions = 30
hours_left = 5
food = "Bread rolls"
print(food, portions, hours_left)
````

អ្នកអាចឱ្យឈ្មោះមួយចង្អុលទៅ value ថ្មី។ Value ចាស់ត្រូវបានបំភ្លេចចោល។

🤔 **ទាយមើល** ថា cell នេះ print អ្វី រួច run វា។

````python
portions = 30
portions = portions + 12
print(portions)
````

### 3 · ការគណនា (arithmetic)

| Operator | អត្ថន័យ | `30 ? 4` |
|---|---|---|
| `+` `-` `*` | បូក ដក គុណ | `34`, `26`, `120` |
| `/` | ចែក (តែងតែផ្ដល់ float) | `7.5` |
| `//` | ចែកយកតែចំនួនគត់ | `7` |
| `%` | សំណល់ | `2` |

````python
portions = 30
volunteers = 4
print(portions / volunteers)
print(portions // volunteers)
print(portions % volunteers)
````

🤔 អ្នកស្ម័គ្រចិត្តមិនអាចដឹកកន្លះចំណែកបានទេ។ ក្នុងចំណោមចម្លើយទាំងបី តើមួយណាមានប្រយោជន៍សម្រាប់អ្នកសម្របសម្រួល ហើយមួយណាអាចធ្វើឱ្យយល់ច្រឡំ? សរសេរមួយប្រយោគ។

### 4 · Text និងលេខ មិនលាយគ្នាដោយចៃដន្យទេ

`+` បូកលេខ ប៉ុន្តែ **តភ្ជាប់** text។ Python មិនព្រមតភ្ជាប់ text ជាមួយលេខទេ ព្រោះវាមិនអាចទាយបានថាអ្នកចង់បានអ្វី។

````python
print(3 + 4)
print("3" + "4")
print(int("30") + 4)      # int() ប្ដូរ text ទៅជាចំនួនគត់
print("Portions: " + str(30))   # str() ប្ដូរលេខទៅជា text
````

### 5 · f-string: text ដែលមាន value នៅខាងក្នុង

ដាក់ `f` នៅមុខសញ្ញាសម្រង់ ហើយសរសេរឈ្មោះនៅក្នុង `{ }`។ Python នឹងបំពេញ value ឱ្យ។
