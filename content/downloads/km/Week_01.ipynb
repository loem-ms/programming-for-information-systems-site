## មុនពេលចាប់ផ្ដើម

1. **រក្សាទុកច្បាប់ចម្លងរបស់អ្នកផ្ទាល់ជាមុនសិន។** នៅក្នុង Colab ចុចលើពាក្យ **Copy to Drive** (ឬ ចូលទៅកាន់ **File** រួចចុច **Save a copy in Drive**)។ ចូរធ្វើការក្នុងច្បាប់ចម្លងនោះ មិនមែនក្នុងច្បាប់ដើមនេះទេ។
2. **Run setup cell ខាងក្រោម** (ចុច ▶)។ Run វាម្ដងទៀត រាល់ពេលដែល Colab ប្រាប់ថា runtime បានចាប់ផ្ដើមឡើងវិញ (restart)។
3. ធ្វើការតាមលំដាប់ពីលើចុះក្រោម។ ផ្នែកនីមួយៗប្រាប់ថាគួរប្រើពេលប៉ុន្មាន។

កម្រិត: **⭐ មូលដ្ឋាន** សម្រាប់គ្រប់គ្នា។ **⭐⭐ បន្ថែម** គឺស្រេចចិត្ត សម្រាប់ពេលអ្នកធ្វើរួចមុនម៉ោង។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាពន្លា (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.8"
_KIT_SHA = "41dafebb7ed52e8114e8bc10c2d9edb023f9aed5e823f1af14589289780e44f5"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

### ✍️ សាកសរសេរទី 2 · តាមដានកូដលើក្រដាសជាមុនសិន

**ការតាមដានកូដ (trace)** មានន័យថា អានកូដម្ដងមួយបន្ទាត់ ដូចដែលកុំព្យូទ័រធ្វើ ហើយសរសេរតម្លៃនីមួយៗទុក។ ចូរអាន cell នេះ ប៉ុន្តែ **កុំទាន់ Run វា**។ នៅលើក្រដាស ចូររកតម្លៃចុងក្រោយរបស់ `card_balance_riel` និងរបស់ `total_riel`។ សរសេរចម្លើយទាំងពីររបស់អ្នក ក្នុងបន្ទាត់ពីរដំបូង រួច Run។

````python
my_card_balance_riel = 0     # TODO: ចម្លើយរបស់អ្នកសម្រាប់ card_balance_riel
my_total_riel = 0            # TODO: ចម្លើយរបស់អ្នកសម្រាប់ total_riel

card_balance_riel = 10000                      # កាតអាហារដ្ឋានរបស់និស្សិត
card_balance_riel = card_balance_riel - 4000   # អាហារថ្ងៃត្រង់
card_balance_riel = card_balance_riel + 2000   # បញ្ចូលលុយបន្ថែម

fare_riel = 1500                               # សំបុត្រឡានក្រុងមួយ
trips = 2
total_riel = fare_riel * trips
fare_riel = 2000                               # ថ្លៃឡើង ក្រោយពេលអ្នកបានបង់រួច

print("✅ card_balance_riel" if my_card_balance_riel == card_balance_riel else f"❌ card_balance_riel ends at {card_balance_riel}: go line by line, right side first")
print("✅ total_riel" if my_total_riel == total_riel else f"❌ total_riel is {total_riel}: it was worked out before fare_riel changed")
````

### 3 · ការគណនា (arithmetic): តើចម្លើយណាមានប្រយោជន៍?

| សញ្ញាប្រមាណវិធី (operator) | អត្ថន័យ | `30 ? 4` |
|---|---|---|
| `+` `-` `*` | បូក ដក គុណ | `34`, `26`, `120` |
| `/` | ចែក (ផ្ដល់ float ជានិច្ច) | `7.5` |
| `//` | ចែកយកតែចំនួនគត់ | `7` |
| `%` | សំណល់ (អ្វីដែលនៅសល់) | `2` |

````python
portions = 30
volunteers = 4
print(portions / volunteers)
print(portions // volunteers)
print(portions % volunteers)
````

### ✍️ សាកសរសេរទី 3 · ប្រអប់ spring rolls

អាហារដ្ឋានវេចខ្ចប់ spring rolls ចំនួន 50 ដាក់ក្នុងប្រអប់ ប្រអប់មួយដាក់បាន 6។ តើមានប្រអប់ **ពេញ** ប៉ុន្មាន ហើយនៅសល់ spring rolls ប៉ុន្មាន? ចូរគណនាដោយប្រើឈ្មោះ ជាមួយ `//` និង `%` មិនមែនដោយប្រើលេខដែលអ្នកបានគិតរួចក្នុងចិត្តទេ។

````python
spring_rolls = 50
rolls_per_box = 6

full_boxes = 0     # TODO: ប្រើ //
rolls_left = 0     # TODO: ប្រើ %

print(f"{full_boxes} full boxes, {rolls_left} rolls left over")
print("✅ full_boxes" if full_boxes == spring_rolls // rolls_per_box else "❌ full_boxes: how many whole boxes? (//)")
print("✅ rolls_left" if rolls_left == spring_rolls % rolls_per_box else "❌ rolls_left: what is left after the full boxes? (%)")
````
