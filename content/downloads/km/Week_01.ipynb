## មុនពេលចាប់ផ្ដើម

1. **រក្សាទុកច្បាប់ចម្លងរបស់អ្នកផ្ទាល់ជាមុនសិន។** នៅក្នុង Colab សូមចុចលើពាក្យ Copy to Drive (ឬ ចូលទៅកាន់ File រួចចុច Save a copy in Drive)។ ធ្វើការក្នុងច្បាប់ចម្លងនោះ មិនមែនក្នុងច្បាប់ដើមនេះទេ។
2. **Run setup cell ខាងក្រោម** (ចុច ▶)។ ប្រសិនបើ Colab ប្រាប់ថា runtime បាន restart សូម Run វាម្ដងទៀត។
3. សូមធ្វើការតាមលំដាប់ពីលើចុះក្រោម។ ផ្នែកនីមួយៗប្រាប់ថាគួរប្រើពេលប៉ុន្មាន។

កម្រិត: **⭐ មូលដ្ឋាន** សម្រាប់គ្រប់គ្នា។ **⭐⭐ បន្ថែម** គឺស្រេចចិត្ត សម្រាប់ពេលអ្នកធ្វើរួចមុនម៉ោង។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.7"
_KIT_SHA = "f2b097068a1ca7d5251963015748526b066b60e0498b0930830bfcb445c723af"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

## វគ្គទី 1 · រៀន: តម្លៃ ឈ្មោះ និងអត្ថបទ (30 នាទី)

### 2 · ឈ្មោះ (variable)

**ឈ្មោះ** មួយចង្អុលទៅតម្លៃមួយ។ សញ្ញា `=` មានន័យថា *"ឱ្យឈ្មោះនេះចង្អុលទៅតម្លៃនេះ"*។ វាមិនមែនជាសញ្ញា `=` (ស្មើ) ក្នុងគណិតវិទ្យាទេ។

````python
portions = 30
hours_left = 5
food = "Bread rolls"
print(food, portions, hours_left)
````

🤔 **ទាយមើល** ថាតើ cell នេះនឹង print អ្វី រួចសូម Run វា។

````python
portions = 30
portions = portions + 12
print(portions)
````

**រំពឹងថា:** …  **លទ្ធផលជាក់ស្ដែង:** …  **ហេតុអ្វី:** …

យើងអាចដាក់ឈ្មោះឱ្យតម្លៃ និងឱ្យឈ្មោះមួយចង្អុលទៅតម្លៃថ្មី។ បញ្ហាបន្ទាប់: អ្នកសម្របសម្រួលត្រូវគណនាដោយប្រើតម្លៃទាំងនោះ — ចែកចំណែក និងរកមើលថានៅសល់ប៉ុន្មាន។

### 3 · ការគណនា (arithmetic)

| សញ្ញាប្រមាណវិធី (operator) | អត្ថន័យ | `30 ? 4` |
|---|---|---|
| `+` `-` `*` | បូក ដក គុណ | `34`, `26`, `120` |
| `/` | ចែក (ផ្ដល់ float ជានិច្ច) | `7.5` |
| `//` | ចែកយកតែចំនួនគត់ | `7` |
| `%` | សំណល់ | `2` |

````python
portions = 30
volunteers = 4
print(portions / volunteers)
print(portions // volunteers)
print(portions % volunteers)
````

🤔 អ្នកស្ម័គ្រចិត្តមិនអាចដឹកកន្លះចំណែកបានទេ។ ក្នុងចំណោមចម្លើយទាំងបី តើមួយណាមានប្រយោជន៍សម្រាប់អ្នកសម្របសម្រួល ហើយតើមួយណាអាចធ្វើឱ្យយល់ច្រឡំ? ចូរសរសេរចម្លើយមួយប្រយោគ។

**ចម្លើយរបស់ខ្ញុំ:** …

យើងអាចគណនាដោយប្រើលេខ។ បញ្ហាបន្ទាប់: សារតាមទូរស័ព្ទមកដល់ជាអត្ថបទ (text) ហើយអត្ថបទ និងលេខ មិនព្រមលាយបញ្ចូលគ្នាទេ។

### 4 · អត្ថបទ និងលេខ មិនលាយបញ្ចូលគ្នាដោយអចេតនាទេ

សញ្ញា `+` បូកលេខ ប៉ុន្តែវា **តភ្ជាប់** អត្ថបទ។ Python មិនព្រមតភ្ជាប់អត្ថបទជាមួយលេខទេ ព្រោះវាមិនដឹងថាអ្នកចង់បានអ្វី។

````python
print(3 + 4)
print("3" + "4")
print(int("30") + 4)      # int() ប្ដូរអត្ថបទ ទៅជាចំនួនគត់
print("Portions: " + str(30))   # str() ប្ដូរលេខ ទៅជាអត្ថបទ
````

យើងអាចប្ដូរលេខទៅជាអត្ថបទ ដោយប្រើ `str()`។ បញ្ហាបន្ទាប់: ការតភ្ជាប់តម្លៃច្រើនតាមរបៀបនោះ ធ្វើឱ្យបន្ទាត់មួយវែង និងពិបាកអាន។

### 5 · f-string: អត្ថបទដែលមានតម្លៃនៅខាងក្នុង

ចូរដាក់អក្សរ `f` នៅមុខសញ្ញាសម្រង់ ហើយសរសេរឈ្មោះនៅក្នុង `{ }`។ Python នឹងបំពេញតម្លៃទាំងនោះឱ្យអ្នក។

````python
food = "Bread rolls"
portions = 30
hours_left = 5
print(f"{food}: {portions} portions, {hours_left} hours left")
````

យើងអាចបង្កើតបន្ទាត់ច្បាស់លាស់មួយ ដោយប្រើ f-string។ បញ្ហាបន្ទាប់: cell ដំណើរការតាមលំដាប់ ហើយនៅពេលមានអ្វីខុសឆ្គង យើងត្រូវអានសារ error។

### 6 · លំដាប់គឺសំខាន់ ហើយសារ error អាចជួយអ្នក

Python ដំណើរការ cell មួយពីលើចុះក្រោម។ នៅក្នុង notebook វាក៏ចងចាំឈ្មោះពី cell ដែលអ្នកបាន Run ពីមុនដែរ **តាមលំដាប់ដែលអ្នកបាន Run វា**។ ប្រសិនបើមានអ្វីមើលទៅចម្លែក សូមប្រើ **Runtime → Restart session and run all** (Colab) ឬ **Kernel → Restart & Run All** (Jupyter)។

នៅពេលដែល Python មិនអាចបន្តដំណើរការបាន វានឹងបង្ហាញ **បញ្ហា (error)** មួយ។ សូមអាន **បន្ទាត់ចុងក្រោយ** មុនគេ ព្រោះវាប្រាប់ថាបញ្ហានោះជាអ្វី។

- លុបសញ្ញា `#` នៅដើមបន្ទាត់ទីពីរ ហើយ Run cell នោះ។
- អានសារ error។
- បន្ទាប់មក ដាក់សញ្ញា `#` នោះវិញ ដើម្បីឱ្យ *Run all* នៅតែដំណើរការបាន។

មុនពេលអ្នកបង្កើត ចូរមើលលំនាំនេះម្ដងសិន លើសំណួរតូចជាង។ អ្នកសម្របសម្រួលសួរថា: *«តើអ្វីកំពុងរង់ចាំការទៅយក មានប៉ុន្មាន ហើយត្រូវទៅយកឱ្យបានឆាប់ប៉ុណ្ណា — ក្នុងមួយបន្ទាត់ខ្លី?»*

អ្នកអាចដាក់ឈ្មោះឱ្យតម្លៃ គណនាដោយប្រើវា និងបង្កើតអត្ថបទ។ បញ្ហាបន្ទាប់: ការប្រើឧបករណ៍ទាំងនោះលើលេខថ្មី — និងការវាយតម្លៃកូដរបស់អ្នកដទៃ ដែលប្រើវាមិនបានល្អ។

### ពិនិត្យបន្ទាត់របស់អ្នក

ការពិនិត្យ (check) គឺជាកូដតូចមួយ ដែលសាកល្បង (test) ការសន្យាមួយ។ សូម Run វា។ សញ្ញា ❌ នីមួយៗប្រាប់អ្នកថាខ្វះអ្វី។ អ្នកមិនទាន់ចាំបាច់យល់កូដរបស់ការពិនិត្យនេះនៅឡើយទេ ហើយអ្នកក៏មិនចាំបាច់សរសេរបន្ទាត់បែបនេះដែរ។

````python
checks = {
    "has the ID": offer_id in summary,
    "has the food": food in summary,
    "portions with the unit": f"{portions} portions" in summary,
    "hours with the unit": f"{hours_left} hours" in summary,
}
for promise, kept in checks.items():
    print("✅" if kept else "❌", promise)
````
