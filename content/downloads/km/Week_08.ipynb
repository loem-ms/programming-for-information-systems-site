## មុនពេលចាប់ផ្ដើម

1. ចូលទៅកាន់ **File → Save a copy in Drive** រួចធ្វើការក្នុងច្បាប់ចម្លងរបស់អ្នក។
2. ចូរ Run setup cell។ នៅក្នុង Colab វានឹងសុំភ្ជាប់ Google Drive៖ សូមយល់ព្រម (yes) ដើម្បីឱ្យ folder `my_foodbridge/` របស់អ្នក នៅរក្សាទុកពីសប្ដាហ៍មួយទៅសប្ដាហ៍មួយ។
3. ⭐ = សម្រាប់គ្រប់គ្នា · ⭐⭐ = សម្រាប់សាកល្បងបន្ថែម។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាពន្លា (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.8"
_KIT_SHA = "41dafebb7ed52e8114e8bc10c2d9edb023f9aed5e823f1af14589289780e44f5"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
bridge.folder.connect()          # រក្សាទុក folder my_foodbridge/ របស់អ្នក (ក្នុង Google Drive នៅក្នុង Colab)
````

### 1 · `sorted` បង្កើត list ថ្មីមួយ ហើយ slice យក item ពីរបីខាងដើម

📖 **វាជាអ្វី?** `sorted(items)` បញ្ជូន list **ថ្មី** មួយមកវិញ (return) ដែលមាន item ដដែល តែរៀបតាមលំដាប់ ពីតូចបំផុតទៅធំ។ `reverse=True` ដាក់តម្លៃធំបំផុតនៅមុនគេ។ **slice** មួយ យកផ្នែកមួយនៃ list៖ `items[start:end]` បង្កើត list ថ្មីមួយ ចាប់ពីទីតាំង `start` រហូតដល់ទីតាំង `end` ប៉ុន្តែ **មិនរាប់បញ្ចូល** ទីតាំង `end` ទេ។ ប្រសិនបើអ្នកមិនសរសេរ `start` វាចាប់ផ្ដើមពីដើម list ហើយប្រសិនបើអ្នកមិនសរសេរ `end` វាទៅដល់ចុង list។ ទីតាំងអវិជ្ជមាន រាប់ពីខាងចុងមកវិញ៖ `[-1]` គឺជា item ចុងក្រោយ។

**ហេតុអ្វីបានជាត្រូវការវា?** អេក្រង់មានទំហំតូច ហើយមនុស្សអានពីលើចុះក្រោម។ របាយការណ៍ត្រូវការលំដាប់ (បន្ទាន់បំផុតនៅមុនគេ) និង item តែពីរបី (បីខាងលើគេ ឬប្រាំដែលថ្មីបំផុត)។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `sorted` ចម្លង item ទាំងអស់ទៅក្នុង list ថ្មីមួយ ហើយរៀបច្បាប់ចម្លងនោះតាមលំដាប់។ list ដើមរបស់អ្នកមិនប្រែប្រួលទេ (ក្នុងឧទាហរណ៍ខាងក្រោម `marks` នៅដដែល)។ slice ក៏បង្កើត list ថ្មីមួយដែរ។ ប្រសិនបើ list ខ្លីជាងអ្វីដែល slice សុំ អ្នកគ្រាន់តែទទួលបានអ្វីដែលមាន៖ គ្មាន error ទេ។

```text
  marks     = [ 72, 45, 90, 61 ]
  position:     0   1   2   3
  from end:    -4  -3  -2  -1

  marks[:3]  → [72, 45, 90]    start at 0, stop before 3
  marks[-2:] → [90, 61]        start 2 from the end, go to the end
```

| កូដ | លទ្ធផល: list ថ្មីមួយ | `marks` ក្រោយមក |
|---|---|---|
| `sorted(marks)` | `[45, 61, 72, 90]` | នៅតែ `[72, 45, 90, 61]` |
| `sorted(marks, reverse=True)` | `[90, 72, 61, 45]` | មិនប្រែប្រួល |
| `sorted(marks, reverse=True)[:3]` | `[90, 72, 61]`: បីខាងលើគេ | មិនប្រែប្រួល |

ប្រវត្តិមួយ រៀបតាមលំដាប់ពេលវេលារួចហើយ ដោយចាស់បំផុតនៅមុនគេ (ព្រឹត្តិការណ៍ (event) សប្ដាហ៍ទី 7 របស់អ្នក) ដូច្នេះ `events[-5:]` យក **ប្រាំដែលថ្មីបំផុត** ដោយមិនចាំបាច់តម្រៀបអ្វីសោះ។

**តើអ្នកជួបវានៅកន្លែងណា?** «បទចម្រៀងពេញនិយម 10 បទប្រចាំសប្ដាហ៍» សារថ្មីបំផុតប្រាំក្នុង app ជជែក (chat) និង «តម្លៃ: ពីទាបទៅខ្ពស់» ក្នុង app ទិញទំនិញ។

🤔 **ទាយមើល** បន្ទាត់នីមួយៗដែលនឹងត្រូវ print មុនពេលអ្នក Run cell ជាពិសេសបន្ទាត់ពីរចុងក្រោយ។

````python
marks = [72, 45, 90, 61]
ranked = sorted(marks, reverse=True)
print(ranked)
print(ranked[:3])          # បីខាងលើគេ
print(marks[-2:])          # ពីរចុងក្រោយ តាមលំដាប់ដើម
print(marks[:10])          # 🤔 មានពិន្ទុតែបួន: តើមានអ្វីកើតឡើង?
print(marks)               # 🤔 តើ sorted បានប្ដូរ marks ដែរឬទេ?
````

### 2 · key function ជ្រើសរើសអ្វីដែលត្រូវប្រៀបធៀប

📖 **វាជាអ្វី?** **key function** ទទួល item មួយ ហើយ return តម្លៃដែលត្រូវប្រើសម្រាប់តម្រៀប។ អ្នកផ្ដល់វាឱ្យ `sorted`៖ `sorted(records, key=hours_left_of)`។

**ហេតុអ្វីបានជាត្រូវការវា?** Python អាចប្រៀបធៀបលេខ និង text បាន ប៉ុន្តែមិនអាចប្រៀបធៀប dictionary ពីរបានទេ៖ រវាងនំបុ័ង និងបាយ តើ record មួយណា "តូចជាង"? ប្រសិនបើអ្នកហៅ `sorted(records)` លើ record គំរូ Python ឈប់ ដោយបង្ហាញ `TypeError: '<' not supported between instances of 'dict' and 'dict'` (`<` មានន័យថា "តូចជាង")។ អ្នកត្រូវប្រាប់ Python ថា ត្រូវប្រៀបធៀប **អ្វី**។

**តើ Python ដំណើរការវាយ៉ាងដូចម្ដេច?** `sorted` ហៅ key function របស់អ្នក **ម្ដង សម្រាប់ record នីមួយៗ** ហើយកត់ចំណាំតម្លៃដែលវា return (តម្លៃនេះហៅថា key)។ វាតម្រៀបតាម key ទាំងនោះ រួចដាក់ **record ទាំងមូល** ទៅក្នុង list ថ្មីមួយ តាមលំដាប់នោះ។ ប្រសិនបើ record ពីរមាន key ដូចគ្នា `sorted` រក្សាលំដាប់ដែលវាចូលមក។ record ទាំងនោះមិនប្រែប្រួលទេ ហើយ list ដែលអ្នកផ្ដល់ឱ្យវា ក៏មិនប្រែប្រួលដែរ។

### 4 · ការបូកបរិមាណចូលទៅក្នុង dictionary: រាប់ (count) ឬបូក (sum)?

````python
# @title Last semester's donors: offers (a count) and portions offered (a sum) { display-mode: "form" }
# cell ដែលរៀបចំរួច: វាអាន semester.csv ហើយ print មួយបន្ទាត់សម្រាប់អ្នកបរិច្ចាគម្នាក់ៗ ដោយចំណែកច្រើនបំផុតនៅមុនគេ។ គ្រាន់តែ Run វា។
import csv

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    semester = list(csv.DictReader(file))
offers_by_donor = {}
portions_by_donor = {}
for row in semester:
    donor = row["donor"]
    offers_by_donor[donor] = offers_by_donor.get(donor, 0) + 1
    portions_by_donor[donor] = portions_by_donor.get(donor, 0) + int(row["portions"])


def portions_of_donor(donor):
    return portions_by_donor[donor]


print("Donor                     offers  portions  per offer")
for donor in sorted(portions_by_donor, key=portions_of_donor, reverse=True):    # dictionary ផ្ដល់ key របស់វា: គឺអ្នកបរិច្ចាគ
    offers = offers_by_donor[donor]
    portions = portions_by_donor[donor]
    print(f"{donor:<24} {offers:>7} {portions:>9} {round(portions / offers):>10}")
print(len(semester), "offers in all")
````

```python
def by_food_name(record):
    return record["food"]

def dashboard(records):
    offered = 0
    picked_up = 0
    delivered = 0
    for record in records:
        offered += record["portions"]
        if record["status"] == "picked_up":
            picked_up += record["portions"]
        if record["status"] == "delivered":
            delivered += record["portions"]
    lines = [f"Meals served: {offered + picked_up + delivered}"]
    for record in sorted(records, key=by_food_name):
        lines.append(f"{record['food']:<26} {record['portions']:>4}  {record['contact']}")
    return "\n".join(lines)
```

ជាដំបូង ចូរអានក្ដារព័ត៌មានសាធារណៈ ដូចជាមនុស្សមិនស្គាល់ម្នាក់នៅក្នុងអាហារដ្ឋាន៖ តើអ្នកអាចដឹងអ្វីខ្លះពីវា ហើយតើអ្វីខ្លះដែលអ្នក **មិន** អាចដឹង? អ្នកណាក៏អាចអានវាបាន។ field បន្ថែមនីមួយៗ ជួយមនុស្សម្នាក់ ប៉ុន្តែអាចប៉ះពាល់មនុស្សម្នាក់ទៀត។ ចូរចាប់ផ្ដើមពីជួរឈរ ក្ដារព័ត៌មានសាធារណៈ នៃតារាងសិទ្ធិ 🔐 របស់អ្នក។

field មួយអាចបង្ហាញថាជានរណា សូម្បីតែនៅពេលឈ្មោះត្រូវបានលាក់ក៏ដោយ។ cell ដែលរៀបចំរួចខាងក្រោម អានការផ្ដល់ជូនពីឆមាសមុន ហើយរាយឈ្មោះអ្នកបរិច្ចាគ ដែលនៅពីក្រោយប្រភេទ (category) នីមួយៗ។ ចូរ Run វា ហើយអានលទ្ធផល។ អ្នកមិនចាំបាច់សរសេរអ្វីទេ។

````python
# @title Last semester: which donors gave each category? { display-mode: "form" }
# cell ដែលរៀបចំរួច: វាអាន semester.csv ហើយរាប់ការផ្ដល់ជូនរបស់អ្នកបរិច្ចាគម្នាក់ៗ ក្នុងប្រភេទនីមួយៗ។ គ្រាន់តែ Run វា។
import csv

with open("foodbridge_kit/data/semester.csv", encoding="utf-8") as file:
    semester = list(csv.DictReader(file))
donors_by_category = {}
for row in semester:
    if row["category"] not in donors_by_category:
        donors_by_category[row["category"]] = {}
    donors = donors_by_category[row["category"]]
    donors[row["donor"]] = donors.get(row["donor"], 0) + 1

print("Category      Donors who gave it (number of offers), from", len(semester), "offers")
for category in donors_by_category:                   # dictionary ផ្ដល់ key របស់វា: គឺប្រភេទនីមួយៗ
    donors = donors_by_category[category]
    parts = []
    for donor in donors:
        parts.append(donor + " " + str(donors[donor]))
    line = f"{category:<13} " + " · ".join(parts)
    if len(donors) == 1:
        line = line + "   ← only one donor"
    print(line)
````

### ⭐⭐ សាកល្បងបន្ថែម · ក្ដារព័ត៌មានសាធារណៈរបស់អ្នក ជាទំព័រវេប

`bridge.show` គូរដោយប្រើកូដរបស់ kit។ នៅទីនេះ អ្នកសរសេរអេក្រង់មួយដោយខ្លួនឯង៖ ទំព័រវេប (web page) `my_foodbridge/public_board.html` ដែលបង្កើតដោយ Python ធម្មតា។ វាគ្រាន់តែជា file មួយប៉ុណ្ណោះ (គ្មាន server គ្មាន internet)៖ ចូរបើកវាក្នុង browser ណាក៏បាន ឬផ្ញើវាតាម Telegram។

ទំព័រវេបមួយ គឺជា text ដែលមាន **tag**៖ `<table>` តារាង · `<tr>` ជួរដេក (row) · `<th>` ចំណងជើងជួរឈរ · `<td>` ប្រអប់មួយ (cell)។ browser អាន `<` ថាជាការចាប់ផ្ដើមនៃ tag មួយ ដូច្នេះឈ្មោះអាហារដែលគេវាយថា `Rice <b>FREE</b>` នឹងធ្វើឱ្យទំព័ររបស់អ្នកប្រែប្រួល។ `html.escape` ប្ដូរ `<`, `>` និង `&` ទៅជាកូដដែលមានសុវត្ថិភាព៖ `html.escape("Rice <b>FREE</b> & soup")` ផ្ដល់ `Rice &lt;b&gt;FREE&lt;/b&gt; &amp; soup` ដែល browser បង្ហាញជា text ធម្មតា។

ចូរចម្លងបន្ទាត់ `PUBLIC_FIELDS` ពី `report.py` របស់អ្នក មកដាក់ក្នុង cell នេះ បំពេញ `TODO` ទាំងពីរ រួច Run វា។ (`not in` គឺផ្ទុយពី `in`។) បន្ទាត់ `assert` គឺជា **តេស្ត** មួយ៖ ប្រសិនបើព័ត៌មានទំនាក់ទំនងណាមួយ ទៅដល់ទំព័រ វានឹងបញ្ឈប់ cell **មុនពេល** file ត្រូវបានរក្សាទុក។ បន្ទាប់មក ចូរទាញយក (download) file នោះ (ផ្ទាំង **Files** → `my_foodbridge` → ចុចកណ្ដុរស្ដាំលើ `public_board.html` → **Download**) ហើយបើកវា។

````python
import html

PUBLIC_FIELDS = ["id", "food", "category", "portions", "status"]     # ✍️ ចម្លងបន្ទាត់នេះពី report.py របស់អ្នក


def hours_left_of(record):
    return record["hours_left"]


header = ""
for field in PUBLIC_FIELDS:
    header = header + "<th>" + field + "</th>"
rows = ""
for record in sorted(bridge.sample_records(), key=hours_left_of):
    if record["status"] not in ["rejected", "cancelled"]:
        cells = ""
        for field in PUBLIC_FIELDS:
            value = str(record.get(field, ""))       # TODO 1: ធ្វើឱ្យតម្លៃមានសុវត្ថិភាព: html.escape(...)
            cells = cells + "<td>" + value + "</td>"
        rows = rows + ""                             # TODO 2: បន្ថែមជួរដេកមួយ: "<tr>" + cells + "</tr>\n"

page = ("<meta charset='utf-8'>\n<h3>Campus FoodBridge · Public board</h3>\n"
        "<table border='1'>\n<tr>" + header + "</tr>\n" + rows + "</table>\n")
assert "@" not in page, "A contact is on the public page: check PUBLIC_FIELDS."
with open("my_foodbridge/public_board.html", "w", encoding="utf-8") as file:
    file.write(page)
print(page)
print("Saved my_foodbridge/public_board.html")
````

ប្រសិនបើអ្នកជ្រើសរើស *ទំព័រសាធារណៈពីសប្ដាហ៍ទី 8* ជាផ្នែកបន្ថែមសម្រាប់ Final project របស់អ្នក ទំព័រនេះនឹងរីកធំក្លាយជា `make_page(records)` ក្នុង `my_foodbridge/screen.py`៖ វា return ទំព័រជា text (`str`) ប្រើ `public_row` ពី `report.py` របស់អ្នក (ដោយ import វា ដូចដែលអ្នកនឹងរៀននៅសប្ដាហ៍ទី 9 មិនមែនចម្លងទេ) និង `html.escape` ហើយមិនបង្ហាញការផ្ដល់ជូនដែលត្រូវបានបដិសេធ ឬលុបចោលទេ។ បន្ទាប់មក `bridge.show("screen")` បង្ហាញវា។
