2. Run setup cell។ Run វាម្ដងទៀត ក្រោយពេល runtime restart។
3. ⭐ = មូលដ្ឋាន សម្រាប់គ្រប់គ្នា, ⭐⭐ = បន្ថែម (ស្រេចចិត្ត)។

````python
# @title ▶ សូម run cell នេះមុនគេ (setup) — វារៀបចំប្រព័ន្ធ FoodBridge
# វាដកយក (unpack) ប្រព័ន្ធរបស់វគ្គសិក្សា ដែលរក្សាទុកនៅក្នុង notebook នេះ។
# មិនត្រូវការ internet ឬ install អ្វីទេ។ សូម run វាម្ដងទៀត ក្រោយពេល Colab restart។
import base64, hashlib, importlib, io, os, sys, zipfile
_KIT_VERSION = "2026.11.3"
_KIT_SHA = "9a7f0e70ab94bb1480312c1ccc2cd08ea21038ab8fa51cc0adbfe520961d0844"
_KIT_ZIP = ""

def _prepare_kit():
    data = base64.b64decode(_KIT_ZIP)
    if hashlib.sha256(data).hexdigest() != _KIT_SHA:
        raise RuntimeError("ឯកសាររបស់វគ្គសិក្សាក្នុង notebook នេះខូច។ សូមទាញយកច្បាប់ថ្មីមួយ។")
    marker = os.path.join("foodbridge_kit", ".kit_version")
    current = open(marker).read().strip() if os.path.exists(marker) else ""
    keep = ("foodbridge_kit/my_settings.py", "foodbridge_kit/my_rules.py")
    with zipfile.ZipFile(io.BytesIO(data)) as archive:
        for name in archive.namelist():
            if current == _KIT_VERSION and os.path.exists(name):
                continue                      # already up to date
            if name in keep and os.path.exists(name):
                continue                      # never overwrite your own files
            os.makedirs(os.path.dirname(name), exist_ok=True)
            with open(name, "wb") as target:
                target.write(archive.read(name))
    with open(marker, "w") as file:
        file.write(_KIT_VERSION)
    folder = os.path.abspath("foodbridge_kit")
    if folder not in sys.path:
        sys.path.insert(0, folder)
    for module in [m for m in sys.modules if m == "foodbridge" or m.startswith("foodbridge.")]:
        del sys.modules[module]             # load the fresh copy

_prepare_kit()
import foodbridge as bridge
bridge.set_language("km")     # screens and checks in Khmer
print("✅ ប្រព័ន្ធ FoodBridge រួចរាល់ហើយ (version " + bridge.KIT_VERSION + ")។ អ្នកអាចបន្តបាន។")
````

### 1 · `sorted()` បង្កើត list ថ្មីមួយតាមលំដាប់

### 2 · ការ sort records ដោយប្រើ key function ដែលមានឈ្មោះ

Python មិនអាចទាយថាត្រូវតម្រៀប dictionary យ៉ាងដូចម្ដេចទេ: តាមឈ្មោះ? តាមពិន្ទុ? អ្នកប្រាប់វាដោយ **key function** មួយ។ វាទទួល item មួយ ហើយ return value ដែលត្រូវប្រើដើម្បី sort។

````python
students = [{"name": "Sokha", "mark": 72},
            {"name": "Dara", "mark": 90},
            {"name": "Vanna", "mark": 61}]

def by_mark(student):
    return student["mark"]

for student in sorted(students, key=by_mark):
    print(student["name"], student["mark"])
````

សរសេរ `key=by_mark` **ដោយគ្មាន** `( )`។ អ្នកផ្ដល់ function ខ្លួនវាផ្ទាល់ទៅ `sorted` ហើយ `sorted` ហៅវាម្ដង សម្រាប់និស្សិតម្នាក់ៗ។ ឈ្មោះល្អប្រាប់ពីលំដាប់: `by_mark`, `most_urgent_first`។

### 3 · ការជ្រើស field ពី dictionary

ទិដ្ឋភាពសាធារណៈ គឺជា **ច្បាប់ចម្លងតូចជាង** នៃ record មួយ។ វិធីមានសុវត្ថិភាពក្នុងការបង្កើតវា គឺចាប់ផ្ដើមពីទទេ ហើយចម្លងតែ field ដែលមាននៅក្នុង **បញ្ជីអនុញ្ញាត (allowed list)** ប៉ុណ្ណោះ។ (លេខទូរស័ព្ទនេះប្រឌិតឡើង។)

🤔 **ទាយមើល:** បន្ទាត់ពីរចុងក្រោយបន្ថែម field ឯកជនថ្មីមួយ។ តើក្ដារព័ត៌មាន (notice board) នឹងបង្ហាញវាទេ?

````python
student = {"name": "Dara", "phone": "012 000 111", "mark": 90, "class": "MIS1"}
NOTICE_BOARD_FIELDS = ["name", "class"]

def notice_board_row(student):
    row = {}
    for field in NOTICE_BOARD_FIELDS:
        row[field] = student[field]
    return row

print(notice_board_row(student))

student["home_address"] = "Street 271"      # នរណាម្នាក់បន្ថែម field ឯកជនថ្មីមួយ នៅពេលក្រោយ
print(notice_board_row(student))
````

### 5 · Menu function ដែល return text

Menu មានការងារពីរ: **សួរ** មនុស្សថាគេចង់បានអ្វី និង **សម្រេច** ថាត្រូវបង្ហាញអ្វី។ ទុកការងារទាំងពីរនេះឱ្យដាច់ពីគ្នា។ ផ្នែកសម្រេច គឺជា function ធម្មតាមួយ ដែល **return** text។ ដូច្នេះអ្នកអាច test វាបានដោយមិនចាំបាច់វាយអក្សរ ហើយ function ដដែលអាចបម្រើ text menu, ការពិនិត្យ ឬទំព័រ web មួយ។

````python
def library_answer(choice):
    if choice == "1":
        return "Open 8:00-17:00, Monday to Saturday."
    if choice == "2":
        return "You can borrow 3 books for 14 days."
    if choice == "q":
        return "Goodbye!"
    return "Unknown choice " + repr(choice) + ". Please type 1, 2 or q."

print(library_answer("1"))
print(library_answer("7"))
````

`repr(choice)` បង្ហាញ text ក្នុងសញ្ញាសម្រង់ ដូច្នេះសូម្បីតែចម្លើយទទេ ឬដកឃ្លាមួយ ក៏អាចមើលឃើញដែរ។
