</details>


```python
def new_db():
    db = sqlite3.connect(":memory:")
    db.execute("PRAGMA foreign_keys=ON")
    db.executescript("""
        CREATE TABLE batches(
            batch_id TEXT PRIMARY KEY, description TEXT NOT NULL,
            ppc INTEGER NOT NULL CHECK(ppc > 0),
            opening INTEGER NOT NULL CHECK(opening >= 0),
            available INTEGER NOT NULL CHECK(available >= 0));
        CREATE TABLE reservations(
            request_id TEXT PRIMARY KEY, label TEXT NOT NULL,
            planned_portions INTEGER NOT NULL CHECK(planned_portions > 0));
        CREATE TABLE lines(
            request_id TEXT REFERENCES reservations(request_id),
            line_no INTEGER, batch_id TEXT REFERENCES batches(batch_id),
            containers INTEGER NOT NULL CHECK(containers > 0),
            ppc_snapshot INTEGER NOT NULL CHECK(ppc_snapshot > 0),
            PRIMARY KEY(request_id, line_no));
        CREATE TABLE events(
            request_id TEXT REFERENCES reservations(request_id),
            event_type TEXT CHECK(event_type IN ('collected', 'delivered')),
            PRIMARY KEY(request_id, event_type));
    """)
    with db:
        db.executemany("INSERT INTO batches VALUES (?, ?, ?, ?, ?)",
                       [(b, name, ppc, opening, opening)
                        for b, name, ppc, opening in BATCHES])
    return db

def snapshot(db):
    return tuple(tuple(db.execute(query).fetchall()) for query in (
        "SELECT * FROM batches ORDER BY batch_id",
        "SELECT * FROM reservations ORDER BY request_id",
        "SELECT * FROM lines ORDER BY request_id, line_no",
        "SELECT * FROM events ORDER BY request_id, event_type",
    ))

print("Setup functions ready / អនុគមន៍រៀបចំរួចរាល់")
```

```python
def checkpoint_stock(db):
    offered, available = db.execute(
        "SELECT SUM(opening * ppc), SUM(available * ppc) FROM batches").fetchone()
    return {"offered": offered, "available": available}


def checkpoint_4(data):
    db = new_db()
    try:
        containers, portions = checkpoint_plan(data)
        notes = (
            "Policy label: " + pickup_label(portions, data["reviewed"]),
            "Stored batch containers: " + str([(row[0], row[4]) for row in snapshot(db)[0]]),
            "Reservation/event tables exist but their actions are still placeholders.",
            "Storage is temporary: this demonstration closes its own in-memory database.",
        )
        return show_checkpoint(4, data, checkpoint_stock(db), notes)
    finally:
        db.close()

checkpoint_4(checkpoint_request(bread=5, rice=2, reviewed=True))
```

```python
def reserve(db, request_id, reviewed, lines, inject_failure=False):
    if type(reviewed) is not bool or not reviewed:
        raise ValueError("Coordinator approval required / ត្រូវការការអនុម័តពីអ្នកសម្របសម្រួល")
    if not isinstance(request_id, str) or not request_id.strip():
        raise ValueError("Request ID required / ត្រូវការលេខសម្គាល់សំណើ")
    if not isinstance(lines, (list, tuple)) or not lines:
        raise ValueError("At least one line required / ត្រូវមានយ៉ាងតិចមួយបន្ទាត់")
    prepared = []
    for line in lines:
        if not isinstance(line, (list, tuple)) or len(line) != 2:
            raise ValueError("Use (batch_id, containers) / ប្រើ (batch_id, containers)")
        batch_id, count = line
        if not isinstance(batch_id, str) or type(count) is not int or count <= 0:
            raise ValueError("Use a batch ID and positive whole count / ប្រើលេខសម្គាល់ និងចំនួនគត់វិជ្ជមាន")
        batch = db.execute("SELECT ppc FROM batches WHERE batch_id=?", (batch_id,)).fetchone()
        if batch is None:
            raise ValueError("Unknown batch / ក្រុមអាហារមិនស្គាល់: " + batch_id)
        prepared.append((batch_id, count, batch[0]))
    total = sum(count * ppc for batch_id, count, ppc in prepared)
    label = pickup_label(total, reviewed)
    with db:
        db.execute("INSERT INTO reservations VALUES (?, ?, ?)", (request_id, label, total))
        for number, (batch_id, count, ppc) in enumerate(prepared, 1):
            change = db.execute(
                "UPDATE batches SET available=available-? WHERE batch_id=? AND available>=?",
                (count, batch_id, count))
            if change.rowcount != 1:
                raise ValueError("Insufficient containers / ឯកតាផ្ទុកមិនគ្រប់គ្រាន់: " + batch_id)
            db.execute("INSERT INTO lines VALUES (?, ?, ?, ?, ?)",
                       (request_id, number, batch_id, count, ppc))
            if inject_failure and number == 1:
                raise RuntimeError("Simulated interruption / ការផ្អាកសន្មត")
    return label


def show_atomic_example():
    db = new_db()
    try:
        reserve(db, *REQUESTS[0])
        reserve(db, *REQUESTS[2])
        before = snapshot(db)
        print("Before / មុន:", [(row[0], row[4]) for row in before[0]])
        try:
            reserve(db, *REQUESTS[3])
        except ValueError as error:
            print("Refused / បដិសេធ:", error)
        after = snapshot(db)
        assert after == before
        print("After / ក្រោយ:", [(row[0], row[4]) for row in after[0]])
        print("P04 headers / កំណត់ត្រាមេ:", db.execute(
            "SELECT * FROM reservations WHERE request_id='P04'").fetchall())
        print("P04 lines / បន្ទាត់:", db.execute(
            "SELECT * FROM lines WHERE request_id='P04'").fetchall())
    finally:
        db.close()

show_atomic_example()
```

```python
def record_event(db, request_id, event_type):
    if event_type not in ("collected", "delivered"):
        raise ValueError("Unknown event / ព្រឹត្តិការណ៍មិនស្គាល់")
    with db:
        if db.execute("SELECT 1 FROM reservations WHERE request_id=?", (request_id,)).fetchone() is None:
            raise ValueError("Accepted reservation required / ត្រូវមានការកក់ដែលទទួលយករួច")
        if event_type == "delivered" and db.execute(
            "SELECT 1 FROM events WHERE request_id=? AND event_type='collected'",
            (request_id,)).fetchone() is None:
            raise ValueError("Record collection first / កត់ត្រាការប្រមូលជាមុន")
        db.execute("INSERT INTO events VALUES (?, ?)", (request_id, event_type))


def show_event_example():
    db = new_db()
    try:
        reserve(db, *REQUESTS[0])
        before = snapshot(db)
        try:
            record_event(db, "P01", "delivered")
        except ValueError as error:
            print("Refused / បដិសេធ:", error)
        assert snapshot(db) == before
        for event in ("collected", "delivered"):
            record_event(db, "P01", event)
        assert snapshot(db)[0] == before[0]
        print("Recorded events / ព្រឹត្តិការណ៍ដែលបានកត់ត្រា:", snapshot(db)[3])
        print("Availability unchanged by events / ព្រឹត្តិការណ៍មិនប្តូរបរិមាណដែលអាចកក់បាន")
    finally:
        db.close()

show_event_example()
```

```python
def checkpoint_events(db, data, accepted):
    notes = []
    if not accepted:
        return ["Events skipped: there is no accepted reservation."]
    for event in data["events"]:
        before = snapshot(db)
        try:
            record_event(db, data["request_id"], event)
        except (ValueError, sqlite3.IntegrityError) as error:
            assert snapshot(db) == before
            notes.append("Event refused: " + str(error))
            break
        notes.append("Event recorded: " + event)
    return notes


def checkpoint_event_totals(db):
    metrics = checkpoint_reserved(db)
    for event in ("collected", "delivered"):
        metrics[event] = db.execute("""
            SELECT COALESCE(SUM(l.containers * l.ppc_snapshot), 0)
            FROM lines l JOIN events e ON l.request_id=e.request_id
            WHERE e.event_type=?""", (event,)).fetchone()[0]
    return metrics


def checkpoint_6(data):
    db = new_db()
    try:
        accepted, decision = checkpoint_reserve(db, data)
        notes = [decision] + checkpoint_events(db, data, accepted)
        notes.append("Recorded event rows: " + str(snapshot(db)[3]))
        notes.append("0 delivery means no delivery recorded, not proof that nothing happened.")
        return show_checkpoint(6, data, checkpoint_event_totals(db), notes)
    finally:
        db.close()

# Try events=("delivered",) to see the real ordering rule refuse it.
# Then try events=("collected", "delivered") to move the delivery total.
checkpoint_6(checkpoint_request(bread=5, rice=2, reviewed=True, events=("collected",)))
```

### ភស្តុតាងរបស់ខ្ញុំ 6 · ត្រូវបំពេញ

សរសេរសរុបដែលរំពឹងទុក និងសំណើរង់ចាំ រួចប្រៀបធៀបនឹងរបាយការណ៍។ បង្ហាញទាំង 94 = 34 + 60 និង 94 = 34 + 18 + 13 + 29 ដោយដាក់ស្លាកឱ្យមានន័យ។ ហេតុអ្វី P05 មិនត្រូវបង្ហាញក្នុងបញ្ជីរង់ចាំការប្រមូល?

**លទ្ធផលដែលបានឃើញ និងការពន្យល់៖** [សរសេរនៅទីនេះ]

```python
def request_rows(db):
    return db.execute("""
        SELECT r.request_id, r.label,
               SUM(l.containers * l.ppc_snapshot) AS planned_portions,
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='collected'),
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='delivered')
        FROM reservations r JOIN lines l ON l.request_id=r.request_id
        GROUP BY r.request_id, r.label ORDER BY r.request_id
    """).fetchall()


def report(db):
    rows = request_rows(db)
    offered, available = db.execute(
        "SELECT SUM(opening * ppc), SUM(available * ppc) FROM batches").fetchone()
    return {
        "offered": offered,
        "available": available,
        "reserved": sum(row[2] for row in rows),
        "collected": sum(row[2] for row in rows if row[3]),
        "delivered": sum(row[2] for row in rows if row[4]),
        "reserved_only": sum(row[2] for row in rows if not row[3]),
        "collected_not_delivered": sum(row[2] for row in rows if row[3] and not row[4]),
    }


def pending_collections(db):
    return [(request_id, portions, label)
            for request_id, label, portions, collected, delivered in request_rows(db)
            if not collected]


def show_report(db):
    labels = {
        "offered": "Opening offered / បានផ្តល់ដំបូង",
        "available": "Available / នៅអាចកក់បាន",
        "reserved": "Reserved / បានកក់",
        "collected": "Collection recorded / បានកត់ត្រាការប្រមូល",
        "delivered": "Delivery recorded / បានកត់ត្រាការប្រគល់",
        "reserved_only": "Reserved, no collection record / បានកក់ គ្មានកំណត់ត្រាប្រមូល",
        "collected_not_delivered": "Collected, no delivery record / បានប្រមូល គ្មានកំណត់ត្រាប្រគល់",
    }
    print("PLANNING PORTIONS / ចំណែកអាហារប៉ាន់ស្មាន")
    for key, value in report(db).items():
        print(labels[key] + ":", value)
    print("Available containers / ឯកតាផ្ទុកដែលអាចកក់បាន:", db.execute(
        "SELECT batch_id, available FROM batches ORDER BY batch_id").fetchall())
    print("Pending collection: ID, portions, label / រង់ចាំការប្រមូល៖ លេខសម្គាល់ ចំណែកអាហារ ស្លាក:",
          pending_collections(db))


def build_day(requests=REQUESTS, events=EVENTS):
    db = new_db()
    decisions = []
    try:
        for request_id, reviewed, lines in requests:
            before = snapshot(db)
            try:
                label = reserve(db, request_id, reviewed, lines)
            except (ValueError, sqlite3.IntegrityError) as error:
                assert snapshot(db) == before
                decisions.append((request_id, "held" if reviewed is False else "rejected", str(error)))
            else:
                decisions.append((request_id, "accepted", label))
        for request_id, event_type in events:
            record_event(db, request_id, event_type)
        return db, decisions
    except Exception:
        db.close()
        raise


def run_canonical_day():
    db, decisions = build_day()
    try:
        print("Request outcomes / លទ្ធផលសំណើ:")
        for row in decisions:
            print(row)
        show_report(db)
    finally:
        db.close()

run_canonical_day()
```

```python
def keyboard_request(input_fn=input):
    request_id = input_fn("Request ID / លេខសម្គាល់សំណើ: ").strip()
    batch_id = input_fn("Batch ID (B01/B02/B03) / លេខសម្គាល់ក្រុមអាហារ: ").strip()
    count = int(input_fn("Containers, whole number / ឯកតាផ្ទុក ជាចំនួនគត់: ").strip())
    approval = input_fn("Approval supplied? yes/no / មានការអនុម័តឬទេ? yes/no: ").strip().lower()
    if approval not in ("yes", "no"):
        raise ValueError("Enter yes or no / បញ្ចូល yes ឬ no")
    return {"request_id": request_id, "reviewed": approval == "yes",
            "lines": [(batch_id, count)], "events": []}


def run_one_request(data):
    db = new_db()
    try:
        before = snapshot(db)
        try:
            label = reserve(db, data["request_id"], data["reviewed"], data["lines"])
        except (ValueError, sqlite3.IntegrityError) as error:
            assert snapshot(db) == before
            print("No reservation / គ្មានការកក់:", error)
        else:
            print("Reserved; label / បានកក់; ស្លាក:", label)
            for event in data["events"]:
                before_event = snapshot(db)
                try:
                    record_event(db, data["request_id"], event)
                except (ValueError, sqlite3.IntegrityError) as error:
                    assert snapshot(db) == before_event
                    print("Event refused / ព្រឹត្តិការណ៍ត្រូវបដិសេធ:", error)
                    break
        show_report(db)
        return report(db)
    finally:
        db.close()


try:
    chosen_input = keyboard_request() if USE_KEYBOARD else USER_INPUT
    sandbox_report = run_one_request(chosen_input)
except (ValueError, EOFError) as error:
    print("Input not accepted; correct it and rerun / ទិន្នន័យមិនត្រឹមត្រូវ; កែហើយដំណើរការឡើងវិញ:", error)
```

```python
class FoodRescueTests(unittest.TestCase):
    def setUp(self):
        self.db = new_db()

    def tearDown(self):
        self.db.close()

    def assert_refused_unchanged(self, exception, action):
        before = snapshot(self.db)
        with self.assertRaises(exception):
            action()
        self.assertEqual(snapshot(self.db), before)

    def test_boundary_and_review(self):
        for total, reviewed, expected in (
            (19, True, "standard"), (20, True, "team"),
            (21, True, "team"), (20, False, "hold")):
            self.assertEqual(pickup_label(total, reviewed), expected)

    def test_invalid_planning_values(self):
        for total, reviewed in ((-1, True), (True, True), (2.5, True), (20, "yes")):
            with self.assertRaises(ValueError):
                pickup_label(total, reviewed)

    def test_normal_request(self):
        self.assertEqual(reserve(self.db, *REQUESTS[0]), "team")
        self.assertEqual(report(self.db)["reserved"], 20)
        self.assertEqual(self.db.execute(
            "SELECT SUM(containers) FROM lines").fetchone()[0], 7)
        self.assertEqual(self.db.execute(
            "SELECT available FROM batches ORDER BY batch_id").fetchall(), [(15,), (4,), (8,)])

    def test_unreviewed_unchanged(self):
        self.assert_refused_unchanged(ValueError, lambda: reserve(self.db, *REQUESTS[1]))

    def test_invalid_counts_unchanged(self):
        for count in (0, -1, True, 1.5, "2"):
            self.assert_refused_unchanged(ValueError,
                lambda: reserve(self.db, "BAD", True, [("B01", count)]))

    def test_empty_and_malformed_inputs(self):
        for lines in ([], [("B01",)], ["B01"], [("B01", 1, 2)]):
            self.assert_refused_unchanged(ValueError,
                lambda: reserve(self.db, "BAD", True, lines))
        self.assert_refused_unchanged(ValueError,
            lambda: reserve(self.db, " ", True, [("B01", 1)]))

    def test_unknown_batch_unchanged(self):
        self.assert_refused_unchanged(ValueError,
            lambda: reserve(self.db, "BAD", True, [("X99", 1)]))

    def test_late_shortage_rolls_back(self):
        reserve(self.db, *REQUESTS[0])
        reserve(self.db, *REQUESTS[2])
        self.assert_refused_unchanged(ValueError, lambda: reserve(self.db, *REQUESTS[3]))
        self.assertEqual(self.db.execute(
            "SELECT available FROM batches WHERE batch_id='B01'").fetchone()[0], 15)

    def test_interruption_rolls_back(self):
        self.assert_refused_unchanged(RuntimeError,
            lambda: reserve(self.db, *REQUESTS[0], inject_failure=True))

    def test_duplicate_request_unchanged(self):
        reserve(self.db, *REQUESTS[0])
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: reserve(self.db, *REQUESTS[0]))

    def test_event_order_and_duplicates(self):
        reserve(self.db, *REQUESTS[0])
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P01", "delivered"))
        batches_before = snapshot(self.db)[0]
        record_event(self.db, "P01", "collected")
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: record_event(self.db, "P01", "collected"))
        record_event(self.db, "P01", "delivered")
        self.assertEqual(snapshot(self.db)[0], batches_before)
        self.assertEqual(report(self.db)["delivered"], 20)
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: record_event(self.db, "P01", "delivered"))

    def test_event_unknown_request_and_type(self):
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P99", "collected"))
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P01", "cancelled"))

    def test_historical_line_snapshot(self):
        reserve(self.db, *REQUESTS[0])
        with self.db:
            self.db.execute("UPDATE batches SET ppc=99 WHERE batch_id='B01'")
        self.assertEqual(report(self.db)["reserved"], 20)
        self.assertEqual(self.db.execute(
            "SELECT ppc_snapshot FROM lines WHERE batch_id='B01'").fetchone()[0], 2)

    def test_empty_report(self):
        self.assertEqual(report(self.db)["reserved"], 0)
        self.assertEqual(report(self.db)["delivered"], 0)
        self.assertEqual(pending_collections(self.db), [])

    def test_canonical_day(self):
        db, decisions = build_day()
        try:
            self.assertEqual([(r[0], r[1]) for r in decisions], [
                ("P01", "accepted"), ("P02", "held"), ("P03", "accepted"),
                ("P04", "rejected"), ("P05", "accepted"), ("P06", "accepted")])
            self.assertEqual(report(db), {"offered": 94, "available": 34, "reserved": 60,
                "collected": 42, "delivered": 29, "reserved_only": 18,
                "collected_not_delivered": 13})
            self.assertEqual(db.execute(
                "SELECT batch_id, available FROM batches ORDER BY batch_id").fetchall(),
                [("B01", 9), ("B02", 2), ("B03", 2)])
            self.assertEqual(db.execute("SELECT SUM(containers) FROM lines").fetchone()[0], 21)
            before = snapshot(db)
            self.assertEqual(pending_collections(db), [("P06", 18, "standard")])
            self.assertEqual(snapshot(db), before)
            result = report(db)
            self.assertEqual(result["offered"], result["available"] + result["reserved"])
            self.assertEqual(result["offered"], result["available"] + result["reserved_only"]
                             + result["collected_not_delivered"] + result["delivered"])
        finally:
            db.close()

    def test_new_collection_preserves_other_measures(self):
        db, decisions = build_day()
        try:
            before = snapshot(db)[0]
            record_event(db, "P06", "collected")
            self.assertEqual(pending_collections(db), [])
            self.assertEqual(report(db)["collected"], 60)
            self.assertEqual(report(db)["delivered"], 29)
            self.assertEqual(snapshot(db)[0], before)
        finally:
            db.close()

    def test_fresh_runs_match(self):
        first, first_decisions = build_day()
        second, second_decisions = build_day()
        try:
            self.assertEqual(snapshot(first), snapshot(second))
            self.assertEqual(first_decisions, second_decisions)
        finally:
            first.close()
            second.close()

    def test_keyboard_input(self):
        answers = iter(["TRY02", "B03", "2", "yes"])
        data = keyboard_request(lambda prompt: next(answers))
        self.assertEqual(data, {"request_id": "TRY02", "reviewed": True,
                               "lines": [("B03", 2)], "events": []})
        for values in (["TRY02", "B03", "many", "yes"],
                       ["TRY02", "B03", "2", "maybe"]):
            answers = iter(values)
            with self.assertRaises(ValueError):
                keyboard_request(lambda prompt: next(answers))


test_log = io.StringIO()
suite = unittest.defaultTestLoader.loadTestsFromTestCase(FoodRescueTests)
test_result = unittest.TextTestRunner(stream=test_log, verbosity=1).run(suite)
if not test_result.wasSuccessful():
    print(test_log.getvalue())
    raise AssertionError("Tests failed; inspect the log / តេស្តបរាជ័យ; ពិនិត្យកំណត់ត្រា")
print("All tests passed / គ្រប់តេស្តបានឆ្លង:", test_result.testsRun)
```
