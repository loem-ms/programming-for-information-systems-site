## សាកល្បងផ្នែកតូចមួយជាមុន
- **ទិន្នន័យបញ្ចូល៖** ចំនួនកញ្ចប់នំប៉័ង និងឈុតបាយ។

កញ្ចប់នំប៉័ងមួយផ្តល់ **2 ចំណែក**; ឈុតបាយមួយផ្តល់ **5 ចំណែក**។ `def` កំណត់មុខងារ; ការហៅវាដំណើរការជំហាន; `return` បញ្ជូនលទ្ធផលត្រឡប់ ហើយ `print` បង្ហាញវា។ ដំណើរការក្រឡាបីបន្ទាប់។ នេះជាផែនការ មិនមែនការកក់ទេ។

```python
# 1. INPUT: change one value, predict, then rerun this cell and cell 3.
# Bread = 2 portions/container; rice = 5 portions/container.
bread_containers = 5
rice_containers = 2
reviewed = True       # Supplied human approval, not a food-safety calculation.
team_threshold = 20   # YOUR CHOICE: when does this fictional team need extra help?
```

```python
# 2. PROCESS: provided working example. Change only the marked part in your copy.
def plan_pickup(bread_containers, rice_containers, reviewed, team_threshold=20):
    # Supplied input checks: keep these unchanged.
    for count in (bread_containers, rice_containers):
        if type(count) is not int or count < 0:
            raise ValueError("ចំនួនឯកតាផ្ទុកត្រូវជាចំនួនគត់មិនអវិជ្ជមាន")
    if type(reviewed) is not bool:
        raise ValueError("ការពិនិត្យត្រូវជា True ឬ False")
    if type(team_threshold) is not int or team_threshold < 1:
        raise ValueError("កម្រិតក្រុមត្រូវជាចំនួនគត់វិជ្ជមាន")

    # YOUR CALCULATION: bread contributes 2 portions/container; rice contributes 5.
    total = bread_containers * 2 + rice_containers * 5
    if not reviewed:              # Keep human review first.
        label = "hold"
    elif total >= team_threshold: # YOUR RULE: explain and test the boundary.
        label = "team"
    else:
        label = "standard"
    return {"portions": total, "label": label}
```

```python
# 3. OUTPUT: call the same function after changing inputs or its marked code.
plan = plan_pickup(bread_containers, rice_containers, reviewed, team_threshold)
planning_labels = {"hold": "រង់ចាំការអនុម័ត", "team": "ក្រុមបន្ថែម", "standard": "ក្រុមធម្មតា"}
print("ចំណែកអាហារប៉ាន់ស្មាន:", plan["portions"])
print("ស្លាកផែនការ:", planning_labels[plan["label"]])
print("នេះជាផែនការ មិនមែនជាការកក់ ឬការប្រគល់ទេ។")
```

```python
import textwrap
from contextlib import redirect_stdout

CHECKPOINT_WIDTH = 60
CHECKPOINT_TITLE = "ប្រព័ន្ធសង្គ្រោះអាហារ"

# Display-only labels: the stored status codes and business rules do not change.
STATUS_KHMER = {
    "hold": "រង់ចាំការអនុម័ត", "held": "រង់ចាំការអនុម័ត",
    "standard": "ក្រុមធម្មតា", "team": "ក្រុមបន្ថែម",
    "accepted": "បានកក់", "rejected": "បានបដិសេធ",
    "collected": "បានប្រមូល", "delivered": "បានប្រគល់",
}

def khmer_label(value):
    return STATUS_KHMER.get(value, value)


def khmer_rows(rows):
    return [tuple(khmer_label(value) for value in row) for row in rows]


def khmer_error(error):
    # Keep known validation details; do not expose raw SQLite diagnostics to learners.
    if isinstance(error, sqlite3.IntegrityError):
        return "កំណត់ត្រាស្ទួន ឬទំនាក់ទំនងទិន្នន័យមិនត្រឹមត្រូវ។ មិនបានរក្សាការកែទេ។"
    return str(error).split(" / ")[-1]


CHECKPOINT_METRICS = (
    ("offered", "បានផ្តល់ដំបូង"), ("available", "នៅអាចកក់បាន"),
    ("reserved", "បានកក់"), ("collected", "បានកត់ត្រាការប្រមូល"),
    ("delivered", "បានកត់ត្រាការប្រគល់"),
)


def checkpoint_request(bread=5, rice=2, reviewed=True, events=()):
    # Editable sample data, not approval or evidence of a real-world event.
    return {"request_id": "TRY01", "reviewed": reviewed,
            "lines": [("B01", bread), ("B02", rice)], "events": list(events)}


def checkpoint_text(text, prefix="  "):
    # Wrap at spaces, never between Khmer letters and their dependent vowels.
    # Only long ASCII data identifiers are split; source data is not changed.
    limit = CHECKPOINT_WIDTH - len(prefix)
    for paragraph in str(text).splitlines() or [""]:
        words = []
        for word in paragraph.split(" "):
            if word.isascii() and len(word) > limit:
                words.extend(word[i:i+limit] for i in range(0, len(word), limit))
            else:
                words.append(word)
        print(textwrap.fill(" ".join(words), width=CHECKPOINT_WIDTH,
                            initial_indent=prefix, subsequent_indent=" " * len(prefix),
                            break_long_words=False, break_on_hyphens=False))


def checkpoint_heading(khmer):
    print("\n" + khmer)
    print("-" * 40)


def show_checkpoint(stage, data=None, metrics=None, notes=()):
    # stage controls which caller can supply real values; it is not an app status.
    metrics = {} if metrics is None else metrics
    print("\n" + "=" * 40)
    print(CHECKPOINT_TITLE)
    print("=" * 40)
    checkpoint_text("ទិន្នន័យសន្មតសម្រាប់ថ្នាក់រៀន។")
    checkpoint_heading("សំណើព្រាងបច្ចុប្បន្ន")
    if data is None:
        checkpoint_text("ទិន្នន័យគំរូពេញមួយថ្ងៃ។")
    else:
        checkpoint_text("លេខសម្គាល់សំណើ៖ " + (str(data["request_id"]) or "មិនទាន់បញ្ចូល"))
        if stage >= 3:
            review = "បានអនុម័ត" if data["reviewed"] is True else "មិនទាន់អនុម័ត" if data["reviewed"] is False else "តម្លៃមិនត្រឹមត្រូវ"
            checkpoint_text("ការពិនិត្យដោយមនុស្ស៖ " + review)
        checkpoint_text("ចំនួនឯកតាផ្ទុកដែលស្នើ៖")
        for batch_id, count in data["lines"]:
            checkpoint_text(str(batch_id) + ": " + str(count), "    - ")
        if not data["lines"]:
            checkpoint_text("មិនទាន់បញ្ចូល", "    - ")
        if data["events"]:
            checkpoint_text("ព្រឹត្តិការណ៍ដែលស្នើកត់ត្រា មិនទាន់ជាភស្តុតាង៖")
            for event in data["events"]:
                checkpoint_text(khmer_label(event), "    - ")
    if metrics:
        checkpoint_heading("របាយការណ៍បច្ចុប្បន្ន")
        checkpoint_text("ឯកតា៖ ចំណែកអាហារប៉ាន់ស្មាន មិនមែនមនុស្សដែលបានទទួលអាហារ។")
        for key, label in CHECKPOINT_METRICS:
            if key in metrics:
                checkpoint_text(label + ": " + str(metrics[key]))
    if notes:
        checkpoint_heading("លទ្ធផល និងកំណត់សម្គាល់")
        for note in notes:
            checkpoint_text(note, "  - ")
    if len(metrics) > 1:
        print()
        checkpoint_text("ចំនួនបានផ្តល់ កក់ ប្រមូល និងប្រគល់រាប់ត្រួតគ្នា។ កុំបូកវាចូលគ្នា។")
    print("=" * 40)
    return dict(metrics)


def checkpoint_1(data):
    return show_checkpoint(1, data, notes=(
        "សំណើព្រាងប៉ុណ្ណោះ៖ បង្ហាញទិន្នន័យ តែមិនទាន់ពិនិត្យបរិមាណ ឬកក់។",
        "ការអនុម័តជាការសម្រេចចិត្តរបស់មនុស្ស។ មិនបានកត់ត្រាព្រឹត្តិការណ៍នៅទីនេះទេ។",))

# RUN THE PROVIDED SAMPLE. You may change the values below to explore.
_ = checkpoint_1(checkpoint_request(bread=5, rice=2, reviewed=True))
```

```python
# PROVIDED INTERFACE: run it, then use the numbered menu. No implementation is required.
# The menu calls only features implemented by the selected section.

class MenuCancelled(Exception):
    pass


def menu_read(prompt, input_fn):
    value = input_fn(prompt).strip()
    if value.lower() == "q":
        raise MenuCancelled()
    return value


def menu_integer(prompt, input_fn, minimum=1, maximum=10000):
    while True:
        value = menu_read(prompt, input_fn)
        try:
            number = int(value)
        except ValueError:
            print("សូមបញ្ចូលចំនួនគត់។")
            continue
        if minimum <= number <= maximum:
            return number
        print("ចំនួនត្រូវនៅចន្លោះ", minimum, "និង", maximum)


def menu_name(prompt, input_fn):
    while True:
        value = menu_read(prompt, input_fn)
        if value:
            return value
        print("សូមបញ្ចូលតម្លៃ មិនត្រូវទុកទំនេរ។")


def menu_database(offers):
    # Only called from section 4 onward, after new_db() has been implemented.
    db = new_db()
    with db:
        db.execute("DELETE FROM batches")
        db.executemany("INSERT INTO batches VALUES (?, ?, ?, ?, ?)",
                       [(batch, name, ppc, count, count) for batch, name, ppc, count in offers])
    return db


def menu_new_state(stage):
    offers = [tuple(row) for row in BATCHES] if stage >= 2 else []
    return {"offers": offers,
            "request": checkpoint_request(reviewed=False) if stage >= 2 else
                       {"request_id": "", "reviewed": False, "lines": [], "events": []},
            "db": menu_database(offers) if stage >= 4 else None,
            "reservations": [], "events": [],
            "notice": "ការសាកល្បងថ្មី។ ទិន្នន័យទាំងអស់ជារឿងសន្មត។"}


def menu_metrics(stage, state):
    if stage >= 7:
        return report(state["db"])
    if stage >= 6:
        return checkpoint_event_totals(state["db"])
    if stage >= 5:
        return checkpoint_reserved(state["db"])
    if stage >= 4:
        return checkpoint_stock(state["db"])
    if stage >= 2:
        return {"offered": sum(ppc * count for _, _, ppc, count in state["offers"])}
    return {}


def menu_estimate(state):
    sizes = {batch: ppc for batch, _, ppc, _ in state["offers"]}
    return sum(count * sizes[batch] for batch, count in state["request"]["lines"])


def menu_view(stage, state):
    notes = [state["notice"], "បរិមាណអាហារផ្តល់ (លេខសម្គាល់ ឈ្មោះ ចំណែកក្នុងមួយឯកតា ចំនួនឯកតា)៖"]
    notes += [f"{batch} · {name}: {count} ឯកតាផ្ទុក; {ppc} ចំណែកក្នុងមួយឯកតា" for batch, name, ppc, count in state["offers"]] or ["មិនទាន់មានសំណើផ្តល់អាហារ។"]
    if stage == 1:
        notes.append("សំណើទាំងនេះជាសំណើព្រាង។ មិនទាន់មានការកក់។")
    elif state["request"]["lines"]:
        portions = menu_estimate(state)
        notes.append("ចំណែកអាហារដែលស្នើ៖ " + str(portions))
        if stage >= 3:
            notes.append("ស្លាកផែនការ៖ " + khmer_label(pickup_label(portions, state["request"]["reviewed"])))
    if stage >= 4:
        notes.append("ស្តុកនៅអាចកក់បាន៖ " + "; ".join(f"{row[0]} = {row[4]} ឯកតាផ្ទុក" for row in snapshot(state["db"])[0]))
    if stage >= 5:
        notes.append("សំណើដែលបានកក់ (ដាច់ពីសំណើព្រាង)៖ " + (", ".join(row[0] for row in state["reservations"]) or "មិនទាន់មាន"))
    if stage >= 6:
        notes.append("ព្រឹត្តិការណ៍បានកត់ត្រាក្នុងគំរូ៖ " + ("; ".join(request + " · " + khmer_label(event) for request, event in state["events"]) or "មិនទាន់មាន"))
    if stage >= 7:
        notes.append("រង់ចាំការប្រមូល៖ " + str(khmer_rows(pending_collections(state["db"]))))
    return show_checkpoint(stage, state["request"], menu_metrics(stage, state), notes)


def menu_offer(stage, state, input_fn):
    if state["reservations"]:
        state["notice"] = "មិនអាចកែបរិមាណដើមក្រោយការកក់ទេ។ ចាប់ផ្តើមការសាកល្បងថ្មីជាមុន។"
        return
    print("បញ្ចូលសំណើផ្តល់អាហារ។ លេខសម្គាល់ដដែលកែតម្លៃដើម; លេខថ្មីបន្ថែមក្រុមអាហារ។")
    batch = menu_name("លេខសម្គាល់ក្រុមអាហារ (ឧ. B04)៖ ", input_fn)
    name = menu_name("ឈ្មោះអាហារសន្មត៖ ", input_fn)
    ppc = menu_integer("ចំណែកអាហារប៉ាន់ស្មានក្នុងមួយឯកតាផ្ទុក៖ ", input_fn)
    count = menu_integer("ចំនួនឯកតាផ្ទុកដែលផ្តល់ (អាចជា 0)៖ ", input_fn, minimum=0)
    offers = [row for row in state["offers"] if row[0] != batch] + [(batch, name, ppc, count)]
    offers.sort(key=lambda row: row[0])
    if stage >= 4:
        new_database = menu_database(offers)
        state["db"].close()
        state["db"] = new_database
    state["offers"] = offers
    state["request"]["reviewed"] = False
    state["notice"] = "បានរក្សាសំណើផ្តល់អាហារក្នុងការសាកល្បងនេះ។ ការកែតម្លៃត្រូវការការអនុម័តសំណើឡើងវិញ។"


def menu_request(state, input_fn):
    if not state["offers"]:
        state["notice"] = "សូមរក្សាទុកអាហារផ្តល់ជាមុន។"
        return False
    request_id = menu_name("លេខសម្គាល់សំណើថ្មី (ឧ. TRY02)៖ ", input_fn)
    line_count = menu_integer("តើស្នើអាហារប៉ុន្មានបន្ទាត់ (1 ដល់ 10)? ", input_fn, maximum=10)
    batches = {row[0] for row in state["offers"]}
    lines = []
    for number in range(1, line_count + 1):
        print("បន្ទាត់", number, "ក្រុមអាហារដែលមាន៖", ", ".join(sorted(batches)))
        while True:
            batch = menu_name("លេខសម្គាល់ក្រុមអាហារ៖ ", input_fn)
            if batch in batches:
                break
            print("មិនមានក្រុមអាហារនេះទេ។ សូមជ្រើសពីបញ្ជី។")
        count = menu_integer("ចំនួនឯកតាផ្ទុកដែលស្នើ៖ ", input_fn)
        lines.append((batch, count))
    state["request"] = {"request_id": request_id, "reviewed": False, "lines": lines, "events": []}
    state["notice"] = "បានរក្សាសំណើព្រាង។ មិនទាន់អនុម័ត ឬកក់ទេ។"
    return True


def menu_review(state, input_fn):
    if not state["request"]["lines"]:
        state["notice"] = "សូមរក្សាទុកសំណើព្រាងជាមុន។"
        return False
    print("នេះកត់ត្រាការសម្រេចចិត្តមនុស្សក្នុងរឿងសន្មត។ កម្មវិធីមិនវាយតម្លៃសុវត្ថិភាពអាហារទេ។")
    approved = menu_integer("តើអ្នកសម្របសម្រួលអនុម័តសំណើព្រាងនេះឬទេ? 1=អនុម័ត, 0=មិនទាន់៖ ", input_fn, minimum=0, maximum=1)
    state["request"]["reviewed"] = bool(approved)
    state["notice"] = "បានកត់ត្រាការពិនិត្យសម្រាប់សំណើព្រាង។ វាមិនកែការកក់ដែលបានទទួលយកមុនទេ។"
    return True


def menu_reserve(state):
    data = state["request"]
    accepted, decision = checkpoint_reserve(state["db"], data)
    state["notice"] = decision
    if accepted:
        state["reservations"].append((data["request_id"], data["reviewed"], tuple(data["lines"])))
        return accepted


def menu_event(state, input_fn):
    print("ការកក់ដែលមាន៖", [row[0] for row in state["reservations"]])
    request_id = menu_name("លេខសម្គាល់ការកក់ដែលត្រូវកត់ត្រាព្រឹត្តិការណ៍៖ ", input_fn)
    choice = menu_integer("ព្រឹត្តិការណ៍៖ 1=ប្រមូល, 2=ប្រគល់៖ ", input_fn, maximum=2)
    event = "collected" if choice == 1 else "delivered"
    before = snapshot(state["db"])
    try:
        record_event(state["db"], request_id, event)
    except (ValueError, sqlite3.IntegrityError) as error:
        assert snapshot(state["db"]) == before
        state["notice"] = "ព្រឹត្តិការណ៍ត្រូវបដិសេធ៖ " + khmer_error(error)
    else:
        state["events"].append((request_id, event))
        state["notice"] = "បានកត់ត្រា៖ " + request_id + " " + khmer_label(event)
    


def menu_output_clearer(keep_history):
    """Use notebook display clearing only when a kernel is actually running."""
    if keep_history:
        return None
    try:
        from IPython import get_ipython
        shell = get_ipython()
        if shell is None or getattr(shell, "kernel", None) is None:
            return None
        from IPython.display import clear_output
    except ImportError:
        # A plain Python script still works without any optional packages.
        return None
    return clear_output


def run_checkpoint_menu(stage, input_fn=None, keep_history=False):
    """A fresh menu session; state lives between choices and closes on exit."""
    if type(stage) is not int or not 1 <= stage <= 10:
        raise ValueError("លេខផ្នែកត្រូវនៅចន្លោះ 1 និង 10។")
    required = {2: "BATCHES", 3: "pickup_label", 4: "new_db", 5: "reserve",
                6: "record_event", 7: "report", 8: "run_one_request",
                9: "FoodRescueTests", 10: "final_baseline_check"}
    if any(name not in globals() for number, name in required.items() if number <= stage):
        print("សូមដំណើរការក្រឡាពីដើមរហូតដល់ផ្នែកនេះជាមុន។")
        return None
    # Close any optional graphical session before opening a text session.
    if "close_checkpoint_form" in globals():
        close_checkpoint_form()
    input_fn = input if input_fn is None else input_fn
    state = menu_new_state(stage)
    clear_screen = menu_output_clearer(keep_history)
    show_details = False
    try:
        while True:
            # Clear only this interactive cell, never static teaching examples.
            # Render before asking for input; never clear an active prompt.
            if clear_screen is not None:
                clear_screen(wait=True)
            menu_view(stage, state)
            if show_details:
                show_report(state["db"])
            print("\nជ្រើសសកម្មភាព")
            print("1 មើលទិន្នន័យបច្ចុប្បន្ន | 2 រក្សាទុកអាហារផ្តល់ | 3 រក្សាទុកសំណើព្រាង")
            if stage >= 2:
                print("4 គណនាចំណែក")
            if stage >= 3:
                print("5 កត់ត្រាការអនុម័ត")
            if stage >= 4:
                print("6 មើលស្តុក")
            if stage >= 5:
                print("7 កក់សំណើទាំងមូល")
            if stage >= 6:
                print("8 កត់ត្រាការប្រមូល ឬការប្រគល់")
            if stage >= 7:
                print("10 មើលរបាយការណ៍")
            if stage >= 8:
                print("11 រក្សាទុកសំណើព្រាង -> កត់ត្រាការអនុម័ត -> កក់សំណើទាំងមូល")
            print("9 ចាប់ផ្តើមថ្មី | 0 បញ្ចប់ | q បោះបង់ការបញ្ចូលបច្ចុប្បន្ន")
            try:
                choice = menu_read("លេខសកម្មភាព៖ ", input_fn)
                if choice == "0":
                    break
                show_details = False
                if choice == "9":
                    if state["db"] is not None:
                        state["db"].close()
                    state = menu_new_state(stage)
                elif choice == "1":
                    pass
                elif choice == "2":
                    menu_offer(stage, state, input_fn)
                elif choice == "3":
                    menu_request(state, input_fn)
                elif choice == "4" and stage >= 2:
                    state["notice"] = "ចំណែកអាហារដែលស្នើ៖ " + str(menu_estimate(state)) + "។ ការប៉ាន់ស្មានមិនមែនជាការកក់ទេ។"
                elif choice == "5" and stage >= 3:
                    menu_review(state, input_fn)
                elif choice == "6" and stage >= 4:
                    state["notice"] = "បរិមាណខាងក្រោមអានពីមូលដ្ឋានទិន្នន័យក្នុងអង្គចងចាំ។"
                elif choice == "7" and stage >= 5:
                    menu_reserve(state)
                elif choice == "8" and stage >= 6:
                    menu_event(state, input_fn)
                elif choice == "10" and stage >= 7:
                    show_details = True
                    state["notice"] = "របាយការណ៍អានពីកំណត់ត្រាបច្ចុប្បន្ន មិនមែនលទ្ធផលគំរូថេរ។"
                elif choice == "11" and stage >= 8:
                    if menu_request(state, input_fn) and menu_review(state, input_fn):
                        menu_reserve(state)
                else:
                    state["notice"] = "ជម្រើសនេះមិនមានទេ។ សូមជ្រើសពីបញ្ជី។"
            except MenuCancelled:
                state["notice"] = "បានបោះបង់ការបញ្ចូល។ ការកក់ និងព្រឹត្តិការណ៍មុននៅដដែល។"
    except (EOFError, KeyboardInterrupt):
        print("\nបានបញ្ឈប់ការបញ្ចូលដោយសុវត្ថិភាព។")
    finally:
        # Capture only synthetic evidence for tests, then always close this session's database.
        result = {"stage": stage, "metrics": menu_metrics(stage, state),
                  "snapshot": snapshot(state["db"]) if state["db"] is not None else None,
                  "offers": tuple(state["offers"]), "request": state["request"],
                  "reservations": tuple(state["reservations"]), "events": tuple(state["events"])}
        if state["db"] is not None:
            state["db"].close()
    print("បានបញ្ចប់ការសាកល្បង។ មូលដ្ឋានទិន្នន័យមិនត្រូវបានរក្សាទុកទេ។ កត់ត្រាភស្តុតាងក្នុងច្បាប់ចម្លងរបស់អ្នក។")
    return result
```

```python
# PROVIDED OPTIONAL INTERFACE. Business rules stay in the functions above/below.
# This source is embedded in the standalone notebook; no file import is needed.

def close_checkpoint_form():
    old = globals().pop("checkpoint_form", None)
    if old is not None:
        old.close()


class CheckpointForm:
    """One temporary practice session, shared by all tabs in this panel."""
    def __init__(self, stage):
        import ipywidgets as widgets
        from IPython.display import display
        self.widgets = widgets
        self.stage = stage
        self.state = menu_new_state(stage)
        self.closed = False
        self.busy = False
        self.controls = {}
        self.owned = []
        self.rows = []
        self.details = False
        self.last_output = ""
        self.output = self.make("Output")
        self.status = self.make("HTML", value="")
        self.history = self.make("Checkbox", value=False, description="រក្សាលទ្ធផលមុន")

        self.offer_choice = self.field("offer_choice", "ជ្រើសអាហារផ្តល់ ឬបន្ថែមថ្មី", "Dropdown", options=[("បន្ថែមថ្មី", None)])
        self.offer_id = self.field("offer_id", "លេខសម្គាល់ក្រុមអាហារ (ឧ. B04)", "Text", value="B04")
        self.offer_name = self.field("offer_name", "ឈ្មោះអាហារ", "Text", value="")
        self.offer_ppc = self.field("offer_ppc", "ចំណែកអាហារក្នុងមួយឯកតាផ្ទុក", "IntText", value=2)
        self.offer_count = self.field("offer_count", "ចំនួនឯកតាផ្ទុកដែលផ្តល់", "IntText", value=0)
        self.offer_choice.children[-1].observe(self.load_offer, names="value")
        offers = self.make("VBox", children=[self.note("លេខដដែលកែអាហារផ្តល់។ លេខថ្មីបន្ថែមអាហារផ្តល់។ ក្រោយកក់ ត្រូវចាប់ផ្តើមថ្មីដើម្បីកែស្តុកដើម។"), self.offer_choice, self.offer_id, self.offer_name, self.offer_ppc, self.offer_count, self.button("offer", "រក្សាទុកអាហារផ្តល់", "Save offer", self.save_offer)])

        self.request_id = self.field("request_id", "លេខសម្គាល់សំណើព្រាង", "Text", value=self.state["request"]["request_id"])
        self.lines_box = self.make("VBox", children=[])
        self.line_buttons = self.make("HBox", children=[self.button("add_line", "បន្ថែមបន្ទាត់", "Add request line", self.add_line), self.button("remove_line", "ដកបន្ទាត់ចុងក្រោយ", "Remove last line", self.remove_line)], layout=widgets.Layout(flex_flow="row wrap"))
        request_children = [self.note("កែប្រអប់ រួចចុច រក្សាទុកសំណើព្រាង។ ការរក្សាទុកមិនមែនជាការអនុម័ត ឬការកក់ទេ។"), self.request_id, self.lines_box, self.line_buttons, self.button("request", "រក្សាទុកសំណើព្រាង", "Save draft request", self.save_request)]
        if stage >= 2:
            request_children.append(self.button("estimate", "គណនាចំណែក", "Estimate portions", self.estimate))
        if stage >= 3:
            self.approval = self.field("approval", "ការសម្រេចរបស់អ្នកសម្របសម្រួលសម្រាប់សំណើព្រាងដែលបានរក្សាទុក", "Checkbox", value=False, description="បានអនុម័ត")
            request_children += [self.approval, self.button("review", "កត់ត្រាការអនុម័ត", "Record coordinator approval", self.review)]
        if stage >= 5:
            request_children += [self.button("reserve", "កក់សំណើទាំងមូល", "Reserve whole request", self.reserve_draft)]
        request = self.make("VBox", children=request_children)
        panels = [offers, request]
        titles = ["អាហារផ្តល់", "សំណើព្រាង"]
        if stage >= 6:
            self.event_request = self.field("event_request", "ជ្រើសសំណើដែលបានកក់", "Dropdown", options=[])
            panels.append(self.make("VBox", children=[self.note("កត់ត្រាអ្វីដែលកើតឡើងក្នុងរឿងសន្មត។ ប៊ូតុងមិនប្រមូល ឬប្រគល់អាហារពិតទេ។"), self.event_request, self.button("collect", "កត់ត្រាការប្រមូល", "Record collection", lambda: self.event(1)), self.button("deliver", "កត់ត្រាការប្រគល់", "Record delivery", lambda: self.event(2))]))
            titles.append("ការប្រមូល និងប្រគល់")
        self.tabs = self.make("Tab", children=panels)
        for i, title in enumerate(titles):
            self.tabs.set_title(i, title)
        self.tabs.selected_index = 0 if stage == 1 else 1
        actions = [self.button("refresh", "មើលទិន្នន័យបច្ចុប្បន្ន", "Show current state", lambda: None)]
        if stage >= 4:
            actions.append(self.button("stock", "មើលស្តុក", "Show available stock", lambda: None))
        if stage >= 7:
            actions.append(self.button("report", "មើលរបាយការណ៍", "Show current report", self.show_report))
        self.confirm_reset = self.make("Checkbox", value=False, description="លុបទិន្នន័យហាត់នេះ ហើយចាប់ផ្តើមថ្មី")
        actions += [self.button("reset", "ចាប់ផ្តើមថ្មី", "Start new practice session", self.reset)]
        self.close_button = self.make("Button", description="បិទទម្រង់", tooltip="Close practice session", layout=widgets.Layout(width="auto", min_width="170px", height="42px"))
        self.close_button.on_click(lambda _: self.close())
        actions.append(self.close_button)
        self.panel = self.make("VBox", children=[self.note("វគ្គហាត់ថ្មី។ ទិន្នន័យរបស់ទម្រង់នេះដាច់ពីលទ្ធផលគំរូខាងលើ។ ប្តូរផ្ទាំងមិនចាប់ផ្តើមថ្មីទេ។"), self.tabs, self.status, self.confirm_reset, self.make("HBox", children=actions, layout=widgets.Layout(flex_flow="row wrap")), self.history, self.output], layout=widgets.Layout(width="100%", max_width="760px"))
        self.sync_offers()
        self.load_request()
        self.refresh()
        display(self.panel)

    def make(self, name, **kwargs):
        control = getattr(self.widgets, name)(**kwargs)
        self.owned.append(control)
        return control

    def note(self, text):
        # Only fixed, authored text goes into HTML. User entries are printed as text.
        return self.make("HTML", value="<p>" + text + "</p>")

    def field(self, key, label, kind, **kwargs):
        control = self.make(kind, layout=self.widgets.Layout(width="100%"), **kwargs)
        self.controls[key] = control
        return self.make("VBox", children=[self.note(label), control])

    def button(self, key, label, english, action):
        button = self.make("Button", description=label, tooltip=english, layout=self.widgets.Layout(width="auto", min_width="170px", height="42px"))
        button.on_click(lambda _: self.act(action))
        self.controls[key] = button
        return button

    def answers(self, values):
        # Existing text handlers validate every field. Invalid input cancels safely
        # instead of waiting for keyboard input or duplicating validation rules.
        pending = iter(str(value) for value in values)
        return lambda _prompt: next(pending, "q")

    def act(self, action):
        if self.closed or self.busy:
            return
        self.busy = True
        messages = io.StringIO() if "io" in globals() else __import__("io").StringIO()
        try:
            with redirect_stdout(messages):
                action()
        except MenuCancelled:
            self.state["notice"] = "មិនបានរក្សាការបញ្ចូលនេះ។ ពិនិត្យតម្លៃ រួចសាកម្តងទៀត។"
        except (ValueError, TypeError) as error:
            self.state["notice"] = "ទិន្នន័យមិនត្រឹមត្រូវ៖ " + str(error)
        finally:
            self.busy = False
            self.refresh(messages.getvalue())

    def sync_offers(self):
        selector = self.controls["offer_choice"]
        previous = selector.value
        selector.options = [("បន្ថែមថ្មី", None)] + [(row[0] + " · " + row[1], row[0]) for row in self.state["offers"]]
        selector.value = previous if previous in [row[0] for row in self.state["offers"]] else None
        self.load_offer({"new": selector.value})
        choices = [(row[0] + " · " + row[1], row[0]) for row in self.state["offers"]]
        for batch, _count, _box in self.rows:
            old = batch.value
            batch.options = choices
            if old in [row[0] for row in self.state["offers"]]:
                batch.value = old

    def load_offer(self, change):
        selected = change["new"]
        row = next((row for row in self.state["offers"] if row[0] == selected), None)
        self.controls["offer_id"].disabled = row is not None
        if row:
            for key, value in zip(("offer_id", "offer_name", "offer_ppc", "offer_count"), row):
                self.controls[key].value = value
        else:
            for key, value in (("offer_id", "B04"), ("offer_name", ""), ("offer_ppc", 2), ("offer_count", 0)):
                self.controls[key].value = value

    def save_offer(self):
        menu_offer(self.stage, self.state, self.answers([self.controls[key].value for key in ("offer_id", "offer_name", "offer_ppc", "offer_count")]))
        self.sync_offers()
        if self.stage >= 3:
            self.controls["approval"].value = self.state["request"]["reviewed"]

    def add_line(self, batch=None, count=1):
        if len(self.rows) >= 10:
            self.state["notice"] = "សំណើមួយមានបានច្រើនបំផុត 10 បន្ទាត់។"
            return
        choices = [(row[0] + " · " + row[1], row[0]) for row in self.state["offers"]]
        choice = self.make("Dropdown", options=choices, layout=self.widgets.Layout(width="100%"))
        if batch in [row[0] for row in self.state["offers"]]:
            choice.value = batch
        quantity = self.make("IntText", value=count, description="ឯកតាផ្ទុក", style={"description_width": "initial"}, layout=self.widgets.Layout(width="100%"))
        box = self.make("VBox", children=[choice, quantity])
        self.rows.append((choice, quantity, box))
        self.lines_box.children = tuple(row[2] for row in self.rows)

    def remove_line(self):
        if self.rows:
            for widget in self.rows.pop():
                widget.close()
            self.lines_box.children = tuple(row[2] for row in self.rows)

    def load_request(self):
        while self.rows:
            self.remove_line()
        self.controls["request_id"].value = self.state["request"]["request_id"]
        for batch, count in self.state["request"]["lines"]:
            self.add_line(batch, count)
        if not self.rows:
            self.add_line()
        if self.stage >= 3:
            self.controls["approval"].value = self.state["request"]["reviewed"]

    def draft_matches(self):
        return (self.controls["request_id"].value.strip() == self.state["request"]["request_id"] and
                [(batch.value, count.value) for batch, count, _box in self.rows] == self.state["request"]["lines"])

    def saved_draft_required(self, action):
        if not self.draft_matches():
            self.state["notice"] = "ប្រអប់មានការកែមិនទាន់រក្សាទុក។ ចុច រក្សាទុកសំណើព្រាង ជាមុន។"
            return
        action()

    def save_request(self):
        values = [self.controls["request_id"].value, len(self.rows)]
        for batch, count, _box in self.rows:
            values += [batch.value or "", count.value]
        menu_request(self.state, self.answers(values))
        if self.stage >= 3:
            self.controls["approval"].value = self.state["request"]["reviewed"]

    def estimate(self):
        self.saved_draft_required(lambda: self.state.update(notice="ចំណែកអាហារដែលស្នើ៖ " + str(menu_estimate(self.state)) + "។ ការប៉ាន់ស្មានមិនមែនជាការកក់ទេ។"))

    def review(self):
        self.saved_draft_required(lambda: menu_review(self.state, self.answers([int(self.controls["approval"].value)])))
        self.controls["approval"].value = self.state["request"]["reviewed"]

    def reserve_draft(self):
        if self.controls["approval"].value != self.state["request"]["reviewed"]:
            self.state["notice"] = "ការអនុម័តក្នុងប្រអប់មិនទាន់កត់ត្រា។ ចុច កត់ត្រាការអនុម័ត ជាមុន។"
            return
        self.saved_draft_required(lambda: menu_reserve(self.state))

    def event(self, number):
        request_id = self.controls["event_request"].value
        if request_id is None:
            self.state["notice"] = "មិនទាន់មានសំណើដែលបានកក់។ កក់សំណើជាមុន។"
            return
        menu_event(self.state, self.answers([request_id, number]))

    def show_report(self):
        self.details = True
        self.state["notice"] = "របាយការណ៍អានពីកំណត់ត្រាបច្ចុប្បន្ន មិនមែនលទ្ធផលគំរូថេរ។"

    def reset(self):
        if not self.confirm_reset.value:
            self.state["notice"] = "សម្គាល់ប្រអប់លុបទិន្នន័យហាត់ ជាមុន។ វាលុបតែវគ្គហាត់ក្នុងទម្រង់នេះ។"
            return
        if self.state["db"] is not None:
            self.state["db"].close()
        self.state = menu_new_state(self.stage)
        self.details = False
        self.confirm_reset.value = False
        self.sync_offers()
        self.load_request()

    def refresh(self, messages=""):
        if self.closed:
            return
        if self.stage >= 6:
            selector = self.controls["event_request"]
            old = selector.value
            selector.options = [row[0] for row in self.state["reservations"]]
            if old in selector.options:
                selector.value = old
        self.controls["offer"].disabled = bool(self.state["reservations"])
        if "reserve" in self.controls:
            # Callback also checks dirty fields; this is not the domain validation.
            self.controls["reserve"].disabled = False
        self.status.value = "<p>លទ្ធផលខាងក្រោមបង្ហាញទិន្នន័យដែលបានរក្សាទុកក្នុងវគ្គហាត់នេះ។ ការកែប្រអប់មិនកែទិន្នន័យដោយស្វ័យប្រវត្តិទេ។</p>"
        import io as output_io
        captured = output_io.StringIO()
        with redirect_stdout(captured):
            if messages:
                print(messages.strip())
            menu_view(self.stage, self.state)
            if self.details:
                show_report(self.state["db"])
        self.last_output = captured.getvalue()
        with self.output:
            if not self.history.value:
                self.output.clear_output(wait=True)
            print(self.last_output, end="")

    def close(self):
        if self.closed:
            return
        self.closed = True
        if self.state["db"] is not None:
            self.state["db"].close()
        for control in reversed(self.owned):
            control.close()
        print("បានបិទវគ្គហាត់។ ទិន្នន័យក្នុងទម្រង់មិនត្រូវបានរក្សាទុកទេ។")


def make_checkpoint_form(stage):
    if type(stage) is not int or not 1 <= stage <= 10:
        raise ValueError("លេខផ្នែកត្រូវនៅចន្លោះ 1 និង 10។")
    required = {2: "BATCHES", 3: "pickup_label", 4: "new_db", 5: "reserve", 6: "record_event", 7: "report", 8: "run_one_request", 9: "FoodRescueTests", 10: "final_baseline_check"}
    if any(name not in globals() for number, name in required.items() if number <= stage):
        print("សូមដំណើរការក្រឡាពីដើមរហូតដល់ផ្នែកនេះជាមុន។")
        return None
    close_checkpoint_form()
    try:
        form = CheckpointForm(stage)
    except ImportError:
        print("ទម្រង់មិនមាន។ ប្រើក្រឡាកូដ ឬម៉ឺនុយអត្ថបទ។ មិនត្រូវដំឡើងអ្វីទេ។")
        return None
    globals()["checkpoint_form"] = form
    return form

# Re-running this definition cell disposes the old panel and its database.
close_checkpoint_form()
```

```python
def reserve(db, request_id, reviewed, lines, inject_failure=False):
    if type(reviewed) is not bool or not reviewed:
        raise ValueError("ត្រូវការការអនុម័តពីអ្នកសម្របសម្រួល")
    if not isinstance(request_id, str) or not request_id.strip():
        raise ValueError("ត្រូវការលេខសម្គាល់សំណើ")
    if not isinstance(lines, (list, tuple)) or not lines:
        raise ValueError("ត្រូវមានយ៉ាងតិចមួយបន្ទាត់")
    prepared = []
    for line in lines:
        if not isinstance(line, (list, tuple)) or len(line) != 2:
            raise ValueError("ប្រើ (batch_id, containers)")
        batch_id, count = line
        if not isinstance(batch_id, str) or type(count) is not int or count <= 0:
            raise ValueError("ប្រើលេខសម្គាល់ និងចំនួនគត់វិជ្ជមាន")
        batch = db.execute("SELECT ppc FROM batches WHERE batch_id=?", (batch_id,)).fetchone()
        if batch is None:
            raise ValueError("ក្រុមអាហារមិនស្គាល់: " + batch_id)
        prepared.append((batch_id, count, batch[0]))
    total = sum(count * ppc for batch_id, count, ppc in prepared)
    label = pickup_label(total, reviewed)
    with db:
        db.execute("INSERT INTO reservations VALUES (?, ?, ?)", (request_id, label, total))
        for number, (batch_id, count, ppc) in enumerate(prepared, 1):
            change = db.execute(
                "UPDATE batches SET available=available-? WHERE batch_id=? AND available>=?",
                (count, batch_id, count))
            if change.rowcount != 1:
                raise ValueError("ឯកតាផ្ទុកមិនគ្រប់គ្រាន់: " + batch_id)
            db.execute("INSERT INTO lines VALUES (?, ?, ?, ?, ?)",
                       (request_id, number, batch_id, count, ppc))
            if inject_failure and number == 1:
                raise RuntimeError("ការផ្អាកសន្មត")
    return label


def show_atomic_example():
    db = new_db()
    try:
        reserve(db, *REQUESTS[0])
        reserve(db, *REQUESTS[2])
        before = snapshot(db)
        print("មុន:", [(row[0], row[4]) for row in before[0]])
        try:
            reserve(db, *REQUESTS[3])
        except ValueError as error:
            print("បដិសេធ:", khmer_error(error))
        after = snapshot(db)
        assert after == before
        print("ក្រោយ:", [(row[0], row[4]) for row in after[0]])
        print("កំណត់ត្រាមេ:", db.execute(
            "SELECT * FROM reservations WHERE request_id='P04'").fetchall())
        print("បន្ទាត់:", db.execute(
            "SELECT * FROM lines WHERE request_id='P04'").fetchall())
    finally:
        db.close()

show_atomic_example()
```

```python
def request_rows(db):
    return db.execute("""
        SELECT r.request_id, r.label,
               SUM(l.containers * l.ppc_snapshot) AS planned_portions,
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='collected'),
               EXISTS(SELECT 1 FROM events e WHERE e.request_id=r.request_id
                      AND e.event_type='delivered')
        FROM reservations r JOIN lines l ON l.request_id=r.request_id
        GROUP BY r.request_id, r.label ORDER BY r.request_id
    """).fetchall()


def report(db):
    rows = request_rows(db)
    offered, available = db.execute(
        "SELECT SUM(opening * ppc), SUM(available * ppc) FROM batches").fetchone()
    return {
        "offered": offered,
        "available": available,
        "reserved": sum(row[2] for row in rows),
        "collected": sum(row[2] for row in rows if row[3]),
        "delivered": sum(row[2] for row in rows if row[4]),
        "reserved_only": sum(row[2] for row in rows if not row[3]),
        "collected_not_delivered": sum(row[2] for row in rows if row[3] and not row[4]),
    }


def pending_collections(db):
    return [(request_id, portions, label)
            for request_id, label, portions, collected, delivered in request_rows(db)
            if not collected]


def show_report(db):
    labels = {
        "offered": "បានផ្តល់ដំបូង",
        "available": "នៅអាចកក់បាន",
        "reserved": "បានកក់",
        "collected": "បានកត់ត្រាការប្រមូល",
        "delivered": "បានកត់ត្រាការប្រគល់",
        "reserved_only": "បានកក់ គ្មានកំណត់ត្រាប្រមូល",
        "collected_not_delivered": "បានប្រមូល គ្មានកំណត់ត្រាប្រគល់",
    }
    print("ចំណែកអាហារប៉ាន់ស្មាន")
    for key, value in report(db).items():
        print(labels[key] + ":", value)
    print("ឯកតាផ្ទុកដែលអាចកក់បាន:", db.execute(
        "SELECT batch_id, available FROM batches ORDER BY batch_id").fetchall())
    print("រង់ចាំការប្រមូល៖ លេខសម្គាល់ ចំណែកអាហារ ស្លាក:",
          khmer_rows(pending_collections(db)))


def build_day(requests=REQUESTS, events=EVENTS):
    db = new_db()
    decisions = []
    try:
        for request_id, reviewed, lines in requests:
            before = snapshot(db)
            try:
                label = reserve(db, request_id, reviewed, lines)
            except (ValueError, sqlite3.IntegrityError) as error:
                assert snapshot(db) == before
                decisions.append((request_id, "held" if reviewed is False else "rejected", str(error)))
            else:
                decisions.append((request_id, "accepted", label))
        for request_id, event_type in events:
            record_event(db, request_id, event_type)
        return db, decisions
    except Exception:
        db.close()
        raise


def run_canonical_day():
    db, decisions = build_day()
    try:
        print("លទ្ធផលសំណើ:")
        for request_id, decision, detail in decisions:
            print(request_id, khmer_label(decision), khmer_label(detail))
        show_report(db)
    finally:
        db.close()

run_canonical_day()
```

```python
class FoodRescueTests(unittest.TestCase):
    def setUp(self):
        self.db = new_db()

    def tearDown(self):
        self.db.close()

    def assert_refused_unchanged(self, exception, action):
        before = snapshot(self.db)
        with self.assertRaises(exception):
            action()
        self.assertEqual(snapshot(self.db), before)

    def test_boundary_and_review(self):
        for total, reviewed, expected in (
            (19, True, "standard"), (20, True, "team"),
            (21, True, "team"), (20, False, "hold")):
            self.assertEqual(pickup_label(total, reviewed), expected)

    def test_invalid_planning_values(self):
        for total, reviewed in ((-1, True), (True, True), (2.5, True), (20, "yes")):
            with self.assertRaises(ValueError):
                pickup_label(total, reviewed)

    def test_normal_request(self):
        self.assertEqual(reserve(self.db, *REQUESTS[0]), "team")
        self.assertEqual(report(self.db)["reserved"], 20)
        self.assertEqual(self.db.execute(
            "SELECT SUM(containers) FROM lines").fetchone()[0], 7)
        self.assertEqual(self.db.execute(
            "SELECT available FROM batches ORDER BY batch_id").fetchall(), [(15,), (4,), (8,)])

    def test_unreviewed_unchanged(self):
        self.assert_refused_unchanged(ValueError, lambda: reserve(self.db, *REQUESTS[1]))

    def test_invalid_counts_unchanged(self):
        for count in (0, -1, True, 1.5, "2"):
            self.assert_refused_unchanged(ValueError,
                lambda: reserve(self.db, "BAD", True, [("B01", count)]))

    def test_empty_and_malformed_inputs(self):
        for lines in ([], [("B01",)], ["B01"], [("B01", 1, 2)]):
            self.assert_refused_unchanged(ValueError,
                lambda: reserve(self.db, "BAD", True, lines))
        self.assert_refused_unchanged(ValueError,
            lambda: reserve(self.db, " ", True, [("B01", 1)]))

    def test_unknown_batch_unchanged(self):
        self.assert_refused_unchanged(ValueError,
            lambda: reserve(self.db, "BAD", True, [("X99", 1)]))

    def test_late_shortage_rolls_back(self):
        reserve(self.db, *REQUESTS[0])
        reserve(self.db, *REQUESTS[2])
        self.assert_refused_unchanged(ValueError, lambda: reserve(self.db, *REQUESTS[3]))
        self.assertEqual(self.db.execute(
            "SELECT available FROM batches WHERE batch_id='B01'").fetchone()[0], 15)

    def test_interruption_rolls_back(self):
        self.assert_refused_unchanged(RuntimeError,
            lambda: reserve(self.db, *REQUESTS[0], inject_failure=True))

    def test_duplicate_request_unchanged(self):
        reserve(self.db, *REQUESTS[0])
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: reserve(self.db, *REQUESTS[0]))

    def test_event_order_and_duplicates(self):
        reserve(self.db, *REQUESTS[0])
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P01", "delivered"))
        batches_before = snapshot(self.db)[0]
        record_event(self.db, "P01", "collected")
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: record_event(self.db, "P01", "collected"))
        record_event(self.db, "P01", "delivered")
        self.assertEqual(snapshot(self.db)[0], batches_before)
        self.assertEqual(report(self.db)["delivered"], 20)
        self.assert_refused_unchanged(sqlite3.IntegrityError,
            lambda: record_event(self.db, "P01", "delivered"))

    def test_event_unknown_request_and_type(self):
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P99", "collected"))
        self.assert_refused_unchanged(ValueError,
            lambda: record_event(self.db, "P01", "cancelled"))

    def test_historical_line_snapshot(self):
        reserve(self.db, *REQUESTS[0])
        with self.db:
            self.db.execute("UPDATE batches SET ppc=99 WHERE batch_id='B01'")
        self.assertEqual(report(self.db)["reserved"], 20)
        self.assertEqual(self.db.execute(
            "SELECT ppc_snapshot FROM lines WHERE batch_id='B01'").fetchone()[0], 2)

    def test_empty_report(self):
        self.assertEqual(report(self.db)["reserved"], 0)
        self.assertEqual(report(self.db)["delivered"], 0)
        self.assertEqual(pending_collections(self.db), [])

    def test_canonical_day(self):
        db, decisions = build_day()
        try:
            self.assertEqual([(r[0], r[1]) for r in decisions], [
                ("P01", "accepted"), ("P02", "held"), ("P03", "accepted"),
                ("P04", "rejected"), ("P05", "accepted"), ("P06", "accepted")])
            self.assertEqual(report(db), {"offered": 94, "available": 34, "reserved": 60,
                "collected": 42, "delivered": 29, "reserved_only": 18,
                "collected_not_delivered": 13})
            self.assertEqual(db.execute(
                "SELECT batch_id, available FROM batches ORDER BY batch_id").fetchall(),
                [("B01", 9), ("B02", 2), ("B03", 2)])
            self.assertEqual(db.execute("SELECT SUM(containers) FROM lines").fetchone()[0], 21)
            before = snapshot(db)
            self.assertEqual(pending_collections(db), [("P06", 18, "standard")])
            self.assertEqual(snapshot(db), before)
            result = report(db)
            self.assertEqual(result["offered"], result["available"] + result["reserved"])
            self.assertEqual(result["offered"], result["available"] + result["reserved_only"]
                             + result["collected_not_delivered"] + result["delivered"])
        finally:
            db.close()

    def test_new_collection_preserves_other_measures(self):
        db, decisions = build_day()
        try:
            before = snapshot(db)[0]
            record_event(db, "P06", "collected")
            self.assertEqual(pending_collections(db), [])
            self.assertEqual(report(db)["collected"], 60)
            self.assertEqual(report(db)["delivered"], 29)
            self.assertEqual(snapshot(db)[0], before)
        finally:
            db.close()

    def test_fresh_runs_match(self):
        first, first_decisions = build_day()
        second, second_decisions = build_day()
        try:
            self.assertEqual(snapshot(first), snapshot(second))
            self.assertEqual(first_decisions, second_decisions)
        finally:
            first.close()
            second.close()

    def test_keyboard_input(self):
        answers = iter(["TRY02", "B03", "2", "yes"])
        data = keyboard_request(lambda prompt: next(answers))
        self.assertEqual(data, {"request_id": "TRY02", "reviewed": True,
                               "lines": [("B03", 2)], "events": []})
        for values in (["TRY02", "B03", "many", "yes"],
                       ["TRY02", "B03", "2", "maybe"]):
            answers = iter(values)
            with self.assertRaises(ValueError):
                keyboard_request(lambda prompt: next(answers))


test_log = io.StringIO()
suite = unittest.defaultTestLoader.loadTestsFromTestCase(FoodRescueTests)
test_result = unittest.TextTestRunner(stream=test_log, verbosity=1).run(suite)
if not test_result.wasSuccessful():
    print(test_log.getvalue())
    raise AssertionError("តេស្តបរាជ័យ; ពិនិត្យកំណត់ត្រា")
print("គ្រប់តេស្តបានឆ្លង:", test_result.testsRun)
```
