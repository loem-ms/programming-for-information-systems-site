### The idea: entities, status and workflow

Think of a parcel you order online: *ordered → shipped → delivered*. The parcel is an **entity**: one real thing the system keeps track of. Its **status** says where it is on its journey. The **workflow** is the set of moves that make sense. A parcel is not delivered before it is shipped, and a delivered parcel does not go back to "ordered".

An offer has the same kind of journey:

Offer lifecycle: new, approved, picked up, delivered, or rejected or cancelled

People make these moves. A **coordinator** approves or rejects a new offer and chooses the partner; volunteers pick up and deliver. The program **records** the decision; whether the food is safe stays a human decision. Rejected, cancelled and delivered are **end states**: nothing moves out of them.

### ⭐⭐ 8 · Read a class: the kit's `Offer`

Inside the kit, each offer is an **object** made from the class `Offer`: a blueprint that keeps an offer's data **and** its actions together. You will not write classes, but you need to read them (for example in AI code). Here is part of the kit's `Offer`:

```python
class Offer:
    """One food offer from a donor. Change its status only with move_to()."""

    def __init__(self, offer_id, donor, food, category, portions, hours_left, contact=""):
        self.id = offer_id
        self.donor = donor
        self.contact = contact
        self.food = food
        self.category = category
        self.portions = portions
        self.hours_left = hours_left
        self.status = "new"
        self.partner = ""

    def move_to(self, new_status, is_allowed):
        """Change status only if is_allowed(old, new) is True. Return True or False."""
        if not is_allowed(self.status, new_status):
            return False
        self.status = new_status
        return True
```

`__init__` runs when a new offer is made; `self` means "this offer"; `move_to` is a **method**. 🤔 Read the class and answer:

### ⚖️ Good or risky?

You asked an AI assistant: *"Volunteers need buttons to update an offer, and the coordinator wants a way to fix mistakes."* It wrote **Version A**. **Version B** is a teammate's rewrite. Both run. Read them in the cell below, then run it. Both try clicks on F004: Lotus Cafe's curry, 12 portions, approved for the Old Market Elders' Group.

````python
# Version A · from an AI assistant
def update_status(record, typed_status):
    record["status"] = typed_status.strip().lower()   # whatever the volunteer types

def mark_delivered(record):
    record["status"] = "delivered"                    # one click, from any status

def fix_mistake(record):
    record["status"] = "new"                          # reset: start again


# Version B · from a teammate
def move(record, new_status, is_allowed):
    old_status = record["status"]
    if is_allowed(old_status, new_status):
        record["status"] = new_status
        return "done: " + record["id"] + " " + old_status + " -> " + new_status
    return "refused: " + record["id"] + " cannot move from " + old_status + " to " + new_status


DEMO_MOVES = {
    "new": ["approved", "rejected"],
    "approved": ["picked_up", "cancelled"],
    "picked_up": ["delivered"],
    "delivered": [],
    "rejected": [],
    "cancelled": [],
}


def lifecycle_rule(old_status, new_status):
    return new_status in DEMO_MOVES.get(old_status, [])


curry_a = bridge.sample_records()[3]      # F004, approved
curry_b = bridge.sample_records()[3]
print("Version A")
mark_delivered(curry_a)
print("  1.", curry_a["id"], curry_a["status"])
fix_mistake(curry_a)
print("  2.", curry_a["id"], curry_a["status"])
update_status(curry_a, "Deliverd ")
print("  3.", curry_a["id"], curry_a["status"])
print("Version B")
print("  1.", move(curry_b, "delivered", lifecycle_rule))
print("  2.", move(curry_b, "new", lifecycle_rule))
print("  3.", move(curry_b, "picked_up", lifecycle_rule))
print("  4.", move(curry_b, "delivered", lifecycle_rule))
````
