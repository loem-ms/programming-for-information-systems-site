### The idea: memory forgets, records must not

Think of a small shop. During the day the owner writes sales on a whiteboard. At night the board is wiped clean. A careful owner also keeps a **sales book**: every sale is a new line in ink, with the time. A mistake is fixed with a new line ("refund, 2 pm"), never by tearing out a page.

A program's memory is the whiteboard: your lists and dictionaries disappear when the program stops or Colab restarts. An information system needs two more things:

- **Persistence:** the records are saved in a file, so they are still there tomorrow.
- **An audit history:** a list of events that says **who did what, and when**. We only ever add to it (it is **append-only**). Then, when a donor asks *"what happened to my bread rolls?"*, someone can answer.

### 6 · JSON Lines: a history you only add to

A history has a different shape from a list of records. We never rewrite it; we **add one line per event**. Each line is one small JSON object. This is called **JSON Lines**, and the files end in `.jsonl`.

````python
def log_event(path, event):
    with open(path, "a", encoding="utf-8") as file:     # "a": add to the end, never empty
        file.write(json.dumps(event) + "\n")             # one event = one line


log_event("phone_history.jsonl", {"time": "09:00", "event": "top up", "usd": 5})
log_event("phone_history.jsonl", {"time": "09:30", "event": "bought data", "usd": 3})

with open("phone_history.jsonl", encoding="utf-8") as file:
    for line in file:
        event = json.loads(line)
        print(event["time"], event["event"], event["usd"])
````

<details><summary>Instructor's view</summary>

1. **It writes straight over the only copy.** `open("offers.json", "w")` empties the file the moment it opens. If Colab disconnects or the power cuts halfway, the file is half-written and cannot be loaded at all (the experiment shows it). All four offers are lost, not only today's change: F004's 12 portions of curry, which the Old Market Elders' Group is still waiting for, and F001's bread on its way to Hope Children's Centre. Version B writes a complete temporary file first; if it stops, the old `offers.json` is still whole. At worst one change is lost, and the history shows what it was.
2. **It rewrites the history and throws old events away.** `"w"` plus `events[-10:]` keeps only the last 10 events. The sample day has 16, so the four "added" events and two refusals are gone today. After one more busy day, the reason F003 was rejected ("no cold storage for fruit today") is gone too. When Green Market Stall asks why its bananas were rejected, nobody can answer. A history must be append-only: Version B opens it with `"a"`.
3. **A rewritten history cannot be trusted.** Code that is allowed to rewrite the file can also change an old line, so nobody can show what really happened at 10:00.
4. Small point: no `encoding="utf-8"`. `json.dump` turns Khmer letters into codes such as `\u1780` by default, so this file survives, but the habit is risky: plain text in Khmer written without it may break on some computers.

**How a test catches it:** save twice and check that the first event is still in the history (the kit's `append_event` check does exactly this); stop a save halfway, as in the experiment, and check that the old file still loads.

**Is Version B perfect?** No. If someone calls it twice with the same `new_events`, the history gets the events twice. Good code still needs checks. AI code often "simplifies" by dropping exactly the lines that protect data on a bad day. Ask: *what happens on a bad day?*

### 🧭 Your decision · what does the history keep?

Every event in the history is a promise: *"we can explain this later."* That is **accountability**. But every event also has a cost: a longer history is harder to read, and the one event that matters is easier to miss (**noise**).

| `"moved"` | a status changes | old → new status, partner, note | nobody can say who got the food, or why F003 was rejected |
