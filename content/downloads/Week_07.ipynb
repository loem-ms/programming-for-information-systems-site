### 1 · Memory and disk: a file and its path

📖 **What it is.** **Memory** holds the values of a running program: every name, list and dictionary you make. The **disk** (a hard disk, an SSD, Google Drive) holds **files**. A file is a named piece of data on disk, often text. A **path** says where a file is: the folders it is in, then its name, joined by `/`. `my_foodbridge/demo_day/offers.json` is the file `offers.json` in the folder `demo_day`, inside your folder `my_foodbridge`. A path with no folder, such as `shopping.txt`, means "in the current folder". In Colab that is the runtime's temporary workspace (`/content`), which disappears when the runtime is deleted. Your notebook itself is saved in Drive, and so is your `my_foodbridge/` folder.

**Why it exists.** Memory is fast, but it forgets: when the program stops or the runtime restarts, memory is empty. A file on disk stays until someone deletes it, so anything you need tomorrow must be written to disk.

| | memory | disk |
|---|---|---|
| holds | the names, lists and dictionaries of a running program | files in folders |
| after a restart | empty | still there |
| in Python | `offers = [...]` | `open("offers.json", "w", encoding="utf-8")` |

Memory and disk: a restart empties the variables in memory; the files on the disk stay

**How Python runs it.** `open(path)` gives you a **file object**: a connection to the file that you read or write through. Without a mode, `open` only reads; `open(path, "w", encoding="utf-8")` opens it for **writing** (concept 2 says more), and `encoding="utf-8"` lets the file hold any language, including Khmer. `file.write(text)` puts exactly that text into the file and adds nothing: a line ends only where you write `"\n"`, the line break. `with open(...) as file:` closes the connection at the end of the block, **even when something goes wrong in the middle**. Closing matters, because it is when Python finishes writing the last text to disk. A `for` loop over a file gives you one line at a time, and each line still ends with `"\n"`.

▶ [Animation: Memory forgets, files stay: `json.dump`, `"w"` and `"a"`](https://loem-ms.github.io/programming-for-information-systems-site/assets/anim/anim_memory_disk.html) (course website)

**Where you meet it.** Your phone keeps your photos when it switches off, because they are files. A game "saves" your progress to a file. A web form that forgets what you typed after a reload kept it only in memory.

### 2 · `"w"` empties the file, `"a"` adds to the end

📖 **What it is.** The second argument of `open` is the **mode**: what you want to do with the file.

| mode | the file does not exist | the file exists | use it for |
|---|---|---|---|
| no mode (or `"r"`) | `FileNotFoundError` | read it from the start | reading |
| `"w"` (write) | creates it | **empties it** the moment it opens | a file you replace as a whole: the records now |
| `"a"` (append) | creates it | keeps everything, adds to the end | a file that only grows: a history |

**Why it exists.** Programs do two different jobs with files. Sometimes the whole content is replaced by a newer version: today's list of offers. Sometimes one new line is added after everything before it: today's events. The mode tells Python which job you mean.

**How Python runs it.** With `"w"`, the file is emptied **when `open` runs**, before you write anything, and Python gives no warning. With `"a"`, every `write` goes to the end, and earlier lines are never touched. Opening a file in the wrong mode is one of the most common ways programs lose data.

File modes: &quot;w&quot; empties the file first, then writes; &quot;a&quot; keeps every line and adds to the end

**Where you meet it.** A bank statement only grows: each payment is a new line (`"a"`). Your phone's settings file is written again as a whole each time you change a setting (`"w"`).

👀 **Example.** 🤔 **Predict** what the cell prints. The file starts with *maths* and *English*. (`file.read()` gives the whole file as one text.)

````python
with open("homework.txt", "w", encoding="utf-8") as file:
    file.write("maths\nEnglish\n")
with open("homework.txt", "w", encoding="utf-8") as file:     # "w" again
    file.write("history\n")
with open("homework.txt", "a", encoding="utf-8") as file:     # "a"
    file.write("Khmer\n")
with open("homework.txt", encoding="utf-8") as file:
    print(file.read())
````

### 3 · JSON: records as text, and "is the file there?"

📖 **What it is.** A **text file** holds only text, not Python lists and dictionaries. Your records are lists and dictionaries, so they must become text first. **JSON** is a standard way to write them as text. It looks almost like Python:

| Python | JSON text |
|---|---|
| `{"id": "F001", "portions": 30}` | `{"id": "F001", "portions": 30}` |
| `True`, `False`, `None` | `true`, `false`, `null` |
| a list of dictionaries `[{...}, {...}]` | a list of objects `[{...}, {...}]` |

**Why it exists.** In Week 5 every value from the CSV file arrived as text, so you needed `int()`. JSON keeps numbers as numbers, and it can hold lists inside dictionaries. People can read it, and almost every programming language can load it. (This notebook is JSON too: an `.ipynb` file is JSON text.)

**How Python runs it.** `json.dump(data, file)` turns your lists and dictionaries into JSON text and writes it into the open file. `json.load(file)` reads the text and builds **new** lists and dictionaries in memory. `indent=2` puts each item on its own line, so a person can read the file. With an `s` at the end, `json.dumps(data)` and `json.loads(text)` do the same with a **string** instead of a file. You need them when you write one line at a time.

JSON: json.dump writes a list of dictionaries as text in offers.json; json.load reads it back

Before you read a file, ask whether it is there: `os.path.exists(path)` answers `True` or `False`. On the first day of a new system there is no file yet, and that is normal, not an error. The pattern you will use all week: **if the file exists, read it; if not, start with an empty list.**

**Where you meet it.** When a weather app asks a server for today's forecast, the answer comes back as JSON text.

👀 **Example.** 🤔 **Predict** the two printed values: is `loaded[1]["students"] + 1` a number, or an error (text + number)? Does `diary.json` exist?

### 4 · JSON Lines: a history you only add to

📖 **What it is.** A history is a list of **events**, and it has a different shape from the records. Each event is one small JSON object on **its own line**. This is called **JSON Lines**, and the files end in `.jsonl`.

**Why it exists.** The records say what is true **now**, so each save replaces the whole file. The history says **what happened**, so it only grows: one new line per event, and earlier lines are never rewritten. A mistake is fixed with a new event, like a new line in the sales book.

| | `offers.json` (the records) | `history.jsonl` (the history) |
|---|---|---|
| says | what is true now | what happened, in order |
| a save | replaces the whole file (`"w"`) | adds one line at the end (`"a"`) |
| inside | one JSON list | one JSON object per line |
| in Python | `json.dump` / `json.load` | `json.dumps(event) + "\n"` / `json.loads(line)` |

**How Python runs it.** To add an event, open the file with `"a"` and write `json.dumps(event) + "\n"`: one dictionary becomes one line of text. To read the history, loop over the file and turn each line back into a dictionary with `json.loads(line)`. Adding a line never touches the lines before it, so a crash can damage at most the last line, never the whole history.

JSON Lines: one event per line in history.jsonl; append_event adds a line, read_events reads them all

**Where you meet it.** A bank statement, the message history of a chat app and the log of a web server are all append-only histories.

👀 **Example.** 🤔 **Predict, then try:** run the cell **twice**. How many lines does the history have then? Is that right for a history? Would it be right for a list of records?

### Part B · Under the hood (≈ 45 min)

#### 🧑‍🏫 How `storage.py` works

Your file has four functions, and the **system** calls them. This is the data flow on the sample day:

```text
 the system (memory)                              my_foodbridge/demo_day/ (disk)
 an offer is added or moved ── save_records ──▶  offers.json.tmp ── os.replace ──▶ offers.json   (replaced)
 an event of a kind in LOG_EVENTS ── append_event ───────────────────────────────▶ history.jsonl (+1 line)

 after a restart:   load_records ◀── offers.json        read_events ◀── history.jsonl
```

**The safe save.** What happens if the power goes off while `offers.json` is being written? With `open("offers.json", "w")`, the file is emptied first. For a moment the **only** copy of every record is half-written. If the power cuts then, the next morning `json.load` meets broken text (the `JSONDecodeError` from Session 1) and the system cannot start. Every offer is lost, not only the last change.

The safe save never writes into the real file:

1. Write the complete new list into a **temporary file**, `offers.json.tmp`, in the same folder.
2. Swap it in with `os.replace("offers.json.tmp", "offers.json")`.

`os.replace` makes the name `offers.json` point to the new file **in one step**. There is no moment when `offers.json` is half old and half new. So a power cut can hit only the temporary file. If it stops during step 1, `offers.json` is still the last complete file: at worst the last change is lost, never everything. The temporary file stays in the same folder because this one-step swap works only on one disk.
